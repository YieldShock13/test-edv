```python
# ============================================================
# CELL 1 — TOTAL-RETURN DATASET
# Maximum common history
# ============================================================

!pip -q install yfinance

import yfinance as yf
import pandas as pd
import numpy as np

START = "1990-01-01"
END   = "2026-09-25"

# ============================================================
# 1. PORTFOLIO — ALBI EXCLUDED
# ============================================================

notionals = {
    "ALSI":        125_000,
    "SP500":       150_000,
    "MSCI_EM":      65_000,
    "Berkshire":    56_000,
    "Gold":          44_000,
    "AGG":           61_000,
    "Exxaro":        22_000,
    "Europe":       125_000
}

TOTAL = sum(notionals.values())

print(f"Portfolio value excluding ALBI: R{TOTAL:,.0f}")


# ============================================================
# 2. DATA SOURCES
#
# ALSI      = J203 price index + assumed 3.28% annual dividend
# SP500     = SPY adjusted close
# MSCI EM   = EEM adjusted close
# Berkshire = BRK.B adjusted close
# Gold      = GLD adjusted close
# AGG       = AGG adjusted close
# Exxaro    = EXX.JO adjusted close
# Europe    = VGK adjusted close
# ============================================================

ticker_map = {
    "ALSI":       "^J203.JO",
    "SP500":      "SPY",
    "MSCI_EM":    "EEM",
    "Berkshire":  "BRK-B",
    "Gold":       "GLD",
    "AGG":        "AGG",
    "Exxaro":     "EXX.JO",
    "Europe":     "VGK"
}


# ============================================================
# 3. DOWNLOAD
# ============================================================

raw = yf.download(
    list(ticker_map.values()),
    start=START,
    end=END,
    auto_adjust=False,
    actions=True,
    progress=False
)

prices = raw["Adj Close"].copy()

reverse_map = {
    ticker: asset
    for asset, ticker in ticker_map.items()
}

prices = prices.rename(columns=reverse_map)
prices = prices[list(ticker_map.keys())]


# ============================================================
# 4. FIX VERIFIED ALSI BAD PRINT
# ============================================================

bad_date = pd.Timestamp("2025-07-23")

if bad_date in prices.index:

    prev_date = pd.Timestamp("2025-07-22")
    next_date = pd.Timestamp("2025-07-24")

    old_value = prices.loc[bad_date, "ALSI"]

    corrected_value = (
        prices.loc[prev_date, "ALSI"] +
        prices.loc[next_date, "ALSI"]
    ) / 2

    prices.loc[bad_date, "ALSI"] = corrected_value

    print("\nALSI bad print corrected:")
    print(f"Old:       {old_value:,.2f}")
    print(f"Corrected: {corrected_value:,.2f}")


# ============================================================
# 5. CONSTRUCT ESTIMATED ALSI TOTAL-RETURN INDEX
#
# Assumption supplied:
# Constant dividend yield = 3.28% p.a.
#
# Convert annual yield to equivalent daily compounded yield:
# (1 + daily_yield)^252 = 1.0328
#
# Total daily return:
# (1 + price return) × (1 + dividend return) - 1
# ============================================================

ALSI_DIVIDEND_YIELD = 0.0328

alsi_price = prices["ALSI"].dropna()

alsi_price_return = (
    alsi_price
    .pct_change(fill_method=None)
)

daily_dividend_return = (
    (1 + ALSI_DIVIDEND_YIELD) ** (1 / 252)
    - 1
)

alsi_total_return = (
    (1 + alsi_price_return) *
    (1 + daily_dividend_return)
    - 1
)

# Construct synthetic total-return index.
# Base value = actual ALSI level at inception.

alsi_tri = pd.Series(
    index=alsi_price.index,
    dtype=float
)

alsi_tri.iloc[0] = alsi_price.iloc[0]

alsi_tri.iloc[1:] = (
    alsi_tri.iloc[0] *
    (1 + alsi_total_return.iloc[1:]).cumprod()
)

# Replace ALSI price index with estimated TRI
prices.loc[alsi_tri.index, "ALSI"] = alsi_tri


# ============================================================
# 6. DATA COVERAGE
# ============================================================

coverage = []

for asset in prices.columns:

    s = prices[asset].dropna()

    coverage.append({
        "Asset": asset,
        "First Date": s.index.min().date(),
        "Last Date": s.index.max().date(),
        "Observations": len(s)
    })

coverage = pd.DataFrame(coverage)

print("\n" + "=" * 75)
print("TOTAL-RETURN DATA COVERAGE")
print("=" * 75)

display(coverage)


# ============================================================
# 7. RETURN SANITY CHECK
# ============================================================

returns = prices.pct_change(fill_method=None)

sanity = []

for asset in returns.columns:

    s = returns[asset].dropna()

    max_abs_date = s.abs().idxmax()

    sanity.append({
        "Asset": asset,
        "N": len(s),
        "Daily Vol": s.std(),
        "Worst Day": s.min(),
        "Best Day": s.max(),
        "Largest Move Date": max_abs_date.date(),
        "Largest Abs Move": s.loc[max_abs_date]
    })

sanity = pd.DataFrame(sanity)

print("\n" + "=" * 75)
print("RETURN SANITY CHECK")
print("=" * 75)

display(
    sanity.style.format({
        "Daily Vol": "{:.2%}",
        "Worst Day": "{:.2%}",
        "Best Day": "{:.2%}",
        "Largest Abs Move": "{:.2%}"
    })
)


# ============================================================
# 8. MAXIMUM COMMON HISTORY
# ============================================================

first_dates = {
    asset: prices[asset].first_valid_index()
    for asset in prices.columns
}

last_dates = {
    asset: prices[asset].last_valid_index()
    for asset in prices.columns
}

common_start = max(first_dates.values())
common_end   = min(last_dates.values())

print("\n" + "=" * 75)
print("MAXIMUM COMMON TOTAL-RETURN SAMPLE")
print("=" * 75)

print("Start:", common_start.date())
print("End:  ", common_end.date())

print("\nLimiting inception asset(s):")

for asset, dt in first_dates.items():

    if dt == common_start:
        print(f"{asset}: {dt.date()}")


# ============================================================
# 9. CONFIRM ALSI ASSUMPTION
# ============================================================

print("\n" + "=" * 75)
print("ALSI TOTAL-RETURN ASSUMPTION")
print("=" * 75)

print("ALSI source: Yahoo ^J203.JO price index")
print("Assumed annual dividend yield: 3.28%")
print("Dividend treatment: continuously reinvested via daily compounding")
print("ALSI series is ESTIMATED total return, not official J203T.")

print("\nCell 1 complete.")
```

```python
# ============================================================
# CELL 2 — FINAL TOTAL-RETURN PORTFOLIO ANALYSIS
# Buy-and-hold | No rebalancing
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

TRADING_DAYS = 252

# Adjustable constant risk-free assumption
RF = 0.05


# ============================================================
# 1. COMMON PORTFOLIO SAMPLE
# ============================================================

portfolio_prices = prices.loc[
    common_start:common_end,
    list(notionals.keys())
].copy()

# Common business-day valuation calendar
calendar = pd.date_range(
    start=common_start,
    end=common_end,
    freq="B"
)

portfolio_prices = portfolio_prices.reindex(calendar)

# Carry previous value when one market is closed
portfolio_prices = portfolio_prices.ffill()

# Remove any initial incomplete rows
portfolio_prices = portfolio_prices.dropna(how="any")


# ============================================================
# 2. BUY-AND-HOLD PORTFOLIO
# NO REBALANCING
# ============================================================

initial_capital = pd.Series(
    notionals,
    dtype=float
)

growth = portfolio_prices.div(
    portfolio_prices.iloc[0]
)

asset_values = growth.mul(
    initial_capital,
    axis=1
)

portfolio_value = asset_values.sum(axis=1)

portfolio_returns = (
    portfolio_value
    .pct_change(fill_method=None)
    .dropna()
)


# ============================================================
# 3. PORTFOLIO RETURN
# ============================================================

start_value = portfolio_value.iloc[0]
end_value   = portfolio_value.iloc[-1]

total_return = (
    end_value / start_value
    - 1
)

years = (
    portfolio_value.index[-1]
    - portfolio_value.index[0]
).days / 365.25

cagr = (
    (end_value / start_value) ** (1 / years)
    - 1
)


# ============================================================
# 4. VOLATILITY + SHARPE
# ============================================================

annual_vol = (
    portfolio_returns.std()
    * np.sqrt(TRADING_DAYS)
)

sharpe_zero = (
    cagr / annual_vol
)

sharpe_rf = (
    (cagr - RF) / annual_vol
)


# ============================================================
# 5. HISTORICAL VaR
# Positive number = loss threshold
# ============================================================

var95 = -portfolio_returns.quantile(0.05)
var99 = -portfolio_returns.quantile(0.01)

var95_zar = var95 * end_value
var99_zar = var99 * end_value


# ============================================================
# 6. MAXIMUM DRAWDOWN
# ============================================================

running_peak = portfolio_value.cummax()

drawdown = (
    portfolio_value / running_peak
    - 1
)

trough_date = drawdown.idxmin()

max_drawdown = drawdown.loc[
    trough_date
]

peak_date = (
    portfolio_value
    .loc[:trough_date]
    .idxmax()
)

peak_value = portfolio_value.loc[
    peak_date
]


# ============================================================
# 7. RECOVERY
# ============================================================

after_trough = portfolio_value.loc[
    trough_date:
]

recovery_candidates = after_trough[
    after_trough >= peak_value
]

if len(recovery_candidates) > 0:

    recovery_date = (
        recovery_candidates.index[0]
    )

    peak_to_recovery = (
        recovery_date - peak_date
    ).days

    trough_to_recovery = (
        recovery_date - trough_date
    ).days

else:

    recovery_date = None
    peak_to_recovery = np.nan
    trough_to_recovery = np.nan


# ============================================================
# 8. ALSI TOTAL-RETURN BENCHMARK
#
# This is the SAME synthetic ALSI total-return series
# constructed in Cell 1 using:
#
# J203 price return + fixed 3.28% annual dividend yield
# ============================================================

alsi = portfolio_prices["ALSI"]

alsi_returns = (
    alsi
    .pct_change(fill_method=None)
    .dropna()
)

alsi_start = alsi.iloc[0]
alsi_end   = alsi.iloc[-1]

alsi_total_return = (
    alsi_end / alsi_start
    - 1
)

alsi_years = (
    alsi.index[-1]
    - alsi.index[0]
).days / 365.25

alsi_cagr = (
    (alsi_end / alsi_start)
    ** (1 / alsi_years)
    - 1
)

alsi_vol = (
    alsi_returns.std()
    * np.sqrt(TRADING_DAYS)
)

alsi_sharpe_zero = (
    alsi_cagr / alsi_vol
)

alsi_sharpe_rf = (
    (alsi_cagr - RF) / alsi_vol
)


# ============================================================
# 9. BETA + ALPHA VS ALSI
# ============================================================

regression = pd.concat(
    [
        portfolio_returns.rename("Portfolio"),
        alsi_returns.rename("ALSI")
    ],
    axis=1
).dropna()

beta = (
    regression["Portfolio"]
    .cov(regression["ALSI"])
    /
    regression["ALSI"].var()
)

# Daily regression intercept
alpha_daily = (
    regression["Portfolio"].mean()
    -
    beta * regression["ALSI"].mean()
)

# Annualised arithmetic alpha
alpha_annual = (
    alpha_daily * TRADING_DAYS
)


# ============================================================
# 10. ENDING WEIGHTS
# ============================================================

weights_over_time = asset_values.div(
    portfolio_value,
    axis=0
)

ending_weights = (
    weights_over_time.iloc[-1]
)


# ============================================================
# 11. SUMMARY TABLE
# ============================================================

summary = pd.DataFrame({

    "Metric": [

        "Start Date",
        "End Date",

        "Initial Portfolio Value",
        "Ending Portfolio Value",

        "Portfolio Total Return",
        "Portfolio CAGR",

        "Portfolio Annualised Vol",

        "Portfolio Sharpe (RF = 0%)",
        f"Portfolio Sharpe (RF = {RF:.0%})",

        "95% Historical VaR (1-day)",
        "95% VaR at Ending Value",

        "99% Historical VaR (1-day)",
        "99% VaR at Ending Value",

        "Maximum Drawdown",

        "Peak Date",
        "Trough Date",
        "Recovery Date",

        "Peak → Recovery",
        "Trough → Recovery",

        "Beta vs ALSI",
        "Annualised Alpha vs ALSI",

        "ALSI Total Return",
        "ALSI CAGR",
        "ALSI Annualised Vol",

        "ALSI Sharpe (RF = 0%)",
        f"ALSI Sharpe (RF = {RF:.0%})"
    ],

    "Value": [

        portfolio_value.index[0].date(),
        portfolio_value.index[-1].date(),

        f"R{start_value:,.0f}",
        f"R{end_value:,.0f}",

        f"{total_return:.2%}",
        f"{cagr:.2%}",

        f"{annual_vol:.2%}",

        f"{sharpe_zero:.3f}",
        f"{sharpe_rf:.3f}",

        f"{var95:.2%}",
        f"R{var95_zar:,.0f}",

        f"{var99:.2%}",
        f"R{var99_zar:,.0f}",

        f"{max_drawdown:.2%}",

        peak_date.date(),
        trough_date.date(),

        (
            recovery_date.date()
            if recovery_date is not None
            else "Not recovered"
        ),

        (
            f"{peak_to_recovery:.0f} days"
            if recovery_date is not None
            else "Not recovered"
        ),

        (
            f"{trough_to_recovery:.0f} days"
            if recovery_date is not None
            else "Not recovered"
        ),

        f"{beta:.3f}",
        f"{alpha_annual:.2%}",

        f"{alsi_total_return:.2%}",
        f"{alsi_cagr:.2%}",
        f"{alsi_vol:.2%}",

        f"{alsi_sharpe_zero:.3f}",
        f"{alsi_sharpe_rf:.3f}"
    ]
})


print("=" * 72)
print("TOTAL-RETURN PORTFOLIO ANALYSIS")
print("=" * 72)

display(summary)


# ============================================================
# 12. ENDING WEIGHTS
# ============================================================

print("\nENDING WEIGHTS — NO REBALANCING")

display(
    ending_weights
    .to_frame("Ending Weight")
    .style.format("{:.2%}")
)


# ============================================================
# 13. PORTFOLIO VALUE CHART
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    portfolio_value.index,
    portfolio_value,
    linewidth=2
)

plt.title(
    "Buy-and-Hold Portfolio Value — Total Return"
)

plt.xlabel("Date")
plt.ylabel("Portfolio Value")

plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()


# ============================================================
# 14. DRAWDOWN CHART
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    drawdown.index,
    drawdown * 100,
    linewidth=1.8
)

plt.axhline(
    0,
    linewidth=1
)

plt.title(
    "Portfolio Drawdown"
)

plt.xlabel("Date")
plt.ylabel("Drawdown (%)")

plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()


# ============================================================
# 15. PORTFOLIO vs ALSI WEALTH INDEX
# ============================================================

portfolio_index = (
    portfolio_value /
    portfolio_value.iloc[0]
) * 100

alsi_index = (
    alsi /
    alsi.iloc[0]
) * 100

plt.figure(figsize=(12, 6))

plt.plot(
    portfolio_index.index,
    portfolio_index,
    label="Portfolio",
    linewidth=2
)

plt.plot(
    alsi_index.index,
    alsi_index,
    label="ALSI Estimated Total Return",
    linewidth=2
)

plt.title(
    "Portfolio vs ALSI — Growth of 100"
)

plt.xlabel("Date")
plt.ylabel("Growth of 100")

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()


# ============================================================
# 16. FINAL SANITY CHECK
# ============================================================

print("\n" + "=" * 72)
print("SANITY CHECK")
print("=" * 72)

print(
    f"Worst portfolio day: "
    f"{portfolio_returns.min():.2%}"
)

print(
    f"Best portfolio day:  "
    f"{portfolio_returns.max():.2%}"
)

print(
    f"Portfolio/ALSI correlation: "
    f"{regression.corr().iloc[0,1]:.3f}"
)

print(
    f"Portfolio daily observations: "
    f"{len(portfolio_returns):,}"
)

print(
    f"ALSI daily observations: "
    f"{len(alsi_returns):,}"
)
```

```python
# ============================================================
# CELL 3 — ZAR-ADJUSTED PORTFOLIO
#
# USD assets → USD/ZAR
# Europe     → EUR/ZAR
# ALSI/Exxaro already ZAR
# ============================================================

import yfinance as yf
import pandas as pd
import numpy as np

# ============================================================
# 1. DOWNLOAD FX SERIES
# ============================================================

fx_tickers = {
    "USDZAR": "ZAR=X",
    "EURZAR": "EURZAR=X"
}

fx_raw = yf.download(
    list(fx_tickers.values()),
    start=common_start.strftime("%Y-%m-%d"),
    end="2026-09-25",
    auto_adjust=False,
    progress=False
)

fx_close = fx_raw["Close"].copy()

fx_close = fx_close.rename(
    columns={
        "ZAR=X": "USDZAR",
        "EURZAR=X": "EURZAR"
    }
)

fx_close = fx_close[
    ["USDZAR", "EURZAR"]
]


# ============================================================
# 2. CLEAN OBVIOUS FX BAD PRINTS
#
# Yahoo previously produced isolated USD/ZAR errors.
# Flag moves >8%, remove the LEVEL, then interpolate.
# Same rule applied to EUR/ZAR.
# ============================================================

fx_clean = fx_close.copy()

for currency in fx_clean.columns:

    r = fx_clean[currency].pct_change(
        fill_method=None
    )

    flagged = r[
        r.abs() > 0.08
    ].index

    print(f"\n{currency} FLAGGED OBSERVATIONS")

    if len(flagged) == 0:

        print("None")

    else:

        for d in flagged:

            print(
                d.date(),
                f"Level = {fx_clean.loc[d, currency]:.4f}",
                f"Return = {r.loc[d]:.2%}"
            )

        # Remove suspicious LEVELS
        fx_clean.loc[
            flagged,
            currency
        ] = np.nan


# Interpolate removed observations
fx_clean = (
    fx_clean
    .interpolate(method="time")
)


# ============================================================
# 3. ALIGN FX WITH PORTFOLIO CALENDAR
# ============================================================

fx_clean = (
    fx_clean
    .reindex(portfolio_prices.index)
    .ffill()
    .bfill()
)


# ============================================================
# 4. FX SANITY CHECK
# ============================================================

print("\n" + "=" * 70)
print("FX SERIES")
print("=" * 70)

for currency in fx_clean.columns:

    start_fx = fx_clean[currency].iloc[0]
    end_fx   = fx_clean[currency].iloc[-1]

    fx_growth = (
        end_fx / start_fx
        - 1
    )

    fx_cagr = (
        (end_fx / start_fx)
        ** (1 / years)
        - 1
    )

    print(
        f"\n{currency}"
        f"\nStart:        {start_fx:.4f}"
        f"\nEnd:          {end_fx:.4f}"
        f"\nTotal change: {fx_growth:.2%}"
        f"\nCAGR:         {fx_cagr:.2%}"
    )


# ============================================================
# 5. CONVERT FOREIGN TOTAL-RETURN SERIES INTO ZAR
# ============================================================

prices_zar = portfolio_prices.copy()

usd_assets = [
    "SP500",
    "MSCI_EM",
    "Berkshire",
    "Gold",
    "AGG"
]

eur_assets = [
    "Europe"
]

# USD-denominated total-return proxies
for asset in usd_assets:

    prices_zar[asset] = (
        portfolio_prices[asset]
        * fx_clean["USDZAR"]
    )

# European equity sleeve:
# treat underlying return as EUR exposure
for asset in eur_assets:

    prices_zar[asset] = (
        portfolio_prices[asset]
        * fx_clean["EURZAR"]
    )

# ALSI and Exxaro remain untouched because they are ZAR series.


# ============================================================
# 6. REBUILD BUY-AND-HOLD PORTFOLIO
# NO REBALANCING
# ============================================================

growth_zar = prices_zar.div(
    prices_zar.iloc[0]
)

asset_values_zar = growth_zar.mul(
    pd.Series(notionals, dtype=float),
    axis=1
)

portfolio_value_zar = (
    asset_values_zar.sum(axis=1)
)

portfolio_returns_zar = (
    portfolio_value_zar
    .pct_change(fill_method=None)
    .dropna()
)


# ============================================================
# 7. ZAR PORTFOLIO RETURN
# ============================================================

start_value_zar = (
    portfolio_value_zar.iloc[0]
)

end_value_zar = (
    portfolio_value_zar.iloc[-1]
)

total_return_zar = (
    end_value_zar /
    start_value_zar
    - 1
)

years_zar = (
    portfolio_value_zar.index[-1]
    - portfolio_value_zar.index[0]
).days / 365.25

cagr_zar = (
    (end_value_zar / start_value_zar)
    ** (1 / years_zar)
    - 1
)


# ============================================================
# 8. ASSET-BY-ASSET FX EFFECT
# ============================================================

comparison = []

for asset in notionals.keys():

    native_growth = (
        portfolio_prices[asset].iloc[-1]
        / portfolio_prices[asset].iloc[0]
    )

    zar_growth = (
        prices_zar[asset].iloc[-1]
        / prices_zar[asset].iloc[0]
    )

    native_cagr = (
        native_growth
        ** (1 / years_zar)
        - 1
    )

    zar_asset_cagr = (
        zar_growth
        ** (1 / years_zar)
        - 1
    )

    if asset in usd_assets:
        currency = "USD/ZAR"

    elif asset in eur_assets:
        currency = "EUR/ZAR"

    else:
        currency = "ZAR"

    comparison.append({

        "Asset": asset,
        "FX": currency,
        "Native CAGR": native_cagr,
        "ZAR CAGR": zar_asset_cagr,
        "FX Impact p.a.": (
            zar_asset_cagr
            - native_cagr
        )
    })

comparison = pd.DataFrame(comparison)


# ============================================================
# 9. PORTFOLIO OUTPUT
# ============================================================

print("\n" + "=" * 70)
print("ZAR-ADJUSTED PORTFOLIO")
print("=" * 70)

print(
    f"\nPeriod: "
    f"{portfolio_value_zar.index[0].date()} "
    f"→ {portfolio_value_zar.index[-1].date()}"
)

print(
    f"\nInitial value: "
    f"R{start_value_zar:,.0f}"
)

print(
    f"Ending value:  "
    f"R{end_value_zar:,.0f}"
)

print(
    f"\nTotal return: "
    f"{total_return_zar:.2%}"
)

print(
    f"ZAR CAGR:     "
    f"{cagr_zar:.2%}"
)

print(
    f"\nPrevious unconverted CAGR: "
    f"{cagr:.2%}"
)

print(
    f"FX effect on portfolio CAGR: "
    f"{cagr_zar - cagr:+.2%} p.a."
)


# ============================================================
# 10. ASSET TABLE
# ============================================================

print("\n" + "=" * 70)
print("ASSET RETURNS — NATIVE vs ZAR")
print("=" * 70)

display(
    comparison.style.format({
        "Native CAGR": "{:.2%}",
        "ZAR CAGR": "{:.2%}",
        "FX Impact p.a.": "{:+.2%}"
    })
)


# ============================================================
# 11. ENDING WEIGHTS AFTER FX
# ============================================================

weights_zar = asset_values_zar.div(
    portfolio_value_zar,
    axis=0
)

print("\nENDING WEIGHTS — ZAR BASIS")

display(
    weights_zar.iloc[-1]
    .to_frame("Ending Weight")
    .style.format("{:.2%}")
)
```

```python
# ============================================================
# S&P 500 — MOMENTUM + ENTROPY / DISORDER REGIME
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Use existing S&P 500 price series
spx = common["S&P 500"].dropna().copy()

# Daily log returns
r = np.log(spx / spx.shift(1)).dropna()


# ============================================================
# 1. MOMENTUM
#
# Conventional 12M momentum excluding most recent month
# Approximation:
# price 21 trading days ago / price 252 days ago - 1
# ============================================================

momentum_12_1 = (
    spx.shift(21) /
    spx.shift(252)
    - 1
)


# ============================================================
# 2. MOMENTUM Z-SCORE — 365 CALENDAR DAYS
# ============================================================

mom_mean = momentum_12_1.rolling(
    "365D",
    min_periods=200
).mean()

mom_std = momentum_12_1.rolling(
    "365D",
    min_periods=200
).std(ddof=1)

momentum_z = (
    momentum_12_1 - mom_mean
) / mom_std


# ============================================================
# 3. SHANNON ENTROPY
#
# Calculate entropy from rolling daily-return distribution.
# 20 fixed bins per rolling window.
#
# Normalize entropy by maximum possible entropy:
# H / log(number of bins)
#
# Range approximately 0 → 1
# Higher = more disorder / uncertainty
# ============================================================

def normalized_entropy(x, bins=20):

    x = np.asarray(x)
    x = x[np.isfinite(x)]

    if len(x) < 50:
        return np.nan

    counts, _ = np.histogram(
        x,
        bins=bins
    )

    p = counts[counts > 0] / counts.sum()

    H = -np.sum(
        p * np.log(p)
    )

    H_max = np.log(bins)

    return H / H_max


entropy_365 = r.rolling(
    "365D",
    min_periods=200
).apply(
    normalized_entropy,
    raw=True
)


# ============================================================
# 4. ENTROPY Z-SCORE
#
# Is current disorder unusually high relative to the
# preceding 365 days?
# ============================================================

entropy_mean = entropy_365.rolling(
    "365D",
    min_periods=200
).mean()

entropy_std = entropy_365.rolling(
    "365D",
    min_periods=200
).std(ddof=1)

entropy_z = (
    entropy_365 - entropy_mean
) / entropy_std


# ============================================================
# 5. ENTROPY TREND
#
# Change over approximately one month.
# Positive = disorder increasing
# ============================================================

entropy_change_21d = (
    entropy_365 -
    entropy_365.shift(21)
)


# ============================================================
# 6. COMBINE
# ============================================================

regime = pd.DataFrame({
    "SP500": spx,
    "Momentum_12_1": momentum_12_1,
    "Momentum_Z": momentum_z,
    "Entropy": entropy_365,
    "Entropy_Z": entropy_z,
    "Entropy_21D_Change": entropy_change_21d
}).dropna()


# ============================================================
# 7. CURRENT STATE
# ============================================================

latest = regime.iloc[-1]

print("=" * 70)
print("S&P 500 MOMENTUM / ENTROPY STATE")
print("=" * 70)

print(
    f"Date:                 {regime.index[-1].date()}"
)

print(
    f"12-1 Momentum:        {latest['Momentum_12_1']:+.2%}"
)

print(
    f"Momentum Z:           {latest['Momentum_Z']:+.3f}"
)

print(
    f"Normalized Entropy:   {latest['Entropy']:.3f}"
)

print(
    f"Entropy Z:            {latest['Entropy_Z']:+.3f}"
)

print(
    f"21D Entropy Change:   {latest['Entropy_21D_Change']:+.4f}"
)


# ============================================================
# 8. SIMPLE REGIME INTERPRETATION
# ============================================================

mom = latest["Momentum_Z"]
ent = latest["Entropy_Z"]
ent_change = latest["Entropy_21D_Change"]

print("\nREGIME")

if mom > 0 and ent_change <= 0:
    print("Positive momentum / disorder stable or falling")

elif mom > 0 and ent_change > 0:
    print("Positive momentum BUT disorder is increasing")

elif mom <= 0 and ent_change > 0:
    print("Momentum weakening AND disorder increasing")

else:
    print("Weak momentum / disorder falling")


# ============================================================
# 9. MOMENTUM CHART
# ============================================================

plt.figure(figsize=(13, 5))

plt.plot(
    regime.index,
    regime["Momentum_Z"],
    linewidth=1.5
)

plt.axhline(0, linewidth=1)
plt.axhline(1, linestyle="--", linewidth=1)
plt.axhline(-1, linestyle="--", linewidth=1)
plt.axhline(2, linestyle=":", linewidth=1)
plt.axhline(-2, linestyle=":", linewidth=1)

plt.title(
    "S&P 500 — 12-1 Momentum Z-Score"
)

plt.ylabel("Z-Score")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()


# ============================================================
# 10. ENTROPY CHART
# ============================================================

plt.figure(figsize=(13, 5))

plt.plot(
    regime.index,
    regime["Entropy_Z"],
    linewidth=1.5
)

plt.axhline(0, linewidth=1)
plt.axhline(1, linestyle="--", linewidth=1)
plt.axhline(-1, linestyle="--", linewidth=1)
plt.axhline(2, linestyle=":", linewidth=1)
plt.axhline(-2, linestyle=":", linewidth=1)

plt.title(
    "S&P 500 — Rolling Return Entropy Z-Score"
)

plt.ylabel("Entropy Z-Score")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()


# ============================================================
# 11. MOMENTUM vs DISORDER — LAST 3 YEARS
# ============================================================

recent = regime.last("1095D")

fig, ax1 = plt.subplots(figsize=(13, 6))

ax1.plot(
    recent.index,
    recent["Momentum_Z"],
    label="Momentum Z",
    linewidth=1.6
)

ax1.set_ylabel("Momentum Z")

ax2 = ax1.twinx()

ax2.plot(
    recent.index,
    recent["Entropy_Z"],
    label="Entropy Z",
    linewidth=1.6,
    linestyle="--"
)

ax2.set_ylabel("Entropy Z")

ax1.axhline(0, linewidth=1)

plt.title(
    "S&P 500 — Momentum vs Market Disorder"
)

ax1.grid(alpha=0.20)

fig.tight_layout()
plt.show()
```

```python
# ============================================================
# S&P 500 — MOMENTUM + MARKET ENTROPY / DISORDER
#
# 12-1 Momentum
# 365-day rolling Momentum Z
# Fixed-bin Shannon Entropy
# 365-day rolling Entropy Z
# 21-day change in Entropy
#
# Charts: Plotly native Google Colab renderer
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"


# ============================================================
# 1. S&P 500 PRICE SERIES
# ============================================================

spx = common["S&P 500"].dropna().copy()

# Daily log returns
r = np.log(spx / spx.shift(1)).dropna()


# ============================================================
# 2. 12-1 MOMENTUM
#
# Price ~1 month ago / price ~12 months ago - 1
# ============================================================

momentum_12_1 = (
    spx.shift(21) /
    spx.shift(252)
) - 1


# ============================================================
# 3. 365-DAY ROLLING MOMENTUM Z-SCORE
#
# Z = (current momentum - 365D mean momentum)
#     / 365D std deviation of momentum
# ============================================================

mom_mean = momentum_12_1.rolling(
    window="365D",
    min_periods=200
).mean()

mom_std = momentum_12_1.rolling(
    window="365D",
    min_periods=200
).std(ddof=1)

momentum_z = (
    momentum_12_1 - mom_mean
) / mom_std


# ============================================================
# 4. FIXED RETURN BINS FOR ENTROPY
#
# IMPORTANT:
# Bin boundaries are calculated ONCE from the full return
# history and then NEVER change.
#
# Robust 0.5% / 99.5% boundaries prevent a few crisis
# observations from determining the entire histogram.
# ============================================================

N_BINS = 20

lower = r.quantile(0.005)
upper = r.quantile(0.995)

fixed_edges = np.linspace(
    lower,
    upper,
    N_BINS + 1
)

# Capture extreme observations in outer bins
fixed_edges[0] = -np.inf
fixed_edges[-1] = np.inf


# ============================================================
# 5. FIXED-BIN NORMALIZED SHANNON ENTROPY
#
# H = -Σ p log(p)
#
# Normalize by maximum entropy:
#
# H_norm = H / log(N bins)
#
# Approximate interpretation:
#
# 0 = highly concentrated / ordered
# 1 = highly dispersed / disordered
# ============================================================

def fixed_bin_entropy(x):

    x = np.asarray(x)

    x = x[
        np.isfinite(x)
    ]

    if len(x) < 50:
        return np.nan

    counts, _ = np.histogram(
        x,
        bins=fixed_edges
    )

    p = (
        counts[counts > 0] /
        counts.sum()
    )

    H = -np.sum(
        p * np.log(p)
    )

    H_max = np.log(N_BINS)

    return H / H_max


# ============================================================
# 6. 365-DAY ROLLING ENTROPY
# ============================================================

entropy_365 = r.rolling(
    window="365D",
    min_periods=200
).apply(
    fixed_bin_entropy,
    raw=True
)


# ============================================================
# 7. 365-DAY ENTROPY Z-SCORE
#
# Is current entropy unusually high/low relative to
# entropy observations during the previous 365 days?
# ============================================================

entropy_mean = entropy_365.rolling(
    window="365D",
    min_periods=200
).mean()

entropy_std = entropy_365.rolling(
    window="365D",
    min_periods=200
).std(ddof=1)

entropy_z = (
    entropy_365 - entropy_mean
) / entropy_std


# ============================================================
# 8. 21-TRADING-DAY CHANGE IN ENTROPY
#
# Positive = disorder increasing
# Negative = disorder decreasing
# ============================================================

entropy_change_21d = (
    entropy_365 -
    entropy_365.shift(21)
)


# ============================================================
# 9. COMBINE
# ============================================================

regime = pd.DataFrame({

    "SP500": spx,

    "Momentum_12_1":
        momentum_12_1,

    "Momentum_Z":
        momentum_z,

    "Entropy":
        entropy_365,

    "Entropy_Z":
        entropy_z,

    "Entropy_21D_Change":
        entropy_change_21d

}).dropna()


# ============================================================
# 10. CURRENT STATE
# ============================================================

latest = regime.iloc[-1]

print("=" * 72)
print("S&P 500 — MOMENTUM / ENTROPY STATE")
print("=" * 72)

print(
    f"Date:                "
    f"{regime.index[-1].date()}"
)

print(
    f"12-1 Momentum:       "
    f"{latest['Momentum_12_1']:+.2%}"
)

print(
    f"Momentum Z:          "
    f"{latest['Momentum_Z']:+.3f}"
)

print(
    f"Normalized Entropy:  "
    f"{latest['Entropy']:.3f}"
)

print(
    f"Entropy Z:           "
    f"{latest['Entropy_Z']:+.3f}"
)

print(
    f"21D Entropy Change:  "
    f"{latest['Entropy_21D_Change']:+.4f}"
)


# ============================================================
# 11. REGIME DESCRIPTION
# ============================================================

mom = latest["Momentum_Z"]

ent_z = latest["Entropy_Z"]

ent_change = latest[
    "Entropy_21D_Change"
]

print("\nREGIME:")

if mom > 0 and ent_change <= 0:

    print(
        "Positive momentum / "
        "disorder stable or falling"
    )

elif mom > 0 and ent_change > 0:

    print(
        "Positive momentum BUT "
        "disorder increasing"
    )

elif mom <= 0 and ent_change > 0:

    print(
        "Momentum weakening AND "
        "disorder increasing"
    )

else:

    print(
        "Weak momentum / "
        "disorder falling"
    )


# ============================================================
# 12. CURRENT METRICS TABLE
# ============================================================

current_table = pd.DataFrame({

    "Metric": [
        "12-1 Momentum",
        "Momentum Z",
        "Normalized Entropy",
        "Entropy Z",
        "21D Entropy Change"
    ],

    "Value": [
        latest["Momentum_12_1"],
        latest["Momentum_Z"],
        latest["Entropy"],
        latest["Entropy_Z"],
        latest["Entropy_21D_Change"]
    ]

})

display(current_table)


# ============================================================
# 13. CHART 1 — MOMENTUM Z
# ============================================================

fig = go.Figure()

fig.add_trace(

    go.Scatter(

        x=regime.index,

        y=regime["Momentum_Z"],

        mode="lines",

        name="12-1 Momentum Z"

    )

)

for level, dash in [

    (0, "solid"),
    (1, "dash"),
    (-1, "dash"),
    (2, "dot"),
    (-2, "dot")

]:

    fig.add_hline(
        y=level,
        line_dash=dash
    )


fig.update_layout(

    title=(
        "S&P 500 — "
        "12-1 Momentum Z-Score"
    ),

    xaxis_title="Date",

    yaxis_title="Momentum Z",

    hovermode="x unified",

    template="plotly_white",

    height=550

)

fig.show()


# ============================================================
# 14. CHART 2 — NORMALIZED ENTROPY LEVEL
# ============================================================

fig = go.Figure()

fig.add_trace(

    go.Scatter(

        x=regime.index,

        y=regime["Entropy"],

        mode="lines",

        name="Normalized Entropy"

    )

)

fig.update_layout(

    title=(
        "S&P 500 — "
        "365-Day Normalized Shannon Entropy"
    ),

    xaxis_title="Date",

    yaxis_title="Entropy (0–1)",

    hovermode="x unified",

    template="plotly_white",

    height=550

)

fig.show()


# ============================================================
# 15. CHART 3 — ENTROPY Z
# ============================================================

fig = go.Figure()

fig.add_trace(

    go.Scatter(

        x=regime.index,

        y=regime["Entropy_Z"],

        mode="lines",

        name="Entropy Z"

    )

)

for level, dash in [

    (0, "solid"),
    (1, "dash"),
    (-1, "dash"),
    (2, "dot"),
    (-2, "dot")

]:

    fig.add_hline(
        y=level,
        line_dash=dash
    )


fig.update_layout(

    title=(
        "S&P 500 — "
        "365-Day Entropy Z-Score"
    ),

    xaxis_title="Date",

    yaxis_title="Entropy Z",

    hovermode="x unified",

    template="plotly_white",

    height=550

)

fig.show()


# ============================================================
# 16. CHART 4 — MOMENTUM vs ENTROPY
# LAST 3 YEARS
# ============================================================

cutoff = (
    regime.index.max()
    - pd.DateOffset(years=3)
)

recent = regime.loc[
    regime.index >= cutoff
]


fig = go.Figure()


fig.add_trace(

    go.Scatter(

        x=recent.index,

        y=recent["Momentum_Z"],

        mode="lines",

        name="Momentum Z",

        yaxis="y"

    )

)


fig.add_trace(

    go.Scatter(

        x=recent.index,

        y=recent["Entropy_Z"],

        mode="lines",

        name="Entropy Z",

        yaxis="y2",

        line=dict(
            dash="dash"
        )

    )

)


fig.update_layout(

    title=(
        "S&P 500 — "
        "Momentum vs Market Disorder"
    ),

    xaxis=dict(
        title="Date"
    ),

    yaxis=dict(
        title="Momentum Z"
    ),

    yaxis2=dict(
        title="Entropy Z",
        overlaying="y",
        side="right"
    ),

    hovermode="x unified",

    template="plotly_white",

    height=600,

    legend=dict(
        orientation="h",
        y=1.10
    )

)

fig.show()


# ============================================================
# 17. CHART 5 — 21D ENTROPY CHANGE
#
# Useful for seeing whether disorder is currently accelerating
# or dissipating.
# ============================================================

fig = go.Figure()

fig.add_trace(

    go.Scatter(

        x=recent.index,

        y=recent["Entropy_21D_Change"],

        mode="lines",

        name="21D Entropy Change"

    )

)

fig.add_hline(
    y=0,
    line_dash="solid"
)

fig.update_layout(

    title=(
        "S&P 500 — "
        "21-Day Change in Market Entropy"
    ),

    xaxis_title="Date",

    yaxis_title="Change in Entropy",

    hovermode="x unified",

    template="plotly_white",

    height=500

)

fig.show()
```

```python
# ============================================================
# S&P 500 — 30-DAY ENTROPY / SHORT-TERM DISORDER
# Fixed bins + native Colab Plotly
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"

spx = common["S&P 500"].dropna().copy()
r = np.log(spx / spx.shift(1)).dropna()


# ============================================================
# 1. FIXED BINS
# ============================================================

N_BINS = 8

lower = r.quantile(0.005)
upper = r.quantile(0.995)

fixed_edges_30 = np.linspace(
    lower,
    upper,
    N_BINS + 1
)

fixed_edges_30[0] = -np.inf
fixed_edges_30[-1] = np.inf


def entropy_fixed_30(x):

    x = np.asarray(x)
    x = x[np.isfinite(x)]

    if len(x) < 15:
        return np.nan

    counts, _ = np.histogram(
        x,
        bins=fixed_edges_30
    )

    p = counts[counts > 0] / counts.sum()

    H = -np.sum(p * np.log(p))

    return H / np.log(N_BINS)


# ============================================================
# 2. 30-CALENDAR-DAY ENTROPY
# ============================================================

entropy_30 = r.rolling(
    "30D",
    min_periods=15
).apply(
    entropy_fixed_30,
    raw=True
)


# ============================================================
# 3. Z-SCORE THE 30D ENTROPY
#
# Compare current short-term entropy with the preceding
# 365 days of 30D entropy observations.
# ============================================================

entropy30_mean = entropy_30.rolling(
    "365D",
    min_periods=150
).mean()

entropy30_std = entropy_30.rolling(
    "365D",
    min_periods=150
).std(ddof=1)

entropy30_z = (
    entropy_30 - entropy30_mean
) / entropy30_std


# ============================================================
# 4. CHANGES
# ============================================================

entropy30_change_5d = (
    entropy_30 - entropy_30.shift(5)
)

entropy30_change_21d = (
    entropy_30 - entropy_30.shift(21)
)


# ============================================================
# 5. CURRENT STATE
# ============================================================

df30 = pd.DataFrame({
    "Entropy_30D": entropy_30,
    "Entropy_30D_Z": entropy30_z,
    "5D_Change": entropy30_change_5d,
    "21D_Change": entropy30_change_21d
}).dropna()

latest = df30.iloc[-1]

print("=" * 65)
print("S&P 500 — 30-DAY ENTROPY STATE")
print("=" * 65)

print(f"Date:              {df30.index[-1].date()}")
print(f"30D Entropy:       {latest['Entropy_30D']:.3f}")
print(f"30D Entropy Z:     {latest['Entropy_30D_Z']:+.3f}")
print(f"5D Change:         {latest['5D_Change']:+.4f}")
print(f"21D Change:        {latest['21D_Change']:+.4f}")


# ============================================================
# 6. ENTROPY LEVEL — LAST 3 YEARS
# ============================================================

cutoff = df30.index.max() - pd.DateOffset(years=3)
recent = df30.loc[df30.index >= cutoff]

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=recent.index,
        y=recent["Entropy_30D"],
        mode="lines",
        name="30D Entropy"
    )
)

fig.update_layout(
    title="S&P 500 — 30-Day Normalized Shannon Entropy",
    xaxis_title="Date",
    yaxis_title="Entropy (0–1)",
    hovermode="x unified",
    template="plotly_white",
    height=550
)

fig.show()


# ============================================================
# 7. ENTROPY Z-SCORE — LAST 3 YEARS
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=recent.index,
        y=recent["Entropy_30D_Z"],
        mode="lines",
        name="30D Entropy Z"
    )
)

for level, dash in [
    (0, "solid"),
    (1, "dash"),
    (-1, "dash"),
    (2, "dot"),
    (-2, "dot")
]:
    fig.add_hline(
        y=level,
        line_dash=dash
    )

fig.update_layout(
    title="S&P 500 — 30-Day Market Disorder Z-Score",
    xaxis_title="Date",
    yaxis_title="Entropy Z",
    hovermode="x unified",
    template="plotly_white",
    height=550
)

fig.show()


# ============================================================
# 8. 30D ENTROPY Z vs EXISTING MOMENTUM Z
# ============================================================

combined = pd.concat(
    [
        momentum_z.rename("Momentum Z"),
        entropy30_z.rename("30D Entropy Z")
    ],
    axis=1
).dropna()

cutoff = combined.index.max() - pd.DateOffset(years=3)
combined = combined.loc[combined.index >= cutoff]

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=combined.index,
        y=combined["Momentum Z"],
        mode="lines",
        name="12-1 Momentum Z"
    )
)

fig.add_trace(
    go.Scatter(
        x=combined.index,
        y=combined["30D Entropy Z"],
        mode="lines",
        name="30D Entropy Z",
        line=dict(dash="dash")
    )
)

fig.add_hline(
    y=0,
    line_dash="solid"
)

fig.update_layout(
    title="S&P 500 — Momentum vs 30-Day Market Disorder",
    xaxis_title="Date",
    yaxis_title="Z-Score",
    hovermode="x unified",
    template="plotly_white",
    height=600
)

fig.show()
```
