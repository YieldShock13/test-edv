```python
# ============================================================
# LIVE PORTFOLIO DASHBOARD — CELL 2
# PERFORMANCE CALCULATION ENGINE
#
# REQUIRES CELL 1 TO HAVE RUN SUCCESSFULLY
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 1. IDENTIFY CURRENT + PREVIOUS TRADING DAY
# ============================================================

valid_intraday_rows = (
    intraday_close
    .notna()
    .any(axis=1)
)

intraday_valid = (
    intraday_close
    .loc[valid_intraday_rows]
    .copy()
)

if intraday_valid.empty:
    raise ValueError("No valid intraday observations.")


current_date = intraday_valid.index[-1].date()

available_dates = sorted(
    set(intraday_valid.index.date)
)

previous_dates = [
    d for d in available_dates
    if d < current_date
]

if not previous_dates:
    raise ValueError(
        "No previous trading date available."
    )

previous_date = previous_dates[-1]


print("=" * 110)
print("LIVE PERFORMANCE ENGINE")
print("=" * 110)

print(
    f"Current trading date:   {current_date}"
)

print(
    f"Previous trading date:  {previous_date}"
)


# ============================================================
# 2. BUILD LIVE HOLDING TABLE
#
# Today's return:
#
# latest intraday price
# ----------------------  - 1
# previous trading close
#
# IMPORTANT:
# We use Yahoo's actual previous trading close.
# ============================================================

live_rows = []


for name, row in master.iterrows():

    ticker = row["ticker"]
    qty = float(row["qty"])
    bucket = row["bucket"]


    if ticker not in intraday_valid.columns:

        live_rows.append({
            "Holding": name,
            "Ticker": ticker,
            "Bucket": bucket,
            "Qty": qty,
            "Status": "NO INTRADAY DATA"
        })

        continue


    s = (
        intraday_valid[ticker]
        .dropna()
    )


    today_s = (
        s[
            s.index.date ==
            current_date
        ]
    )


    previous_s = (
        s[
            s.index.date ==
            previous_date
        ]
    )


    if len(today_s) == 0:

        live_rows.append({
            "Holding": name,
            "Ticker": ticker,
            "Bucket": bucket,
            "Qty": qty,
            "Status": "NO CURRENT-DAY DATA"
        })

        continue


    if len(previous_s) == 0:

        live_rows.append({
            "Holding": name,
            "Ticker": ticker,
            "Bucket": bucket,
            "Qty": qty,
            "Status": "NO PREVIOUS CLOSE"
        })

        continue


    latest_cents = float(
        today_s.iloc[-1]
    )

    previous_cents = float(
        previous_s.iloc[-1]
    )


    # Yahoo JSE prices -> cents
    latest_price = (
        latest_cents / 100
    )

    previous_price = (
        previous_cents / 100
    )


    today_return = (
        latest_price /
        previous_price
        - 1
    )


    opening_value = (
        qty *
        previous_price
    )


    current_value = (
        qty *
        latest_price
    )


    pnl = (
        current_value -
        opening_value
    )


    latest_timestamp = (
        today_s.index[-1]
    )


    live_rows.append({

        "Holding": name,
        "Ticker": ticker,
        "Bucket": bucket,
        "Qty": qty,

        "Previous_Close_R":
            previous_price,

        "Latest_R":
            latest_price,

        "Today_Return":
            today_return,

        "Opening_Value_R":
            opening_value,

        "Current_Value_R":
            current_value,

        "Today_PnL_R":
            pnl,

        "Last_Update":
            latest_timestamp,

        "Status":
            "OK"
    })


live = pd.DataFrame(
    live_rows
)


# ============================================================
# 3. ENSURE COMPLETE LIVE COVERAGE
# ============================================================

live_valid = (
    live[
        live["Status"] == "OK"
    ]
    .copy()
)


if len(live_valid) != len(master):

    print("\nWARNING — LIVE DATA INCOMPLETE")

    print(
        live[
            live["Status"] != "OK"
        ][
            [
                "Holding",
                "Ticker",
                "Status"
            ]
        ]
        .to_string(index=False)
    )


# ============================================================
# 4. PORTFOLIO-LEVEL CALCULATION
#
# Combined return is NOT the average of Main + Subset.
#
# It is:
#
# total portfolio R P&L
# ---------------------
# total opening value
# ============================================================

def calculate_portfolio(bucket=None):

    if bucket is None:

        x = (
            live_valid
            .copy()
        )

        label = "COMBINED"

    else:

        x = (
            live_valid[
                live_valid["Bucket"] ==
                bucket
            ]
            .copy()
        )

        label = bucket.upper()


    opening = (
        x["Opening_Value_R"]
        .sum()
    )


    current = (
        x["Current_Value_R"]
        .sum()
    )


    pnl = (
        x["Today_PnL_R"]
        .sum()
    )


    ret = (
        pnl / opening
        if opening != 0
        else np.nan
    )


    return {
        "Portfolio": label,
        "Opening_Value_R": opening,
        "Current_Value_R": current,
        "Today_PnL_R": pnl,
        "Today_Return": ret
    }


combined_stats = calculate_portfolio()

main_stats = calculate_portfolio(
    "Main"
)

subset_stats = calculate_portfolio(
    "Subset"
)


portfolio_summary = pd.DataFrame([
    combined_stats,
    main_stats,
    subset_stats
])


# ============================================================
# 5. PORTFOLIO WEIGHTS
#
# Current live economic weights.
# ============================================================

combined_current = (
    live_valid["Current_Value_R"]
    .sum()
)


live_valid["Combined_Weight"] = (
    live_valid["Current_Value_R"]
    /
    combined_current
)


for bucket in ["Main", "Subset"]:

    bucket_total = (
        live_valid.loc[
            live_valid["Bucket"] ==
            bucket,
            "Current_Value_R"
        ]
        .sum()
    )

    mask = (
        live_valid["Bucket"] ==
        bucket
    )

    live_valid.loc[
        mask,
        "Bucket_Weight"
    ] = (
        live_valid.loc[
            mask,
            "Current_Value_R"
        ]
        /
        bucket_total
    )


# ============================================================
# 6. HISTORICAL TOTAL RETURN HELPERS
#
# Historical returns use Adjusted Close.
#
# We require actual data coverage back to the required
# starting period.
#
# If insufficient -> NaN -> displayed as "–".
# ============================================================

adj = (
    adj_close
    .copy()
    .sort_index()
)


# Remove rows where everything is missing
adj = adj.loc[
    adj.notna().any(axis=1)
]


latest_hist_date = (
    adj.index.max()
)


def first_valid_on_or_after(
    series,
    target_date,
    tolerance_days=10
):

    """
    Find first valid observation on/after target date,
    but reject it if it is too far from target.

    Prevents a 3-year return being calculated from
    only 1 year of available history.
    """

    s = (
        series
        .dropna()
        .sort_index()
    )

    if s.empty:
        return None


    candidates = (
        s[
            s.index >= target_date
        ]
    )


    if candidates.empty:
        return None


    date = candidates.index[0]


    distance = (
        date.normalize() -
        target_date.normalize()
    ).days


    if distance > tolerance_days:
        return None


    return (
        date,
        float(candidates.iloc[0])
    )


def latest_valid_price(series):

    s = (
        series
        .dropna()
        .sort_index()
    )

    if s.empty:
        return None

    return (
        s.index[-1],
        float(s.iloc[-1])
    )


def total_return_from_target(
    series,
    target_date,
    tolerance_days=10
):

    start = first_valid_on_or_after(
        series,
        target_date,
        tolerance_days
    )

    end = latest_valid_price(
        series
    )


    if start is None or end is None:
        return np.nan


    start_date, start_price = start
    end_date, end_price = end


    if end_date <= start_date:
        return np.nan


    if start_price <= 0:
        return np.nan


    return (
        end_price /
        start_price
        - 1
    )


# ============================================================
# 7. PERIOD START DATES
# ============================================================

hist_end = pd.Timestamp(
    latest_hist_date
).normalize()


# Month-to-date:
# previous month-end / first trading observation
# around beginning of current month.

month_start = (
    hist_end
    .replace(day=1)
)


# YTD:
year_start = pd.Timestamp(
    year=hist_end.year,
    month=1,
    day=1
)


one_year_start = (
    hist_end -
    pd.DateOffset(years=1)
)


three_year_start = (
    hist_end -
    pd.DateOffset(years=3)
)


five_year_start = (
    hist_end -
    pd.DateOffset(years=5)
)


# ============================================================
# 8. HISTORICAL ASSET RETURNS
# ============================================================

historical_rows = []


for name, row in master.iterrows():

    ticker = row["ticker"]


    if ticker not in adj.columns:

        historical_rows.append({
            "Holding": name,
            "Ticker": ticker,
            "Bucket": row["bucket"],
            "M/M": np.nan,
            "YTD": np.nan,
            "1Y": np.nan,
            "3Y": np.nan,
            "5Y": np.nan
        })

        continue


    s = (
        adj[ticker]
        .dropna()
    )


    # ----------------------------------------
    # M/M
    #
    # User requested M/M.
    # Here interpreted as month-to-date.
    # ----------------------------------------

    mm = total_return_from_target(
        s,
        month_start,
        tolerance_days=10
    )


    # ----------------------------------------
    # YTD
    # ----------------------------------------

    ytd = total_return_from_target(
        s,
        year_start,
        tolerance_days=10
    )


    # ----------------------------------------
    # 1 YEAR
    # ----------------------------------------

    one_y = total_return_from_target(
        s,
        one_year_start,
        tolerance_days=10
    )


    # ----------------------------------------
    # 3 YEARS
    # ----------------------------------------

    three_y = total_return_from_target(
        s,
        three_year_start,
        tolerance_days=10
    )


    # ----------------------------------------
    # 5 YEARS
    # ----------------------------------------

    five_y = total_return_from_target(
        s,
        five_year_start,
        tolerance_days=10
    )


    historical_rows.append({

        "Holding":
            name,

        "Ticker":
            ticker,

        "Bucket":
            row["bucket"],

        "M/M":
            mm,

        "YTD":
            ytd,

        "1Y":
            one_y,

        "3Y":
            three_y,

        "5Y":
            five_y
    })


historical_returns = pd.DataFrame(
    historical_rows
)


# ============================================================
# 9. MERGE LIVE + HISTORICAL
# ============================================================

asset_table = (
    live_valid[
        [
            "Holding",
            "Ticker",
            "Bucket",
            "Today_Return",
            "Current_Value_R",
            "Today_PnL_R",
            "Last_Update"
        ]
    ]
    .merge(
        historical_returns,
        on=[
            "Holding",
            "Ticker",
            "Bucket"
        ],
        how="left"
    )
)


# ============================================================
# 10. FORMAT HELPERS
# ============================================================

def pct_string(x):

    if pd.isna(x):
        return "–"

    return f"{x:+.2%}"


def rand_string(x):

    if pd.isna(x):
        return "–"

    return f"R{x:,.2f}"


# ============================================================
# 11. PORTFOLIO SUMMARY OUTPUT
# ============================================================

print("\n" + "=" * 110)
print("PORTFOLIO — TODAY")
print("=" * 110)


summary_display = (
    portfolio_summary
    .copy()
)


summary_display["D/D Return"] = (
    summary_display["Today_Return"]
    .apply(pct_string)
)


summary_display["D/D P&L"] = (
    summary_display["Today_PnL_R"]
    .apply(rand_string)
)


summary_display["Opening Value"] = (
    summary_display["Opening_Value_R"]
    .apply(rand_string)
)


summary_display["Current Value"] = (
    summary_display["Current_Value_R"]
    .apply(rand_string)
)


print(
    summary_display[
        [
            "Portfolio",
            "D/D Return",
            "D/D P&L",
            "Opening Value",
            "Current Value"
        ]
    ]
    .to_string(
        index=False
    )
)


# ============================================================
# 12. ALL-ASSET PERFORMANCE TABLE
#
# NO WEIGHTS — as requested.
# ============================================================

print("\n" + "=" * 125)
print("ALL ASSETS — RETURN DASHBOARD")
print("=" * 125)


asset_display = (
    asset_table[
        [
            "Holding",
            "Bucket",
            "Today_Return",
            "M/M",
            "YTD",
            "1Y",
            "3Y",
            "5Y",
            "Last_Update"
        ]
    ]
    .copy()
)


for col in [
    "Today_Return",
    "M/M",
    "YTD",
    "1Y",
    "3Y",
    "5Y"
]:

    asset_display[col] = (
        asset_display[col]
        .apply(pct_string)
    )


asset_display = (
    asset_display
    .rename(
        columns={
            "Today_Return":
                "Today"
        }
    )
)


print(
    asset_display
    .to_string(
        index=False
    )
)


# ============================================================
# 13. LIVE DATA FRESHNESS
# ============================================================

latest_update = (
    live_valid["Last_Update"]
    .max()
)


print("\n" + "=" * 110)
print("DATA STATUS")
print("=" * 110)


print(
    f"Latest portfolio observation: "
    f"{latest_update}"
)


stale_cutoff = (
    latest_update -
    pd.Timedelta(minutes=30)
)


stale = (
    live_valid[
        live_valid["Last_Update"] <
        stale_cutoff
    ][
        [
            "Holding",
            "Ticker",
            "Last_Update"
        ]
    ]
)


if len(stale):

    print(
        "\nHoldings with last trade "
        ">30 minutes older than latest portfolio observation:"
    )

    print(
        stale.to_string(
            index=False
        )
    )

else:

    print(
        "\nNo holding has a last trade "
        ">30 minutes older than the latest portfolio observation."
    )


# ============================================================
# 14. RECONCILIATION
# ============================================================

main_plus_subset_pnl = (
    main_stats["Today_PnL_R"]
    +
    subset_stats["Today_PnL_R"]
)


combined_pnl = (
    combined_stats["Today_PnL_R"]
)


pnl_difference = (
    combined_pnl -
    main_plus_subset_pnl
)


main_plus_subset_value = (
    main_stats["Current_Value_R"]
    +
    subset_stats["Current_Value_R"]
)


combined_value = (
    combined_stats["Current_Value_R"]
)


value_difference = (
    combined_value -
    main_plus_subset_value
)


print("\n" + "=" * 110)
print("RECONCILIATION")
print("=" * 110)


print(
    f"Combined P&L less Main + Subset: "
    f"R{pnl_difference:,.6f}"
)

print(
    f"Combined value less Main + Subset: "
    f"R{value_difference:,.6f}"
)


if (
    abs(pnl_difference) < 0.01
    and
    abs(value_difference) < 0.01
):

    print(
        "\nPASS: Combined portfolio reconciles "
        "exactly to Main + Subset."
    )

else:

    print(
        "\nFAIL: Portfolio reconciliation error."
    )


# ============================================================
# 15. CELL STATUS
# ============================================================

print("\n" + "=" * 110)
print("CELL 2 COMPLETE")
print("=" * 110)

print("""
This calculation engine now provides:

• Combined portfolio D/D return
• Main portfolio D/D return
• Subset D/D return
• Nominal daily P&L
• Current market values
• Every asset's Today return
• M/M total return
• YTD total return
• 1Y total return
• 3Y total return
• 5Y total return
• Missing history shown as "–"
• Live-price freshness checks
• Combined = Main + Subset reconciliation

Historical period returns use Yahoo Adjusted Close.
Intraday returns use ordinary intraday prices versus prior close.

Next step after validation:
build the actual Streamlit dashboard UI and persistent daily NAV history.
""")
```

```python
# ============================================================
# LIVE PORTFOLIO DASHBOARD — CELL 3
# HISTORICAL DATA INTEGRITY AUDIT
#
# PURPOSE:
# 1. Audit all 24 Adjusted Close series.
# 2. Detect extreme one-day returns / scale breaks.
# 3. Examine STXCAP specifically.
# 4. Do NOT repair anything yet.
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 1. SETTINGS
# ============================================================

# Anything above 40% absolute daily Adjusted Close return
# is sufficiently unusual to demand inspection.
EXTREME_THRESHOLD = 0.40

# Secondary warning threshold
WARNING_THRESHOLD = 0.20


# ============================================================
# 2. AUDIT EACH SECURITY
# ============================================================

audit_summary = []
extreme_events = []


for name, row in master.iterrows():

    ticker = row["ticker"]

    if ticker not in adj_close.columns:

        audit_summary.append({
            "Holding": name,
            "Ticker": ticker,
            "Observations": 0,
            "First_Date": None,
            "Last_Date": None,
            "Max_Abs_Daily_Return": np.nan,
            "Days_Over_20pct": np.nan,
            "Days_Over_40pct": np.nan,
            "Status": "NO HISTORY"
        })

        continue


    s = (
        adj_close[ticker]
        .dropna()
        .sort_index()
    )


    if len(s) < 2:

        audit_summary.append({
            "Holding": name,
            "Ticker": ticker,
            "Observations": len(s),
            "First_Date": (
                s.index[0].date()
                if len(s)
                else None
            ),
            "Last_Date": (
                s.index[-1].date()
                if len(s)
                else None
            ),
            "Max_Abs_Daily_Return": np.nan,
            "Days_Over_20pct": 0,
            "Days_Over_40pct": 0,
            "Status": "INSUFFICIENT HISTORY"
        })

        continue


    returns = (
        s.pct_change(
            fill_method=None
        )
        .dropna()
    )


    abs_returns = (
        returns.abs()
    )


    over_20 = (
        abs_returns >
        WARNING_THRESHOLD
    )


    over_40 = (
        abs_returns >
        EXTREME_THRESHOLD
    )


    max_abs = (
        abs_returns.max()
    )


    if over_40.any():

        status = "FAIL — EXTREME JUMP"

    elif over_20.any():

        status = "REVIEW — >20% DAY"

    else:

        status = "PASS"


    audit_summary.append({

        "Holding":
            name,

        "Ticker":
            ticker,

        "Observations":
            len(s),

        "First_Date":
            s.index[0].date(),

        "Last_Date":
            s.index[-1].date(),

        "Max_Abs_Daily_Return":
            max_abs,

        "Days_Over_20pct":
            int(over_20.sum()),

        "Days_Over_40pct":
            int(over_40.sum()),

        "Status":
            status
    })


    # --------------------------------------------------------
    # Store every >20% event for inspection
    # --------------------------------------------------------

    flagged_dates = (
        returns[
            abs_returns >
            WARNING_THRESHOLD
        ]
        .index
    )


    for dt in flagged_dates:

        loc = (
            s.index.get_loc(dt)
        )


        previous_dt = (
            s.index[loc - 1]
        )


        previous_price = (
            float(
                s.loc[previous_dt]
            )
        )


        current_price = (
            float(
                s.loc[dt]
            )
        )


        event_return = (
            current_price /
            previous_price
            - 1
        )


        extreme_events.append({

            "Holding":
                name,

            "Ticker":
                ticker,

            "Previous_Date":
                previous_dt.date(),

            "Date":
                dt.date(),

            "Previous_Adj_Close":
                previous_price,

            "Current_Adj_Close":
                current_price,

            "Daily_Return":
                event_return
        })


audit_df = pd.DataFrame(
    audit_summary
)


events_df = pd.DataFrame(
    extreme_events
)


# ============================================================
# 3. SUMMARY TABLE
# ============================================================

print(
    "=" * 130
)

print(
    "HISTORICAL ADJUSTED-CLOSE INTEGRITY AUDIT"
)

print(
    "=" * 130
)


audit_display = (
    audit_df.copy()
)


audit_display[
    "Max_Abs_Daily_Return"
] = (
    audit_display[
        "Max_Abs_Daily_Return"
    ] * 100
)


audit_display = (
    audit_display.rename(
        columns={
            "Max_Abs_Daily_Return":
                "Max_Abs_Daily_Return_%"
        }
    )
)


print(

    audit_display[
        [
            "Holding",
            "Ticker",
            "Observations",
            "First_Date",
            "Last_Date",
            "Max_Abs_Daily_Return_%",
            "Days_Over_20pct",
            "Days_Over_40pct",
            "Status"
        ]
    ]

    .sort_values(
        "Max_Abs_Daily_Return_%",
        ascending=False,
        na_position="last"
    )

    .round({
        "Max_Abs_Daily_Return_%": 2
    })

    .to_string(
        index=False
    )
)


# ============================================================
# 4. ALL FLAGGED >20% DAILY EVENTS
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "ALL >20% ABSOLUTE DAILY ADJUSTED-CLOSE MOVES"
)

print(
    "=" * 130
)


if len(events_df):

    event_display = (
        events_df.copy()
    )


    event_display[
        "Daily_Return_%"
    ] = (
        event_display[
            "Daily_Return"
        ] * 100
    )


    print(

        event_display[
            [
                "Holding",
                "Ticker",
                "Previous_Date",
                "Date",
                "Previous_Adj_Close",
                "Current_Adj_Close",
                "Daily_Return_%"
            ]
        ]

        .sort_values(
            "Daily_Return_%",
            key=lambda x: x.abs(),
            ascending=False
        )

        .round({
            "Previous_Adj_Close": 4,
            "Current_Adj_Close": 4,
            "Daily_Return_%": 2
        })

        .to_string(
            index=False
        )
    )

else:

    print(
        "No >20% daily Adjusted Close moves detected."
    )


# ============================================================
# 5. STXCAP — DETAILED INSPECTION
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "STXCAP.JO — DETAILED DIAGNOSTIC"
)

print(
    "=" * 130
)


ticker = "STXCAP.JO"


if ticker not in adj_close.columns:

    print(
        "STXCAP.JO not found."
    )

else:

    stx = (
        adj_close[ticker]
        .dropna()
        .sort_index()
    )


    stx_ret = (
        stx.pct_change(
            fill_method=None
        )
    )


    stx_diag = pd.DataFrame({

        "Adj_Close":
            stx,

        "Daily_Return":
            stx_ret
    })


    stx_diag[
        "Daily_Return_%"
    ] = (
        stx_diag[
            "Daily_Return"
        ] * 100
    )


    # --------------------------------------------
    # Largest absolute daily changes
    # --------------------------------------------

    largest = (
        stx_diag
        .dropna()
        .assign(
            Abs_Return=lambda x:
                x["Daily_Return"].abs()
        )
        .sort_values(
            "Abs_Return",
            ascending=False
        )
        .head(20)
    )


    print(
        "\n20 LARGEST STXCAP ADJUSTED-CLOSE MOVES"
    )

    print(
        "-" * 130
    )


    print(

        largest[
            [
                "Adj_Close",
                "Daily_Return_%"
            ]
        ]

        .round({
            "Adj_Close": 4,
            "Daily_Return_%": 2
        })

        .to_string()
    )


# ============================================================
# 6. COMPARE RAW CLOSE VS ADJUSTED CLOSE FOR STXCAP
#
# This tells us whether the break exists in:
# - actual raw price data
# - Yahoo's adjustment factor
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "STXCAP — RAW CLOSE VS ADJUSTED CLOSE"
)

print(
    "=" * 130
)


if (
    isinstance(
        history_raw.columns,
        pd.MultiIndex
    )
    and
    "Close" in
    history_raw.columns.get_level_values(0)
):

    raw_close = (
        history_raw["Close"][
            "STXCAP.JO"
        ]
        .dropna()
        .sort_index()
    )


    adjusted = (
        adj_close[
            "STXCAP.JO"
        ]
        .dropna()
        .sort_index()
    )


    comparison = pd.concat(
        [
            raw_close.rename(
                "Raw_Close"
            ),

            adjusted.rename(
                "Adj_Close"
            )
        ],
        axis=1
    )


    comparison[
        "Adjustment_Factor"
    ] = (
        comparison[
            "Adj_Close"
        ]
        /
        comparison[
            "Raw_Close"
        ]
    )


    comparison[
        "Raw_Return"
    ] = (
        comparison[
            "Raw_Close"
        ]
        .pct_change(
            fill_method=None
        )
    )


    comparison[
        "Adjusted_Return"
    ] = (
        comparison[
            "Adj_Close"
        ]
        .pct_change(
            fill_method=None
        )
    )


    # Flag either:
    # raw >20%
    # adjusted >20%
    # adjustment factor changes >10%

    factor_change = (
        comparison[
            "Adjustment_Factor"
        ]
        .pct_change(
            fill_method=None
        )
    )


    suspect_mask = (

        comparison[
            "Raw_Return"
        ]
        .abs()
        .gt(0.20)

        |

        comparison[
            "Adjusted_Return"
        ]
        .abs()
        .gt(0.20)

        |

        factor_change
        .abs()
        .gt(0.10)
    )


    suspect_dates = (
        comparison[
            suspect_mask
        ]
        .index
    )


    if len(suspect_dates):

        # Include surrounding observations
        positions = set()


        for dt in suspect_dates:

            pos = (
                comparison
                .index
                .get_loc(dt)
            )


            for offset in range(
                -3,
                4
            ):

                p = (
                    pos +
                    offset
                )


                if (
                    p >= 0
                    and
                    p < len(comparison)
                ):

                    positions.add(
                        p
                    )


        positions = sorted(
            positions
        )


        detailed = (
            comparison.iloc[
                positions
            ]
            .copy()
        )


        detailed[
            "Raw_Return_%"
        ] = (
            detailed[
                "Raw_Return"
            ] * 100
        )


        detailed[
            "Adjusted_Return_%"
        ] = (
            detailed[
                "Adjusted_Return"
            ] * 100
        )


        print(
            detailed[
                [
                    "Raw_Close",
                    "Adj_Close",
                    "Adjustment_Factor",
                    "Raw_Return_%",
                    "Adjusted_Return_%"
                ]
            ]
            .round({
                "Raw_Close": 4,
                "Adj_Close": 4,
                "Adjustment_Factor": 6,
                "Raw_Return_%": 2,
                "Adjusted_Return_%": 2
            })
            .to_string()
        )


    else:

        print(
            "No obvious STXCAP raw/adjusted discontinuity found."
        )


# ============================================================
# 7. AUTOMATIC HISTORICAL RETURN ELIGIBILITY
#
# IMPORTANT:
# This does NOT repair data.
#
# Any security with a >40% Adjusted Close one-day move is
# blocked from historical horizon calculations until reviewed.
# ============================================================

failed_tickers = set(

    audit_df.loc[
        audit_df[
            "Days_Over_40pct"
        ] > 0,
        "Ticker"
    ]
)


historical_eligibility = (
    audit_df[
        [
            "Holding",
            "Ticker",
            "Status"
        ]
    ]
    .copy()
)


historical_eligibility[
    "Historical_Returns_Allowed"
] = (
    ~historical_eligibility[
        "Ticker"
    ]
    .isin(
        failed_tickers
    )
)


print(
    "\n" +
    "=" * 130
)

print(
    "HISTORICAL RETURN ELIGIBILITY"
)

print(
    "=" * 130
)


print(

    historical_eligibility
    .to_string(
        index=False
    )
)


# ============================================================
# 8. FINAL RESULT
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "AUDIT RESULT"
)

print(
    "=" * 130
)


print(
    f"Securities audited: "
    f"{len(audit_df)}"
)


print(
    f"Securities with >20% daily moves: "
    f"{int((audit_df['Days_Over_20pct'] > 0).sum())}"
)


print(
    f"Securities with >40% daily moves: "
    f"{int((audit_df['Days_Over_40pct'] > 0).sum())}"
)


if len(failed_tickers):

    print(
        "\nBLOCKED FROM HISTORICAL RETURN DISPLAY:"
    )

    for ticker in sorted(
        failed_tickers
    ):

        name = (
            master[
                master["ticker"] ==
                ticker
            ]
            .index[0]
        )

        print(
            f"  {name} ({ticker})"
        )

else:

    print(
        "\nNo series automatically blocked."
    )


print(
    "\nNo data have been repaired or altered by this cell."
)

print(
    "Send me this output before we change the historical return engine."
)

print(
    "=" * 130
)
```

```python
# ============================================================
# LIVE PORTFOLIO DASHBOARD — CELL 4
# JSE UNIT-NORMALISATION + POST-REPAIR AUDIT
#
# Requires Cells 1–3.
# ============================================================

import pandas as pd
import numpy as np


# ============================================================
# 1. WORK ON COPIES ONLY
# ============================================================

adj_clean = adj_close.copy().sort_index()

raw_close_all = (
    history_raw["Close"]
    .copy()
    .sort_index()
)

repair_log = []


# ============================================================
# 2. NORMALISE 100x / 0.01x UNIT BREAKS
#
# Logic:
# Compare each observation with the previous CLEAN observation.
#
# If ratio is approximately 100:
# divide current observation by 100.
#
# If ratio is approximately 0.01:
# multiply current observation by 100.
#
# We require a very large ratio:
# >50x or <1/50x.
#
# This deliberately does NOT touch ordinary large market moves.
# ============================================================

def repair_unit_series(series, ticker, field):

    s = (
        series
        .dropna()
        .sort_index()
        .copy()
    )

    if len(s) < 2:
        return s


    cleaned = s.copy()

    previous_clean = float(
        cleaned.iloc[0]
    )


    for i in range(1, len(cleaned)):

        original = float(
            cleaned.iloc[i]
        )

        current = original


        if (
            previous_clean <= 0
            or
            current <= 0
        ):
            previous_clean = current
            continue


        ratio = (
            current /
            previous_clean
        )


        action = None


        # ----------------------------------------
        # Current point appears 100x too large
        # ----------------------------------------

        if ratio > 50:

            candidate = (
                current / 100
            )

            candidate_ratio = (
                candidate /
                previous_clean
            )


            # Only accept if scaling produces
            # a remotely plausible adjacent move.
            if (
                0.50 <=
                candidate_ratio <=
                1.50
            ):

                current = candidate

                action = "DIVIDE_BY_100"


        # ----------------------------------------
        # Current point appears 100x too small
        # ----------------------------------------

        elif ratio < (1 / 50):

            candidate = (
                current * 100
            )

            candidate_ratio = (
                candidate /
                previous_clean
            )


            if (
                0.50 <=
                candidate_ratio <=
                1.50
            ):

                current = candidate

                action = "MULTIPLY_BY_100"


        # ----------------------------------------
        # Save repair
        # ----------------------------------------

        if action is not None:

            cleaned.iloc[i] = current

            repair_log.append({

                "Ticker":
                    ticker,

                "Field":
                    field,

                "Date":
                    cleaned.index[i].date(),

                "Original":
                    original,

                "Cleaned":
                    current,

                "Action":
                    action,

                "Previous_Clean":
                    previous_clean
            })


        previous_clean = float(
            cleaned.iloc[i]
        )


    return cleaned


# ============================================================
# 3. APPLY TO ALL SECURITIES
#
# We repair Close and Adjusted Close independently using
# exactly the same strict unit-break logic.
# ============================================================

raw_clean = raw_close_all.copy()


for ticker in adj_clean.columns:

    # Adjusted Close
    repaired_adj = repair_unit_series(
        adj_clean[ticker],
        ticker,
        "Adj Close"
    )

    adj_clean.loc[
        repaired_adj.index,
        ticker
    ] = repaired_adj


    # Raw Close
    if ticker in raw_clean.columns:

        repaired_raw = repair_unit_series(
            raw_clean[ticker],
            ticker,
            "Close"
        )

        raw_clean.loc[
            repaired_raw.index,
            ticker
        ] = repaired_raw


repair_df = pd.DataFrame(
    repair_log
)


# ============================================================
# 4. PRINT EXACTLY WHAT WAS CHANGED
# ============================================================

print("=" * 130)
print("JSE UNIT NORMALISATION — REPAIR LOG")
print("=" * 130)


if len(repair_df):

    print(
        repair_df
        .sort_values(
            [
                "Ticker",
                "Date",
                "Field"
            ]
        )
        .round({
            "Original": 4,
            "Cleaned": 4,
            "Previous_Clean": 4
        })
        .to_string(
            index=False
        )
    )

else:

    print(
        "No 100x / 0.01x unit breaks repaired."
    )


print(
    f"\nTotal individual field repairs: "
    f"{len(repair_df)}"
)


# ============================================================
# 5. POST-REPAIR AUDIT
# ============================================================

post_rows = []
post_events = []


for name, row in master.iterrows():

    ticker = row["ticker"]


    if ticker not in adj_clean.columns:

        continue


    s = (
        adj_clean[ticker]
        .dropna()
        .sort_index()
    )


    if len(s) < 2:

        post_rows.append({

            "Holding":
                name,

            "Ticker":
                ticker,

            "Observations":
                len(s),

            "Max_Abs_Daily_Return":
                np.nan,

            "Days_Over_20pct":
                0,

            "Days_Over_40pct":
                0,

            "Status":
                "INSUFFICIENT HISTORY"
        })

        continue


    r = (
        s.pct_change(
            fill_method=None
        )
        .dropna()
    )


    abs_r = (
        r.abs()
    )


    n20 = int(
        (abs_r > 0.20)
        .sum()
    )


    n40 = int(
        (abs_r > 0.40)
        .sum()
    )


    max_abs = (
        abs_r.max()
    )


    if n40 > 0:

        status = (
            "FAIL — >40% REMAINS"
        )

    elif n20 > 0:

        status = (
            "REVIEW — >20% REMAINS"
        )

    else:

        status = "PASS"


    post_rows.append({

        "Holding":
            name,

        "Ticker":
            ticker,

        "Observations":
            len(s),

        "Max_Abs_Daily_Return":
            max_abs,

        "Days_Over_20pct":
            n20,

        "Days_Over_40pct":
            n40,

        "Status":
            status
    })


    flagged = (
        r[
            abs_r >
            0.20
        ]
    )


    for dt, ret in flagged.items():

        pos = (
            s.index.get_loc(dt)
        )

        previous_dt = (
            s.index[pos - 1]
        )


        post_events.append({

            "Holding":
                name,

            "Ticker":
                ticker,

            "Previous_Date":
                previous_dt.date(),

            "Date":
                dt.date(),

            "Previous_Adj_Close":
                float(
                    s.loc[
                        previous_dt
                    ]
                ),

            "Current_Adj_Close":
                float(
                    s.loc[
                        dt
                    ]
                ),

            "Daily_Return":
                float(ret)
        })


post_audit = pd.DataFrame(
    post_rows
)

post_events_df = pd.DataFrame(
    post_events
)


# ============================================================
# 6. POST-REPAIR AUDIT TABLE
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "POST-REPAIR HISTORICAL AUDIT"
)

print(
    "=" * 130
)


post_display = (
    post_audit.copy()
)


post_display[
    "Max_Abs_Daily_Return_%"
] = (
    post_display[
        "Max_Abs_Daily_Return"
    ] * 100
)


print(

    post_display[
        [
            "Holding",
            "Ticker",
            "Observations",
            "Max_Abs_Daily_Return_%",
            "Days_Over_20pct",
            "Days_Over_40pct",
            "Status"
        ]
    ]

    .sort_values(
        "Max_Abs_Daily_Return_%",
        ascending=False,
        na_position="last"
    )

    .round({
        "Max_Abs_Daily_Return_%":
            2
    })

    .to_string(
        index=False
    )
)


# ============================================================
# 7. REMAINING >20% EVENTS
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "REMAINING >20% DAILY MOVES AFTER UNIT REPAIR"
)

print(
    "=" * 130
)


if len(post_events_df):

    temp = (
        post_events_df.copy()
    )

    temp[
        "Daily_Return_%"
    ] = (
        temp[
            "Daily_Return"
        ] * 100
    )


    print(

        temp[
            [
                "Holding",
                "Ticker",
                "Previous_Date",
                "Date",
                "Previous_Adj_Close",
                "Current_Adj_Close",
                "Daily_Return_%"
            ]
        ]

        .sort_values(
            "Daily_Return_%",
            key=lambda x: x.abs(),
            ascending=False
        )

        .round({
            "Previous_Adj_Close":
                4,

            "Current_Adj_Close":
                4,

            "Daily_Return_%":
                2
        })

        .to_string(
            index=False
        )
    )

else:

    print(
        "No >20% daily moves remain."
    )


# ============================================================
# 8. HISTORICAL RETURN ELIGIBILITY
#
# Hard rule:
# >40% remaining = BLOCK.
#
# >20% remaining = REVIEW, but not automatically altered.
#
# This allows genuine March-2020 moves to remain visible
# for inspection.
# ============================================================

blocked = set(

    post_audit.loc[
        post_audit[
            "Days_Over_40pct"
        ] > 0,
        "Ticker"
    ]
)


print(
    "\n" +
    "=" * 130
)

print(
    "POST-REPAIR ELIGIBILITY"
)

print(
    "=" * 130
)


eligibility = (
    post_audit[
        [
            "Holding",
            "Ticker",
            "Status"
        ]
    ]
    .copy()
)


eligibility[
    "Historical_Returns_Allowed"
] = (
    ~eligibility[
        "Ticker"
    ]
    .isin(
        blocked
    )
)


print(
    eligibility
    .to_string(
        index=False
    )
)


# ============================================================
# 9. CLEAN HISTORICAL RETURN FUNCTION
# ============================================================

def clean_period_return(
    ticker,
    target_date,
    tolerance_days=10
):

    if ticker in blocked:
        return np.nan


    if ticker not in adj_clean.columns:
        return np.nan


    s = (
        adj_clean[ticker]
        .dropna()
        .sort_index()
    )


    if len(s) < 2:
        return np.nan


    candidates = (
        s[
            s.index >=
            target_date
        ]
    )


    if candidates.empty:
        return np.nan


    start_date = (
        candidates.index[0]
    )


    gap = (
        start_date.normalize()
        -
        target_date.normalize()
    ).days


    if gap > tolerance_days:
        return np.nan


    start_price = float(
        candidates.iloc[0]
    )


    end_price = float(
        s.iloc[-1]
    )


    if (
        start_price <= 0
        or
        s.index[-1] <= start_date
    ):
        return np.nan


    return (
        end_price /
        start_price
        - 1
    )


# ============================================================
# 10. REBUILD HISTORICAL RETURN TABLE
# ============================================================

hist_end = (
    pd.Timestamp(
        adj_clean.index.max()
    )
    .normalize()
)


month_start = (
    hist_end.replace(
        day=1
    )
)


year_start = pd.Timestamp(
    year=hist_end.year,
    month=1,
    day=1
)


one_year_start = (
    hist_end -
    pd.DateOffset(
        years=1
    )
)


three_year_start = (
    hist_end -
    pd.DateOffset(
        years=3
    )
)


five_year_start = (
    hist_end -
    pd.DateOffset(
        years=5
    )
)


clean_history_rows = []


for name, row in master.iterrows():

    ticker = (
        row["ticker"]
    )


    clean_history_rows.append({

        "Holding":
            name,

        "Ticker":
            ticker,

        "Bucket":
            row["bucket"],

        "M/M":
            clean_period_return(
                ticker,
                month_start
            ),

        "YTD":
            clean_period_return(
                ticker,
                year_start
            ),

        "1Y":
            clean_period_return(
                ticker,
                one_year_start
            ),

        "3Y":
            clean_period_return(
                ticker,
                three_year_start
            ),

        "5Y":
            clean_period_return(
                ticker,
                five_year_start
            )
    })


historical_returns_clean = (
    pd.DataFrame(
        clean_history_rows
    )
)


# ============================================================
# 11. ADD TODAY'S LIVE RETURN
# ============================================================

final_asset_returns = (

    live_valid[
        [
            "Holding",
            "Ticker",
            "Bucket",
            "Today_Return",
            "Last_Update"
        ]
    ]

    .merge(
        historical_returns_clean,

        on=[
            "Holding",
            "Ticker",
            "Bucket"
        ],

        how="left"
    )
)


# ============================================================
# 12. DISPLAY
# ============================================================

def pct_display(x):

    if pd.isna(x):
        return "–"

    return (
        f"{x:+.2%}"
    )


final_display = (
    final_asset_returns.copy()
)


for col in [
    "Today_Return",
    "M/M",
    "YTD",
    "1Y",
    "3Y",
    "5Y"
]:

    final_display[col] = (
        final_display[col]
        .apply(
            pct_display
        )
    )


final_display = (
    final_display.rename(
        columns={
            "Today_Return":
                "Today"
        }
    )
)


print(
    "\n" +
    "=" * 130
)

print(
    "CLEANED ALL-ASSET RETURN TABLE"
)

print(
    "=" * 130
)


print(

    final_display[
        [
            "Holding",
            "Bucket",
            "Today",
            "M/M",
            "YTD",
            "1Y",
            "3Y",
            "5Y",
            "Last_Update"
        ]
    ]

    .to_string(
        index=False
    )
)


# ============================================================
# 13. STXCAP SPECIFIC BEFORE / AFTER CHECK
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "STXCAP — BEFORE / AFTER SANITY CHECK"
)

print(
    "=" * 130
)


stxcap_original = (
    adj_close[
        "STXCAP.JO"
    ]
    .dropna()
)


stxcap_clean = (
    adj_clean[
        "STXCAP.JO"
    ]
    .dropna()
)


stxcap_check = pd.DataFrame({

    "Original":
        stxcap_original,

    "Clean":
        stxcap_clean
})


changed = (
    stxcap_check[
        ~np.isclose(
            stxcap_check[
                "Original"
            ],
            stxcap_check[
                "Clean"
            ],
            equal_nan=True
        )
    ]
)


if len(changed):

    print(
        changed
        .round(4)
        .to_string()
    )

else:

    print(
        "No STXCAP observations changed."
    )


# ============================================================
# 14. FINAL STATUS
# ============================================================

print(
    "\n" +
    "=" * 130
)

print(
    "CELL 4 RESULT"
)

print(
    "=" * 130
)


print(
    f"Remaining securities with >40% daily moves: "
    f"{len(blocked)}"
)


if len(blocked):

    print(
        "\nSTILL BLOCKED:"
    )

    for ticker in sorted(
        blocked
    ):

        name = (
            master[
                master["ticker"] ==
                ticker
            ]
            .index[0]
        )

        print(
            f"  {name} ({ticker})"
        )

else:

    print(
        "\nPASS: No >40% daily Adjusted Close "
        "moves remain after unit normalisation."
    )


print(
    "\nDo not proceed to Streamlit until "
    "this post-repair output has been checked."
)

print(
    "=" * 130
)
```

```python
# ============================================================
# LIVE PORTFOLIO DASHBOARD — CELL 5
# BHETNC HISTORICAL RECONSTRUCTION
#
# Historical:
# Berkshire Hathaway B × USD/ZAR × fee drag
#
# Live:
# remains actual BHETNC.JO
# ============================================================

import yfinance as yf
import pandas as pd
import numpy as np


# ============================================================
# 1. SETTINGS
# ============================================================

BH_TICKER = "BRK-B"
FX_TICKER = "USDZAR=X"
BHETNC_TICKER = "BHETNC.JO"

# Actual BHETNC listing date
LISTING_DATE = pd.Timestamp("2020-11-27")

# Annual management fee
ANNUAL_FEE = 0.01


# ============================================================
# 2. DOWNLOAD BRK.B + USD/ZAR FROM YAHOO
# ============================================================

print("=" * 120)
print("DOWNLOADING BHETNC RECONSTRUCTION INPUTS")
print("=" * 120)

proxy_raw = yf.download(
    [BH_TICKER, FX_TICKER],
    start="2020-11-20",
    interval="1d",
    auto_adjust=False,
    progress=False,
    threads=True,
    group_by="column"
)

if not isinstance(proxy_raw.columns, pd.MultiIndex):
    raise ValueError(
        "Unexpected Yahoo proxy-data format."
    )


# ============================================================
# 3. EXTRACT ADJUSTED CLOSE
# ============================================================

if "Adj Close" not in proxy_raw.columns.get_level_values(0):
    raise ValueError(
        "Adjusted Close unavailable for proxy inputs."
    )

proxy_adj = (
    proxy_raw["Adj Close"]
    .copy()
    .sort_index()
)

brk = (
    proxy_adj[BH_TICKER]
    .dropna()
    .rename("BRK")
)

usdzar = (
    proxy_adj[FX_TICKER]
    .dropna()
    .rename("USDZAR")
)


# ============================================================
# 4. ALIGN CALENDARS
#
# Use union of dates.
# Forward-fill PRICE information only.
#
# This is necessary because:
# US and FX calendars differ slightly.
#
# Returns are calculated AFTER alignment.
# ============================================================

proxy_prices = pd.concat(
    [brk, usdzar],
    axis=1
).sort_index()

proxy_prices = (
    proxy_prices
    .ffill()
)

proxy_prices = (
    proxy_prices
    .loc[
        proxy_prices.index >= LISTING_DATE
    ]
)

proxy_prices = (
    proxy_prices
    .dropna(
        subset=["BRK", "USDZAR"]
    )
)


# ============================================================
# 5. CONSTRUCT ZAR ECONOMIC PRICE INDEX
#
# BRK.B in USD × ZAR per USD
#
# gives a ZAR-denominated Berkshire exposure.
# ============================================================

proxy_prices["BRK_ZAR"] = (
    proxy_prices["BRK"]
    *
    proxy_prices["USDZAR"]
)


# ============================================================
# 6. CALCULATE DAILY ECONOMIC RETURNS
# ============================================================

proxy_prices["Gross_Return"] = (
    proxy_prices["BRK_ZAR"]
    .pct_change(
        fill_method=None
    )
)


# ============================================================
# 7. APPLY 1% ANNUAL FEE DRAG
#
# Convert 1% effective annual fee to daily equivalent.
# ============================================================

daily_fee = (
    (1 - ANNUAL_FEE) ** (1 / 365.25)
    - 1
)

proxy_prices["Net_Return"] = (
    (1 + proxy_prices["Gross_Return"])
    *
    (1 + daily_fee)
    - 1
)


# First observation has no return
proxy_returns = (
    proxy_prices["Net_Return"]
    .dropna()
)


# ============================================================
# 8. BUILD SYNTHETIC ADJUSTED-CLOSE INDEX
#
# Starting level = 100.
#
# Price level itself is arbitrary.
# Returns are what matter.
# ============================================================

synthetic_index = (
    100
    *
    (1 + proxy_returns)
    .cumprod()
)

synthetic_index.name = (
    BHETNC_TICKER
)


# Add starting observation
first_proxy_date = (
    proxy_prices.index[0]
)

synthetic_index = pd.concat(
    [
        pd.Series(
            [100.0],
            index=[
                first_proxy_date
            ],
            name=BHETNC_TICKER
        ),
        synthetic_index
    ]
)

synthetic_index = (
    synthetic_index[
        ~synthetic_index
        .index
        .duplicated(
            keep="last"
        )
    ]
    .sort_index()
)


# ============================================================
# 9. INSERT INTO CLEAN HISTORICAL DATA
#
# IMPORTANT:
# We replace ONLY BHETNC historical data.
#
# Actual BHETNC intraday pricing remains untouched.
# ============================================================

# Expand adj_clean index if necessary
combined_index = (
    adj_clean.index
    .union(
        synthetic_index.index
    )
    .sort_values()
)

adj_clean = (
    adj_clean
    .reindex(
        combined_index
    )
)

adj_clean.loc[
    synthetic_index.index,
    BHETNC_TICKER
] = (
    synthetic_index.values
)


# Remove any Yahoo BHETNC values outside our synthetic
# history to ensure the series is internally consistent.

mask_outside_proxy = (
    ~adj_clean.index.isin(
        synthetic_index.index
    )
)

adj_clean.loc[
    mask_outside_proxy,
    BHETNC_TICKER
] = np.nan


# ============================================================
# 10. SANITY CHECK BHETNC SERIES
# ============================================================

bh_series = (
    adj_clean[
        BHETNC_TICKER
    ]
    .dropna()
    .sort_index()
)

bh_returns = (
    bh_series
    .pct_change(
        fill_method=None
    )
    .dropna()
)

print("\n" + "=" * 120)
print("BHETNC SYNTHETIC HISTORY")
print("=" * 120)

print(
    f"First observation:   "
    f"{bh_series.index[0].date()}"
)

print(
    f"Last observation:    "
    f"{bh_series.index[-1].date()}"
)

print(
    f"Observations:        "
    f"{len(bh_series):,}"
)

print(
    f"Maximum daily gain:  "
    f"{bh_returns.max():+.2%}"
)

print(
    f"Maximum daily loss:  "
    f"{bh_returns.min():+.2%}"
)

print(
    f"Daily fee equivalent:"
    f" {daily_fee:.8%}"
)


# ============================================================
# 11. RECOMPUTE HORIZON RETURNS
# ============================================================

def bh_period_return(target_date, tolerance_days=10):

    s = bh_series

    candidates = (
        s[
            s.index >= target_date
        ]
    )

    if candidates.empty:
        return np.nan

    start_date = (
        candidates.index[0]
    )

    gap = (
        start_date.normalize()
        -
        target_date.normalize()
    ).days

    if gap > tolerance_days:
        return np.nan

    start_value = float(
        candidates.iloc[0]
    )

    end_value = float(
        s.iloc[-1]
    )

    if (
        start_value <= 0
        or
        s.index[-1] <= start_date
    ):
        return np.nan

    return (
        end_value /
        start_value
        - 1
    )


hist_end = (
    bh_series.index[-1]
    .normalize()
)

month_start = (
    hist_end.replace(day=1)
)

year_start = pd.Timestamp(
    hist_end.year,
    1,
    1
)

one_year_start = (
    hist_end -
    pd.DateOffset(years=1)
)

three_year_start = (
    hist_end -
    pd.DateOffset(years=3)
)

five_year_start = (
    hist_end -
    pd.DateOffset(years=5)
)


bh_periods = {
    "M/M":
        bh_period_return(
            month_start
        ),

    "YTD":
        bh_period_return(
            year_start
        ),

    "1Y":
        bh_period_return(
            one_year_start
        ),

    "3Y":
        bh_period_return(
            three_year_start
        ),

    "5Y":
        bh_period_return(
            five_year_start
        )
}


print("\n" + "=" * 120)
print("BHETNC HISTORICAL RETURN CHECK")
print("=" * 120)

for period, value in bh_periods.items():

    if pd.isna(value):

        display = "–"

    else:

        display = (
            f"{value:+.2%}"
        )

    print(
        f"{period:<5}: {display}"
    )


# ============================================================
# 12. UPDATE CLEAN HISTORICAL RETURN TABLE
#
# Recalculate ALL assets so downstream dashboard object
# is internally consistent.
# ============================================================

clean_history_rows = []


for name, row in master.iterrows():

    ticker = row["ticker"]


    clean_history_rows.append({

        "Holding":
            name,

        "Ticker":
            ticker,

        "Bucket":
            row["bucket"],

        "M/M":
            clean_period_return(
                ticker,
                month_start
            ),

        "YTD":
            clean_period_return(
                ticker,
                year_start
            ),

        "1Y":
            clean_period_return(
                ticker,
                one_year_start
            ),

        "3Y":
            clean_period_return(
                ticker,
                three_year_start
            ),

        "5Y":
            clean_period_return(
                ticker,
                five_year_start
            )
    })


historical_returns_clean = pd.DataFrame(
    clean_history_rows
)


# ============================================================
# 13. MERGE WITH ACTUAL LIVE RETURNS
#
# BHETNC Today remains the REAL BHETNC.JO quote.
# ============================================================

final_asset_returns = (

    live_valid[
        [
            "Holding",
            "Ticker",
            "Bucket",
            "Today_Return",
            "Last_Update"
        ]
    ]

    .merge(
        historical_returns_clean,

        on=[
            "Holding",
            "Ticker",
            "Bucket"
        ],

        how="left"
    )
)


# ============================================================
# 14. DISPLAY FINAL TABLE
# ============================================================

def pct_display(x):

    if pd.isna(x):
        return "–"

    return f"{x:+.2%}"


final_display = (
    final_asset_returns.copy()
)


for col in [
    "Today_Return",
    "M/M",
    "YTD",
    "1Y",
    "3Y",
    "5Y"
]:

    final_display[col] = (
        final_display[col]
        .apply(
            pct_display
        )
    )


final_display = (
    final_display.rename(
        columns={
            "Today_Return":
                "Today"
        }
    )
)


print("\n" + "=" * 135)
print("UPDATED CLEAN RETURN DASHBOARD")
print("=" * 135)


print(

    final_display[
        [
            "Holding",
            "Bucket",
            "Today",
            "M/M",
            "YTD",
            "1Y",
            "3Y",
            "5Y",
            "Last_Update"
        ]
    ]

    .to_string(
        index=False
    )
)


# ============================================================
# 15. BHETNC FINAL CONFIRMATION
# ============================================================

print("\n" + "=" * 120)
print("BHETNC DATA SOURCING")
print("=" * 120)

print("""
TODAY:
    Actual BHETNC.JO Yahoo intraday market price.

HISTORICAL:
    Berkshire Hathaway Class B (BRK-B)
    × USD/ZAR (USDZAR=X)
    less 1.00% annual fee drag.

Synthetic history begins at BHETNC's actual listing date.

No proxy is used for today's BHETNC return or portfolio valuation.
""")


print("=" * 120)
print("CELL 5 COMPLETE")
print("=" * 120)
```

```python
# ============================================================
# CELL 8 — FINAL BHETNC HISTORICAL HORIZON ENGINE
#
# TODAY:
#   Actual BHETNC.JO intraday return
#
# HISTORICAL HORIZONS:
#   BRK-B return × USD/ZAR return × fee drag
#
# No synthetic daily BHETNC series.
# ============================================================

import yfinance as yf
import pandas as pd
import numpy as np


# ============================================================
# 1. DOWNLOAD INPUTS
# ============================================================

proxy = yf.download(
    ["BRK-B", "USDZAR=X"],
    start="2020-11-20",
    interval="1d",
    auto_adjust=False,
    progress=False,
    threads=True
)

if not isinstance(proxy.columns, pd.MultiIndex):
    raise ValueError("Unexpected Yahoo data structure.")

proxy_adj = (
    proxy["Adj Close"]
    .copy()
    .sort_index()
)

brk = proxy_adj["BRK-B"].dropna()
fx  = proxy_adj["USDZAR=X"].dropna()


# ============================================================
# 2. ENDPOINT VALIDATION
#
# We do NOT clean/interpolate FX.
#
# A selected FX endpoint is rejected if its level represents
# an implausible >10% move from BOTH nearby valid observations.
# ============================================================

def endpoint_is_valid(series, date, threshold=0.10):

    s = series.dropna().sort_index()

    if date not in s.index:
        return False

    pos = s.index.get_loc(date)

    # Need surrounding observations
    if pos == 0 or pos == len(s) - 1:
        return True

    value = float(s.iloc[pos])
    prev_value = float(s.iloc[pos - 1])
    next_value = float(s.iloc[pos + 1])

    move_prev = abs(value / prev_value - 1)
    move_next = abs(value / next_value - 1)

    # Reject isolated bad print
    if (
        move_prev > threshold
        and
        move_next > threshold
    ):
        return False

    return True


# ============================================================
# 3. FIND FIRST VALID OBSERVATION ON/AFTER TARGET
# ============================================================

def get_endpoint(series, target, tolerance_days=10,
                 validate_fx=False):

    s = series.dropna().sort_index()

    candidates = s[
        (s.index >= target) &
        (s.index <= target + pd.Timedelta(days=tolerance_days))
    ]

    for dt, value in candidates.items():

        if validate_fx:

            if not endpoint_is_valid(
                s,
                dt,
                threshold=0.10
            ):
                continue

        return dt, float(value)

    return None


# ============================================================
# 4. BHETNC PERIOD RETURN
#
# (1 + BRK return)
# × (1 + FX return)
# × fee survival factor
# - 1
# ============================================================

ANNUAL_FEE = 0.01


def bhetnc_period_return(target_start, target_end):

    # --------------------------
    # START
    # --------------------------

    brk_start = get_endpoint(
        brk,
        target_start,
        tolerance_days=10
    )

    fx_start = get_endpoint(
        fx,
        target_start,
        tolerance_days=10,
        validate_fx=True
    )


    # --------------------------
    # END
    # --------------------------

    brk_end = get_endpoint(
        brk,
        target_end,
        tolerance_days=10
    )

    fx_end = get_endpoint(
        fx,
        target_end,
        tolerance_days=10,
        validate_fx=True
    )


    if any(
        x is None
        for x in [
            brk_start,
            fx_start,
            brk_end,
            fx_end
        ]
    ):
        return np.nan


    # Require BRK and FX observations to be reasonably aligned
    start_dates = [
        brk_start[0],
        fx_start[0]
    ]

    end_dates = [
        brk_end[0],
        fx_end[0]
    ]


    if (
        abs(
            (start_dates[0] - start_dates[1]).days
        ) > 3
    ):
        return np.nan


    if (
        abs(
            (end_dates[0] - end_dates[1]).days
        ) > 3
    ):
        return np.nan


    # --------------------------
    # RETURNS
    # --------------------------

    brk_factor = (
        brk_end[1] /
        brk_start[1]
    )

    fx_factor = (
        fx_end[1] /
        fx_start[1]
    )


    # Actual elapsed time
    start_date = max(start_dates)
    end_date = max(end_dates)

    years = (
        (end_date - start_date).days /
        365.25
    )


    fee_factor = (
        (1 - ANNUAL_FEE) ** years
    )


    total_factor = (
        brk_factor
        *
        fx_factor
        *
        fee_factor
    )


    return (
        total_factor - 1
    )


# ============================================================
# 5. PERIOD DATES
# ============================================================

# Historical data currently ends on previous completed session.
hist_end = min(
    brk.index[-1],
    fx.index[-1]
).normalize()


month_start = (
    hist_end.replace(day=1)
)

year_start = pd.Timestamp(
    hist_end.year,
    1,
    1
)

one_year_start = (
    hist_end -
    pd.DateOffset(years=1)
)

three_year_start = (
    hist_end -
    pd.DateOffset(years=3)
)

five_year_start = (
    hist_end -
    pd.DateOffset(years=5)
)


# ============================================================
# 6. CALCULATE BHETNC HORIZONS
# ============================================================

bhetnc_results = {

    "M/M":
        bhetnc_period_return(
            month_start,
            hist_end
        ),

    "YTD":
        bhetnc_period_return(
            year_start,
            hist_end
        ),

    "1Y":
        bhetnc_period_return(
            one_year_start,
            hist_end
        ),

    "3Y":
        bhetnc_period_return(
            three_year_start,
            hist_end
        ),

    "5Y":
        bhetnc_period_return(
            five_year_start,
            hist_end
        )
}


# ============================================================
# 7. PRINT BHETNC RESULTS
# ============================================================

print("=" * 100)
print("BHETNC — FINAL HISTORICAL HORIZON CALCULATION")
print("=" * 100)

print(f"Historical end date: {hist_end.date()}")
print()

for period, value in bhetnc_results.items():

    display = (
        "–"
        if pd.isna(value)
        else f"{value:+.2%}"
    )

    print(
        f"{period:<4}: {display}"
    )


# ============================================================
# 8. UPDATE EXISTING HISTORICAL TABLE
#
# Everything except BHETNC remains exactly as Cell 4 produced.
# ============================================================

historical_returns_final = (
    historical_returns_clean.copy()
)

mask = (
    historical_returns_final["Ticker"]
    ==
    "BHETNC.JO"
)

for period, value in bhetnc_results.items():

    historical_returns_final.loc[
        mask,
        period
    ] = value


# ============================================================
# 9. REBUILD FINAL ASSET TABLE
#
# Today still comes from live_valid:
# i.e. REAL BHETNC.JO.
# ============================================================

final_asset_returns = (

    live_valid[
        [
            "Holding",
            "Ticker",
            "Bucket",
            "Today_Return",
            "Last_Update"
        ]
    ]

    .merge(
        historical_returns_final,

        on=[
            "Holding",
            "Ticker",
            "Bucket"
        ],

        how="left"
    )
)


# ============================================================
# 10. DISPLAY
# ============================================================

def pct_display(x):

    if pd.isna(x):
        return "–"

    return f"{x:+.2%}"


display = final_asset_returns.copy()

for col in [
    "Today_Return",
    "M/M",
    "YTD",
    "1Y",
    "3Y",
    "5Y"
]:

    display[col] = (
        display[col]
        .apply(pct_display)
    )


display = display.rename(
    columns={
        "Today_Return":
            "Today"
    }
)


print("\n" + "=" * 135)
print("FINAL ASSET RETURN TABLE")
print("=" * 135)

print(
    display[
        [
            "Holding",
            "Bucket",
            "Today",
            "M/M",
            "YTD",
            "1Y",
            "3Y",
            "5Y",
            "Last_Update"
        ]
    ]
    .to_string(index=False)
)


# ============================================================
# 11. FINAL BHETNC METHOD
# ============================================================

print("\n" + "=" * 100)
print("BHETNC METHOD")
print("=" * 100)

print("""
Today:
    Actual BHETNC.JO market return.

M/M, YTD, 1Y, 3Y, 5Y:
    Point-to-point BRK-B × USD/ZAR return,
    less 1.00% annual fee drag.

No synthetic daily BHETNC series is used.

USD/ZAR endpoints are checked for isolated >10% bad prints.
No FX observations are modified or interpolated.
""")
```

```python
# ============================================================
# CREATE STREAMLIT PORTFOLIO MONITOR — app.py
# ============================================================

app_code = r'''
import streamlit as st
import pandas as pd
import numpy as np
import yfinance as yf
from pathlib import Path
from datetime import datetime
from zoneinfo import ZoneInfo
import plotly.express as px
import plotly.graph_objects as go


# ============================================================
# PAGE
# ============================================================

st.set_page_config(
    page_title="Portfolio Monitor",
    page_icon="📈",
    layout="wide",
    initial_sidebar_state="collapsed"
)

SAST = ZoneInfo("Africa/Johannesburg")
NOW = datetime.now(SAST)

PORTFOLIO_START = pd.Timestamp("2026-09-28")
NAV_FILE = Path("portfolio_history.csv")


# ============================================================
# STYLE
# ============================================================

st.markdown("""
<style>

.block-container {
    padding-top: 1.4rem;
    padding-bottom: 3rem;
    max-width: 1450px;
}

[data-testid="stMetric"] {
    border: 1px solid rgba(128,128,128,0.25);
    border-radius: 10px;
    padding: 15px 18px;
}

.small-note {
    opacity: 0.70;
    font-size: 0.84rem;
}

.section-note {
    opacity: 0.70;
    margin-top: -10px;
    margin-bottom: 15px;
}

</style>
""", unsafe_allow_html=True)


# ============================================================
# PORTFOLIO
# ============================================================

# Quantities / portfolio composition established at inception.
#
# IMPORTANT:
# Historical individual-security returns below are NOT used
# to create historical portfolio performance.

portfolio = {
    # ---------------- MAIN ----------------

    "Anglo American": {
        "ticker": "AGL.JO",
        "qty": 2491,
        "bucket": "Main"
    },

    "AB InBev": {
        "ticker": "ANH.JO",
        "qty": 500,
        "bucket": "Main"
    },

    "BHP Group": {
        "ticker": "BHG.JO",
        "qty": 600,
        "bucket": "Main"
    },

    "Bidcorp": {
        "ticker": "BID.JO",
        "qty": 3000,
        "bucket": "Main"
    },

    "British American Tobacco": {
        "ticker": "BTI.JO",
        "qty": 1840,
        "bucket": "Main"
    },

    "Bidvest": {
        "ticker": "BVT.JO",
        "qty": 5000,
        "bucket": "Main"
    },

    "Richemont": {
        "ticker": "CFR.JO",
        "qty": 565,
        "bucket": "Main"
    },

    "eMedia Holdings N": {
        "ticker": "EMN.JO",
        "qty": 2517,
        "bucket": "Main"
    },

    "Remgro": {
        "ticker": "REM.JO",
        "qty": 6000,
        "bucket": "Main"
    },

    "Reinet": {
        "ticker": "RNI.JO",
        "qty": 2800,
        "bucket": "Main"
    },

    "Shoprite": {
        "ticker": "SHP.JO",
        "qty": 5500,
        "bucket": "Main"
    },

    "Tiger Brands": {
        "ticker": "TBS.JO",
        "qty": 3000,
        "bucket": "Main"
    },

    "Valterra Platinum": {
        "ticker": "VAL.JO",
        "qty": 250,
        "bucket": "Main"
    },

    # ---------------- SUBSET ----------------

    "FNB Berkshire ETN": {
        "ticker": "BHETNC.JO",
        "qty": 2500,
        "bucket": "Subset"
    },

    "10X S&P 500": {
        "ticker": "CSP500.JO",
        "qty": 855,
        "bucket": "Subset"
    },

    "Exxaro": {
        "ticker": "EXX.JO",
        "qty": 120,
        "bucket": "Subset"
    },

    "NewGold": {
        "ticker": "GLD.JO",
        "qty": 77,
        "bucket": "Subset"
    },

    "10X Active Income": {
        "ticker": "INCOME.JO",
        "qty": 11300,
        "bucket": "Subset"
    },

    "Satrix Capped All Share": {
        "ticker": "STXCAP.JO",
        "qty": 2100,
        "bucket": "Subset"
    },

    "Satrix MSCI EM ESG": {
        "ticker": "STXEME.JO",
        "qty": 840,
        "bucket": "Subset"
    },

    "Satrix STOXX Europe": {
        "ticker": "STXEUR.JO",
        "qty": 2060,
        "bucket": "Subset"
    },

    "Satrix Global Bond": {
        "ticker": "STXGBD.JO",
        "qty": 1730,
        "bucket": "Subset"
    },

    "Satrix SA Bond": {
        "ticker": "STXGOV.JO",
        "qty": 13000,
        "bucket": "Subset"
    },

    "Satrix S&P 500 Feeder": {
        "ticker": "STX500.JO",
        "qty": 210,
        "bucket": "Subset"
    }
}


master = pd.DataFrame(portfolio).T
master.index.name = "Holding"

tickers = master["ticker"].tolist()


# ============================================================
# YAHOO DOWNLOAD
#
# 15-minute cache = dashboard refresh cadence.
# ============================================================

@st.cache_data(ttl=900)
def download_market_data(tickers):

    intraday = yf.download(
        tickers,
        period="5d",
        interval="5m",
        auto_adjust=False,
        progress=False,
        threads=True
    )

    history = yf.download(
        tickers,
        period="5y",
        interval="1d",
        auto_adjust=False,
        progress=False,
        threads=True
    )

    return intraday, history


intraday_raw, history_raw = download_market_data(tickers)


# ============================================================
# EXTRACT DATA
# ============================================================

def get_field(df, field):

    if isinstance(df.columns, pd.MultiIndex):

        if field not in df.columns.get_level_values(0):
            return pd.DataFrame()

        return df[field].copy()

    return pd.DataFrame()


intraday_close = get_field(
    intraday_raw,
    "Close"
)

history_close = get_field(
    history_raw,
    "Close"
)

adj_close = get_field(
    history_raw,
    "Adj Close"
)


# ============================================================
# JSE UNIT NORMALISATION
#
# Yahoo occasionally mixes cents and rand for JSE history.
#
# Repair ONLY obvious ~100x discontinuities.
# ============================================================

def normalise_jse_units(series):

    s = series.copy().sort_index()

    valid = s.dropna()

    if len(valid) < 2:
        return s

    cleaned = valid.copy()

    previous = float(cleaned.iloc[0])

    for i in range(1, len(cleaned)):

        current = float(cleaned.iloc[i])

        if previous <= 0 or current <= 0:
            previous = current
            continue

        ratio = current / previous

        if ratio > 50:

            candidate = current / 100

            if 0.50 <= candidate / previous <= 1.50:
                cleaned.iloc[i] = candidate

        elif ratio < 0.02:

            candidate = current * 100

            if 0.50 <= candidate / previous <= 1.50:
                cleaned.iloc[i] = candidate

        previous = float(cleaned.iloc[i])

    s.loc[cleaned.index] = cleaned

    return s


for ticker in adj_close.columns:

    adj_close[ticker] = normalise_jse_units(
        adj_close[ticker]
    )


# ============================================================
# LIVE PERFORMANCE
# ============================================================

live_rows = []

for holding, row in master.iterrows():

    ticker = row["ticker"]
    qty = float(row["qty"])
    bucket = row["bucket"]

    if ticker not in intraday_close.columns:
        continue

    s = intraday_close[ticker].dropna()

    if s.empty:
        continue

    # Yahoo timestamps
    idx = s.index

    if idx.tz is None:
        idx = idx.tz_localize("UTC")

    local_dates = idx.tz_convert(SAST).date

    unique_dates = sorted(set(local_dates))

    if len(unique_dates) < 2:
        continue

    today = unique_dates[-1]
    prior_day = unique_dates[-2]

    today_mask = local_dates == today
    prior_mask = local_dates == prior_day

    today_s = s.iloc[np.where(today_mask)[0]]
    prior_s = s.iloc[np.where(prior_mask)[0]]

    if today_s.empty or prior_s.empty:
        continue

    latest_raw = float(today_s.iloc[-1])
    prior_raw = float(prior_s.iloc[-1])

    # JSE Yahoo quotes are generally in cents.
    latest_price = latest_raw / 100
    prior_price = prior_raw / 100

    ret = latest_price / prior_price - 1

    opening_value = qty * prior_price
    current_value = qty * latest_price

    pnl = current_value - opening_value

    timestamp = idx[np.where(today_mask)[0][-1]]

    timestamp = timestamp.tz_convert(SAST)

    live_rows.append({
        "Holding": holding,
        "Ticker": ticker,
        "Bucket": bucket,
        "Qty": qty,
        "Prior_Close": prior_price,
        "Latest": latest_price,
        "Today_Return": ret,
        "Opening_Value": opening_value,
        "Current_Value": current_value,
        "PnL": pnl,
        "Last_Update": timestamp
    })


live = pd.DataFrame(live_rows)


# ============================================================
# PORTFOLIO SUMMARY
# ============================================================

def portfolio_stats(bucket=None):

    if bucket is None:
        x = live.copy()
        name = "Combined"

    else:
        x = live[live["Bucket"] == bucket].copy()
        name = bucket

    opening = x["Opening_Value"].sum()
    current = x["Current_Value"].sum()
    pnl = x["PnL"].sum()

    ret = (
        pnl / opening
        if opening
        else np.nan
    )

    return {
        "Portfolio": name,
        "Opening": opening,
        "Current": current,
        "PnL": pnl,
        "Return": ret
    }


combined = portfolio_stats()
main = portfolio_stats("Main")
subset = portfolio_stats("Subset")


# ============================================================
# HEADER
# ============================================================

st.title("Portfolio Monitor")

latest_timestamp = (
    live["Last_Update"].max()
    if not live.empty
    else None
)

if latest_timestamp is not None:

    st.markdown(
        f"""
        <div class="small-note">
        Latest market observation:
        {latest_timestamp.strftime("%d %b %Y %H:%M SAST")}
        · Market data cached for 15 minutes
        </div>
        """,
        unsafe_allow_html=True
    )


# ============================================================
# LIVE CARDS
# ============================================================

st.subheader("Live / Intraday Estimate")

st.markdown(
    """
    <div class="section-note">
    Ordinary intraday prices versus previous market close.
    Official daily portfolio performance is recorded from
    Adjusted Close after EOD.
    </div>
    """,
    unsafe_allow_html=True
)


cols = st.columns(3)

stats = [
    combined,
    main,
    subset
]


for col, stat in zip(cols, stats):

    with col:

        st.metric(
            label=stat["Portfolio"],
            value=f"R{stat['Current']:,.0f}",
            delta=f"{stat['Return']:+.2%}"
        )

        st.caption(
            f"Today P&L: R{stat['PnL']:+,.0f}"
        )


# ============================================================
# PORTFOLIO PERFORMANCE
#
# ONLY ACTUAL OBSERVATIONS FROM 28 SEP 2026 FORWARD.
# ============================================================

st.divider()

st.subheader(
    "Actual Portfolio Performance"
)

st.markdown(
    """
    <div class="section-note">
    Performance record begins 28 September 2026.
    No historical portfolio performance is backfilled.
    </div>
    """,
    unsafe_allow_html=True
)


# ============================================================
# LOAD STORED DAILY RETURNS
# ============================================================

if NAV_FILE.exists():

    nav = pd.read_csv(
        NAV_FILE,
        parse_dates=["Date"]
    )

else:

    nav = pd.DataFrame(
        columns=[
            "Date",
            "Combined",
            "Main",
            "Subset"
        ]
    )


if not nav.empty:

    nav = (
        nav[
            nav["Date"] >= PORTFOLIO_START
        ]
        .sort_values("Date")
        .drop_duplicates(
            subset=["Date"],
            keep="last"
        )
    )


# ============================================================
# CUMULATIVE RETURN CHART
# ============================================================

if nav.empty:

    st.info(
        "The official portfolio track record begins at today's "
        "EOD close. The cumulative-return chart will populate "
        "as official daily returns are recorded."
    )

else:

    cumulative = nav.copy()

    for col in [
        "Combined",
        "Main",
        "Subset"
    ]:

        cumulative[col] = (
            (1 + cumulative[col])
            .cumprod()
            - 1
        )


    selected = st.radio(
        "Portfolio",
        ["Combined", "Main", "Subset"],
        horizontal=True
    )


    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=cumulative["Date"],
            y=cumulative[selected] * 100,
            mode="lines",
            name=selected,
            hovertemplate=(
                "%{x|%d %b %Y}<br>"
                "Cumulative return: %{y:.2f}%"
                "<extra></extra>"
            )
        )
    )

    fig.update_layout(
        height=430,
        margin=dict(
            l=20,
            r=20,
            t=20,
            b=20
        ),
        yaxis_title="Cumulative Return (%)",
        xaxis_title=None,
        hovermode="x unified"
    )

    st.plotly_chart(
        fig,
        use_container_width=True
    )


    # ========================================================
    # DAILY RETURN HISTORY
    # ========================================================

    with st.expander(
        "Daily Portfolio Return History"
    ):

        daily_display = nav.copy()

        for col in [
            "Combined",
            "Main",
            "Subset"
        ]:

            daily_display[col] = (
                daily_display[col]
                .map(
                    lambda x:
                    f"{x:+.2%}"
                )
            )

        st.dataframe(
            daily_display.sort_values(
                "Date",
                ascending=False
            ),
            use_container_width=True,
            hide_index=True
        )


# ============================================================
# TODAY'S ASSET RETURNS — VISUAL
# ============================================================

st.divider()

st.subheader(
    "Today's Individual Asset Performance"
)


if not live.empty:

    chart_data = (
        live[
            [
                "Holding",
                "Bucket",
                "Today_Return"
            ]
        ]
        .copy()
        .sort_values(
            "Today_Return"
        )
    )


    fig_assets = px.bar(
        chart_data,
        x="Today_Return",
        y="Holding",
        orientation="h",
        color="Bucket",
        labels={
            "Today_Return":
                "Today's Return",
            "Holding":
                ""
        },
        hover_data={
            "Today_Return":
                ":.2%",
            "Bucket":
                True
        }
    )


    fig_assets.update_layout(
        height=700,
        xaxis_tickformat=".1%",
        legend_title_text="Portfolio",
        margin=dict(
            l=20,
            r=20,
            t=20,
            b=20
        )
    )


    st.plotly_chart(
        fig_assets,
        use_container_width=True
    )


# ============================================================
# LIVE HOLDING TABLE
# ============================================================

with st.expander(
    "Live Holding Detail"
):

    live_display = (
        live[
            [
                "Holding",
                "Bucket",
                "Today_Return",
                "Current_Value",
                "PnL",
                "Last_Update"
            ]
        ]
        .copy()
    )


    live_display[
        "Today"
    ] = (
        live_display[
            "Today_Return"
        ]
        .map(
            lambda x:
            f"{x:+.2%}"
        )
    )


    live_display[
        "Value"
    ] = (
        live_display[
            "Current_Value"
        ]
        .map(
            lambda x:
            f"R{x:,.0f}"
        )
    )


    live_display[
        "P&L"
    ] = (
        live_display[
            "PnL"
        ]
        .map(
            lambda x:
            f"R{x:+,.0f}"
        )
    )


    live_display[
        "Updated"
    ] = (
        live_display[
            "Last_Update"
        ]
        .map(
            lambda x:
            x.strftime(
                "%H:%M"
            )
        )
    )


    st.dataframe(
        live_display[
            [
                "Holding",
                "Bucket",
                "Today",
                "Value",
                "P&L",
                "Updated"
            ]
        ],
        use_container_width=True,
        hide_index=True
    )


# ============================================================
# INDIVIDUAL SECURITY HISTORICAL RETURNS
#
# IMPORTANT:
# COMPLETELY SEPARATE FROM PORTFOLIO PERFORMANCE.
# ============================================================

st.divider()

st.subheader(
    "Individual Asset Historical Returns"
)

st.markdown(
    """
    <div class="section-note">
    Reference information for individual securities only.
    These figures are not historical portfolio returns and
    are not used to construct the portfolio track record.
    </div>
    """,
    unsafe_allow_html=True
)


# ============================================================
# HISTORICAL PERIOD HELPERS
# ============================================================

def endpoint_return(
    series,
    start,
    end,
    tolerance=10
):

    s = (
        series
        .dropna()
        .sort_index()
    )

    if s.empty:
        return np.nan

    starts = s[
        (s.index >= start) &
        (
            s.index <=
            start +
            pd.Timedelta(days=tolerance)
        )
    ]

    ends = s[
        s.index <= end
    ]

    if starts.empty or ends.empty:
        return np.nan

    start_date = starts.index[0]
    end_date = ends.index[-1]

    if end_date <= start_date:
        return np.nan

    return (
        float(ends.iloc[-1])
        /
        float(starts.iloc[0])
        - 1
    )


hist_end = adj_close.index.max().normalize()

periods = {
    "M/M":
        hist_end.replace(day=1),

    "YTD":
        pd.Timestamp(
            hist_end.year,
            1,
            1
        ),

    "1Y":
        hist_end -
        pd.DateOffset(years=1),

    "3Y":
        hist_end -
        pd.DateOffset(years=3),

    "5Y":
        hist_end -
        pd.DateOffset(years=5)
}


# ============================================================
# BHETNC HISTORICAL ENDPOINT PROXY
# ============================================================

@st.cache_data(ttl=86400)
def download_bhetnc_inputs():

    x = yf.download(
        [
            "BRK-B",
            "USDZAR=X"
        ],
        start="2020-11-20",
        interval="1d",
        auto_adjust=False,
        progress=False
    )

    return x


bh_inputs = download_bhetnc_inputs()

if isinstance(
    bh_inputs.columns,
    pd.MultiIndex
):

    bh_adj = (
        bh_inputs["Adj Close"]
        .copy()
    )

else:

    bh_adj = pd.DataFrame()


def valid_fx_endpoint(
    s,
    date,
    threshold=0.10
):

    s = s.dropna().sort_index()

    if date not in s.index:
        return False

    pos = s.index.get_loc(date)

    if (
        pos == 0
        or
        pos == len(s) - 1
    ):
        return True

    value = float(s.iloc[pos])
    prev = float(s.iloc[pos - 1])
    nxt = float(s.iloc[pos + 1])

    return not (
        abs(value / prev - 1) > threshold
        and
        abs(value / nxt - 1) > threshold
    )


def get_valid_endpoint(
    s,
    target,
    tolerance=10,
    validate_fx=False
):

    s = s.dropna().sort_index()

    candidates = s[
        (s.index >= target) &
        (
            s.index <=
            target +
            pd.Timedelta(days=tolerance)
        )
    ]

    for dt, value in candidates.items():

        if (
            validate_fx
            and
            not valid_fx_endpoint(
                s,
                dt
            )
        ):
            continue

        return dt, float(value)

    return None


def bhetnc_return(start, end):

    if bh_adj.empty:
        return np.nan

    brk = bh_adj["BRK-B"]
    fx = bh_adj["USDZAR=X"]

    bs = get_valid_endpoint(
        brk,
        start
    )

    fs = get_valid_endpoint(
        fx,
        start,
        validate_fx=True
    )

    be = get_valid_endpoint(
        brk,
        end
    )

    fe = get_valid_endpoint(
        fx,
        end,
        validate_fx=True
    )

    if any(
        x is None
        for x in [
            bs,
            fs,
            be,
            fe
        ]
    ):
        return np.nan

    if (
        abs(
            (bs[0] - fs[0]).days
        ) > 3
        or
        abs(
            (be[0] - fe[0]).days
        ) > 3
    ):
        return np.nan

    years = (
        (
            max(be[0], fe[0])
            -
            max(bs[0], fs[0])
        ).days
        /
        365.25
    )

    factor = (
        (be[1] / bs[1])
        *
        (fe[1] / fs[1])
        *
        ((1 - 0.01) ** years)
    )

    return factor - 1


# ============================================================
# BUILD INDIVIDUAL-ASSET REFERENCE TABLE
# ============================================================

reference_rows = []

for holding, row in master.iterrows():

    ticker = row["ticker"]

    result = {
        "Asset": holding,
        "Portfolio": row["bucket"]
    }

    for period, start in periods.items():

        if ticker == "BHETNC.JO":

            result[period] = (
                bhetnc_return(
                    start,
                    hist_end
                )
            )

        elif ticker in adj_close.columns:

            result[period] = (
                endpoint_return(
                    adj_close[ticker],
                    start,
                    hist_end
                )
            )

        else:

            result[period] = np.nan

    reference_rows.append(result)


reference = pd.DataFrame(
    reference_rows
)


# ============================================================
# REFERENCE TABLE DISPLAY
# ============================================================

reference_display = (
    reference.copy()
)


for col in [
    "M/M",
    "YTD",
    "1Y",
    "3Y",
    "5Y"
]:

    reference_display[col] = (
        reference_display[col]
        .map(
            lambda x:
            "–"
            if pd.isna(x)
            else f"{x:+.2%}"
        )
    )


st.dataframe(
    reference_display,
    use_container_width=True,
    hide_index=True
)


st.caption(
    "BHETNC historical reference returns are reconstructed "
    "from Berkshire Hathaway Class B × USD/ZAR less the "
    "1.00% annual fee. Current/intraday BHETNC performance "
    "uses the actual BHETNC.JO market instrument."
)


# ============================================================
# FOOTER
# ============================================================

st.divider()

st.caption(
    "Portfolio performance record inception: "
    "28 September 2026. Historical individual-security "
    "statistics are reference information only."
)
'''

with open("app.py", "w") as f:
    f.write(app_code)

print("app.py created successfully.")
print("Lines:", len(app_code.splitlines()))
```

```python
%%writefile eod_update.py

import pandas as pd
import numpy as np
import yfinance as yf
from pathlib import Path
from datetime import datetime
from zoneinfo import ZoneInfo

SAST = ZoneInfo("Africa/Johannesburg")

PORTFOLIO_START = pd.Timestamp("2026-09-28")
NAV_FILE = Path("portfolio_history.csv")


# ============================================================
# PORTFOLIO HELD FOR 28 SEP 2026
# ============================================================

portfolio = {
    # MAIN
    "Anglo American": ("AGL.JO", 2491, "Main"),
    "AB InBev": ("ANH.JO", 500, "Main"),
    "BHP Group": ("BHG.JO", 600, "Main"),
    "Bidcorp": ("BID.JO", 3000, "Main"),
    "British American Tobacco": ("BTI.JO", 1840, "Main"),
    "Bidvest": ("BVT.JO", 5000, "Main"),
    "Richemont": ("CFR.JO", 565, "Main"),
    "eMedia Holdings N": ("EMN.JO", 2517, "Main"),
    "Remgro": ("REM.JO", 6000, "Main"),
    "Reinet": ("RNI.JO", 2800, "Main"),
    "Shoprite": ("SHP.JO", 5500, "Main"),
    "Tiger Brands": ("TBS.JO", 3000, "Main"),
    "Valterra Platinum": ("VAL.JO", 250, "Main"),

    # SUBSET
    "FNB Berkshire ETN": ("BHETNC.JO", 2500, "Subset"),
    "10X S&P 500": ("CSP500.JO", 855, "Subset"),
    "Exxaro": ("EXX.JO", 120, "Subset"),
    "NewGold": ("GLD.JO", 77, "Subset"),
    "10X Active Income": ("INCOME.JO", 11300, "Subset"),
    "Satrix Capped All Share": ("STXCAP.JO", 2100, "Subset"),
    "Satrix MSCI EM ESG": ("STXEME.JO", 840, "Subset"),
    "Satrix STOXX Europe": ("STXEUR.JO", 2060, "Subset"),
    "Satrix Global Bond": ("STXGBD.JO", 1730, "Subset"),
    "Satrix SA Bond": ("STXGOV.JO", 13000, "Subset"),
    "Satrix S&P 500 Feeder": ("STX500.JO", 210, "Subset"),
}

holdings = pd.DataFrame(
    [
        {
            "Holding": name,
            "Ticker": values[0],
            "Qty": values[1],
            "Bucket": values[2]
        }
        for name, values in portfolio.items()
    ]
)

tickers = holdings["Ticker"].tolist()


# ============================================================
# DOWNLOAD DAILY DATA
# ============================================================

print("Downloading daily Yahoo data...")

raw = yf.download(
    tickers,
    period="10d",
    interval="1d",
    auto_adjust=False,
    progress=False,
    threads=True
)

if not isinstance(raw.columns, pd.MultiIndex):
    raise RuntimeError("Unexpected Yahoo column structure.")

if "Adj Close" not in raw.columns.get_level_values(0):
    raise RuntimeError("Adjusted Close unavailable.")

adj = raw["Adj Close"].copy()
close = raw["Close"].copy()


# ============================================================
# FIND TWO MOST RECENT COMPLETE TRADING DATES
# ============================================================

dates = adj.dropna(how="all").index.sort_values()

if len(dates) < 2:
    raise RuntimeError("Not enough daily observations.")

current_date = pd.Timestamp(dates[-1]).normalize()
previous_date = pd.Timestamp(dates[-2]).normalize()

print("Previous trading date:", previous_date.date())
print("Current trading date: ", current_date.date())


if current_date < PORTFOLIO_START:
    raise RuntimeError(
        "Current observation predates portfolio inception."
    )


# ============================================================
# CALCULATE SECURITY TOTAL RETURNS
# ============================================================

rows = []

for _, h in holdings.iterrows():

    ticker = h["Ticker"]

    if ticker not in adj.columns:
        print(f"WARNING: {ticker} missing.")
        continue

    prev_adj = adj.loc[previous_date, ticker]
    curr_adj = adj.loc[current_date, ticker]

    prev_close = close.loc[previous_date, ticker]
    curr_close = close.loc[current_date, ticker]

    if pd.isna(prev_adj) or pd.isna(curr_adj):
        print(f"WARNING: {ticker} missing adjusted close.")
        continue

    # Yahoo JSE quote units are immaterial for returns.
    total_return = float(curr_adj / prev_adj - 1)

    # Beginning-of-day economic value.
    #
    # We use previous ordinary close × quantity to establish
    # actual portfolio capital at risk at the beginning of day.
    #
    # Unit conversion /100 because Yahoo JSE prices are quoted
    # in cents for these instruments.
    opening_value = float(h["Qty"] * prev_close / 100)

    rows.append({
        "Holding": h["Holding"],
        "Ticker": ticker,
        "Bucket": h["Bucket"],
        "Qty": h["Qty"],
        "Opening_Value": opening_value,
        "Total_Return": total_return
    })


detail = pd.DataFrame(rows)

if detail.empty:
    raise RuntimeError("No valid holdings.")


# ============================================================
# COVERAGE CHECK
# ============================================================

expected = set(tickers)
received = set(detail["Ticker"])

missing = sorted(expected - received)

if missing:
    raise RuntimeError(
        "STOP — missing holdings: " + ", ".join(missing)
    )


# ============================================================
# PORTFOLIO RETURN
#
# Beginning-of-day market-value weighting.
# ============================================================

def calc_portfolio_return(df):

    capital = df["Opening_Value"].sum()

    if capital <= 0:
        return np.nan

    pnl_equivalent = (
        df["Opening_Value"]
        *
        df["Total_Return"]
    ).sum()

    return float(pnl_equivalent / capital)


combined_return = calc_portfolio_return(detail)

main_return = calc_portfolio_return(
    detail[detail["Bucket"] == "Main"]
)

subset_return = calc_portfolio_return(
    detail[detail["Bucket"] == "Subset"]
)


# ============================================================
# OUTPUT
# ============================================================

print()
print("=" * 90)
print("OFFICIAL EOD PORTFOLIO RETURN")
print("=" * 90)

print(f"Date:      {current_date.date()}")
print(f"Combined:  {combined_return:+.4%}")
print(f"Main:      {main_return:+.4%}")
print(f"Subset:    {subset_return:+.4%}")

print()
print("These are Adjusted-Close total returns.")
print()


# ============================================================
# SAVE / UPDATE HISTORY
# ============================================================

new_row = pd.DataFrame([{
    "Date": current_date,
    "Combined": combined_return,
    "Main": main_return,
    "Subset": subset_return
}])


if NAV_FILE.exists():

    history = pd.read_csv(
        NAV_FILE,
        parse_dates=["Date"]
    )

    history = pd.concat(
        [history, new_row],
        ignore_index=True
    )

else:

    history = new_row.copy()


history = (
    history[
        history["Date"] >= PORTFOLIO_START
    ]
    .sort_values("Date")
    .drop_duplicates(
        subset=["Date"],
        keep="last"
    )
    .reset_index(drop=True)
)


history.to_csv(
    NAV_FILE,
    index=False
)


# ============================================================
# CUMULATIVE RETURN
# ============================================================

cum = history.copy()

for col in ["Combined", "Main", "Subset"]:

    cum[col] = (
        (1 + cum[col])
        .cumprod()
        - 1
    )


print("=" * 90)
print("STORED PORTFOLIO HISTORY")
print("=" * 90)

display_history = history.copy()

for col in ["Combined", "Main", "Subset"]:
    display_history[col] = display_history[col].map(
        lambda x: f"{x:+.4%}"
    )

print(display_history.to_string(index=False))


print()
print("=" * 90)
print("CUMULATIVE RETURN SINCE 28 SEP 2026")
print("=" * 90)

latest = cum.iloc[-1]

print(f"Combined: {latest['Combined']:+.4%}")
print(f"Main:     {latest['Main']:+.4%}")
print(f"Subset:   {latest['Subset']:+.4%}")

print()
print("portfolio_history.csv updated successfully.")
```

```python
from pathlib import Path
import py_compile

REPO = Path("/content/test-edv")
APP = REPO / "app.py"
EOD = REPO / "eod_update.py"

assert APP.exists(), f"Missing: {APP}"
assert EOD.exists(), f"Missing: {EOD}"

app = APP.read_text()
eod = EOD.read_text()

# ============================================================
# 1. EOD_UPDATE.PY
# Save each holding's official Adjusted Close as tomorrow's basis
# ============================================================

if 'BASIS_FILE = Path("portfolio_basis.csv")' not in eod:
    eod = eod.replace(
        'NAV_FILE = Path("portfolio_history.csv")',
        'NAV_FILE = Path("portfolio_history.csv")\n'
        'BASIS_FILE = Path("portfolio_basis.csv")'
    )

basis_marker = "# SAVE OFFICIAL EOD BASIS FOR NEXT TRADING DAY"

if basis_marker not in eod:

    history_marker = """# ============================================================
# SAVE / UPDATE HISTORY
# ============================================================"""

    assert history_marker in eod, (
        "STOP: Could not locate SAVE / UPDATE HISTORY "
        "section in eod_update.py"
    )

    basis_code = r'''
# ============================================================
# SAVE OFFICIAL EOD BASIS FOR NEXT TRADING DAY
# ============================================================

basis_rows = []

for _, h in holdings.iterrows():

    ticker = h["Ticker"]

    if ticker not in adj.columns:
        raise RuntimeError(
            f"STOP: {ticker} missing from adjusted-close data."
        )

    curr_adj = adj.loc[current_date, ticker]

    if pd.isna(curr_adj):
        raise RuntimeError(
            f"STOP: No adjusted close for {ticker} on {current_date}."
        )

    # Yahoo JSE prices are quoted in cents.
    basis_rows.append({
        "Date": pd.Timestamp(current_date).date(),
        "Holding": h["Holding"],
        "Ticker": ticker,
        "Bucket": h["Bucket"],
        "Qty": float(h["Qty"]),
        "Adjusted_Basis": float(curr_adj) / 100.0
    })


basis = pd.DataFrame(basis_rows)

if len(basis) != len(holdings):
    raise RuntimeError(
        "STOP: EOD basis does not contain every holding."
    )

if basis["Ticker"].duplicated().any():
    raise RuntimeError(
        "STOP: Duplicate tickers in EOD basis."
    )

if basis["Adjusted_Basis"].isna().any():
    raise RuntimeError(
        "STOP: Missing adjusted-close basis."
    )

if (basis["Adjusted_Basis"] <= 0).any():
    raise RuntimeError(
        "STOP: Invalid adjusted-close basis."
    )

basis.to_csv(BASIS_FILE, index=False)

print()
print("=" * 90)
print("NEXT-DAY EOD BASIS SAVED")
print("=" * 90)
print(f"Basis date: {pd.Timestamp(current_date).date()}")
print(f"Holdings:   {len(basis)}")
print(f"File:       {BASIS_FILE}")
print()

'''

    eod = eod.replace(
        history_marker,
        basis_code + history_marker
    )

EOD.write_text(eod)


# ============================================================
# 2. APP.PY
# Force live dashboard to use persisted Adjusted Close basis
# ============================================================

app = APP.read_text()

if 'BASIS_FILE = Path("portfolio_basis.csv")' not in app:
    app = app.replace(
        'NAV_FILE = Path("portfolio_history.csv")',
        'NAV_FILE = Path("portfolio_history.csv")\n'
        'BASIS_FILE = Path("portfolio_basis.csv")'
    )

start_marker = """# ============================================================
# LIVE PERFORMANCE
# ============================================================"""

end_marker = """# ============================================================
# PORTFOLIO SUMMARY
# ============================================================"""

assert start_marker in app, (
    "STOP: LIVE PERFORMANCE section not found in app.py"
)

assert end_marker in app, (
    "STOP: PORTFOLIO SUMMARY section not found in app.py"
)

start = app.index(start_marker)
end = app.index(end_marker)

new_live_section = r'''# ============================================================
# LIVE PERFORMANCE
#
# ACCOUNTING CHAIN:
#
# Previous trading day's Adjusted Close
#           ↓
#     portfolio_basis.csv
#           ↓
# Today's ordinary intraday market price
#
# Historical asset-return table is NOT used here.
# ============================================================

basis = pd.DataFrame()

if BASIS_FILE.exists():

    basis = pd.read_csv(
        BASIS_FILE,
        parse_dates=["Date"]
    )

    required = {
        "Date",
        "Holding",
        "Ticker",
        "Bucket",
        "Qty",
        "Adjusted_Basis"
    }

    missing = required - set(basis.columns)

    if missing:
        raise RuntimeError(
            "portfolio_basis.csv missing columns: "
            + ", ".join(sorted(missing))
        )

    basis["Date"] = pd.to_datetime(
        basis["Date"]
    ).dt.normalize()


live_rows = []

for holding, row in master.iterrows():

    ticker = row["ticker"]
    qty = float(row["qty"])
    bucket = row["bucket"]

    if ticker not in intraday_close.columns:
        continue

    s = intraday_close[ticker].dropna()

    if s.empty:
        continue

    idx = s.index

    if idx.tz is None:
        idx = idx.tz_localize("UTC")

    local_idx = idx.tz_convert(SAST)
    local_dates = local_idx.date

    unique_dates = sorted(set(local_dates))

    if not unique_dates:
        continue

    trading_date = unique_dates[-1]

    today_positions = np.where(
        local_dates == trading_date
    )[0]

    if len(today_positions) == 0:
        continue

    latest_raw = float(
        s.iloc[today_positions[-1]]
    )

    # Yahoo JSE intraday prices: cents -> rand
    latest_price = latest_raw / 100.0

    basis_price = None
    basis_date = None
    basis_source = None

    # --------------------------------------------------------
    # NORMAL OPERATION:
    # Use latest persisted Adjusted Close strictly BEFORE today.
    # --------------------------------------------------------

    if not basis.empty:

        b = basis[
            basis["Ticker"] == ticker
        ].copy()

        b = b[
            b["Date"].dt.date < trading_date
        ]

        if not b.empty:

            b = b.sort_values("Date").iloc[-1]

            basis_price = float(
                b["Adjusted_Basis"]
            )

            basis_date = pd.Timestamp(
                b["Date"]
            )

            basis_source = (
                "Prior EOD Adjusted Close"
            )

    # --------------------------------------------------------
    # INITIAL DEPLOYMENT FALLBACK ONLY
    #
    # Before portfolio_basis.csv has ever been created,
    # use previous ordinary intraday close.
    #
    # After the first successful EOD run this fallback should
    # no longer be used.
    # --------------------------------------------------------

    if basis_price is None:

        prior_dates = [
            d for d in unique_dates
            if d < trading_date
        ]

        if not prior_dates:
            continue

        prior_day = prior_dates[-1]

        prior_positions = np.where(
            local_dates == prior_day
        )[0]

        if len(prior_positions) == 0:
            continue

        prior_raw = float(
            s.iloc[prior_positions[-1]]
        )

        basis_price = prior_raw / 100.0
        basis_date = pd.Timestamp(prior_day)

        basis_source = (
            "Prior ordinary close — inception fallback"
        )

    if basis_price <= 0:
        raise RuntimeError(
            f"Invalid basis for {ticker}: {basis_price}"
        )

    today_return = (
        latest_price / basis_price - 1.0
    )

    opening_value = (
        qty * basis_price
    )

    current_value = (
        qty * latest_price
    )

    pnl = (
        current_value - opening_value
    )

    timestamp = local_idx[
        today_positions[-1]
    ]

    live_rows.append({
        "Holding": holding,
        "Ticker": ticker,
        "Bucket": bucket,
        "Qty": qty,

        # Keep this field name because the existing dashboard
        # may reference it downstream.
        "Prior_Close": basis_price,

        "Basis_Date": basis_date,
        "Basis_Source": basis_source,

        "Latest": latest_price,
        "Today_Return": today_return,

        "Opening_Value": opening_value,
        "Current_Value": current_value,
        "PnL": pnl,

        "Last_Update": timestamp
    })


live = pd.DataFrame(live_rows)

'''

app = (
    app[:start]
    + new_live_section
    + app[end:]
)

APP.write_text(app)


# ============================================================
# 3. COMPILE BOTH FILES
# ============================================================

py_compile.compile(
    str(APP),
    doraise=True
)

py_compile.compile(
    str(EOD),
    doraise=True
)


# ============================================================
# 4. VERIFY PATCH EXISTS
# ============================================================

app_check = APP.read_text()
eod_check = EOD.read_text()

assert 'BASIS_FILE = Path("portfolio_basis.csv")' in app_check
assert 'BASIS_FILE = Path("portfolio_basis.csv")' in eod_check

assert "Prior EOD Adjusted Close" in app_check
assert "NEXT-DAY EOD BASIS SAVED" in eod_check


print("=" * 90)
print("PATCH COMPLETE")
print("=" * 90)

print("PASS: app.py compiles")
print("PASS: eod_update.py compiles")
print("PASS: app.py reads persisted EOD adjusted-close basis")
print("PASS: eod_update.py writes persisted EOD adjusted-close basis")

print()
print("ACCOUNTING CHAIN:")
print(
    "EOD Adjusted Close -> portfolio_basis.csv "
    "-> next trading day's ordinary intraday price"
)

print()
print("Nothing has been pushed to GitHub yet.")
```

```python
from pathlib import Path

path = Path("/content/test-edv/eod_update.py")
s = path.read_text()

start = s.index("# ============================================================\n# FIND TWO MOST RECENT COMPLETE TRADING DATES")
end = s.index("# ============================================================\n# SAVE / UPDATE HISTORY")

replacement = r'''# ============================================================
# CURRENT EOD DATE
# ============================================================

dates = adj.dropna(how="all").index.sort_values()

if len(dates) < 1:
    raise RuntimeError("No daily observations available.")

current_date = pd.Timestamp(dates[-1]).normalize()

print("Current trading date:", current_date.date())

if current_date < PORTFOLIO_START:
    raise RuntimeError(
        "Current observation predates portfolio inception."
    )


# ============================================================
# CURRENT EOD ADJUSTED-CLOSE BASIS
# ============================================================

basis_rows = []

for _, h in holdings.iterrows():

    ticker = h["Ticker"]

    if ticker not in adj.columns:
        raise RuntimeError(
            f"STOP — {ticker} absent from Yahoo response."
        )

    curr_adj = adj.loc[current_date, ticker]

    if pd.isna(curr_adj):
        raise RuntimeError(
            f"STOP — no EOD adjusted close for {ticker}"
        )

    # Yahoo JSE instruments quoted in cents.
    basis_price = float(curr_adj / 100)

    basis_rows.append({
        "Date": current_date,
        "Holding": h["Holding"],
        "Ticker": ticker,
        "Bucket": h["Bucket"],
        "Qty": float(h["Qty"]),
        "Adjusted_Basis": basis_price
    })

current_basis = pd.DataFrame(basis_rows)

if len(current_basis) != len(holdings):
    raise RuntimeError(
        "STOP — current EOD basis does not contain every holding."
    )


# ============================================================
# INCEPTION DAY
#
# 28 Sep 2026 establishes the official portfolio basis.
# There is deliberately NO portfolio return on inception day.
# ============================================================

if not BASIS_FILE.exists():

    current_basis.to_csv(BASIS_FILE, index=False)

    print()
    print("=" * 90)
    print("PORTFOLIO INCEPTION BASIS ESTABLISHED")
    print("=" * 90)
    print(f"Basis date: {current_date.date()}")
    print(f"Holdings:   {len(current_basis)}")
    print()
    print("No portfolio return recorded on inception day.")
    print("Tomorrow's return will be measured from this adjusted-close basis.")

    raise SystemExit(0)


# ============================================================
# LOAD PRIOR OFFICIAL EOD BASIS
# ============================================================

prior_basis = pd.read_csv(
    BASIS_FILE,
    parse_dates=["Date"]
)

required_cols = {
    "Date", "Holding", "Ticker",
    "Bucket", "Qty", "Adjusted_Basis"
}

missing_cols = required_cols - set(prior_basis.columns)

if missing_cols:
    raise RuntimeError(
        "STOP — portfolio_basis.csv missing columns: "
        + ", ".join(sorted(missing_cols))
    )

prior_basis["Date"] = pd.to_datetime(
    prior_basis["Date"]
).dt.normalize()

prior_date = prior_basis["Date"].max()

prior_basis = prior_basis[
    prior_basis["Date"] == prior_date
].copy()

print("Prior official basis date:", prior_date.date())


# Do not create a duplicate daily return if rerun on same EOD date.
if current_date <= prior_date:

    print()
    print(
        f"No new trading day to record. "
        f"Current EOD={current_date.date()}, "
        f"stored basis={prior_date.date()}."
    )

    raise SystemExit(0)


# ============================================================
# COVERAGE CHECK
# ============================================================

expected = set(holdings["Ticker"])

prior_received = set(prior_basis["Ticker"])
current_received = set(current_basis["Ticker"])

missing_prior = sorted(expected - prior_received)
missing_current = sorted(expected - current_received)

if missing_prior:
    raise RuntimeError(
        "STOP — holdings missing from prior basis: "
        + ", ".join(missing_prior)
    )

if missing_current:
    raise RuntimeError(
        "STOP — holdings missing from current basis: "
        + ", ".join(missing_current)
    )


# ============================================================
# DAILY SECURITY RETURNS
#
# Official return:
# current EOD adjusted close / prior official adjusted basis - 1
# ============================================================

detail = prior_basis.merge(
    current_basis[
        ["Ticker", "Adjusted_Basis"]
    ].rename(
        columns={"Adjusted_Basis": "Current_Adjusted_Basis"}
    ),
    on="Ticker",
    how="left",
    validate="one_to_one"
)

detail["Total_Return"] = (
    detail["Current_Adjusted_Basis"]
    /
    detail["Adjusted_Basis"]
    - 1
)

# Beginning-of-day economic value based on prior adjusted basis.
detail["Opening_Value"] = (
    detail["Qty"]
    *
    detail["Adjusted_Basis"]
)


# ============================================================
# PORTFOLIO RETURN
# ============================================================

def calc_portfolio_return(df):

    capital = df["Opening_Value"].sum()

    if capital <= 0:
        raise RuntimeError(
            "STOP — non-positive opening portfolio value."
        )

    pnl = (
        df["Opening_Value"]
        *
        df["Total_Return"]
    ).sum()

    return float(pnl / capital)


combined_return = calc_portfolio_return(detail)

main_return = calc_portfolio_return(
    detail[detail["Bucket"] == "Main"]
)

subset_return = calc_portfolio_return(
    detail[detail["Bucket"] == "Subset"]
)


print()
print("=" * 90)
print("OFFICIAL EOD PORTFOLIO RETURN")
print("=" * 90)

print(f"Prior basis: {prior_date.date()}")
print(f"Date:        {current_date.date()}")
print(f"Combined:    {combined_return:+.4%}")
print(f"Main:        {main_return:+.4%}")
print(f"Subset:      {subset_return:+.4%}")

print()
print("Returns use the stored prior-EOD adjusted-close basis.")
print()


# ============================================================
# SAVE CURRENT ADJUSTED CLOSE AS NEXT-DAY BASIS
# ============================================================

current_basis.to_csv(
    BASIS_FILE,
    index=False
)

print("Next-day adjusted-close basis saved.")
print(
    f"Basis date: {current_date.date()} | "
    f"Holdings: {len(current_basis)}"
)
print()

'''

new_s = s[:start] + replacement + s[end:]

path.write_text(new_s)

# Compile before doing anything else.
compile(new_s, str(path), "exec")

print("PASS: eod_update.py patched and compiles.")
```

```python
import yfinance as yf
import pandas as pd

tickers = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

rows = []

for ticker in tickers:

    print(f"Checking {ticker}...")

    try:
        x = yf.Ticker(ticker).history(
            period="5y",
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=True
        )

        if x.empty:
            rows.append({
                "Ticker": ticker,
                "Observations": 0,
                "Dividend_Events": None,
                "Split_Events": None,
                "Total_Dividends": None,
                "Last_Dividend_Date": None,
                "Last_Dividend": None
            })
            continue

        div = x["Dividends"] if "Dividends" in x.columns else pd.Series(dtype=float)
        spl = x["Stock Splits"] if "Stock Splits" in x.columns else pd.Series(dtype=float)

        div_events = div[div != 0]
        split_events = spl[spl != 0]

        rows.append({
            "Ticker": ticker,
            "Observations": len(x),
            "Dividend_Events": len(div_events),
            "Split_Events": len(split_events),
            "Total_Dividends": div_events.sum() if len(div_events) else 0,
            "Last_Dividend_Date":
                div_events.index[-1].date() if len(div_events) else None,
            "Last_Dividend":
                div_events.iloc[-1] if len(div_events) else None
        })

    except Exception as e:

        rows.append({
            "Ticker": ticker,
            "Observations": "ERROR",
            "Dividend_Events": "ERROR",
            "Split_Events": "ERROR",
            "Total_Dividends": "ERROR",
            "Last_Dividend_Date": None,
            "Last_Dividend": str(e)
        })


audit = pd.DataFrame(rows)

print("\n" + "=" * 120)
print("CORPORATE ACTION AUDIT")
print("=" * 120)
print(audit.to_string(index=False))


print("\n" + "=" * 120)
print("ALL SPLIT EVENTS FOUND")
print("=" * 120)

for ticker in tickers:

    try:
        x = yf.Ticker(ticker).history(
            period="5y",
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=True
        )

        if "Stock Splits" in x.columns:

            events = x[x["Stock Splits"] != 0][
                ["Close", "Dividends", "Stock Splits"]
            ]

            if not events.empty:
                print(f"\n{ticker}")
                print(events.to_string())

    except:
        pass


print("\n" + "=" * 120)
print("MOST RECENT DIVIDEND FOR EACH HOLDING")
print("=" * 120)

for ticker in tickers:

    try:
        x = yf.Ticker(ticker).history(
            period="5y",
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=True
        )

        if "Dividends" in x.columns:

            events = x[x["Dividends"] != 0]

            if not events.empty:

                i = events.index[-1]
                loc = x.index.get_loc(i)

                lo = max(0, loc - 2)
                hi = min(len(x), loc + 3)

                print(f"\n--- {ticker} ---")
                print(
                    x.iloc[lo:hi][
                        ["Close", "Adj Close",
                         "Dividends", "Stock Splits"]
                    ].to_string()
                )

    except Exception as e:
        print(ticker, "ERROR:", e)
```

```python
import yfinance as yf
import pandas as pd

tickers = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

raw = yf.download(
    tickers,
    period="5y",
    interval="1d",
    auto_adjust=False,
    actions=True,
    repair=False,
    group_by="column",
    progress=False,
    threads=False
)

print("=" * 110)
print("DIVIDEND / DISTRIBUTION EVENTS")
print("=" * 110)

for ticker in tickers:
    try:
        d = raw[("Dividends", ticker)].dropna()
        d = d[d != 0]

        print(f"\n{ticker}: {len(d)} events")

        if len(d):
            print(d.tail(10).to_string())

    except Exception as e:
        print(f"\n{ticker}: ERROR — {e}")


print("\n" + "=" * 110)
print("STOCK SPLIT EVENTS")
print("=" * 110)

for ticker in tickers:
    try:
        s = raw[("Stock Splits", ticker)].dropna()
        s = s[s != 0]

        print(f"\n{ticker}: {len(s)} events")

        if len(s):
            print(s.to_string())

    except Exception as e:
        print(f"\n{ticker}: ERROR — {e}")
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","CSP500.JO","EXX.JO","GLD.JO","INCOME.JO",
    "STXCAP.JO","STXEME.JO","STXEUR.JO","STXGBD.JO",
    "STXGOV.JO","STX500.JO"
]

raw = yf.download(
    tickers,
    period="5y",
    interval="1d",
    auto_adjust=False,
    actions=True,
    repair=False,
    group_by="column",
    progress=False,
    threads=False
)

results = []

for ticker in tickers:

    try:
        df = pd.DataFrame({
            "Close": raw[("Close", ticker)],
            "AdjClose": raw[("Adj Close", ticker)],
            "Dividend": raw[("Dividends", ticker)],
            "Split": raw[("Stock Splits", ticker)]
        }).dropna(subset=["Close", "AdjClose"])

        # Yahoo JSE prices/dividends are both in cents.
        df["PrevClose"] = df["Close"].shift(1)

        # No-split dividend dates only for clean comparison.
        events = df[
            (df["Dividend"] != 0) &
            ((df["Split"] == 0) | df["Split"].isna())
        ].copy()

        if events.empty:
            continue

        events["Explicit_TR"] = (
            (events["Close"] + events["Dividend"])
            / events["PrevClose"]
            - 1
        )

        events["Yahoo_Adj_TR"] = (
            events["AdjClose"]
            / df["AdjClose"].shift(1).loc[events.index]
            - 1
        )

        events["Difference_bp"] = (
            events["Explicit_TR"] - events["Yahoo_Adj_TR"]
        ) * 10000

        for date, row in events.iterrows():
            results.append({
                "Ticker": ticker,
                "Date": date.date(),
                "Dividend": row["Dividend"],
                "Explicit_TR_%": row["Explicit_TR"] * 100,
                "Yahoo_Adj_TR_%": row["Yahoo_Adj_TR"] * 100,
                "Difference_bp": row["Difference_bp"]
            })

    except Exception as e:
        print(ticker, "ERROR:", e)

check = pd.DataFrame(results)

print("=" * 120)
print("DIVIDEND TOTAL-RETURN RECONSTRUCTION")
print("=" * 120)

print(
    check.tail(60).to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n" + "=" * 120)
print("SUMMARY")
print("=" * 120)

print("Dividend events tested:", len(check))

if len(check):
    print(
        "Median absolute difference:",
        f"{check['Difference_bp'].abs().median():.4f} bp"
    )
    print(
        "Maximum absolute difference:",
        f"{check['Difference_bp'].abs().max():.4f} bp"
    )
```
