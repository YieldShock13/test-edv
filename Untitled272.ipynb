```python
# ============================================================
# CELL 1 — SPY OPTIONS DATA: DOWNLOAD + VALIDATION
# ============================================================

!pip -q install yfinance plotly

import yfinance as yf
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from datetime import datetime, timezone

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

TICKER = "SPY"

# ------------------------------------------------------------
# DOWNLOAD UNDERLYING
# ------------------------------------------------------------

spy = yf.Ticker(TICKER)

px = spy.history(period="5d", auto_adjust=False)

if px.empty:
    raise RuntimeError("Could not retrieve SPY price data.")

spot = float(px["Close"].dropna().iloc[-1])

print(f"SPY spot: ${spot:,.2f}")

# ------------------------------------------------------------
# AVAILABLE OPTION EXPIRIES
# ------------------------------------------------------------

expiries = list(spy.options)

if len(expiries) == 0:
    raise RuntimeError("Yahoo returned no SPY option expiries.")

print(f"\nAvailable expiries: {len(expiries)}")
print("First 10:")
print(expiries[:10])

# ------------------------------------------------------------
# PULL NEAREST EXPIRY
# ------------------------------------------------------------

expiry = expiries[0]

chain = spy.option_chain(expiry)

calls = chain.calls.copy()
puts  = chain.puts.copy()

calls["optionType"] = "CALL"
puts["optionType"]   = "PUT"

calls["expiry"] = expiry
puts["expiry"]  = expiry

options = pd.concat(
    [calls, puts],
    ignore_index=True
)

# ------------------------------------------------------------
# BASIC CLEANING / DERIVED FIELDS
# ------------------------------------------------------------

options["mid"] = (options["bid"] + options["ask"]) / 2

options["spread"] = options["ask"] - options["bid"]

options["spread_pct"] = np.where(
    options["mid"] > 0,
    options["spread"] / options["mid"],
    np.nan
)

options["distance_from_spot_pct"] = (
    options["strike"] / spot - 1
) * 100

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("CHAIN VALIDATION")
print("=" * 65)

print(f"Expiry:          {expiry}")
print(f"Calls:           {len(calls):,}")
print(f"Puts:            {len(puts):,}")
print(f"Total contracts: {len(options):,}")

print(
    f"Strike range:    "
    f"${options['strike'].min():,.0f} – "
    f"${options['strike'].max():,.0f}"
)

print(f"\nTotal call OI: {calls['openInterest'].fillna(0).sum():,.0f}")
print(f"Total put OI:  {puts['openInterest'].fillna(0).sum():,.0f}")

print("\nColumns returned:")
print(options.columns.tolist())

# ------------------------------------------------------------
# DISPLAY LIQUID AREA AROUND SPOT
# ------------------------------------------------------------

near_spot = options[
    options["strike"].between(
        spot * 0.90,
        spot * 1.10
    )
].copy()

cols = [
    "optionType",
    "strike",
    "bid",
    "ask",
    "mid",
    "volume",
    "openInterest",
    "impliedVolatility",
    "distance_from_spot_pct"
]

display(
    near_spot[cols]
    .sort_values(["strike", "optionType"])
    .reset_index(drop=True)
)
```

```python
# ============================================================
# CELL 2 — MULTI-EXPIRY GAMMA ENGINE
# ============================================================

from scipy.stats import norm

# Risk-free rate assumption
# Can replace later with actual Treasury curve by maturity
R = 0.04

# SPY dividend yield approximation
Q = 0.011

now = pd.Timestamp.now(tz="UTC")

all_chains = []

# Use first 10 expiries
for expiry in expiries[:10]:

    chain = spy.option_chain(expiry)

    for df, opt_type in [
        (chain.calls.copy(), "CALL"),
        (chain.puts.copy(), "PUT")
    ]:

        df["optionType"] = opt_type
        df["expiry"] = expiry

        expiry_dt = pd.Timestamp(expiry, tz="UTC") + pd.Timedelta(hours=20)

        df["T"] = (
            expiry_dt - now
        ).total_seconds() / (365.25 * 24 * 3600)

        all_chains.append(df)

options_all = pd.concat(all_chains, ignore_index=True)

# ------------------------------------------------------------
# CLEAN
# ------------------------------------------------------------

options_all["openInterest"] = (
    options_all["openInterest"]
    .fillna(0)
)

options_all["volume"] = (
    options_all["volume"]
    .fillna(0)
)

options_all["IV"] = options_all["impliedVolatility"]

options_all = options_all[
    (options_all["T"] > 0) &
    (options_all["IV"] > 0) &
    (options_all["strike"] > 0)
].copy()

# ------------------------------------------------------------
# BLACK-SCHOLES GAMMA
# ------------------------------------------------------------

S = spot
K = options_all["strike"]
T = options_all["T"]
sigma = options_all["IV"]

d1 = (
    np.log(S / K)
    + (R - Q + 0.5 * sigma**2) * T
) / (
    sigma * np.sqrt(T)
)

options_all["gamma"] = (
    np.exp(-Q * T)
    * norm.pdf(d1)
    / (S * sigma * np.sqrt(T))
)

# ------------------------------------------------------------
# GAMMA EXPOSURE
#
# Dollar gamma for a 1% move in SPY
# ------------------------------------------------------------

options_all["gamma_exposure"] = (
    options_all["gamma"]
    * options_all["openInterest"]
    * 100
    * S**2
    * 0.01
)

# Keep calls/puts separate.
# DO NOT yet pretend puts = negative dealer gamma.
options_all["call_gamma"] = np.where(
    options_all["optionType"] == "CALL",
    options_all["gamma_exposure"],
    0
)

options_all["put_gamma"] = np.where(
    options_all["optionType"] == "PUT",
    options_all["gamma_exposure"],
    0
)

# ------------------------------------------------------------
# AGGREGATE BY STRIKE
# ------------------------------------------------------------

gamma_by_strike = (
    options_all
    .groupby("strike", as_index=False)
    .agg(
        call_gamma=("call_gamma", "sum"),
        put_gamma=("put_gamma", "sum"),
        call_oi=("openInterest",
                 lambda x: x[
                     options_all.loc[x.index, "optionType"] == "CALL"
                 ].sum()),
        put_oi=("openInterest",
                lambda x: x[
                    options_all.loc[x.index, "optionType"] == "PUT"
                ].sum())
    )
)

# Only useful area around spot
gamma_plot = gamma_by_strike[
    gamma_by_strike["strike"].between(
        spot * 0.90,
        spot * 1.10
    )
].copy()

print(f"Contracts analysed: {len(options_all):,}")
print(f"Expiries analysed:   {options_all['expiry'].nunique()}")
print(f"SPY spot:             ${spot:.2f}")

display(
    gamma_plot.sort_values("strike")
)
```

```python
# ============================================================
# CELL 3 — GAMMA CONCENTRATION LANDSCAPE
# ============================================================

import plotly.graph_objects as go

# Convert to $ millions of gamma per 1% SPY move
gamma_plot["call_gamma_mm"] = gamma_plot["call_gamma"] / 1e6
gamma_plot["put_gamma_mm"]  = gamma_plot["put_gamma"] / 1e6

gamma_plot["distance_pct"] = (
    gamma_plot["strike"] / spot - 1
) * 100

# ------------------------------------------------------------
# INTERACTIVE CHART
# ------------------------------------------------------------

fig = go.Figure()

fig.add_bar(
    x=gamma_plot["strike"],
    y=gamma_plot["call_gamma_mm"],
    name="Call gamma concentration"
)

fig.add_bar(
    x=gamma_plot["strike"],
    y=-gamma_plot["put_gamma_mm"],
    name="Put gamma concentration"
)

# Spot
fig.add_vline(
    x=spot,
    line_dash="dash",
    annotation_text=f"SPY {spot:.2f}",
    annotation_position="top"
)

fig.update_layout(
    title="SPY Gamma Concentration by Strike",
    xaxis_title="Strike",
    yaxis_title="$mm gamma exposure per 1% SPY move",
    barmode="relative",
    template="plotly_white",
    height=650,
    hovermode="x unified"
)

fig.show()


# ============================================================
# ANALYTICAL METRICS
# ============================================================

total_gamma = (
    options_all["gamma_exposure"].sum()
)

# Gamma within ±1% and ±2% of spot
near_1 = options_all[
    options_all["strike"].between(
        spot * 0.99,
        spot * 1.01
    )
]["gamma_exposure"].sum()

near_2 = options_all[
    options_all["strike"].between(
        spot * 0.98,
        spot * 1.02
    )
]["gamma_exposure"].sum()

gamma_1pct_share = near_1 / total_gamma
gamma_2pct_share = near_2 / total_gamma


# Gamma-weighted absolute distance from spot
options_all["abs_distance_pct"] = abs(
    options_all["strike"] / spot - 1
)

gamma_weighted_distance = np.average(
    options_all["abs_distance_pct"],
    weights=options_all["gamma_exposure"]
)


# Call / put gamma concentration
call_total = options_all.loc[
    options_all["optionType"] == "CALL",
    "gamma_exposure"
].sum()

put_total = options_all.loc[
    options_all["optionType"] == "PUT",
    "gamma_exposure"
].sum()

call_put_gamma_ratio = call_total / put_total


print("=" * 55)
print("GAMMA STATE")
print("=" * 55)

print(f"Gamma within ±1% of spot: {gamma_1pct_share:.2%}")
print(f"Gamma within ±2% of spot: {gamma_2pct_share:.2%}")

print(
    f"Gamma-weighted distance:  "
    f"{gamma_weighted_distance:.2%}"
)

print(
    f"Call / Put gamma ratio:   "
    f"{call_put_gamma_ratio:.3f}"
)
```

```python
# ============================================================
# CELL 4 — IV SKEW + TERM STRUCTURE
# ============================================================

# ------------------------------------------------------------
# BLACK-SCHOLES DELTA
# ------------------------------------------------------------

options_all["delta"] = np.where(
    options_all["optionType"] == "CALL",
    np.exp(-Q * options_all["T"]) * norm.cdf(d1),
    -np.exp(-Q * options_all["T"]) * norm.cdf(-d1)
)

# ------------------------------------------------------------
# EXTRACT VOL STATE FOR EACH EXPIRY
# ------------------------------------------------------------

vol_rows = []

for expiry, df in options_all.groupby("expiry"):

    calls_e = df[df["optionType"] == "CALL"].copy()
    puts_e  = df[df["optionType"] == "PUT"].copy()

    if len(calls_e) == 0 or len(puts_e) == 0:
        continue

    # ATM options: closest strike to spot
    calls_e["atm_dist"] = abs(calls_e["strike"] - spot)
    puts_e["atm_dist"]  = abs(puts_e["strike"] - spot)

    atm_call = calls_e.loc[calls_e["atm_dist"].idxmin()]
    atm_put  = puts_e.loc[puts_e["atm_dist"].idxmin()]

    # Average ATM call/put IV
    atm_iv = np.mean([
        atm_call["IV"],
        atm_put["IV"]
    ])

    # 25-delta call
    calls_e["delta25_dist"] = abs(calls_e["delta"] - 0.25)
    call25 = calls_e.loc[calls_e["delta25_dist"].idxmin()]

    # -25-delta put
    puts_e["delta25_dist"] = abs(puts_e["delta"] + 0.25)
    put25 = puts_e.loc[puts_e["delta25_dist"].idxmin()]

    call25_iv = call25["IV"]
    put25_iv  = put25["IV"]

    # Equity convention:
    # positive number = downside puts richer than upside calls
    risk_reversal = put25_iv - call25_iv

    vol_rows.append({
        "expiry": expiry,
        "DTE": df["T"].iloc[0] * 365.25,

        "ATM_IV": atm_iv,

        "Put25_IV": put25_iv,
        "Call25_IV": call25_iv,

        "Put25_Delta": put25["delta"],
        "Call25_Delta": call25["delta"],

        "Put25_Strike": put25["strike"],
        "Call25_Strike": call25["strike"],

        "Risk_Reversal_25D": risk_reversal
    })

vol_state = pd.DataFrame(vol_rows)

vol_state = vol_state.sort_values("DTE").reset_index(drop=True)

display(
    vol_state.round({
        "DTE": 1,
        "ATM_IV": 4,
        "Put25_IV": 4,
        "Call25_IV": 4,
        "Put25_Delta": 4,
        "Call25_Delta": 4,
        "Risk_Reversal_25D": 4
    })
)

# ============================================================
# PLOT 1 — ATM IV TERM STRUCTURE
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=vol_state["DTE"],
        y=vol_state["ATM_IV"] * 100,
        mode="lines+markers",
        name="ATM IV"
    )
)

fig.update_layout(
    title="SPY ATM Implied Volatility Term Structure",
    xaxis_title="Days to Expiry",
    yaxis_title="ATM Implied Volatility (%)",
    template="plotly_white",
    height=500
)

fig.show()


# ============================================================
# PLOT 2 — 25 DELTA VOL SKEW
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=vol_state["DTE"],
        y=vol_state["Put25_IV"] * 100,
        mode="lines+markers",
        name="25Δ Put IV"
    )
)

fig.add_trace(
    go.Scatter(
        x=vol_state["DTE"],
        y=vol_state["Call25_IV"] * 100,
        mode="lines+markers",
        name="25Δ Call IV"
    )
)

fig.update_layout(
    title="SPY 25Δ Put / Call Implied Volatility",
    xaxis_title="Days to Expiry",
    yaxis_title="Implied Volatility (%)",
    template="plotly_white",
    height=500
)

fig.show()


# ============================================================
# PLOT 3 — 25 DELTA RISK REVERSAL
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=vol_state["DTE"],
        y=vol_state["Risk_Reversal_25D"] * 100,
        mode="lines+markers",
        name="25Δ RR"
    )
)

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.update_layout(
    title="SPY 25Δ Downside Volatility Premium",
    xaxis_title="Days to Expiry",
    yaxis_title="25Δ Put IV − 25Δ Call IV (vol points)",
    template="plotly_white",
    height=500
)

fig.show()
```

```python
# ============================================================
# CELL 5 — INTERPOLATED 25-DELTA SKEW
# ============================================================

from scipy.interpolate import interp1d

rows = []

for expiry, df in options_all.groupby("expiry"):

    calls = df[
        (df["optionType"] == "CALL") &
        (df["delta"] > 0.05) &
        (df["delta"] < 0.50)
    ].copy()

    puts = df[
        (df["optionType"] == "PUT") &
        (df["delta"] < -0.05) &
        (df["delta"] > -0.50)
    ].copy()

    if len(calls) < 3 or len(puts) < 3:
        continue

    # Sort by delta
    calls = calls.sort_values("delta")
    puts  = puts.sort_values("delta")

    # Remove duplicate delta values
    calls = calls.drop_duplicates("delta")
    puts  = puts.drop_duplicates("delta")

    try:

        call_interp = interp1d(
            calls["delta"],
            calls["IV"],
            kind="linear",
            bounds_error=True
        )

        put_interp = interp1d(
            puts["delta"],
            puts["IV"],
            kind="linear",
            bounds_error=True
        )

        call25_iv = float(call_interp(0.25))
        put25_iv  = float(put_interp(-0.25))

    except ValueError:
        continue

    T = df["T"].iloc[0]

    # ATM IV from nearest call/put strike
    atm = df.loc[
        (df["strike"] - spot).abs().nsmallest(2).index,
        "IV"
    ].mean()

    rows.append({
        "expiry": expiry,
        "DTE": T * 365.25,
        "ATM_IV": atm,
        "Put25_IV": put25_iv,
        "Call25_IV": call25_iv,
        "RR_25D": put25_iv - call25_iv
    })


skew = pd.DataFrame(rows).sort_values("DTE")

display(skew.round(4))


# ------------------------------------------------------------
# PLOT
# ------------------------------------------------------------

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=skew["DTE"],
        y=skew["RR_25D"] * 100,
        mode="lines+markers",
        name="25Δ skew"
    )
)

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.update_layout(
    title="SPY Interpolated 25Δ Volatility Skew",
    xaxis_title="Days to Expiry",
    yaxis_title="25Δ Put IV − 25Δ Call IV (vol points)",
    template="plotly_white",
    height=500
)

fig.show()
```

```python
# ============================================================
# CELL 6 — INDEPENDENT IV FROM MARKET MID-PRICES
# ============================================================

from scipy.optimize import brentq

# ------------------------------------------------------------
# BLACK-SCHOLES PRICE
# ------------------------------------------------------------

def bs_price(S, K, T, r, q, sigma, option_type):

    if T <= 0 or sigma <= 0:
        return np.nan

    d1 = (
        np.log(S / K)
        + (r - q + 0.5 * sigma**2) * T
    ) / (sigma * np.sqrt(T))

    d2 = d1 - sigma * np.sqrt(T)

    if option_type == "CALL":

        return (
            S * np.exp(-q*T) * norm.cdf(d1)
            - K * np.exp(-r*T) * norm.cdf(d2)
        )

    else:

        return (
            K * np.exp(-r*T) * norm.cdf(-d2)
            - S * np.exp(-q*T) * norm.cdf(-d1)
        )


# ------------------------------------------------------------
# IMPLIED VOL SOLVER
# ------------------------------------------------------------

def solve_iv(row):

    bid = row["bid"]
    ask = row["ask"]

    # Require an actual two-sided market
    if (
        pd.isna(bid) or pd.isna(ask) or
        bid <= 0 or ask <= 0 or
        ask < bid
    ):
        return np.nan

    market_price = (bid + ask) / 2

    try:

        return brentq(
            lambda sigma:
                bs_price(
                    spot,
                    row["strike"],
                    row["T"],
                    R,
                    Q,
                    sigma,
                    row["optionType"]
                ) - market_price,
            0.001,
            5.0
        )

    except:
        return np.nan


# ------------------------------------------------------------
# CALCULATE OUR IV
# ------------------------------------------------------------

options_all["IV_mid"] = options_all.apply(
    solve_iv,
    axis=1
)

valid = options_all.dropna(
    subset=["IV_mid"]
).copy()

print(
    f"Valid independently solved IVs: "
    f"{len(valid):,} / {len(options_all):,}"
)


# ------------------------------------------------------------
# COMPARE YAHOO IV VS OUR IV
# ------------------------------------------------------------

valid["IV_difference"] = (
    valid["IV_mid"] -
    valid["IV"]
)

print("\nIV comparison")

print(
    f"Median absolute difference: "
    f"{valid['IV_difference'].abs().median()*100:.3f} vol points"
)

print(
    f"Mean absolute difference:   "
    f"{valid['IV_difference'].abs().mean()*100:.3f} vol points"
)


# ------------------------------------------------------------
# RECALCULATE DELTA USING OUR IV
# ------------------------------------------------------------

sigma = valid["IV_mid"]

d1_mid = (
    np.log(spot / valid["strike"])
    + (R - Q + 0.5 * sigma**2) * valid["T"]
) / (
    sigma * np.sqrt(valid["T"])
)

valid["delta_mid"] = np.where(
    valid["optionType"] == "CALL",

    np.exp(-Q * valid["T"]) *
    norm.cdf(d1_mid),

    -np.exp(-Q * valid["T"]) *
    norm.cdf(-d1_mid)
)


# ------------------------------------------------------------
# EXACT 25 DELTA INTERPOLATION
# ------------------------------------------------------------

rows = []

for expiry, df in valid.groupby("expiry"):

    calls = df[
        (df["optionType"] == "CALL") &
        (df["delta_mid"] > 0.05) &
        (df["delta_mid"] < 0.50)
    ].sort_values("delta_mid").drop_duplicates("delta_mid")

    puts = df[
        (df["optionType"] == "PUT") &
        (df["delta_mid"] < -0.05) &
        (df["delta_mid"] > -0.50)
    ].sort_values("delta_mid").drop_duplicates("delta_mid")

    if len(calls) < 3 or len(puts) < 3:
        continue

    try:

        call_interp = interp1d(
            calls["delta_mid"],
            calls["IV_mid"],
            bounds_error=True
        )

        put_interp = interp1d(
            puts["delta_mid"],
            puts["IV_mid"],
            bounds_error=True
        )

        call25 = float(call_interp(0.25))
        put25  = float(put_interp(-0.25))

    except:
        continue

    rows.append({
        "expiry": expiry,
        "DTE": df["T"].iloc[0] * 365.25,
        "Put25_IV": put25,
        "Call25_IV": call25,
        "RR_25D": put25 - call25
    })


skew_mid = pd.DataFrame(rows).sort_values("DTE")

display(skew_mid.round(4))


# ------------------------------------------------------------
# CHART
# ------------------------------------------------------------

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=skew_mid["DTE"],
        y=skew_mid["RR_25D"] * 100,
        mode="lines+markers",
        name="Mid-price derived skew"
    )
)

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.update_layout(
    title="SPY 25Δ Skew — Independently Derived from Market Prices",
    xaxis_title="Days to Expiry",
    yaxis_title="25Δ Put IV − 25Δ Call IV (vol points)",
    template="plotly_white",
    height=500
)

fig.show()
```

```python
# ============================================================
# CELL 7 — REBUILD GAMMA USING VALIDATED MID-PRICE IV
# ============================================================

# From now on, use only contracts where IV was successfully
# solved from a valid two-sided bid/ask market.

clean = valid.copy()

# ------------------------------------------------------------
# RECOMPUTE GAMMA USING IV_mid
# ------------------------------------------------------------

sigma = clean["IV_mid"]
T     = clean["T"]
K     = clean["strike"]

d1_clean = (
    np.log(spot / K)
    + (R - Q + 0.5 * sigma**2) * T
) / (
    sigma * np.sqrt(T)
)

clean["gamma"] = (
    np.exp(-Q * T)
    * norm.pdf(d1_clean)
    / (spot * sigma * np.sqrt(T))
)

# ------------------------------------------------------------
# DOLLAR GAMMA EXPOSURE
# Dollar change in delta for approximately a 1% SPY move
# ------------------------------------------------------------

clean["gamma_exposure"] = (
    clean["gamma"]
    * clean["openInterest"]
    * 100
    * spot**2
    * 0.01
)

clean["call_gamma"] = np.where(
    clean["optionType"] == "CALL",
    clean["gamma_exposure"],
    0.0
)

clean["put_gamma"] = np.where(
    clean["optionType"] == "PUT",
    clean["gamma_exposure"],
    0.0
)

# ------------------------------------------------------------
# DISTANCE FROM SPOT
# ------------------------------------------------------------

clean["distance_pct"] = (
    clean["strike"] / spot - 1
)

clean["abs_distance_pct"] = clean["distance_pct"].abs()

# ------------------------------------------------------------
# REBUILD GAMMA BY STRIKE
# ------------------------------------------------------------

gamma_clean = (
    clean
    .groupby("strike", as_index=False)
    .agg(
        call_gamma=("call_gamma", "sum"),
        put_gamma=("put_gamma", "sum")
    )
)

gamma_clean = gamma_clean[
    gamma_clean["strike"].between(
        spot * 0.90,
        spot * 1.10
    )
].copy()

gamma_clean["call_gamma_mm"] = (
    gamma_clean["call_gamma"] / 1e6
)

gamma_clean["put_gamma_mm"] = (
    gamma_clean["put_gamma"] / 1e6
)

# ------------------------------------------------------------
# ANALYTICAL STATE VARIABLES
# ------------------------------------------------------------

total_gamma = clean["gamma_exposure"].sum()

gamma_1 = clean[
    clean["strike"].between(
        spot * 0.99,
        spot * 1.01
    )
]["gamma_exposure"].sum()

gamma_2 = clean[
    clean["strike"].between(
        spot * 0.98,
        spot * 1.02
    )
]["gamma_exposure"].sum()

gamma_1_share = gamma_1 / total_gamma
gamma_2_share = gamma_2 / total_gamma

gamma_weighted_distance = np.average(
    clean["abs_distance_pct"],
    weights=clean["gamma_exposure"]
)

call_gamma_total = clean.loc[
    clean["optionType"] == "CALL",
    "gamma_exposure"
].sum()

put_gamma_total = clean.loc[
    clean["optionType"] == "PUT",
    "gamma_exposure"
].sum()

call_put_gamma_ratio = (
    call_gamma_total / put_gamma_total
)

# ------------------------------------------------------------
# OUTPUT
# ------------------------------------------------------------

print("=" * 60)
print("VALIDATED GAMMA STATE")
print("=" * 60)

print(f"Contracts used:             {len(clean):,}")
print(f"Gamma within ±1% of spot:  {gamma_1_share:.2%}")
print(f"Gamma within ±2% of spot:  {gamma_2_share:.2%}")
print(f"Gamma-weighted distance:   {gamma_weighted_distance:.2%}")
print(f"Call / Put gamma ratio:    {call_put_gamma_ratio:.3f}")

# ------------------------------------------------------------
# UPDATED GAMMA CHART
# ------------------------------------------------------------

fig = go.Figure()

fig.add_bar(
    x=gamma_clean["strike"],
    y=gamma_clean["call_gamma_mm"],
    name="Call gamma concentration"
)

fig.add_bar(
    x=gamma_clean["strike"],
    y=-gamma_clean["put_gamma_mm"],
    name="Put gamma concentration"
)

fig.add_vline(
    x=spot,
    line_dash="dash",
    annotation_text=f"SPY {spot:.2f}",
    annotation_position="top"
)

fig.update_layout(
    title="SPY Gamma Concentration — Validated IV",
    xaxis_title="Strike",
    yaxis_title="$mm gamma exposure per 1% SPY move",
    barmode="relative",
    template="plotly_white",
    height=650,
    hovermode="x unified"
)

fig.show()
```

```python
# ============================================================
# CELL 11 — HISTORICAL DAILY GAMMA STATE
# ============================================================

import numpy as np
import pandas as pd

df = hist_opt.copy()

df["date"] = pd.to_datetime(df["date"])
df["expiration"] = pd.to_datetime(df["expiration"])

# ------------------------------------------------------------
# GET SPY DAILY PRICE
# Needed to express strikes relative to spot
# ------------------------------------------------------------

spy_px = yf.download(
    "SPY",
    start="2025-01-01",
    end="2025-12-15",
    auto_adjust=True,
    progress=False
)

spy_close = spy_px["Close"].squeeze()

spot_df = (
    spy_close
    .rename("spot")
    .reset_index()
    .rename(columns={"Date": "date"})
)

spot_df["date"] = pd.to_datetime(spot_df["date"]).dt.tz_localize(None)

df = df.merge(
    spot_df,
    on="date",
    how="inner"
)

# ------------------------------------------------------------
# CONTRACT VARIABLES
# ------------------------------------------------------------

df["DTE"] = (
    df["expiration"] - df["date"]
).dt.days

df["distance"] = (
    df["strike"] / df["spot"] - 1
)

df["abs_distance"] = df["distance"].abs()

# Dollar gamma for a 1% SPY move
df["gamma_dollar"] = (
    df["gamma"]
    * df["open_interest"]
    * 100
    * df["spot"]**2
    * 0.01
)

# ------------------------------------------------------------
# RESEARCH UNIVERSE
#
# 1–60 DTE:
# removes same-day expiry noise while retaining short/medium
# dated positioning.
#
# ±20% moneyness:
# removes irrelevant deep wings.
# ------------------------------------------------------------

research = df[
    (df["DTE"] >= 1) &
    (df["DTE"] <= 60) &
    (df["abs_distance"] <= 0.20) &
    (df["open_interest"] > 0) &
    (df["gamma"] > 0)
].copy()

# ------------------------------------------------------------
# DAILY STATE FUNCTION
# ------------------------------------------------------------

def gamma_state(x):

    total = x["gamma_dollar"].sum()

    if total <= 0:
        return pd.Series(dtype=float)

    near1 = x.loc[
        x["abs_distance"] <= 0.01,
        "gamma_dollar"
    ].sum()

    near2 = x.loc[
        x["abs_distance"] <= 0.02,
        "gamma_dollar"
    ].sum()

    call_gamma = x.loc[
        x["type"] == "call",
        "gamma_dollar"
    ].sum()

    put_gamma = x.loc[
        x["type"] == "put",
        "gamma_dollar"
    ].sum()

    weighted_distance = np.average(
        x["abs_distance"],
        weights=x["gamma_dollar"]
    )

    # Herfindahl concentration across strikes
    strike_gamma = (
        x.groupby("strike")["gamma_dollar"]
        .sum()
    )

    weights = strike_gamma / strike_gamma.sum()

    gamma_hhi = np.sum(weights**2)

    # Strike carrying largest gamma mass
    max_gamma_strike = strike_gamma.idxmax()

    max_strike_distance = (
        max_gamma_strike / x["spot"].iloc[0] - 1
    )

    return pd.Series({

        "spot":
            x["spot"].iloc[0],

        "total_gamma":
            total,

        "gamma_1pct_share":
            near1 / total,

        "gamma_2pct_share":
            near2 / total,

        "gamma_weighted_distance":
            weighted_distance,

        "call_put_gamma_ratio":
            call_gamma / put_gamma
            if put_gamma > 0 else np.nan,

        "gamma_hhi":
            gamma_hhi,

        "max_gamma_strike":
            max_gamma_strike,

        "max_gamma_strike_distance":
            max_strike_distance,

        "contracts":
            len(x)
    })


daily_gamma = (
    research
    .groupby("date")
    .apply(gamma_state)
    .reset_index()
)

print("=" * 65)
print("HISTORICAL DAILY GAMMA STATE")
print("=" * 65)

print(f"Trading days: {len(daily_gamma):,}")

display(
    daily_gamma.tail(15).round(4)
)
```

```python
# ============================================================
# CELL 12 — CORRECTED HISTORICAL VOLATILITY-SURFACE STATE
# ============================================================

from scipy.interpolate import interp1d
import numpy as np
import pandas as pd


def interpolate_delta_iv(x, target_delta, option_type):

    if option_type == "call":
        z = x[
            (x["type"] == "call") &
            (x["delta"] >= 0.10) &
            (x["delta"] <= 0.45)
        ].copy()

    else:
        z = x[
            (x["type"] == "put") &
            (x["delta"] >= -0.45) &
            (x["delta"] <= -0.10)
        ].copy()

    z = (
        z[["delta", "implied_volatility"]]
        .dropna()
        .sort_values("delta")
        .drop_duplicates("delta")
    )

    if len(z) < 3:
        return np.nan

    if not (
        z["delta"].min()
        <= target_delta
        <= z["delta"].max()
    ):
        return np.nan

    return float(
        np.interp(
            target_delta,
            z["delta"],
            z["implied_volatility"]
        )
    )


def interpolate_atm_iv(chain, spot):

    # Use strikes immediately surrounding spot
    z = chain[
        (chain["strike"] >= spot * 0.95) &
        (chain["strike"] <= spot * 1.05) &
        (chain["implied_volatility"] > 0.03) &
        (chain["implied_volatility"] < 1.00)
    ].copy()

    # Average call/put IV at each strike
    z = (
        z.groupby("strike", as_index=False)
        ["implied_volatility"]
        .mean()
        .sort_values("strike")
    )

    below = z[z["strike"] <= spot]
    above = z[z["strike"] >= spot]

    if len(below) == 0 or len(above) == 0:
        return np.nan

    lo = below.iloc[-1]
    hi = above.iloc[0]

    # Spot exactly equals listed strike
    if lo["strike"] == hi["strike"]:
        return float(lo["implied_volatility"])

    # Linear interpolation to exact spot
    weight = (
        (spot - lo["strike"]) /
        (hi["strike"] - lo["strike"])
    )

    return float(
        lo["implied_volatility"]
        + weight *
        (
            hi["implied_volatility"]
            - lo["implied_volatility"]
        )
    )


# ------------------------------------------------------------
# CLEAN SURFACE UNIVERSE
# ------------------------------------------------------------

surface = df[
    (df["DTE"] >= 5) &
    (df["DTE"] <= 60) &
    (df["implied_volatility"] > 0.03) &
    (df["implied_volatility"] < 1.00) &
    (df["ask"] > 0)
].copy()


# ------------------------------------------------------------
# BUILD DAILY STATE
# ------------------------------------------------------------

surface_rows = []

for date, day in surface.groupby("date"):

    spot_today = day["spot"].iloc[0]

    expiries = (
        day[["expiration", "DTE"]]
        .drop_duplicates()
        .copy()
    )

    if len(expiries) == 0:
        continue


    # ==========================
    # ~30 DAY EXPIRY
    # ==========================

    idx30 = (
        expiries["DTE"] - 30
    ).abs().idxmin()

    expiry30 = expiries.loc[idx30, "expiration"]
    dte30 = expiries.loc[idx30, "DTE"]

    chain30 = day[
        day["expiration"] == expiry30
    ].copy()

    atm30 = interpolate_atm_iv(
        chain30,
        spot_today
    )

    put25 = interpolate_delta_iv(
        chain30,
        -0.25,
        "put"
    )

    call25 = interpolate_delta_iv(
        chain30,
        0.25,
        "call"
    )


    # ==========================
    # ~7 DAY EXPIRY
    # ==========================

    idx7 = (
        expiries["DTE"] - 7
    ).abs().idxmin()

    expiry7 = expiries.loc[idx7, "expiration"]
    dte7 = expiries.loc[idx7, "DTE"]

    chain7 = day[
        day["expiration"] == expiry7
    ].copy()

    atm7 = interpolate_atm_iv(
        chain7,
        spot_today
    )


    # ==========================
    # STORE STATE
    # ==========================

    surface_rows.append({

        "date":
            date,

        "DTE_30":
            dte30,

        "ATM_IV_30":
            atm30,

        "Put25_IV_30":
            put25,

        "Call25_IV_30":
            call25,

        "RR_25D_30":
            put25 - call25
            if np.isfinite(put25)
            and np.isfinite(call25)
            else np.nan,

        "DTE_7":
            dte7,

        "ATM_IV_7":
            atm7,

        "IV_term_slope":
            atm7 - atm30
            if np.isfinite(atm7)
            and np.isfinite(atm30)
            else np.nan
    })


daily_surface = (
    pd.DataFrame(surface_rows)
    .sort_values("date")
    .reset_index(drop=True)
)


print("=" * 65)
print("CORRECTED HISTORICAL VOLATILITY STATE")
print("=" * 65)

print(f"Trading days: {len(daily_surface):,}")

print("\nMissing observations:")

print(
    daily_surface[
        [
            "ATM_IV_30",
            "RR_25D_30",
            "ATM_IV_7",
            "IV_term_slope"
        ]
    ]
    .isna()
    .sum()
)

display(
    daily_surface.tail(15).round(4)
)
```

```python
# ============================================================
# CELL 13 — MASTER RESEARCH PANEL + FORWARD OUTCOMES
# ============================================================

# ------------------------------------------------------------
# MERGE GAMMA + VOL STATE
# ------------------------------------------------------------

panel = (
    daily_gamma
    .merge(
        daily_surface,
        on="date",
        how="inner"
    )
    .sort_values("date")
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# DAILY SPY LOG RETURNS
# ------------------------------------------------------------

panel["log_return"] = np.log(
    panel["spot"] / panel["spot"].shift(1)
)


# ------------------------------------------------------------
# FORWARD ABSOLUTE RETURNS
# ------------------------------------------------------------

panel["fwd_abs_return_1d"] = (
    np.log(
        panel["spot"].shift(-1)
        / panel["spot"]
    ).abs()
)

panel["fwd_abs_return_5d"] = (
    np.log(
        panel["spot"].shift(-5)
        / panel["spot"]
    ).abs()
)

panel["fwd_abs_return_10d"] = (
    np.log(
        panel["spot"].shift(-10)
        / panel["spot"]
    ).abs()
)


# ------------------------------------------------------------
# FORWARD REALIZED VOLATILITY
#
# Uses NEXT h daily returns, excluding today's return.
# Annualized using sqrt(252).
# ------------------------------------------------------------

returns = panel["log_return"]

def forward_rv(series, horizon):

    values = series.to_numpy()

    result = np.full(
        len(values),
        np.nan
    )

    for i in range(
        len(values) - horizon
    ):

        future_returns = values[
            i + 1:
            i + horizon + 1
        ]

        if np.isfinite(
            future_returns
        ).all():

            result[i] = (
                np.sqrt(
                    np.mean(
                        future_returns**2
                    )
                )
                * np.sqrt(252)
            )

    return result


panel["fwd_RV_5d"] = forward_rv(
    returns,
    5
)

panel["fwd_RV_10d"] = forward_rv(
    returns,
    10
)

panel["fwd_RV_20d"] = forward_rv(
    returns,
    20
)


# ------------------------------------------------------------
# EX-POST VOLATILITY RISK PREMIUM
#
# Positive:
# implied vol exceeded subsequent realized vol.
# ------------------------------------------------------------

panel["VRP_30_vs_fwd20"] = (
    panel["ATM_IV_30"]
    - panel["fwd_RV_20d"]
)


# ------------------------------------------------------------
# NORMALIZE TOTAL GAMMA
#
# Raw dollar gamma grows mechanically with SPY price,
# contract activity and market size.
#
# Divide by spot² for a more comparable state variable.
# ------------------------------------------------------------

panel["gamma_normalized"] = (
    panel["total_gamma"]
    / panel["spot"]**2
)


# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print("=" * 70)
print("MASTER GAMMA–VOL RESEARCH PANEL")
print("=" * 70)

print(f"Observations: {len(panel):,}")

print("\nDate range:")
print(
    panel["date"].min(),
    "→",
    panel["date"].max()
)

print("\nForward outcome availability:")

for col in [
    "fwd_abs_return_1d",
    "fwd_abs_return_5d",
    "fwd_abs_return_10d",
    "fwd_RV_5d",
    "fwd_RV_10d",
    "fwd_RV_20d"
]:
    print(
        f"{col:<22}",
        panel[col].notna().sum()
    )


# ------------------------------------------------------------
# SHOW FINAL RESEARCH VARIABLES
# ------------------------------------------------------------

research_cols = [

    "date",
    "spot",

    # Gamma state
    "gamma_normalized",
    "gamma_1pct_share",
    "gamma_2pct_share",
    "gamma_weighted_distance",
    "gamma_hhi",
    "call_put_gamma_ratio",
    "max_gamma_strike_distance",

    # Vol state
    "ATM_IV_30",
    "RR_25D_30",
    "ATM_IV_7",
    "IV_term_slope",

    # Future outcomes
    "fwd_abs_return_1d",
    "fwd_abs_return_5d",
    "fwd_abs_return_10d",
    "fwd_RV_5d",
    "fwd_RV_10d",
    "fwd_RV_20d",
    "VRP_30_vs_fwd20"
]

display(
    panel[
        research_cols
    ].tail(15).round(4)
)
```

```python
# ============================================================
# CELL 14 — GAMMA CONCENTRATION vs FORWARD 5D REALIZED VOL
# ============================================================

import statsmodels.api as sm
import plotly.graph_objects as go
import numpy as np
import pandas as pd


# ------------------------------------------------------------
# CLEAN SAMPLE
# ------------------------------------------------------------

test = panel[
    [
        "date",
        "gamma_1pct_share",
        "gamma_2pct_share",
        "gamma_weighted_distance",
        "gamma_hhi",
        "ATM_IV_30",
        "fwd_RV_5d"
    ]
].dropna().copy()


# ------------------------------------------------------------
# SIMPLE OLS
# Forward RV ~ gamma concentration
# ------------------------------------------------------------

X = sm.add_constant(
    test["gamma_1pct_share"]
)

model = sm.OLS(
    test["fwd_RV_5d"],
    X
).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 4}
)

print("=" * 65)
print("GAMMA CONCENTRATION → FORWARD 5D REALIZED VOL")
print("=" * 65)

print(
    f"Observations:          {len(test)}"
)

print(
    f"Gamma coefficient:    "
    f"{model.params['gamma_1pct_share']:.4f}"
)

print(
    f"HAC t-statistic:       "
    f"{model.tvalues['gamma_1pct_share']:.3f}"
)

print(
    f"HAC p-value:           "
    f"{model.pvalues['gamma_1pct_share']:.4f}"
)

print(
    f"R²:                    "
    f"{model.rsquared:.3f}"
)

print(
    f"Correlation:           "
    f"{test['gamma_1pct_share'].corr(test['fwd_RV_5d']):.3f}"
)


# ------------------------------------------------------------
# REGRESSION LINE
# ------------------------------------------------------------

x_line = np.linspace(
    test["gamma_1pct_share"].min(),
    test["gamma_1pct_share"].max(),
    100
)

y_line = (
    model.params["const"]
    + model.params["gamma_1pct_share"]
    * x_line
)


# ------------------------------------------------------------
# SCATTER
# ------------------------------------------------------------

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=test["gamma_1pct_share"] * 100,
        y=test["fwd_RV_5d"] * 100,
        mode="markers",
        name="Daily observations",
        marker=dict(
            size=7,
            opacity=0.50
        ),
        customdata=test["date"],
        hovertemplate=(
            "Date: %{customdata|%Y-%m-%d}<br>"
            "Gamma within ±1%: %{x:.1f}%<br>"
            "Forward 5D RV: %{y:.1f}%"
            "<extra></extra>"
        )
    )
)

fig.add_trace(
    go.Scatter(
        x=x_line * 100,
        y=y_line * 100,
        mode="lines",
        name="OLS fit",
        line=dict(width=3)
    )
)

fig.update_layout(
    title="Gamma Concentration vs Subsequent 5-Day SPY Realized Volatility",
    xaxis_title="Share of gamma within ±1% of spot (%)",
    yaxis_title="Forward 5-day realized volatility (%)",
    template="plotly_white",
    height=600,
    hovermode="closest"
)

fig.show()


# ------------------------------------------------------------
# QUINTILE TEST
# ------------------------------------------------------------

test["gamma_quintile"] = pd.qcut(
    test["gamma_1pct_share"],
    5,
    labels=[
        "Q1 — Lowest",
        "Q2",
        "Q3",
        "Q4",
        "Q5 — Highest"
    ]
)

quintiles = (
    test.groupby(
        "gamma_quintile",
        observed=True
    )
    .agg(
        mean_gamma=("gamma_1pct_share", "mean"),
        mean_fwd_RV=("fwd_RV_5d", "mean"),
        median_fwd_RV=("fwd_RV_5d", "median"),
        observations=("fwd_RV_5d", "size")
    )
    .reset_index()
)

print("\n" + "=" * 65)
print("GAMMA CONCENTRATION QUINTILES")
print("=" * 65)

display(
    quintiles.assign(
        mean_gamma=
            quintiles["mean_gamma"] * 100,
        mean_fwd_RV=
            quintiles["mean_fwd_RV"] * 100,
        median_fwd_RV=
            quintiles["median_fwd_RV"] * 100
    ).round(2)
)


fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=quintiles["gamma_quintile"],
        y=quintiles["mean_fwd_RV"] * 100,
        name="Mean forward 5D RV",
        text=(
            quintiles["mean_fwd_RV"] * 100
        ).round(1),
        texttemplate="%{text:.1f}%",
        textposition="outside"
    )
)

fig.update_layout(
    title="Subsequent 5-Day Realized Volatility by Gamma-Concentration Quintile",
    xaxis_title="Gamma concentration quintile",
    yaxis_title="Mean forward 5-day realized volatility (%)",
    template="plotly_white",
    height=550,
    showlegend=False
)

fig.show()
```

```python
# ============================================================
# CELL 16 — GAMMA PINNING TEST
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go


pin = panel[
    [
        "date",
        "spot",
        "max_gamma_strike",
        "max_gamma_strike_distance",
        "gamma_hhi",
        "gamma_1pct_share",
        "ATM_IV_30",
        "fwd_abs_return_5d"
    ]
].dropna().copy()


# ------------------------------------------------------------
# DISTANCE TO DOMINANT GAMMA STRIKE
# ------------------------------------------------------------

pin["gamma_strike_distance"] = (
    pin["max_gamma_strike_distance"].abs()
)


# ------------------------------------------------------------
# IV-IMPLIED 5D MOVE
#
# sigma * sqrt(5/252)
# ------------------------------------------------------------

pin["implied_move_5d"] = (
    pin["ATM_IV_30"]
    * np.sqrt(5 / 252)
)


# ------------------------------------------------------------
# REALIZED MOVEMENT RELATIVE TO IMPLIED
#
# < 1 = moved less than implied
# > 1 = moved more than implied
# ------------------------------------------------------------

pin["move_vs_implied"] = (
    pin["fwd_abs_return_5d"]
    / pin["implied_move_5d"]
)


# ------------------------------------------------------------
# REGRESSION
#
# If proximity to dominant gamma strike matters:
#
# larger distance -> larger move_vs_implied
#
# Therefore expected beta > 0.
# ------------------------------------------------------------

X = sm.add_constant(
    pin[
        [
            "gamma_strike_distance",
            "gamma_hhi",
            "gamma_1pct_share"
        ]
    ]
)

model_pin = sm.OLS(
    pin["move_vs_implied"],
    X
).fit(
    cov_type="HAC",
    cov_kwds={"maxlags": 4}
)


print("=" * 70)
print("GAMMA PINNING TEST")
print("=" * 70)

print(
    f"Observations:              {len(pin)}"
)

print(
    f"Distance coefficient:      "
    f"{model_pin.params['gamma_strike_distance']:.4f}"
)

print(
    f"HAC t-statistic:           "
    f"{model_pin.tvalues['gamma_strike_distance']:.3f}"
)

print(
    f"HAC p-value:               "
    f"{model_pin.pvalues['gamma_strike_distance']:.4f}"
)

print(
    f"R²:                        "
    f"{model_pin.rsquared:.3f}"
)


# ------------------------------------------------------------
# DISTANCE QUINTILES
# ------------------------------------------------------------

pin["distance_quintile"] = pd.qcut(
    pin["gamma_strike_distance"],
    5,
    labels=[
        "Q1 — Closest",
        "Q2",
        "Q3",
        "Q4",
        "Q5 — Furthest"
    ]
)

pin_q = (
    pin.groupby(
        "distance_quintile",
        observed=True
    )
    .agg(
        mean_distance=(
            "gamma_strike_distance",
            "mean"
        ),
        mean_move_vs_implied=(
            "move_vs_implied",
            "mean"
        ),
        median_move_vs_implied=(
            "move_vs_implied",
            "median"
        ),
        observations=(
            "move_vs_implied",
            "size"
        )
    )
    .reset_index()
)

print("\n" + "=" * 70)
print("DISTANCE FROM DOMINANT GAMMA STRIKE")
print("=" * 70)

display(
    pin_q.assign(
        mean_distance=
            pin_q["mean_distance"] * 100
    ).round(3)
)


# ------------------------------------------------------------
# PRESENTATION
#
# Bars are appropriate here because we're comparing
# discrete distance regimes.
# 1.0 line = realized move equalled implied move.
# ------------------------------------------------------------

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=pin_q["distance_quintile"],
        y=pin_q["mean_move_vs_implied"],
        text=pin_q["mean_move_vs_implied"].round(2),
        textposition="outside",
        name="Realized / implied move"
    )
)

fig.add_hline(
    y=1,
    line_dash="dash",
    annotation_text="Implied move",
    annotation_position="top right"
)

fig.update_layout(
    title="SPY Movement Relative to Implied Move by Distance from Dominant Gamma Strike",
    xaxis_title="Starting distance from dominant gamma strike",
    yaxis_title="5-day absolute move / IV-implied 5-day move",
    template="plotly_white",
    height=550,
    showlegend=False
)

fig.show()
```

```python
# ============================================================
# CELL 19 — CURRENT DEALER GAMMA PROFILE / GAMMA FLIP
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go
from scipy.stats import norm


gp = clean.copy()

gp = gp[
    (gp["openInterest"] > 0) &
    (gp["IV_mid"] > 0) &
    (gp["T"] > 0)
].copy()


# ------------------------------------------------------------
# BLACK-SCHOLES GAMMA
# Recalculate gamma at hypothetical SPY levels.
#
# We hold:
#   OI, IV and time-to-expiry constant
#
# and vary only spot.
# ------------------------------------------------------------

def bs_gamma(S, K, T, sigma, r=0.0):

    d1 = (
        np.log(S / K)
        + (r + 0.5 * sigma**2) * T
    ) / (
        sigma * np.sqrt(T)
    )

    return (
        norm.pdf(d1)
        / (S * sigma * np.sqrt(T))
    )


# ------------------------------------------------------------
# SPOT GRID
# ------------------------------------------------------------

spot_grid = np.linspace(
    spot * 0.90,
    spot * 1.10,
    201
)


profile = []


for S_test in spot_grid:

    gamma_test = bs_gamma(
        S=S_test,
        K=gp["strike"].values,
        T=gp["T"].values,
        sigma=gp["IV_mid"].values
    )

    dollar_gamma = (
        gamma_test
        * gp["openInterest"].values
        * 100
        * S_test**2
        * 0.01
    )

    # Same sign convention as Cell 18:
    # CALL + / PUT -

    signs = np.where(
        gp["optionType"].str.lower().eq("call"),
        1,
        -1
    )

    net_gex = np.sum(
        dollar_gamma * signs
    )

    profile.append(
        {
            "spot_level": S_test,
            "net_gex": net_gex
        }
    )


profile = pd.DataFrame(profile)

profile["net_gex_bn"] = (
    profile["net_gex"] / 1e9
)


# ------------------------------------------------------------
# FIND ZERO-GAMMA CROSSINGS
# ------------------------------------------------------------

crossings = []

for i in range(len(profile) - 1):

    y1 = profile.loc[i, "net_gex"]
    y2 = profile.loc[i + 1, "net_gex"]

    if y1 * y2 < 0:

        x1 = profile.loc[i, "spot_level"]
        x2 = profile.loc[i + 1, "spot_level"]

        # linear interpolation between grid points

        zero = (
            x1
            - y1 * (x2 - x1) / (y2 - y1)
        )

        crossings.append(zero)


print("=" * 68)
print("CURRENT SPY DEALER GAMMA PROFILE")
print("=" * 68)

print(
    f"Current SPY spot:       ${spot:,.2f}"
)

if crossings:

    print(
        "Gamma flip level(s):   "
        + ", ".join(
            f"${x:,.2f}" for x in crossings
        )
    )

else:

    print(
        "Gamma flip level:      "
        "No zero crossing within ±10% of spot"
    )


# ------------------------------------------------------------
# CURRENT GEX FROM PROFILE
# ------------------------------------------------------------

current_idx = (
    profile["spot_level"] - spot
).abs().idxmin()

current_profile_gex = (
    profile.loc[current_idx, "net_gex_bn"]
)

print(
    f"Current net GEX proxy:  "
    f"${current_profile_gex:,.3f}bn / 1% move"
)


# ------------------------------------------------------------
# CHART
# ------------------------------------------------------------

fig = go.Figure()


fig.add_trace(
    go.Scatter(
        x=profile["spot_level"],
        y=profile["net_gex_bn"],
        mode="lines",
        name="Net dealer GEX",
        line=dict(width=3),

        hovertemplate=(
            "Hypothetical SPY: $%{x:.2f}<br>"
            "Net GEX: $%{y:.2f}bn / 1% move"
            "<extra></extra>"
        )
    )
)


# zero gamma line

fig.add_hline(
    y=0,
    line_dash="dash",
    line_width=1
)


# current spot

fig.add_vline(
    x=spot,
    line_dash="dash",
    line_width=2,
    annotation_text=f"Current SPY {spot:.2f}",
    annotation_position="top"
)


# gamma flip(s)

for zero in crossings:

    fig.add_vline(
        x=zero,
        line_dash="dot",
        line_width=2,
        annotation_text=f"Gamma flip {zero:.2f}",
        annotation_position="bottom"
    )


fig.update_layout(

    title={
        "text":
            "SPY Estimated Dealer Gamma Profile"
            "<br>"
            "<sup>Net GEX as SPY spot changes | "
            "Today's OI and IV surface held constant | CALL + / PUT −</sup>",

        "x": 0.5,
        "xanchor": "center"
    },

    xaxis_title="Hypothetical SPY spot",

    yaxis_title="Estimated net dealer GEX ($bn / 1% move)",

    template="plotly_white",

    height=600,

    hovermode="x unified"
)

fig.show()
```

```python
# ============================================================
# CELL 20 — ESTIMATED DEALER DELTA EXPOSURE BY STRIKE
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go


dx = clean.copy()

dx = dx[
    (dx["openInterest"] > 0) &
    (dx["T"] > 0) &
    (dx["IV_mid"] > 0) &
    (dx["strike"] >= spot * 0.94) &
    (dx["strike"] <= spot * 1.06)
].copy()


# ------------------------------------------------------------
# USE OUR MID-IV DELTA
#
# Cell 6 calculated delta from independently solved IV.
# Handle likely column names automatically.
# ------------------------------------------------------------

delta_candidates = [
    "delta_mid",
    "Delta_mid",
    "delta_calc",
    "delta"
]

delta_col = next(
    (c for c in delta_candidates if c in dx.columns),
    None
)

if delta_col is None:
    raise KeyError(
        "No calculated delta column found. "
        f"Available columns: {list(dx.columns)}"
    )


# ------------------------------------------------------------
# DEALER DELTA PROXY
#
# Maintain Cell 18 inventory convention:
#
# calls: + option exposure
# puts:  - option exposure
#
# DEX = assumed dealer position × option delta ×
#       OI × 100 × spot
#
# This is dollar-equivalent delta exposure.
# ------------------------------------------------------------

position_sign = np.where(
    dx["optionType"].str.lower().eq("call"),
    1.0,
    -1.0
)

dx["dealer_dex"] = (
    position_sign
    * dx[delta_col]
    * dx["openInterest"]
    * 100
    * spot
)


# ------------------------------------------------------------
# AGGREGATE BY STRIKE
# ------------------------------------------------------------

strike_dex = (
    dx
    .groupby("strike", as_index=False)
    .agg(
        net_dex=("dealer_dex", "sum")
    )
    .sort_values("strike")
)

strike_dex["net_dex_bn"] = (
    strike_dex["net_dex"] / 1e9
)

total_dex = strike_dex["net_dex"].sum()


print("=" * 68)
print("CURRENT ESTIMATED SPY DEALER DELTA STATE")
print("=" * 68)

print(f"SPY spot:              ${spot:,.2f}")
print(f"Delta column used:     {delta_col}")
print(
    f"Net DEX proxy:         "
    f"${total_dex / 1e9:,.3f}bn equivalent"
)

print("\nDealer inventory is inferred, not observed.")


# ------------------------------------------------------------
# CHART
# ------------------------------------------------------------

positive = strike_dex[
    strike_dex["net_dex_bn"] >= 0
]

negative = strike_dex[
    strike_dex["net_dex_bn"] < 0
]


fig = go.Figure()


fig.add_trace(
    go.Bar(
        x=positive["strike"],
        y=positive["net_dex_bn"],
        name="Positive DEX",
        marker_color="#2E8B57",

        hovertemplate=(
            "Strike: %{x:.0f}<br>"
            "DEX: $%{y:.2f}bn"
            "<extra></extra>"
        )
    )
)


fig.add_trace(
    go.Bar(
        x=negative["strike"],
        y=negative["net_dex_bn"],
        name="Negative DEX",
        marker_color="#C94C4C",

        hovertemplate=(
            "Strike: %{x:.0f}<br>"
            "DEX: $%{y:.2f}bn"
            "<extra></extra>"
        )
    )
)


fig.add_vline(
    x=spot,
    line_dash="dash",
    line_width=2,
    annotation_text=f"SPY {spot:.2f}",
    annotation_position="top"
)


fig.add_hline(
    y=0,
    line_width=1
)


fig.update_layout(

    title={
        "text":
            "SPY Estimated Dealer Delta Exposure by Strike"
            "<br>"
            "<sup>Current dollar-equivalent delta exposure | "
            "Dealer inventory inferred from positioning convention</sup>",
        "x": 0.5,
        "xanchor": "center"
    },

    xaxis_title="Strike",

    yaxis_title="Estimated dealer DEX ($bn equivalent)",

    template="plotly_white",

    height=600,

    barmode="relative",

    hovermode="closest",

    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="left",
        x=0
    )
)


fig.update_xaxes(
    range=[
        spot * 0.94,
        spot * 1.06
    ]
)

fig.show()


# ------------------------------------------------------------
# LARGEST DEX NODES
# ------------------------------------------------------------

print("\nLARGEST POSITIVE DEX STRIKES")

display(
    strike_dex
    .nlargest(5, "net_dex")
    [["strike", "net_dex_bn"]]
    .round(3)
)


print("\nLARGEST NEGATIVE DEX STRIKES")

display(
    strike_dex
    .nsmallest(5, "net_dex")
    [["strike", "net_dex_bn"]]
    .round(3)
)
```

```python
# ============================================================
# CELL 21 — CURRENT SPY IMPLIED VOLATILITY SURFACE
# Corrected: Delta × Maturity with robust ATM construction
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go


# ------------------------------------------------------------
# DATA
# ------------------------------------------------------------

vs = clean.copy()

vs = vs[
    (vs["IV_mid"] > 0) &
    (vs["T"] > 0)
].copy()

vs["DTE"] = vs["T"] * 365.25


# ------------------------------------------------------------
# FIND CALCULATED DELTA COLUMN
# ------------------------------------------------------------

delta_candidates = [
    "delta_mid",
    "Delta_mid",
    "delta_calc",
    "delta"
]

delta_col = next(
    (c for c in delta_candidates if c in vs.columns),
    None
)

if delta_col is None:
    raise KeyError("No calculated delta column found.")


vs["abs_delta"] = vs[delta_col].abs()


# ------------------------------------------------------------
# DELTA GRID
#
# We deliberately stop at 45Δ.
# ATM is calculated independently from strike-space IV.
# ------------------------------------------------------------

delta_grid = np.array([
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.35,
    0.40,
    0.45
])


surface_rows = []


# ------------------------------------------------------------
# BUILD EACH EXPIRY'S SMILE
# ------------------------------------------------------------

for expiry, chain in vs.groupby("expiry"):

    dte = chain["DTE"].median()

    row = {
        "expiry": expiry,
        "DTE": dte
    }


    # --------------------------------------------------------
    # ATM IV
    #
    # Use IV at strikes immediately surrounding current spot.
    # Average call/put IV at each strike first, then interpolate
    # in strike space to exact spot.
    # --------------------------------------------------------

    atm_chain = (
        chain[
            (chain["strike"] >= spot * 0.95) &
            (chain["strike"] <= spot * 1.05)
        ]
        .groupby("strike", as_index=False)
        .agg(
            IV_mid=("IV_mid", "mean")
        )
        .sort_values("strike")
    )


    below = atm_chain[
        atm_chain["strike"] <= spot
    ].tail(1)

    above = atm_chain[
        atm_chain["strike"] >= spot
    ].head(1)


    if len(below) == 1 and len(above) == 1:

        k1 = below["strike"].iloc[0]
        v1 = below["IV_mid"].iloc[0]

        k2 = above["strike"].iloc[0]
        v2 = above["IV_mid"].iloc[0]

        if k1 == k2:

            atm_iv = v1

        else:

            atm_iv = (
                v1
                + (spot - k1)
                * (v2 - v1)
                / (k2 - k1)
            )

    else:

        atm_iv = np.nan


    row["ATM"] = atm_iv


    # --------------------------------------------------------
    # PUT / CALL DELTA SMILES
    # --------------------------------------------------------

    for opt_type, prefix in [
        ("call", "Call"),
        ("put", "Put")
    ]:

        side = chain[
            chain["optionType"]
            .str.lower()
            .eq(opt_type)
        ].copy()


        side = side[
            (side["abs_delta"] >= 0.08) &
            (side["abs_delta"] <= 0.48)
        ]


        side = (
            side
            .sort_values("abs_delta")
            .drop_duplicates(
                subset="abs_delta",
                keep="last"
            )
        )


        if len(side) < 2:
            continue


        for d in delta_grid:

            if (
                d >= side["abs_delta"].min()
                and
                d <= side["abs_delta"].max()
            ):

                row[f"{prefix}_{int(d*100)}"] = (
                    np.interp(
                        d,
                        side["abs_delta"],
                        side["IV_mid"]
                    )
                )


    surface_rows.append(row)


surface = (
    pd.DataFrame(surface_rows)
    .sort_values("DTE")
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# X-AXIS
#
# Downside wing -> ATM -> upside wing
# ------------------------------------------------------------

labels = [
    "10Δ Put",
    "15Δ Put",
    "20Δ Put",
    "25Δ Put",
    "30Δ Put",
    "35Δ Put",
    "40Δ Put",
    "45Δ Put",

    "ATM",

    "45Δ Call",
    "40Δ Call",
    "35Δ Call",
    "30Δ Call",
    "25Δ Call",
    "20Δ Call",
    "15Δ Call",
    "10Δ Call"
]


matrix = []


for _, row in surface.iterrows():

    values = [

        row.get("Put_10", np.nan),
        row.get("Put_15", np.nan),
        row.get("Put_20", np.nan),
        row.get("Put_25", np.nan),
        row.get("Put_30", np.nan),
        row.get("Put_35", np.nan),
        row.get("Put_40", np.nan),
        row.get("Put_45", np.nan),

        row.get("ATM", np.nan),

        row.get("Call_45", np.nan),
        row.get("Call_40", np.nan),
        row.get("Call_35", np.nan),
        row.get("Call_30", np.nan),
        row.get("Call_25", np.nan),
        row.get("Call_20", np.nan),
        row.get("Call_15", np.nan),
        row.get("Call_10", np.nan)
    ]

    matrix.append(
        np.array(values, dtype=float) * 100
    )


matrix = np.array(matrix)


# ------------------------------------------------------------
# Y-AXIS
# ------------------------------------------------------------

y_labels = [
    f"{r['expiry']} | {r['DTE']:.1f}D"
    for _, r in surface.iterrows()
]


# ------------------------------------------------------------
# HEATMAP
# ------------------------------------------------------------

fig = go.Figure()


fig.add_trace(
    go.Heatmap(

        x=labels,
        y=y_labels,
        z=matrix,

        colorscale="Viridis",

        colorbar=dict(
            title="Implied vol (%)"
        ),

        hovertemplate=(
            "%{x}<br>"
            "Expiry: %{y}<br>"
            "IV: %{z:.2f}%"
            "<extra></extra>"
        )
    )
)


fig.update_layout(

    title={
        "text":
            "SPY Current Implied Volatility Surface"
            "<br>"
            "<sup>Mid-price solved IV | Delta × expiry</sup>",

        "x": 0.5,
        "xanchor": "center"
    },

    xaxis_title="Option delta / moneyness",

    yaxis_title="Expiry | Days to expiry",

    template="plotly_white",

    height=620,

    margin=dict(
        l=120,
        r=120,
        t=100,
        b=90
    )
)


fig.show()


# ------------------------------------------------------------
# QUICK VALIDATION
# ------------------------------------------------------------

print("=" * 65)
print("ATM IV BY EXPIRY")
print("=" * 65)

display(
    surface[
        [
            "expiry",
            "DTE",
            "ATM"
        ]
    ]
    .assign(
        ATM=lambda x: x["ATM"] * 100
    )
    .round(
        {
            "DTE": 1,
            "ATM": 2
        }
    )
)
```

```python
# ============================================================
# CELL 22 — CURRENT SPY VOLATILITY PREMIUM CURVE
# ATM IV minus trailing realized volatility
# ============================================================

import numpy as np
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go


# ------------------------------------------------------------
# GET SPY PRICE HISTORY
# Adjusted prices appropriate here for realized returns
# ------------------------------------------------------------

px = yf.download(
    "SPY",
    period="6mo",
    auto_adjust=True,
    progress=False
)

close = px["Close"].squeeze().dropna()

returns = np.log(
    close / close.shift(1)
).dropna()


# ------------------------------------------------------------
# REALIZED VOL FUNCTION
# ------------------------------------------------------------

def trailing_rv(n):

    n = max(int(n), 2)

    if len(returns) < n:
        return np.nan

    r = returns.tail(n)

    return (
        np.sqrt(np.mean(r**2))
        * np.sqrt(252)
    )


# ------------------------------------------------------------
# USE ATM SURFACE FROM CELL 21
#
# Convert calendar DTE approximately into trading days:
# DTE × 252 / 365
# ------------------------------------------------------------

vp = surface[
    [
        "expiry",
        "DTE",
        "ATM"
    ]
].copy()


vp["RV_window"] = (
    vp["DTE"]
    * 252 / 365.25
).round().astype(int)

vp["RV_window"] = (
    vp["RV_window"]
    .clip(lower=2)
)


vp["Trailing_RV"] = (
    vp["RV_window"]
    .apply(trailing_rv)
)


# ------------------------------------------------------------
# VOLATILITY PREMIUM
#
# Percentage-point spread:
#
# ATM IV - trailing RV
# ------------------------------------------------------------

vp["Vol_Premium"] = (
    vp["ATM"]
    - vp["Trailing_RV"]
)


# Convert for presentation

vp["ATM_pct"] = vp["ATM"] * 100
vp["RV_pct"] = vp["Trailing_RV"] * 100
vp["Premium_pct"] = vp["Vol_Premium"] * 100


# ------------------------------------------------------------
# OUTPUT TABLE
# ------------------------------------------------------------

print("=" * 72)
print("CURRENT SPY VOLATILITY PREMIUM")
print("ATM IV minus matched trailing realized volatility")
print("=" * 72)

display(
    vp[
        [
            "expiry",
            "DTE",
            "RV_window",
            "ATM_pct",
            "RV_pct",
            "Premium_pct"
        ]
    ]
    .rename(
        columns={
            "ATM_pct": "ATM_IV_%",
            "RV_pct": "Trailing_RV_%",
            "Premium_pct": "Vol_Premium_pts"
        }
    )
    .round(2)
)


# ------------------------------------------------------------
# LINE CHART
#
# Main lines = IV and RV
# Shaded distance between them = descriptive vol premium
# ------------------------------------------------------------

fig = go.Figure()


fig.add_trace(
    go.Scatter(
        x=vp["DTE"],
        y=vp["ATM_pct"],

        mode="lines+markers",

        name="ATM implied volatility",

        line=dict(width=3),

        hovertemplate=(
            "DTE: %{x:.1f}<br>"
            "ATM IV: %{y:.2f}%"
            "<extra></extra>"
        )
    )
)


fig.add_trace(
    go.Scatter(
        x=vp["DTE"],
        y=vp["RV_pct"],

        mode="lines+markers",

        name="Matched trailing realized volatility",

        line=dict(width=3),

        hovertemplate=(
            "DTE: %{x:.1f}<br>"
            "Trailing RV: %{y:.2f}%"
            "<extra></extra>"
        )
    )
)


fig.update_layout(

    title={
        "text":
            "SPY Current Volatility Premium Across Maturity"
            "<br>"
            "<sup>ATM implied volatility vs matched trailing realized volatility</sup>",

        "x": 0.5,
        "xanchor": "center"
    },

    xaxis_title="Days to expiry",

    yaxis_title="Annualized volatility (%)",

    template="plotly_white",

    height=600,

    hovermode="x unified",

    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="left",
        x=0
    )
)

fig.show()
```
