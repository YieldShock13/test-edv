```python
import yfinance as yf
import pandas as pd
import numpy as np

portfolio = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

controls = ["O", "CVX", "SPY"]

tickers = portfolio + controls

results = []

for ticker in tickers:

    print("Processing:", ticker)

    try:
        x = yf.Ticker(ticker).history(
            period="5y",
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=False
        )

        if x.empty or len(x) < 2:
            results.append({
                "Ticker": ticker,
                "Group": "CONTROL" if ticker in controls else "PORTFOLIO",
                "Status": "INSUFFICIENT DATA"
            })
            continue

        # ----------------------------------------------------
        # CLEAN FIELDS
        # ----------------------------------------------------

        close = x["Close"].astype(float)

        adj = x["Adj Close"].astype(float)

        div = (
            x["Dividends"].fillna(0).astype(float)
            if "Dividends" in x.columns
            else pd.Series(0.0, index=x.index)
        )

        splits = (
            x["Stock Splits"].fillna(0).astype(float)
            if "Stock Splits" in x.columns
            else pd.Series(0.0, index=x.index)
        )

        # ----------------------------------------------------
        # 1. RAW PRICE RETURN
        # ----------------------------------------------------

        price_return = (
            close.iloc[-1] / close.iloc[0] - 1
        )

        # ----------------------------------------------------
        # 2. YAHOO ADJUSTED-CLOSE RETURN
        # ----------------------------------------------------

        yahoo_adj_return = (
            adj.iloc[-1] / adj.iloc[0] - 1
        )

        # ----------------------------------------------------
        # 3. INDEPENDENT TOTAL-RETURN RECONSTRUCTION
        #
        # Dividend:
        #     (P_t + D_t) / P_(t-1)
        #
        # Split:
        # Yahoo's split field is shares-after / shares-before.
        # Therefore economic ending value is:
        #
        #     P_t * split_factor
        #
        # when a split occurs.
        #
        # Normal day => factor = 1
        # ----------------------------------------------------

        split_factor = splits.copy()

        split_factor = split_factor.where(
            split_factor != 0,
            1.0
        )

        previous_close = close.shift(1)

        daily_total_return = (
            (
                close * split_factor + div
            )
            / previous_close
            - 1
        )

        daily_total_return = daily_total_return.dropna()

        reconstructed_return = (
            (1 + daily_total_return).prod() - 1
        )

        # ----------------------------------------------------
        # EVENT INFORMATION
        # ----------------------------------------------------

        dividend_events = int((div != 0).sum())
        split_events = int((splits != 0).sum())

        total_dividends = float(div.sum())

        # ----------------------------------------------------
        # COMPARISONS
        # ----------------------------------------------------

        reconstruction_vs_adj = (
            reconstructed_return -
            yahoo_adj_return
        )

        adj_vs_price = (
            yahoo_adj_return -
            price_return
        )

        reconstruction_vs_price = (
            reconstructed_return -
            price_return
        )

        results.append({
            "Ticker": ticker,
            "Group":
                "CONTROL" if ticker in controls
                else "PORTFOLIO",

            "Start":
                close.index[0].date(),

            "End":
                close.index[-1].date(),

            "Price_Return_%":
                price_return * 100,

            "Reconstructed_TR_%":
                reconstructed_return * 100,

            "Yahoo_Adj_Return_%":
                yahoo_adj_return * 100,

            "Recon_minus_Adj_pp":
                reconstruction_vs_adj * 100,

            "Adj_minus_Price_pp":
                adj_vs_price * 100,

            "Recon_minus_Price_pp":
                reconstruction_vs_price * 100,

            "Dividend_Events":
                dividend_events,

            "Total_Dividends":
                total_dividends,

            "Split_Events":
                split_events,

            "Status": "OK"
        })

    except Exception as e:

        results.append({
            "Ticker": ticker,
            "Group":
                "CONTROL" if ticker in controls
                else "PORTFOLIO",
            "Status": f"ERROR: {e}"
        })


out = pd.DataFrame(results)


# ============================================================
# OUTPUT
# ============================================================

cols = [
    "Ticker",
    "Group",
    "Price_Return_%",
    "Reconstructed_TR_%",
    "Yahoo_Adj_Return_%",
    "Recon_minus_Adj_pp",
    "Adj_minus_Price_pp",
    "Dividend_Events",
    "Split_Events",
    "Status"
]

print("\n" + "=" * 145)
print("TOTAL RETURN VALIDATION")
print("=" * 145)

print(
    out[cols].to_string(
        index=False,
        float_format=lambda v: f"{v:,.2f}"
    )
)


# ============================================================
# CONTROLS
# ============================================================

print("\n" + "=" * 145)
print("CONTROL VALIDATION — O / CVX / SPY")
print("=" * 145)

control_output = out[out["Group"] == "CONTROL"]

print(
    control_output[cols].to_string(
        index=False,
        float_format=lambda v: f"{v:,.2f}"
    )
)


# ============================================================
# PORTFOLIO — BIGGEST DISAGREEMENTS
# ============================================================

print("\n" + "=" * 145)
print("PORTFOLIO — LARGEST RECONSTRUCTED vs YAHOO ADJUSTED DIFFERENCES")
print("=" * 145)

portfolio_output = (
    out[out["Group"] == "PORTFOLIO"]
    .copy()
)

portfolio_output["ABS_DIFF"] = (
    portfolio_output["Recon_minus_Adj_pp"].abs()
)

portfolio_output = portfolio_output.sort_values(
    "ABS_DIFF",
    ascending=False
)

print(
    portfolio_output[cols].to_string(
        index=False,
        float_format=lambda v: f"{v:,.2f}"
    )
)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

rows = []

for ticker in tickers:

    x = yf.Ticker(ticker).history(
        period="5y",
        interval="1d",
        auto_adjust=False,
        actions=True,
        repair=False
    )

    if x.empty or "Dividends" not in x.columns:
        continue

    x = x.copy()

    # --------------------------------------------------------
    # DAILY PRICE RETURN
    # --------------------------------------------------------

    x["Price_Return"] = (
        x["Close"] / x["Close"].shift(1) - 1
    )

    # --------------------------------------------------------
    # OUR RECONSTRUCTED RETURN
    # --------------------------------------------------------

    split = x["Stock Splits"].fillna(0)

    split_factor = split.where(
        split != 0,
        1.0
    )

    x["Recon_Return"] = (
        (
            x["Close"] * split_factor
            + x["Dividends"].fillna(0)
        )
        / x["Close"].shift(1)
        - 1
    )

    # --------------------------------------------------------
    # YAHOO ADJUSTED RETURN
    # --------------------------------------------------------

    x["Adj_Return"] = (
        x["Adj Close"] / x["Adj Close"].shift(1) - 1
    )

    # Dividend-event observations only
    events = x[x["Dividends"].fillna(0) != 0].copy()

    for date, r in events.iterrows():

        recon_adj_diff_bp = (
            r["Recon_Return"] - r["Adj_Return"]
        ) * 10000

        price_adj_diff_bp = (
            r["Price_Return"] - r["Adj_Return"]
        ) * 10000

        rows.append({
            "Ticker": ticker,
            "Date": date.date(),
            "Dividend": r["Dividends"],

            "Price_Return_%":
                r["Price_Return"] * 100,

            "Reconstructed_Return_%":
                r["Recon_Return"] * 100,

            "Adjusted_Return_%":
                r["Adj_Return"] * 100,

            "Recon_vs_Adj_bp":
                recon_adj_diff_bp,

            "Price_vs_Adj_bp":
                price_adj_diff_bp
        })


events = pd.DataFrame(rows)

# ============================================================
# SEMANTIC CLASSIFICATION
#
# Recon ≈ Adjusted: within 10 bp
# Adjusted materially differs from Price: > 25 bp
# ============================================================

events["RECON_MATCHES_ADJ"] = (
    events["Recon_vs_Adj_bp"].abs() <= 10
)

events["ADJ_DIFFERS_FROM_PRICE"] = (
    events["Price_vs_Adj_bp"].abs() >= 25
)

events["VALIDATION_CASE"] = (
    events["RECON_MATCHES_ADJ"]
    &
    events["ADJ_DIFFERS_FROM_PRICE"]
)


print("=" * 125)
print("VALIDATION CASES")
print("Reconstructed ≈ Adjusted, while Adjusted differs from Price")
print("=" * 125)

valid = events[events["VALIDATION_CASE"]].copy()

if valid.empty:
    print("NONE FOUND")
else:
    print(
        valid.to_string(
            index=False,
            float_format=lambda x: f"{x:,.4f}"
        )
    )


print("\n" + "=" * 125)
print("VALIDATION SUMMARY BY PORTFOLIO HOLDING")
print("=" * 125)

summary = (
    events.groupby("Ticker")
    .agg(
        Dividend_Events=("Date", "count"),
        Validation_Cases=("VALIDATION_CASE", "sum"),
        Median_Recon_vs_Adj_bp=("Recon_vs_Adj_bp", lambda x: x.abs().median()),
        Median_Price_vs_Adj_bp=("Price_vs_Adj_bp", lambda x: x.abs().median())
    )
    .reset_index()
)

print(
    summary.to_string(
        index=False,
        float_format=lambda x: f"{x:,.2f}"
    )
)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

START = "2021-09-28"
END   = "2026-09-29"

results = []
duplicates = []

for ticker in tickers:

    try:
        df = yf.Ticker(ticker).history(
            start=START,
            end=END,
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=False
        ).sort_index()

        if df.empty or len(df) < 2:
            results.append({
                "Ticker": ticker,
                "Status": "INSUFFICIENT DATA"
            })
            continue

        close = pd.to_numeric(df["Close"], errors="coerce")
        adj = pd.to_numeric(df["Adj Close"], errors="coerce")

        div = pd.to_numeric(
            df.get("Dividends", 0),
            errors="coerce"
        ).fillna(0)

        splits = pd.to_numeric(
            df.get("Stock Splits", 0),
            errors="coerce"
        ).fillna(0)

        # =========================================================
        # CLEAN DUPLICATE DIVIDEND RECORDS
        #
        # Suspected Yahoo pattern:
        # identical amount recorded once at period-end and again
        # on the later actual ex-distribution date.
        #
        # Rule:
        # same amount (within tiny tolerance), occurring again
        # within 45 days -> keep LATER observation.
        # =========================================================

        clean_div = div.copy()

        events = div[div != 0]

        event_items = list(events.items())

        for i in range(len(event_items) - 1):

            date1, amount1 = event_items[i]
            date2, amount2 = event_items[i + 1]

            days_apart = (date2 - date1).days

            same_amount = np.isclose(
                amount1,
                amount2,
                rtol=0,
                atol=0.01
            )

            if same_amount and 0 < days_apart <= 45:

                duplicates.append({
                    "Ticker": ticker,
                    "Removed_Date": date1.date(),
                    "Removed_Amount": amount1,
                    "Kept_Date": date2.date(),
                    "Kept_Amount": amount2,
                    "Days_Apart": days_apart
                })

                clean_div.loc[date1] = 0.0

        # =========================================================
        # PRICE RETURN
        # =========================================================

        price_return = (
            close.iloc[-1] / close.iloc[0] - 1
        )

        # =========================================================
        # YAHOO ADJUSTED RETURN
        # =========================================================

        adjusted_return = (
            adj.iloc[-1] / adj.iloc[0] - 1
        )

        # =========================================================
        # CLEAN RECONSTRUCTED RETURN
        # =========================================================

        split_factor = splits.where(
            splits != 0,
            1.0
        )

        daily_reconstructed = (
            (close * split_factor + clean_div)
            / close.shift(1)
            - 1
        ).dropna()

        reconstructed_return = (
            (1 + daily_reconstructed).prod() - 1
        )

        results.append({
            "Ticker": ticker,
            "Start": close.index[0].date(),
            "End": close.index[-1].date(),

            "5Y Price Return":
                price_return * 100,

            "5Y Reconstructed Return":
                reconstructed_return * 100,

            "5Y Adjusted Return":
                adjusted_return * 100,

            "Raw Dividend Events":
                int((div != 0).sum()),

            "Clean Dividend Events":
                int((clean_div != 0).sum()),

            "Duplicates Removed":
                int((div != 0).sum() - (clean_div != 0).sum()),

            "Status": "OK"
        })

    except Exception as e:

        results.append({
            "Ticker": ticker,
            "Status": f"ERROR: {e}"
        })


# ================================================================
# SHOW EXACTLY WHAT WAS REMOVED
# ================================================================

dup_df = pd.DataFrame(duplicates)

print("=" * 115)
print("SUSPECTED DUPLICATE DIVIDENDS REMOVED")
print("=" * 115)

if dup_df.empty:
    print("NONE")
else:
    print(
        dup_df.to_string(
            index=False,
            float_format=lambda x: f"{x:,.4f}"
        )
    )


# ================================================================
# FIVE-YEAR RESULTS
# ================================================================

result = pd.DataFrame(results)

result["Order"] = result["Ticker"].map(
    {ticker: i for i, ticker in enumerate(tickers)}
)

result = (
    result
    .sort_values("Order")
    .drop(columns="Order")
)

print("\n" + "=" * 145)
print("FIVE-YEAR RETURNS AFTER DUPLICATE-DIVIDEND CLEANING")
print("=" * 145)

cols = [
    "Ticker",
    "Start",
    "End",
    "5Y Price Return",
    "5Y Reconstructed Return",
    "5Y Adjusted Return",
    "Raw Dividend Events",
    "Clean Dividend Events",
    "Duplicates Removed",
    "Status"
]

print(
    result[cols].to_string(
        index=False,
        float_format=lambda x: f"{x:,.2f}"
    )
)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

TICKERS = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

START = "2021-09-28"
END   = "2026-09-29"

rows = []
repair_log = []
duplicate_log = []

# ================================================================
# PRICE UNIT REPAIR
# Repairs obvious temporary 100x / 0.01x Yahoo observations.
# Uses neighbouring observations to determine the normal regime.
# ================================================================

def repair_isolated_100x(series, ticker):
    s = series.copy().astype(float)

    for _ in range(5):
        changed = False

        for i in range(1, len(s) - 1):
            prev = s.iloc[i - 1]
            curr = s.iloc[i]
            nxt  = s.iloc[i + 1]

            if not all(np.isfinite([prev, curr, nxt])):
                continue
            if min(prev, curr, nxt) <= 0:
                continue

            neighbour_ratio = prev / nxt

            # Neighbours broadly agree with one another.
            if 0.5 <= neighbour_ratio <= 2.0:

                # Current observation is ~100x too small.
                if curr / prev < 0.02 and curr / nxt < 0.02:
                    old = curr
                    s.iloc[i] = curr * 100

                    repair_log.append({
                        "Ticker": ticker,
                        "Date": s.index[i].date(),
                        "Old": old,
                        "New": s.iloc[i],
                        "Repair": "x100"
                    })
                    changed = True

                # Current observation is ~100x too large.
                elif curr / prev > 50 and curr / nxt > 50:
                    old = curr
                    s.iloc[i] = curr / 100

                    repair_log.append({
                        "Ticker": ticker,
                        "Date": s.index[i].date(),
                        "Old": old,
                        "New": s.iloc[i],
                        "Repair": "/100"
                    })
                    changed = True

        if not changed:
            break

    return s


# ================================================================
# DUPLICATE DISTRIBUTION CLEANING
# ================================================================

def clean_duplicate_dividends(dividends, ticker):

    d = dividends.copy()
    events = list(d[d != 0].items())

    removed = 0

    for i in range(len(events) - 1):

        d1, a1 = events[i]
        d2, a2 = events[i + 1]

        if (
            np.isclose(a1, a2, atol=0.01, rtol=0)
            and 0 < (d2 - d1).days <= 45
        ):
            d.loc[d1] = 0.0
            removed += 1

            duplicate_log.append({
                "Ticker": ticker,
                "Removed": d1.date(),
                "Amount": a1,
                "Kept": d2.date()
            })

    return d, removed


# ================================================================
# PROCESS ALL HOLDINGS
# ================================================================

for ticker in TICKERS:

    try:

        hist = yf.Ticker(ticker).history(
            start=START,
            end=END,
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=False
        ).sort_index()

        if hist.empty or "Close" not in hist:
            rows.append({
                "Ticker": ticker,
                "Status": "—"
            })
            continue

        if hist.index.tz is not None:
            hist.index = hist.index.tz_localize(None)

        raw_close = pd.to_numeric(
            hist["Close"], errors="coerce"
        ).dropna()

        if len(raw_close) < 2:
            rows.append({
                "Ticker": ticker,
                "Status": "—"
            })
            continue

        close = raw_close.copy()

        # ========================================================
        # GENERIC ISOLATED UNIT REPAIRS
        # ========================================================

        close = repair_isolated_100x(close, ticker)


        # ========================================================
        # STXCAP STRUCTURAL REGIME
        #
        # Yahoo switches from ~R30-60 representation to
        # ~3,000-6,000 representation in May 2026.
        #
        # Put PRE-switch prices into the later cents convention.
        # Its pre-switch dividend records must be put into the
        # same convention as well.
        # ========================================================

        STXCAP_SWITCH = pd.Timestamp("2026-05-18")

        if ticker == "STXCAP.JO":

            mask = close.index < STXCAP_SWITCH

            close.loc[mask] = close.loc[mask] * 100

            repair_log.append({
                "Ticker": ticker,
                "Date": "pre-2026-05-18",
                "Old": "R-style regime",
                "New": "x100",
                "Repair": "STRUCTURAL NORMALIZATION"
            })


        # ========================================================
        # DIVIDENDS
        # ========================================================

        if "Dividends" in hist:

            dividends = (
                pd.to_numeric(
                    hist["Dividends"],
                    errors="coerce"
                )
                .fillna(0)
                .reindex(close.index, fill_value=0)
            )

        else:
            dividends = pd.Series(
                0.0,
                index=close.index
            )

        # STXCAP dividends are already expressed in cents-like
        # amounts (e.g. 60.94), so after converting the historical
        # price from ~58 to ~5800, DO NOT multiply dividends by 100.

        dividends, duplicates_removed = (
            clean_duplicate_dividends(
                dividends,
                ticker
            )
        )


        # ========================================================
        # SPLITS
        # ========================================================

        if "Stock Splits" in hist:

            splits = (
                pd.to_numeric(
                    hist["Stock Splits"],
                    errors="coerce"
                )
                .fillna(0)
                .reindex(close.index, fill_value=0)
            )

        else:
            splits = pd.Series(
                0.0,
                index=close.index
            )


        # ========================================================
        # REMAINING PRICE DISCONTINUITIES
        # ========================================================

        ratios = close / close.shift(1)

        remaining_bad = (
            (ratios > 5) |
            (ratios < 0.20)
        ).fillna(False)

        remaining_discontinuities = int(
            remaining_bad.sum()
        )


        # ========================================================
        # RETURNS
        # ========================================================

        P0 = float(close.iloc[0])
        P1 = float(close.iloc[-1])

        price_return = P1 / P0 - 1

        cash_dividends = float(
            dividends.sum()
        )

        # NON-REINVESTED
        non_reinvested = (
            (P1 + cash_dividends) / P0 - 1
        )

        # REINVESTED — validation only
        shares = 1.0

        for date, dividend in dividends[dividends != 0].items():

            if date not in close.index:
                continue

            price = float(close.loc[date])

            if price > 0:
                shares *= (
                    1 + float(dividend) / price
                )

        reinvested = (
            shares * P1 / P0 - 1
        )


        # ========================================================
        # FIVE-YEAR AVAILABILITY
        # ========================================================

        actual_start = close.index[0]
        requested_start = pd.Timestamp(START)

        full_5y = (
            actual_start <=
            requested_start + pd.Timedelta(days=10)
        )

        if full_5y:
            price_display = price_return * 100
            nonre_display = non_reinvested * 100
            reinvest_display = reinvested * 100
        else:
            price_display = np.nan
            nonre_display = np.nan
            reinvest_display = np.nan


        # ========================================================
        # FINAL SANITY CHECK
        # ========================================================

        max_distribution_pct = 0.0

        for date, dividend in dividends[dividends != 0].items():

            if date in close.index and close.loc[date] > 0:

                y = (
                    float(dividend) /
                    float(close.loc[date])
                ) * 100

                max_distribution_pct = max(
                    max_distribution_pct,
                    y
                )

        if remaining_discontinuities > 0:
            status = "REVIEW PRICE"

        elif max_distribution_pct > 25:
            status = "REVIEW DISTRIBUTION"

        elif not full_5y:
            status = "SHORT HISTORY — 5Y = —"

        else:
            status = "PASS"


        rows.append({

            "Ticker": ticker,
            "Start": actual_start.date(),
            "End": close.index[-1].date(),

            "5Y Price (%)":
                price_display,

            "5Y Non-Reinvested (%)":
                nonre_display,

            "5Y Reinvested Validation (%)":
                reinvest_display,

            "Cash Distributions":
                cash_dividends,

            "Dividend Events":
                int((dividends != 0).sum()),

            "Duplicates Removed":
                duplicates_removed,

            "Remaining Price Errors":
                remaining_discontinuities,

            "Max Distribution / Price (%)":
                max_distribution_pct,

            "Status":
                status
        })


    except Exception as e:

        rows.append({
            "Ticker": ticker,
            "Status": f"ERROR: {e}"
        })


# ================================================================
# FINAL TABLE
# ================================================================

result = pd.DataFrame(rows)

result["Order"] = result["Ticker"].map(
    {t: i for i, t in enumerate(TICKERS)}
)

result = (
    result
    .sort_values("Order")
    .drop(columns="Order")
)

print("=" * 170)
print("FINAL CONTROLLED RETURN RECONSTRUCTION")
print("=" * 170)

print(
    result.to_string(
        index=False,
        na_rep="—",
        float_format=lambda x: f"{x:,.2f}"
    )
)


# ================================================================
# REPAIR LOG
# ================================================================

print("\n" + "=" * 110)
print("PRICE UNIT REPAIRS")
print("=" * 110)

repairs = pd.DataFrame(repair_log)

if repairs.empty:
    print("NONE")
else:
    print(repairs.to_string(index=False))


# ================================================================
# DUPLICATE LOG
# ================================================================

print("\n" + "=" * 110)
print("DUPLICATE DISTRIBUTIONS REMOVED")
print("=" * 110)

dups = pd.DataFrame(duplicate_log)

if dups.empty:
    print("NONE")
else:
    print(
        dups.to_string(
            index=False,
            float_format=lambda x: f"{x:,.4f}"
        )
    )
```

```python
import yfinance as yf
import pandas as pd
import numpy as np

# ==============================================================
# CONFIG
# ==============================================================

TICKERS = [
    "AGL.JO","ANH.JO","BHG.JO","BID.JO","BTI.JO","BVT.JO",
    "CFR.JO","EMN.JO","REM.JO","RNI.JO","SHP.JO","TBS.JO",
    "VAL.JO","BHETNC.JO","CSP500.JO","EXX.JO","GLD.JO",
    "INCOME.JO","STXCAP.JO","STXEME.JO","STXEUR.JO",
    "STXGBD.JO","STXGOV.JO","STX500.JO"
]

START = "2021-09-28"
END   = "2026-09-29"

rows = []
repair_log = []
duplicate_log = []


# ==============================================================
# GENERIC 100x PRICE-UNIT REGIME REPAIR
# ==============================================================

def repair_100x_regimes(series, ticker):

    s = series.copy().astype(float)

    if len(s) < 2:
        return s

    scale = 1.0
    repaired = s.copy()

    for i in range(1, len(s)):

        raw_prev = s.iloc[i - 1]
        raw_curr = s.iloc[i]

        if (
            not np.isfinite(raw_prev)
            or not np.isfinite(raw_curr)
            or raw_prev <= 0
            or raw_curr <= 0
        ):
            repaired.iloc[i] = raw_curr * scale
            continue

        raw_ratio = raw_curr / raw_prev

        # Yahoo suddenly changes from normal units to /100 units
        if 0.005 <= raw_ratio <= 0.02:

            scale *= 100.0

            repair_log.append({
                "Ticker": ticker,
                "Date": s.index[i].date(),
                "Event": "RAW PRICE /100",
                "New Scale": scale
            })

        # Yahoo suddenly changes from /100 units back to normal units
        elif 50 <= raw_ratio <= 200:

            scale /= 100.0

            repair_log.append({
                "Ticker": ticker,
                "Date": s.index[i].date(),
                "Event": "RAW PRICE x100",
                "New Scale": scale
            })

        repaired.iloc[i] = raw_curr * scale

    return repaired


# ==============================================================
# DUPLICATE DIVIDEND CLEANER
# ==============================================================

def clean_duplicate_dividends(dividends, ticker):

    d = dividends.copy()

    events = list(
        d[d != 0].items()
    )

    removed = 0

    for i in range(len(events) - 1):

        d1, a1 = events[i]
        d2, a2 = events[i + 1]

        same_amount = np.isclose(
            a1,
            a2,
            atol=0.01,
            rtol=0
        )

        close_dates = (
            0 < (d2 - d1).days <= 45
        )

        if same_amount and close_dates:

            d.loc[d1] = 0.0
            removed += 1

            duplicate_log.append({
                "Ticker": ticker,
                "Removed": d1.date(),
                "Amount": a1,
                "Kept": d2.date()
            })

    return d, removed


# ==============================================================
# PROCESS ENTIRE HOLDING SET
# ==============================================================

for ticker in TICKERS:

    try:

        hist = yf.Ticker(ticker).history(
            start=START,
            end=END,
            interval="1d",
            auto_adjust=False,
            actions=True,
            repair=False
        ).sort_index()

        if hist.empty or "Close" not in hist.columns:

            rows.append({
                "Ticker": ticker,
                "Status": "NO DATA"
            })

            continue

        if hist.index.tz is not None:
            hist.index = hist.index.tz_localize(None)

        close_raw = pd.to_numeric(
            hist["Close"],
            errors="coerce"
        ).dropna()

        if len(close_raw) < 2:

            rows.append({
                "Ticker": ticker,
                "Status": "NO DATA"
            })

            continue


        # ======================================================
        # PRICE NORMALISATION
        # ======================================================
        #
        # STXCAP IS EXEMPT FROM THE GENERIC REPAIR.
        #
        # We already identified its specific structural regime:
        #
        # pre-2026-05-18  : ~30-60
        # post-2026-05-18 : ~3000-6000
        #
        # Therefore normalize the old regime ONCE by x100.
        #
        # All other securities use the generic regime detector.
        # ======================================================

        if ticker == "STXCAP.JO":

            close = close_raw.copy()

            switch = pd.Timestamp("2026-05-18")

            mask = close.index < switch

            close.loc[mask] *= 100.0

            repair_log.append({
                "Ticker": ticker,
                "Date": "pre-2026-05-18",
                "Event": "STXCAP STRUCTURAL x100",
                "New Scale": 100.0
            })

        else:

            close = repair_100x_regimes(
                close_raw,
                ticker
            )


        # ======================================================
        # DIVIDENDS / CASH DISTRIBUTIONS
        # ======================================================

        if "Dividends" in hist.columns:

            dividends = (
                pd.to_numeric(
                    hist["Dividends"],
                    errors="coerce"
                )
                .fillna(0)
                .reindex(
                    close.index,
                    fill_value=0
                )
            )

        else:

            dividends = pd.Series(
                0.0,
                index=close.index
            )


        # ------------------------------------------------------
        # IMPORTANT:
        #
        # STXCAP dividends are NOT multiplied by 100.
        #
        # The problem was its historical PRICE convention.
        # Once historical prices are x100, distributions such
        # as ~60 become economically sensible relative to price.
        # ------------------------------------------------------

        dividends, duplicates_removed = (
            clean_duplicate_dividends(
                dividends,
                ticker
            )
        )


        # ======================================================
        # FINAL PRICE SANITY CHECK
        # ======================================================

        ratios = (
            close /
            close.shift(1)
        )

        bad_price = (
            (ratios > 5.0)
            |
            (ratios < 0.20)
        ).fillna(False)

        remaining_price_errors = int(
            bad_price.sum()
        )


        # ======================================================
        # DISTRIBUTION SANITY CHECK
        # ======================================================

        distribution_ratios = []

        for date, amount in dividends[
            dividends != 0
        ].items():

            if (
                date in close.index
                and close.loc[date] > 0
            ):

                distribution_ratios.append(
                    float(amount)
                    /
                    float(close.loc[date])
                )

        max_distribution_pct = (
            max(distribution_ratios) * 100
            if distribution_ratios
            else 0.0
        )


        # ======================================================
        # RETURNS
        # ======================================================

        P0 = float(
            close.iloc[0]
        )

        P1 = float(
            close.iloc[-1]
        )

        cash_distributions = float(
            dividends.sum()
        )


        # ------------------------------------------------------
        # PRICE RETURN
        # ------------------------------------------------------

        price_return = (
            P1 / P0
        ) - 1


        # ------------------------------------------------------
        # OUR PORTFOLIO RETURN
        #
        # CASH DIVIDENDS ARE NOT REINVESTED.
        #
        # Return =
        #
        # Ending price
        # - Starting price
        # + Cash distributions
        # ----------------------
        # Starting price
        # ------------------------------------------------------

        non_reinvested_return = (
            P1
            - P0
            + cash_distributions
        ) / P0


        # ======================================================
        # FIVE-YEAR HISTORY CHECK
        # ======================================================

        actual_start = (
            close.index[0]
        )

        full_5y = (
            actual_start
            <=
            pd.Timestamp(START)
            + pd.Timedelta(days=10)
        )


        # ======================================================
        # STATUS
        # ======================================================

        if remaining_price_errors > 0:

            status = "REVIEW PRICE"

        elif max_distribution_pct > 25:

            status = "REVIEW DISTRIBUTION"

        elif not full_5y:

            status = "SHORT HISTORY"

        else:

            status = "PASS"


        # ======================================================
        # STORE RESULT
        # ======================================================

        rows.append({

            "Ticker":
                ticker,

            "Start":
                actual_start.date(),

            "End":
                close.index[-1].date(),

            "5Y Price Return (%)":
                price_return * 100
                if full_5y
                else np.nan,

            "5Y Non-Reinvested Return (%)":
                non_reinvested_return * 100
                if full_5y
                else np.nan,

            "Cash Distributions":
                cash_distributions,

            "Dividend Events":
                int(
                    (dividends != 0).sum()
                ),

            "Duplicates Removed":
                duplicates_removed,

            "Remaining Price Errors":
                remaining_price_errors,

            "Max Distribution / Price (%)":
                max_distribution_pct,

            "Status":
                status
        })


    except Exception as e:

        rows.append({
            "Ticker": ticker,
            "Status": f"ERROR: {e}"
        })


# ==============================================================
# FINAL TABLE
# ==============================================================

result = pd.DataFrame(rows)

result["Order"] = result["Ticker"].map(
    {
        ticker: i
        for i, ticker in enumerate(TICKERS)
    }
)

result = (
    result
    .sort_values("Order")
    .drop(columns="Order")
)


print("=" * 160)
print("FINAL NON-REINVESTED PORTFOLIO RETURN AUDIT")
print("=" * 160)

print(
    result.to_string(
        index=False,
        na_rep="—",
        float_format=lambda x: f"{x:,.2f}"
    )
)


# ==============================================================
# PRICE REPAIR LOG
# ==============================================================

print("\n" + "=" * 110)
print("PRICE REPAIR LOG")
print("=" * 110)

repairs = pd.DataFrame(
    repair_log
)

if repairs.empty:

    print("NONE")

else:

    print(
        repairs.to_string(
            index=False
        )
    )


# ==============================================================
# DUPLICATE DISTRIBUTION LOG
# ==============================================================

print("\n" + "=" * 110)
print("DUPLICATE DISTRIBUTIONS REMOVED")
print("=" * 110)

dups = pd.DataFrame(
    duplicate_log
)

if dups.empty:

    print("NONE")

else:

    print(
        dups.to_string(
            index=False,
            float_format=lambda x: f"{x:,.4f}"
        )
    )


# ==============================================================
# FINAL FAILURES ONLY
# ==============================================================

print("\n" + "=" * 110)
print("FINAL DATA-QUALITY FAILURES")
print("=" * 110)

failures = result[
    result["Status"].isin([
        "REVIEW PRICE",
        "REVIEW DISTRIBUTION"
    ])
]

if failures.empty:

    print("NONE — DATA QUALITY CONTROLS PASSED")

else:

    print(
        failures.to_string(
            index=False,
            na_rep="—",
            float_format=lambda x: f"{x:,.2f}"
        )
    )


# ==============================================================
# SHORT-HISTORY / NO-DATA NAMES
# ==============================================================

print("\n" + "=" * 110)
print("INCOMPLETE FIVE-YEAR HISTORY")
print("=" * 110)

incomplete = result[
    result["Status"].isin([
        "SHORT HISTORY",
        "NO DATA"
    ])
]

if incomplete.empty:

    print("NONE")

else:

    print(
        incomplete[
            ["Ticker", "Start", "End", "Status"]
        ].to_string(
            index=False,
            na_rep="—"
        )
    )
```

```python
%%writefile eod_update.py

import pandas as pd
import numpy as np
import yfinance as yf
from pathlib import Path
from zoneinfo import ZoneInfo

SAST = ZoneInfo("Africa/Johannesburg")

PORTFOLIO_START = pd.Timestamp("2026-09-28")
NAV_FILE = Path("portfolio_history.csv")


# ============================================================
# PORTFOLIO HELD FOR 28 SEP 2026
# ============================================================

portfolio = {
    # MAIN
    "Anglo American": ("AGL.JO", 2491, "Main"),
    "AB InBev": ("ANH.JO", 500, "Main"),
    "BHP Group": ("BHG.JO", 600, "Main"),
    "Bidcorp": ("BID.JO", 3000, "Main"),
    "British American Tobacco": ("BTI.JO", 1840, "Main"),
    "Bidvest": ("BVT.JO", 5000, "Main"),
    "Richemont": ("CFR.JO", 565, "Main"),
    "eMedia Holdings N": ("EMN.JO", 2517, "Main"),
    "Remgro": ("REM.JO", 6000, "Main"),
    "Reinet": ("RNI.JO", 2800, "Main"),
    "Shoprite": ("SHP.JO", 5500, "Main"),
    "Tiger Brands": ("TBS.JO", 3000, "Main"),
    "Valterra Platinum": ("VAL.JO", 250, "Main"),

    # SUBSET
    "FNB Berkshire ETN": ("BHETNC.JO", 2500, "Subset"),
    "10X S&P 500": ("CSP500.JO", 855, "Subset"),
    "Exxaro": ("EXX.JO", 120, "Subset"),
    "NewGold": ("GLD.JO", 77, "Subset"),
    "10X Active Income": ("INCOME.JO", 11300, "Subset"),
    "Satrix Capped All Share": ("STXCAP.JO", 2100, "Subset"),
    "Satrix MSCI EM ESG": ("STXEME.JO", 840, "Subset"),
    "Satrix STOXX Europe": ("STXEUR.JO", 2060, "Subset"),
    "Satrix Global Bond": ("STXGBD.JO", 1730, "Subset"),
    "Satrix SA Bond": ("STXGOV.JO", 13000, "Subset"),
    "Satrix S&P 500 Feeder": ("STX500.JO", 210, "Subset"),
}


holdings = pd.DataFrame([
    {
        "Holding": name,
        "Ticker": values[0],
        "Qty": values[1],
        "Bucket": values[2],
    }
    for name, values in portfolio.items()
])

tickers = holdings["Ticker"].tolist()


# ============================================================
# PRICE-UNIT REPAIR
#
# Same regime logic validated in historical audit.
# ============================================================

def repair_100x_regimes(series):

    s = series.copy().astype(float)

    if len(s) < 2:
        return s

    repaired = s.copy()
    scale = 1.0

    for i in range(1, len(s)):

        raw_prev = s.iloc[i - 1]
        raw_curr = s.iloc[i]

        if (
            not np.isfinite(raw_prev)
            or not np.isfinite(raw_curr)
            or raw_prev <= 0
            or raw_curr <= 0
        ):
            repaired.iloc[i] = raw_curr * scale
            continue

        raw_ratio = raw_curr / raw_prev

        if 0.005 <= raw_ratio <= 0.02:
            scale *= 100.0

        elif 50 <= raw_ratio <= 200:
            scale /= 100.0

        repaired.iloc[i] = raw_curr * scale

    return repaired


# ============================================================
# DOWNLOAD EACH HOLDING
#
# We deliberately use:
#
#   auto_adjust=False
#   actions=True
#   repair=False
#
# because OUR validated engine controls the reconstruction.
# ============================================================

print("Downloading raw Yahoo prices and corporate actions...")

security_data = {}

for ticker in tickers:

    hist = yf.Ticker(ticker).history(
        period="10d",
        interval="1d",
        auto_adjust=False,
        actions=True,
        repair=False
    ).sort_index()

    if hist.empty:
        raise RuntimeError(
            f"STOP — no Yahoo history for {ticker}"
        )

    if hist.index.tz is not None:
        hist.index = hist.index.tz_localize(None)

    if "Close" not in hist.columns:
        raise RuntimeError(
            f"STOP — Close unavailable for {ticker}"
        )

    close_raw = pd.to_numeric(
        hist["Close"],
        errors="coerce"
    )

    # --------------------------------------------------------
    # PRICE NORMALISATION
    # --------------------------------------------------------

    # STXCAP's historical structural problem predates the
    # progression portfolio. From portfolio inception onward
    # it is already in the current price convention.
    #
    # Therefore NO old-history x100 adjustment is required
    # for the live EOD engine.

    close = repair_100x_regimes(
        close_raw
    )

    # --------------------------------------------------------
    # CASH DISTRIBUTIONS
    # --------------------------------------------------------

    if "Dividends" in hist.columns:

        dividends = pd.to_numeric(
            hist["Dividends"],
            errors="coerce"
        ).fillna(0.0)

    else:

        dividends = pd.Series(
            0.0,
            index=hist.index
        )

    # --------------------------------------------------------
    # SPLITS
    # --------------------------------------------------------

    if "Stock Splits" in hist.columns:

        splits = pd.to_numeric(
            hist["Stock Splits"],
            errors="coerce"
        ).fillna(0.0)

    else:

        splits = pd.Series(
            0.0,
            index=hist.index
        )

    security_data[ticker] = pd.DataFrame({
        "Close": close,
        "Dividend": dividends,
        "Split": splits,
    })


# ============================================================
# FIND COMMON TWO MOST RECENT TRADING DATES
# ============================================================

common_dates = None

for ticker, df in security_data.items():

    valid_dates = set(
        df.index[df["Close"].notna()]
    )

    if common_dates is None:
        common_dates = valid_dates
    else:
        common_dates &= valid_dates


if not common_dates or len(common_dates) < 2:
    raise RuntimeError(
        "STOP — fewer than two common trading dates "
        "across the portfolio."
    )


dates = sorted(common_dates)

previous_date = pd.Timestamp(
    dates[-2]
).normalize()

current_date = pd.Timestamp(
    dates[-1]
).normalize()


print("Previous trading date:", previous_date.date())
print("Current trading date: ", current_date.date())


if current_date < PORTFOLIO_START:
    raise RuntimeError(
        "Current observation predates portfolio inception."
    )


# ============================================================
# SECURITY RETURNS
#
# NON-REINVESTED:
#
# P&L =
#     Qty × (P1 - P0 + cash distribution)
#
# Return =
#     P&L / opening market value
#
# Because Yahoo JSE prices and dividends are both in the same
# quote convention, the /100 conversion cancels from the
# security return.
# ============================================================

rows = []


for _, h in holdings.iterrows():

    ticker = h["Ticker"]
    qty = float(h["Qty"])

    df = security_data[ticker]

    if (
        previous_date not in df.index
        or current_date not in df.index
    ):
        raise RuntimeError(
            f"STOP — required dates unavailable for {ticker}"
        )

    prev_close = float(
        df.loc[previous_date, "Close"]
    )

    curr_close = float(
        df.loc[current_date, "Close"]
    )

    dividend = float(
        df.loc[current_date, "Dividend"]
    )


    # ========================================================
    # HARD DATA CONTROLS
    # ========================================================

    if (
        not np.isfinite(prev_close)
        or not np.isfinite(curr_close)
        or prev_close <= 0
        or curr_close <= 0
    ):
        raise RuntimeError(
            f"STOP — invalid price for {ticker}"
        )


    price_ratio = (
        curr_close / prev_close
    )

    if (
        price_ratio > 5.0
        or price_ratio < 0.20
    ):
        raise RuntimeError(
            f"STOP — unresolved price discontinuity "
            f"for {ticker}: "
            f"{prev_close:.4f} -> {curr_close:.4f}"
        )


    if dividend < 0:
        raise RuntimeError(
            f"STOP — negative distribution for {ticker}"
        )


    if dividend > 0:

        distribution_ratio = (
            dividend / curr_close
        )

        if distribution_ratio > 0.25:
            raise RuntimeError(
                f"STOP — suspicious distribution for {ticker}: "
                f"{dividend:.4f} vs price {curr_close:.4f}"
            )


    # ========================================================
    # SPLIT CONTROL
    # ========================================================

    split = float(
        df.loc[current_date, "Split"]
    )

    if split != 0:

        raise RuntimeError(
            f"STOP — stock split detected for {ticker} "
            f"on {current_date.date()}: {split}. "
            f"Manual quantity verification required."
        )


    # ========================================================
    # NON-REINVESTED RETURN
    # ========================================================

    price_return = (
        curr_close / prev_close
    ) - 1.0

    total_return = (
        curr_close
        - prev_close
        + dividend
    ) / prev_close


    # ========================================================
    # OPENING ECONOMIC VALUE
    #
    # Yahoo JSE securities are quoted in cents.
    # ========================================================

    opening_value = (
        qty * prev_close / 100.0
    )

    price_pnl = (
        qty
        * (curr_close - prev_close)
        / 100.0
    )

    dividend_cash = (
        qty
        * dividend
        / 100.0
    )

    total_pnl = (
        price_pnl
        + dividend_cash
    )


    rows.append({
        "Holding": h["Holding"],
        "Ticker": ticker,
        "Bucket": h["Bucket"],
        "Qty": qty,

        "Previous_Close": prev_close,
        "Current_Close": curr_close,

        "Dividend_per_Unit": dividend,

        "Opening_Value": opening_value,

        "Price_Return": price_return,
        "Total_Return": total_return,

        "Price_PnL": price_pnl,
        "Dividend_Cash": dividend_cash,
        "Total_PnL": total_pnl,
    })


detail = pd.DataFrame(rows)


# ============================================================
# COVERAGE
# ============================================================

expected = set(tickers)
received = set(detail["Ticker"])

missing = sorted(
    expected - received
)

if missing:
    raise RuntimeError(
        "STOP — missing holdings: "
        + ", ".join(missing)
    )


# ============================================================
# PORTFOLIO RETURN
#
# Explicit P&L / opening capital.
#
# This is economically identical to beginning-value weighting
# the individual non-reinvested returns, but easier to audit.
# ============================================================

def calc_portfolio_return(df):

    capital = float(
        df["Opening_Value"].sum()
    )

    pnl = float(
        df["Total_PnL"].sum()
    )

    if capital <= 0:
        return np.nan

    return pnl / capital


combined_return = calc_portfolio_return(
    detail
)

main_return = calc_portfolio_return(
    detail[
        detail["Bucket"] == "Main"
    ]
)

subset_return = calc_portfolio_return(
    detail[
        detail["Bucket"] == "Subset"
    ]
)


# ============================================================
# AUDIT TABLE
# ============================================================

audit = detail[[
    "Holding",
    "Ticker",
    "Opening_Value",
    "Previous_Close",
    "Current_Close",
    "Dividend_per_Unit",
    "Price_Return",
    "Total_Return",
    "Price_PnL",
    "Dividend_Cash",
    "Total_PnL",
]].copy()


print()
print("=" * 140)
print("SECURITY RETURN AUDIT — NON-REINVESTED")
print("=" * 140)

display_audit = audit.copy()

for col in [
    "Price_Return",
    "Total_Return",
]:
    display_audit[col] = (
        display_audit[col]
        .map(lambda x: f"{x:+.4%}")
    )

for col in [
    "Opening_Value",
    "Previous_Close",
    "Current_Close",
    "Dividend_per_Unit",
    "Price_PnL",
    "Dividend_Cash",
    "Total_PnL",
]:
    display_audit[col] = (
        display_audit[col]
        .map(lambda x: f"{x:,.2f}")
    )

print(
    display_audit.to_string(
        index=False
    )
)


# ============================================================
# OFFICIAL EOD OUTPUT
# ============================================================

print()
print("=" * 90)
print("OFFICIAL EOD PORTFOLIO RETURN")
print("=" * 90)

print(f"Date:      {current_date.date()}")
print(f"Combined:  {combined_return:+.4%}")
print(f"Main:      {main_return:+.4%}")
print(f"Subset:    {subset_return:+.4%}")

print()
print(
    "Method: raw price change + cash distributions; "
    "distributions are NOT reinvested."
)
print()


# ============================================================
# SAVE / UPDATE HISTORY
# ============================================================

new_row = pd.DataFrame([{
    "Date": current_date,
    "Combined": combined_return,
    "Main": main_return,
    "Subset": subset_return,
}])


if NAV_FILE.exists():

    history = pd.read_csv(
        NAV_FILE,
        parse_dates=["Date"]
    )

    history = pd.concat(
        [history, new_row],
        ignore_index=True
    )

else:

    history = new_row.copy()


history = (
    history[
        history["Date"] >= PORTFOLIO_START
    ]
    .sort_values("Date")
    .drop_duplicates(
        subset=["Date"],
        keep="last"
    )
    .reset_index(drop=True)
)


history.to_csv(
    NAV_FILE,
    index=False
)


# ============================================================
# CUMULATIVE PROGRESSION RETURN
#
# Daily portfolio returns compound through time.
#
# This does NOT mean individual cash dividends are assumed
# reinvested into the securities.
# ============================================================

cum = history.copy()

for col in [
    "Combined",
    "Main",
    "Subset",
]:

    cum[col] = (
        (1 + cum[col])
        .cumprod()
        - 1
    )


print("=" * 90)
print("STORED PORTFOLIO HISTORY")
print("=" * 90)

display_history = history.copy()

for col in [
    "Combined",
    "Main",
    "Subset",
]:

    display_history[col] = (
        display_history[col]
        .map(lambda x: f"{x:+.4%}")
    )


print(
    display_history.to_string(
        index=False
    )
)


print()
print("=" * 90)
print("CUMULATIVE RETURN SINCE 28 SEP 2026")
print("=" * 90)

latest = cum.iloc[-1]

print(
    f"Combined: {latest['Combined']:+.4%}"
)

print(
    f"Main:     {latest['Main']:+.4%}"
)

print(
    f"Subset:   {latest['Subset']:+.4%}"
)

print()
print(
    "portfolio_history.csv updated successfully."
)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# CONFIG
# ============================================================

NAV_FILE = Path("portfolio_history.csv")

ANNUALISATION = 252
ROLLING_WINDOW = 20

SERIES = ["Combined", "Main", "Subset"]


# ============================================================
# LOAD ACTUAL LIVE TRACK RECORD
# ============================================================

if not NAV_FILE.exists():
    raise FileNotFoundError(
        "portfolio_history.csv not found."
    )

df = pd.read_csv(
    NAV_FILE,
    parse_dates=["Date"]
)

df = (
    df
    .sort_values("Date")
    .drop_duplicates("Date", keep="last")
    .reset_index(drop=True)
)

if df.empty:
    raise RuntimeError("Portfolio history is empty.")

for col in SERIES:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# ============================================================
# NAV INDEX
#
# First actual daily return is included.
# Starting capital immediately before the first observation = 100.
# ============================================================

for col in SERIES:

    df[f"{col}_NAV"] = (
        100.0 *
        (1.0 + df[col]).cumprod()
    )


# ============================================================
# CUMULATIVE RETURN
# ============================================================

for col in SERIES:

    df[f"{col}_Cumulative"] = (
        df[f"{col}_NAV"] / 100.0
    ) - 1.0


# ============================================================
# RUNNING PEAK + DRAWDOWN
# ============================================================

for col in SERIES:

    nav = df[f"{col}_NAV"]

    peak = nav.cummax()

    df[f"{col}_Drawdown"] = (
        nav / peak
    ) - 1.0


# ============================================================
# 20-DAY ROLLING ANNUALISED VOLATILITY
# ============================================================

for col in SERIES:

    df[f"{col}_Vol20"] = (
        df[col]
        .rolling(
            ROLLING_WINDOW,
            min_periods=ROLLING_WINDOW
        )
        .std(ddof=1)
        * np.sqrt(ANNUALISATION)
    )


# ============================================================
# 20-DAY ROLLING SHARPE
#
# Uses 0% risk-free rate for clean portfolio-performance
# measurement, consistent with the earlier portfolio analytics.
# ============================================================

for col in SERIES:

    rolling_mean = (
        df[col]
        .rolling(
            ROLLING_WINDOW,
            min_periods=ROLLING_WINDOW
        )
        .mean()
    )

    rolling_std = (
        df[col]
        .rolling(
            ROLLING_WINDOW,
            min_periods=ROLLING_WINDOW
        )
        .std(ddof=1)
    )

    df[f"{col}_Sharpe20"] = (
        np.sqrt(ANNUALISATION)
        * rolling_mean
        / rolling_std
    )


# ============================================================
# MAIN vs SUBSET RELATIVE PERFORMANCE
# ============================================================

df["Main_minus_Subset_Daily"] = (
    df["Main"] - df["Subset"]
)

df["Main_vs_Subset_Cumulative"] = (
    (
        df["Main_NAV"]
        /
        df["Subset_NAV"]
    )
    - 1.0
)


# ============================================================
# SUMMARY
# ============================================================

summary_rows = []

for col in SERIES:

    returns = df[col].dropna()

    nav = df.loc[
        returns.index,
        f"{col}_NAV"
    ]

    cumulative = (
        nav.iloc[-1] / 100.0
    ) - 1.0


    # --------------------------------------------------------
    # Annualised volatility
    # --------------------------------------------------------

    if len(returns) >= 2:

        ann_vol = (
            returns.std(ddof=1)
            * np.sqrt(ANNUALISATION)
        )

    else:

        ann_vol = np.nan


    # --------------------------------------------------------
    # Sharpe
    # --------------------------------------------------------

    if (
        len(returns) >= 2
        and returns.std(ddof=1) > 0
    ):

        sharpe = (
            np.sqrt(ANNUALISATION)
            * returns.mean()
            / returns.std(ddof=1)
        )

    else:

        sharpe = np.nan


    # --------------------------------------------------------
    # Maximum drawdown
    # --------------------------------------------------------

    if len(nav) >= 2:

        running_peak = nav.cummax()

        drawdown = (
            nav / running_peak
        ) - 1.0

        max_dd = drawdown.min()

    else:

        max_dd = np.nan


    summary_rows.append({
        "Portfolio": col,
        "Observations": len(returns),
        "Start": df["Date"].iloc[0].date(),
        "End": df["Date"].iloc[-1].date(),
        "NAV": nav.iloc[-1],
        "Cumulative Return": cumulative,
        "Annualised Volatility": ann_vol,
        "Sharpe (0% RF)": sharpe,
        "Maximum Drawdown": max_dd,
    })


summary = pd.DataFrame(summary_rows)


# ============================================================
# DISPLAY SUMMARY
# ============================================================

display_summary = summary.copy()

display_summary["NAV"] = (
    display_summary["NAV"]
    .map(lambda x: f"{x:,.4f}")
)

for col in [
    "Cumulative Return",
    "Annualised Volatility",
    "Maximum Drawdown",
]:

    display_summary[col] = (
        display_summary[col]
        .map(
            lambda x:
            "—"
            if pd.isna(x)
            else f"{x:+.2%}"
        )
    )

display_summary["Sharpe (0% RF)"] = (
    display_summary["Sharpe (0% RF)"]
    .map(
        lambda x:
        "—"
        if pd.isna(x)
        else f"{x:.3f}"
    )
)


print("=" * 125)
print("PROGRESSION PORTFOLIO — LIVE TRACK RECORD")
print("=" * 125)

print(
    display_summary.to_string(
        index=False
    )
)


# ============================================================
# LATEST OBSERVATION
# ============================================================

latest = df.iloc[-1]

print()
print("=" * 80)
print("LATEST PROGRESSION")
print("=" * 80)

print(f"Date: {latest['Date'].date()}")

print(
    f"Combined NAV: {latest['Combined_NAV']:.4f} | "
    f"Return: {latest['Combined_Cumulative']:+.4%}"
)

print(
    f"Main NAV:     {latest['Main_NAV']:.4f} | "
    f"Return: {latest['Main_Cumulative']:+.4%}"
)

print(
    f"Subset NAV:   {latest['Subset_NAV']:.4f} | "
    f"Return: {latest['Subset_Cumulative']:+.4%}"
)

print(
    f"Main vs Subset: "
    f"{latest['Main_vs_Subset_Cumulative']:+.4%}"
)


# ============================================================
# SAVE ANALYTICS
# ============================================================

df.to_csv(
    "progression_analytics.csv",
    index=False
)

summary.to_csv(
    "progression_summary.csv",
    index=False
)

print()
print("Saved:")
print("  progression_analytics.csv")
print("  progression_summary.csv")
```

```python
from pathlib import Path

p = Path("/content/app.py")
text = p.read_text()

start_marker = "# HISTORICAL PERIOD HELPERS"
end_marker = "# FOOTER"

start = text.find(start_marker)
end = text.find(end_marker)

if start == -1 or end == -1 or end <= start:
    raise RuntimeError("STOP — historical block boundaries unavailable.")


replacement = r'''# HISTORICAL PERIOD HELPERS — CONTROLLED NON-REINVESTED
#
# Return methodology:
#
#     (P1 - P0 + cumulative cash distributions) / P0
#
# Cash distributions are NOT reinvested.
# Raw Close is used; Yahoo Adjusted Close is NOT used.
# ============================================================

@st.cache_data(ttl=86400)
def download_historical_reference(ticker):

    hist = yf.Ticker(ticker).history(
        start="2020-09-01",
        interval="1d",
        auto_adjust=False,
        actions=True,
        repair=False
    ).sort_index()

    if hist.empty:
        return pd.DataFrame()

    if hist.index.tz is not None:
        hist.index = hist.index.tz_localize(None)

    return hist


def historical_price_repair(ticker, series):

    s = (
        pd.to_numeric(series, errors="coerce")
        .astype(float)
        .copy()
    )

    # --------------------------------------------------------
    # STXCAP STRUCTURAL HISTORICAL REGIME
    # Validated in historical return audit:
    # pre-18 May 2026 observations require x100 normalization.
    # --------------------------------------------------------

    if ticker == "STXCAP.JO":

        mask = (
            s.index <
            pd.Timestamp("2026-05-18")
        )

        s.loc[mask] = (
            s.loc[mask] * 100.0
        )

    # --------------------------------------------------------
    # Other isolated/regime x100 errors.
    # Uses the same transition logic as the validated audit.
    # --------------------------------------------------------

    s = repair_100x_regimes(s)

    return s


def clean_historical_distributions(
    ticker,
    dividends
):

    d = (
        pd.to_numeric(
            dividends,
            errors="coerce"
        )
        .fillna(0.0)
    )

    d = d[d > 0].copy()

    if d.empty:
        return d

    # --------------------------------------------------------
    # Validated duplicate distribution records.
    #
    # Yahoo contains pairs with identical cash amounts where
    # one date is an artificial month-end record and the other
    # is the actual distribution record.
    #
    # Remove ONLY the explicitly validated duplicates.
    # --------------------------------------------------------

    duplicate_dates = {

        "STXCAP.JO": [
            "2021-12-31",
            "2022-02-28",
            "2022-04-29",
            "2022-06-30",
            "2022-08-31",
            "2022-10-31",
            "2022-12-30",
            "2023-02-28",
            "2023-04-28",
            "2023-08-31",
            "2023-10-31",
            "2023-12-29",
            "2024-02-29",
            "2024-04-30",
            "2024-06-28",
            "2024-08-30",
            "2024-10-31",
        ],

        "STXGBD.JO": [
            "2021-09-30",
            "2022-03-31",
            "2022-09-30",
            "2023-03-31",
            "2024-03-28",
            "2024-09-30",
        ],

        "STXGOV.JO": [
            "2022-06-30",
            "2022-09-30",
            "2022-12-30",
            "2023-03-31",
            "2023-06-30",
            "2023-09-29",
            "2023-12-29",
            "2024-03-28",
            "2024-06-28",
            "2024-09-30",
        ],
    }

    for dt in duplicate_dates.get(
        ticker,
        []
    ):

        timestamp = pd.Timestamp(dt)

        if timestamp in d.index:
            d = d.drop(timestamp)

    return d


def non_reinvested_endpoint_return(
    ticker,
    hist,
    target_start,
    target_end,
    tolerance=10
):

    if hist.empty or "Close" not in hist.columns:
        return np.nan

    close = historical_price_repair(
        ticker,
        hist["Close"]
    )

    close = (
        close
        .dropna()
        .sort_index()
    )

    if close.empty:
        return np.nan

    # --------------------------------------------------------
    # Require actual history around requested start.
    # This prevents short-history securities from being given
    # artificial 3Y/5Y results.
    # --------------------------------------------------------

    starts = close[
        (close.index >= target_start) &
        (
            close.index <=
            target_start +
            pd.Timedelta(days=tolerance)
        )
    ]

    ends = close[
        close.index <= target_end
    ]

    if starts.empty or ends.empty:
        return np.nan

    start_date = starts.index[0]
    end_date = ends.index[-1]

    if end_date <= start_date:
        return np.nan

    p0 = float(
        close.loc[start_date]
    )

    p1 = float(
        close.loc[end_date]
    )

    if (
        not np.isfinite(p0)
        or not np.isfinite(p1)
        or p0 <= 0
        or p1 <= 0
    ):
        return np.nan

    if "Dividends" in hist.columns:

        dividends = (
            clean_historical_distributions(
                ticker,
                hist["Dividends"]
            )
        )

        cash = float(
            dividends[
                (dividends.index > start_date) &
                (dividends.index <= end_date)
            ].sum()
        )

    else:
        cash = 0.0

    return (
        p1 - p0 + cash
    ) / p0


# ============================================================
# HISTORICAL END DATE
# ============================================================

hist_end = pd.Timestamp(
    adj_close.index.max()
).normalize()


periods = {

    "M/M":
        hist_end.replace(day=1),

    "YTD":
        pd.Timestamp(
            hist_end.year,
            1,
            1
        ),

    "1Y":
        hist_end -
        pd.DateOffset(years=1),

    "3Y":
        hist_end -
        pd.DateOffset(years=3),

    "5Y":
        hist_end -
        pd.DateOffset(years=5),
}


# ============================================================
# BHETNC HISTORICAL REFERENCE
#
# Yahoo BHETNC.JO history is unavailable before portfolio
# inception, so retain the established Berkshire × USD/ZAR
# proxy less the 1.00% annual ETN fee.
# ============================================================

@st.cache_data(ttl=86400)
def download_bhetnc_inputs():

    return yf.download(
        [
            "BRK-B",
            "USDZAR=X"
        ],
        start="2020-09-01",
        interval="1d",
        auto_adjust=False,
        progress=False
    )


bh_inputs = download_bhetnc_inputs()

if isinstance(
    bh_inputs.columns,
    pd.MultiIndex
):

    bh_adj = (
        bh_inputs["Adj Close"]
        .copy()
    )

else:

    bh_adj = pd.DataFrame()


def get_proxy_endpoint(
    s,
    target,
    tolerance=10
):

    s = s.dropna().sort_index()

    x = s[
        (s.index >= target) &
        (
            s.index <=
            target +
            pd.Timedelta(days=tolerance)
        )
    ]

    if x.empty:
        return None

    return (
        x.index[0],
        float(x.iloc[0])
    )


def bhetnc_return(start, end):

    if bh_adj.empty:
        return np.nan

    bs = get_proxy_endpoint(
        bh_adj["BRK-B"],
        start
    )

    fs = get_proxy_endpoint(
        bh_adj["USDZAR=X"],
        start
    )

    be_candidates = (
        bh_adj["BRK-B"]
        .dropna()
        .loc[:end]
    )

    fe_candidates = (
        bh_adj["USDZAR=X"]
        .dropna()
        .loc[:end]
    )

    if (
        bs is None
        or fs is None
        or be_candidates.empty
        or fe_candidates.empty
    ):
        return np.nan

    be = (
        be_candidates.index[-1],
        float(be_candidates.iloc[-1])
    )

    fe = (
        fe_candidates.index[-1],
        float(fe_candidates.iloc[-1])
    )

    years = (
        (
            max(be[0], fe[0])
            -
            max(bs[0], fs[0])
        ).days
        /
        365.25
    )

    if years <= 0:
        return np.nan

    factor = (
        (be[1] / bs[1])
        *
        (fe[1] / fs[1])
        *
        ((1 - 0.01) ** years)
    )

    return factor - 1


# ============================================================
# BUILD INDIVIDUAL-ASSET REFERENCE TABLE
# ============================================================

reference_rows = []

for holding, row in master.iterrows():

    ticker = row["ticker"]

    result = {
        "Asset": holding,
        "Portfolio": row["bucket"]
    }

    if ticker != "BHETNC.JO":

        hist = download_historical_reference(
            ticker
        )

    else:

        hist = pd.DataFrame()

    for period, period_start in periods.items():

        if ticker == "BHETNC.JO":

            result[period] = (
                bhetnc_return(
                    period_start,
                    hist_end
                )
            )

        else:

            result[period] = (
                non_reinvested_endpoint_return(
                    ticker,
                    hist,
                    period_start,
                    hist_end
                )
            )

    reference_rows.append(result)


reference = pd.DataFrame(
    reference_rows
)


# ============================================================
# REFERENCE TABLE DISPLAY
# ============================================================

reference_display = reference.copy()

for col in [
    "M/M",
    "YTD",
    "1Y",
    "3Y",
    "5Y"
]:

    reference_display[col] = (
        reference_display[col]
        .map(
            lambda x:
            "–"
            if pd.isna(x)
            else f"{x:+.2%}"
        )
    )


st.dataframe(
    reference_display,
    use_container_width=True,
    hide_index=True
)


st.caption(
    "Historical individual-security returns use raw price "
    "change plus cumulative cash distributions. Cash "
    "distributions are not reinvested. A dash indicates "
    "insufficient history for the requested period. "
    "BHETNC historical reference returns are reconstructed "
    "from Berkshire Hathaway Class B × USD/ZAR less the "
    "1.00% annual fee."
)


'''

# ============================================================
# REPLACE ONLY HISTORICAL BLOCK
# ============================================================

patched = (
    text[:start]
    + replacement
    + text[end:]
)

p.write_text(patched)


# ============================================================
# VALIDATION
# ============================================================

check = p.read_text()

new_start = check.find(start_marker)
new_end = check.find(end_marker)
new_block = check[new_start:new_end]

print("=" * 80)
print("HISTORICAL RETURN PATCH COMPLETE")
print("=" * 80)

print(
    "Non-reinvested formula present:",
    "p1 - p0 + cash" in new_block
)

print(
    "Duplicate controls present:",
    "duplicate_dates" in new_block
)

print(
    "STXCAP normalization present:",
    'ticker == "STXCAP.JO"' in new_block
)

print(
    "Old adj_close[ticker] method removed:",
    "adj_close[ticker]" not in new_block
)

print(
    "Dash for insufficient history present:",
    '"–"' in new_block
)

print()
print("app.py patched locally — NOT pushed to GitHub yet.")
```

```python
from pathlib import Path

p = Path("/content/app.py")
text = p.read_text()

marker = "def historical_price_repair(ticker, series):"

if marker not in text:
    raise RuntimeError("STOP — historical_price_repair not found.")

helper = r'''
def repair_100x_regimes(series, ticker):

    s = series.copy().astype(float)

    if len(s) < 2:
        return s

    scale = 1.0
    repaired = s.copy()

    for i in range(1, len(s)):

        raw_prev = s.iloc[i - 1]
        raw_curr = s.iloc[i]

        if (
            not np.isfinite(raw_prev)
            or not np.isfinite(raw_curr)
            or raw_prev <= 0
            or raw_curr <= 0
        ):
            repaired.iloc[i] = raw_curr * scale
            continue

        raw_ratio = raw_curr / raw_prev

        if 0.005 <= raw_ratio <= 0.02:
            scale *= 100.0

        elif 50 <= raw_ratio <= 200:
            scale /= 100.0

        repaired.iloc[i] = raw_curr * scale

    return repaired


'''

# Prevent accidental duplicate insertion.
if "def repair_100x_regimes(series, ticker):" in text:
    raise RuntimeError("STOP — helper already exists in app.py.")

text = text.replace(marker, helper + marker, 1)
p.write_text(text)

# ------------------------------------------------------------
# IMPORTANT: validate in a CLEAN namespace.
# This prevents the exact false-positive we just had.
# ------------------------------------------------------------

clean_namespace = {
    "__name__": "__main__",
    "__file__": "/content/app.py",
}

exec(
    compile(
        p.read_text(),
        "/content/app.py",
        "exec"
    ),
    clean_namespace
)

reference_clean = clean_namespace["reference"]

check = reference_clean[
    ["Asset", "M/M", "YTD", "1Y", "3Y", "5Y"]
].copy()

for c in ["M/M", "YTD", "1Y", "3Y", "5Y"]:
    check[c] = check[c].map(
        lambda x: "—" if pd.isna(x) else f"{x:+.2%}"
    )

print()
print("=" * 100)
print("CLEAN-NAMESPACE VALIDATION")
print("=" * 100)
print(check.to_string(index=False))
print()
print("PASS — app.py executed without relying on notebook globals.")
print("NOT PUSHED.")
```

```python
from pathlib import Path
import shutil

p = Path("/content/app.py")

if not p.exists():
    raise FileNotFoundError("STOP — /content/app.py not found.")

# Permanent local safety copy before this change
backup = Path("/content/app_before_performance_analytics.py")
shutil.copy2(p, backup)

text = p.read_text()

anchor = '''    # ========================================================
    # DAILY RETURN HISTORY
    # ========================================================
'''

if anchor not in text:
    raise RuntimeError("STOP — Daily Return History anchor not found.")

if "PORTFOLIO ANALYTICS — FROM INCEPTION" in text:
    raise RuntimeError("STOP — analytics patch already exists.")

analytics = r'''
    # ========================================================
    # PORTFOLIO ANALYTICS — FROM INCEPTION
    #
    # Calculated ONLY from official daily portfolio returns
    # stored in portfolio_history.csv.
    #
    # Benchmark analytics use STXCAP.JO on matching dates.
    # Nothing here modifies the official portfolio track record.
    # ========================================================

    RF_ANNUAL = 0.0725
    TRADING_DAYS = 252
    WEEKS_PER_YEAR = 52

    r = (
        pd.to_numeric(
            nav[selected],
            errors="coerce"
        )
        .dropna()
    )

    # --------------------------------------------
    # Safe display helpers
    # --------------------------------------------

    def pct_or_dash(x):
        return (
            "–"
            if x is None or not np.isfinite(x)
            else f"{x:+.2%}"
        )

    def num_or_dash(x):
        return (
            "–"
            if x is None or not np.isfinite(x)
            else f"{x:.2f}"
        )

    def days_or_dash(x):
        return (
            "–"
            if x is None or not np.isfinite(x)
            else f"{int(x)}"
        )

    # --------------------------------------------
    # Cumulative return
    # --------------------------------------------

    cumulative_return = (
        (1.0 + r).prod() - 1.0
        if len(r) >= 1
        else np.nan
    )

    # --------------------------------------------
    # Daily annualised volatility
    # --------------------------------------------

    annualised_vol = (
        r.std(ddof=1)
        * np.sqrt(TRADING_DAYS)
        if len(r) >= 2
        else np.nan
    )

    # --------------------------------------------
    # Weekly annualised volatility
    #
    # Compound official daily observations into
    # calendar weeks, then annualise weekly sigma.
    # --------------------------------------------

    weekly_frame = (
        nav[["Date", selected]]
        .dropna()
        .set_index("Date")
        .sort_index()
    )

    weekly_returns = (
        (1.0 + weekly_frame[selected])
        .resample("W-FRI")
        .prod()
        - 1.0
    )

    # Do not count an empty resampled week.
    weekly_counts = (
        weekly_frame[selected]
        .resample("W-FRI")
        .count()
    )

    weekly_returns = weekly_returns[
        weekly_counts > 0
    ]

    weekly_vol = (
        weekly_returns.std(ddof=1)
        * np.sqrt(WEEKS_PER_YEAR)
        if len(weekly_returns) >= 2
        else np.nan
    )

    # --------------------------------------------
    # Drawdown and recovery
    # --------------------------------------------

    if len(r) >= 1:

        nav_index = (
            (1.0 + r)
            .cumprod()
        )

        running_peak = (
            nav_index.cummax()
        )

        drawdown = (
            nav_index / running_peak
            - 1.0
        )

        max_drawdown = float(
            drawdown.min()
        )

        trough_label = (
            drawdown.idxmin()
        )

        trough_pos = (
            list(nav_index.index)
            .index(trough_label)
        )

        peak_value = float(
            running_peak.loc[
                trough_label
            ]
        )

        post_trough = (
            nav_index.iloc[
                trough_pos + 1:
            ]
        )

        recovered = (
            post_trough[
                post_trough >= peak_value
            ]
        )

        if (
            max_drawdown < 0
            and not recovered.empty
        ):

            recovery_label = (
                recovered.index[0]
            )

            recovery_pos = (
                list(nav_index.index)
                .index(recovery_label)
            )

            days_to_recover = (
                recovery_pos
                - trough_pos
            )

        else:
            days_to_recover = np.nan

    else:

        max_drawdown = np.nan
        days_to_recover = np.nan

    # --------------------------------------------
    # Maximum upside
    # Largest positive official daily return.
    # --------------------------------------------

    positive_days = r[r > 0]

    max_upside = (
        float(positive_days.max())
        if not positive_days.empty
        else np.nan
    )

    # --------------------------------------------
    # Win / Loss %
    #
    # Zero-return days are neither wins nor losses.
    # Percentages use all valid portfolio days as
    # denominator.
    # --------------------------------------------

    if len(r) >= 1:

        win_pct = float(
            (r > 0).sum()
            / len(r)
        )

        loss_pct = float(
            (r < 0).sum()
            / len(r)
        )

    else:

        win_pct = np.nan
        loss_pct = np.nan

    # --------------------------------------------
    # Annualised portfolio return + Sharpe
    #
    # Geometric annualisation of actual official
    # daily observations from inception.
    # --------------------------------------------

    if len(r) >= 2:

        growth = float(
            (1.0 + r).prod()
        )

        annualised_return = (
            growth ** (
                TRADING_DAYS
                / len(r)
            )
            - 1.0
        )

    else:
        annualised_return = np.nan

    sharpe = (
        (
            annualised_return
            - RF_ANNUAL
        )
        / annualised_vol
        if (
            np.isfinite(annualised_return)
            and np.isfinite(annualised_vol)
            and annualised_vol > 0
        )
        else np.nan
    )

    # --------------------------------------------
    # STXCAP benchmark for Beta / Alpha
    #
    # Downloaded independently for analytics only.
    # Does NOT alter portfolio_history.csv.
    # --------------------------------------------

    beta = np.nan
    alpha = np.nan

    if len(nav) >= 2:

        benchmark_start = (
            nav["Date"].min()
            - pd.Timedelta(days=7)
        )

        benchmark_end = (
            nav["Date"].max()
            + pd.Timedelta(days=2)
        )

        try:

            benchmark_hist = (
                yf.Ticker("STXCAP.JO")
                .history(
                    start=benchmark_start,
                    end=benchmark_end,
                    interval="1d",
                    auto_adjust=False,
                    actions=False,
                    repair=False
                )
            )

            if not benchmark_hist.empty:

                if benchmark_hist.index.tz is not None:
                    benchmark_hist.index = (
                        benchmark_hist.index
                        .tz_localize(None)
                    )

                benchmark_close = (
                    pd.to_numeric(
                        benchmark_hist["Close"],
                        errors="coerce"
                    )
                    .dropna()
                    .sort_index()
                )

                # Apply same validated STXCAP
                # structural price normalization.
                benchmark_close = (
                    historical_price_repair(
                        "STXCAP.JO",
                        benchmark_close
                    )
                )

                benchmark_return = (
                    benchmark_close
                    .pct_change()
                    .dropna()
                    .rename("Benchmark")
                )

                portfolio_for_beta = (
                    nav[
                        ["Date", selected]
                    ]
                    .copy()
                    .rename(
                        columns={
                            selected:
                            "Portfolio"
                        }
                    )
                    .set_index("Date")
                )

                aligned = (
                    portfolio_for_beta
                    .join(
                        benchmark_return,
                        how="inner"
                    )
                    .dropna()
                )

                # Need enough aligned observations
                # and non-zero benchmark variance.
                if (
                    len(aligned) >= 2
                    and
                    aligned["Benchmark"]
                    .var(ddof=1) > 0
                ):

                    beta = float(
                        aligned[
                            ["Portfolio",
                             "Benchmark"]
                        ]
                        .cov()
                        .loc[
                            "Portfolio",
                            "Benchmark"
                        ]
                        /
                        aligned[
                            "Benchmark"
                        ]
                        .var(ddof=1)
                    )

                    rf_daily = (
                        (1.0 + RF_ANNUAL)
                        ** (1.0 / TRADING_DAYS)
                        - 1.0
                    )

                    portfolio_excess = (
                        aligned["Portfolio"]
                        - rf_daily
                    )

                    benchmark_excess = (
                        aligned["Benchmark"]
                        - rf_daily
                    )

                    alpha_daily = float(
                        portfolio_excess.mean()
                        -
                        beta
                        * benchmark_excess.mean()
                    )

                    # Geometric annualisation of
                    # estimated daily CAPM alpha.
                    if alpha_daily > -1:

                        alpha = (
                            (1.0 + alpha_daily)
                            ** TRADING_DAYS
                            - 1.0
                        )

        except Exception:
            # Analytics must never break the
            # official portfolio dashboard.
            beta = np.nan
            alpha = np.nan

    # --------------------------------------------
    # DISPLAY
    # --------------------------------------------

    st.markdown(
        "#### Performance & Risk Since Inception"
    )

    metric_row_1 = st.columns(4)

    metric_row_1[0].metric(
        "Cumulative Return",
        pct_or_dash(cumulative_return)
    )

    metric_row_1[1].metric(
        "Annualised Volatility",
        pct_or_dash(annualised_vol)
    )

    metric_row_1[2].metric(
        "Weekly Volatility",
        pct_or_dash(weekly_vol)
    )

    metric_row_1[3].metric(
        "Maximum Drawdown",
        pct_or_dash(max_drawdown)
    )

    metric_row_2 = st.columns(4)

    metric_row_2[0].metric(
        "Maximum Upside",
        pct_or_dash(max_upside)
    )

    metric_row_2[1].metric(
        "Days to Recover",
        days_or_dash(days_to_recover)
    )

    metric_row_2[2].metric(
        "Beta vs STXCAP",
        num_or_dash(beta)
    )

    metric_row_2[3].metric(
        "Alpha",
        pct_or_dash(alpha)
    )

    metric_row_3 = st.columns(3)

    metric_row_3[0].metric(
        "Sharpe Ratio",
        num_or_dash(sharpe)
    )

    metric_row_3[1].metric(
        "Win %",
        pct_or_dash(win_pct)
    )

    metric_row_3[2].metric(
        "Loss %",
        pct_or_dash(loss_pct)
    )

    st.caption(
        "Statistics use official portfolio observations from "
        "28 September 2026 onward. Volatility is annualised "
        "from daily or weekly returns as labelled. Beta and "
        "CAPM alpha use STXCAP as benchmark and a 7.25% "
        "annual risk-free rate."
    )


'''

patched = text.replace(
    anchor,
    analytics + anchor,
    1
)

p.write_text(patched)

print("=" * 80)
print("PERFORMANCE ANALYTICS PATCH COMPLETE")
print("=" * 80)
print("Backup:", backup)
print(
    "Analytics inserted:",
    "PORTFOLIO ANALYTICS — FROM INCEPTION"
    in p.read_text()
)
print(
    "Daily-history section retained:",
    "Daily Portfolio Return History"
    in p.read_text()
)
print(
    "Historical-security section retained:",
    "Individual Asset Historical Returns"
    in p.read_text()
)
print()
print("NOT PUSHED TO GITHUB.")
```

```python
# ============================================================
# UPCOMING DIVIDENDS + PORTFOLIO NEWS
#
# Yahoo calendar/news information only.
# This section is informational and does not feed into any
# portfolio valuation, return, risk, or historical calculations.
# ============================================================

@st.cache_data(ttl=3600)
def get_yahoo_portfolio_updates(tickers):

    today = pd.Timestamp.now().normalize()

    dividend_rows = []
    news_rows = []

    for ticker in tickers:

        try:
            stock = yf.Ticker(ticker)

            # ------------------------------------------------
            # UPCOMING EX-DIVIDEND DATE
            # ------------------------------------------------

            try:
                calendar = stock.calendar

                if isinstance(calendar, dict):

                    ex_date = calendar.get(
                        "Ex-Dividend Date"
                    )

                    if ex_date is not None:

                        ex_date = pd.Timestamp(
                            ex_date
                        ).normalize()

                        if ex_date >= today:

                            dividend_rows.append(
                                {
                                    "Ticker": ticker,
                                    "Ex-Dividend Date": ex_date
                                }
                            )

            except Exception:
                pass

            # ------------------------------------------------
            # NEWS
            # ------------------------------------------------

            try:
                news = stock.news or []

                for item in news[:3]:

                    content = item.get(
                        "content",
                        {}
                    )

                    title = content.get(
                        "title"
                    )

                    pub_date = content.get(
                        "pubDate"
                    )

                    provider = (
                        content
                        .get(
                            "provider",
                            {}
                        )
                        .get(
                            "displayName"
                        )
                    )

                    url = (
                        content
                        .get(
                            "clickThroughUrl",
                            {}
                        )
                        .get(
                            "url"
                        )
                    )

                    if not url:

                        url = (
                            content
                            .get(
                                "canonicalUrl",
                                {}
                            )
                            .get(
                                "url"
                            )
                        )

                    if title:

                        news_rows.append(
                            {
                                "Ticker": ticker,
                                "Published": pub_date,
                                "Source": provider,
                                "Headline": title,
                                "URL": url
                            }
                        )

            except Exception:
                pass

        except Exception:
            pass

    dividends = pd.DataFrame(
        dividend_rows
    )

    news = pd.DataFrame(
        news_rows
    )

    return dividends, news


portfolio_tickers = (
    master["ticker"]
    .dropna()
    .astype(str)
    .unique()
    .tolist()
)


upcoming_dividends, portfolio_news = (
    get_yahoo_portfolio_updates(
        portfolio_tickers
    )
)


# ============================================================
# UPCOMING DIVIDEND CALENDAR
# ============================================================

st.subheader(
    "Upcoming Dividend Calendar"
)

if not upcoming_dividends.empty:

    upcoming_dividends = (
        upcoming_dividends
        .drop_duplicates()
        .sort_values(
            "Ex-Dividend Date"
        )
        .reset_index(
            drop=True
        )
    )

    upcoming_dividends[
        "Ex-Dividend Date"
    ] = (
        upcoming_dividends[
            "Ex-Dividend Date"
        ]
        .dt.strftime(
            "%d %b %Y"
        )
    )

    st.dataframe(
        upcoming_dividends,
        use_container_width=True,
        hide_index=True
    )

else:

    st.caption(
        "No upcoming ex-dividend dates are currently "
        "available from Yahoo Finance."
    )


st.caption(
    "Only future ex-dividend dates explicitly supplied by "
    "Yahoo Finance are shown. No dividend dates or amounts "
    "are estimated."
)


# ============================================================
# PORTFOLIO NEWS
# ============================================================

st.subheader(
    "Portfolio News"
)

if not portfolio_news.empty:

    portfolio_news[
        "Published"
    ] = pd.to_datetime(
        portfolio_news[
            "Published"
        ],
        errors="coerce",
        utc=True
    )

    portfolio_news = (
        portfolio_news
        .drop_duplicates(
            subset=[
                "Headline"
            ]
        )
        .sort_values(
            "Published",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )

    portfolio_news[
        "Published"
    ] = (
        portfolio_news[
            "Published"
        ]
        .dt.strftime(
            "%d %b %Y %H:%M UTC"
        )
    )

    st.dataframe(
        portfolio_news,
        use_container_width=True,
        hide_index=True,
        column_config={
            "URL":
                st.column_config.LinkColumn(
                    "Article"
                )
        }
    )

else:

    st.caption(
        "No portfolio news is currently available "
        "from Yahoo Finance."
    )


st.caption(
    "News is supplied by Yahoo Finance and its underlying "
    "publishers. Headlines are informational only."
)
```
