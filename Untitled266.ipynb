```python
# ============================================================
# WTI - BRENT SPREAD
# 60-DAY TRAILING OU / Z-SCORE / PDF / CDF / FULL SAMPLE
# + MULTIVARIATE OU EIGENMODES
# ============================================================

import pandas as pd
import numpy as np
import scipy.stats as stats

from scipy.linalg import logm, eig
from statsmodels.tsa.ar_model import AutoReg

import plotly.graph_objects as go
from plotly.subplots import make_subplots
from plotly.offline import iplot, init_notebook_mode

init_notebook_mode(connected=True)


# ============================================================
# SETTINGS
# ============================================================

FILE_NAME = "WTI Crude - Brent Spread c1 Historical Data (5).csv"

TRAILING_WINDOW = 60
YEAR_START = "2026-01-01"

# Plot colours
BG = "#E5E5E5"
GRID = "#BDBDBD"
TEXT = "#222222"


# ============================================================
# LOAD + CLEAN
# ============================================================

df = pd.read_csv(FILE_NAME)

df["Date"] = pd.to_datetime(df["Date"])

df["Price"] = pd.to_numeric(
    df["Price"],
    errors="coerce"
)

df = (
    df[["Date", "Price"]]
    .dropna()
    .sort_values("Date")
    .drop_duplicates("Date")
    .set_index("Date")
)

df.rename(
    columns={"Price": "Spread"},
    inplace=True
)

print("Sample:")
print(df.index.min().date(), "to", df.index.max().date())
print("Observations:", len(df))


# ============================================================
# GRAPH STYLE HELPER
# ============================================================

def style_fig(fig, title, ytitle="", xtitle="Date"):

    fig.update_layout(
        title=title,
        plot_bgcolor=BG,
        paper_bgcolor=BG,
        font=dict(color=TEXT),
        hovermode="x unified",
        legend=dict(
            bgcolor="rgba(255,255,255,0.65)"
        ),
        margin=dict(
            l=70,
            r=90,
            t=80,
            b=60
        )
    )

    fig.update_xaxes(
        title=xtitle,
        showgrid=True,
        gridcolor=GRID
    )

    fig.update_yaxes(
        title=ytitle,
        showgrid=True,
        gridcolor=GRID
    )

    return fig


# ============================================================
# FINAL POINT LABEL HELPER
# ============================================================

def label_last(
    fig,
    x,
    y,
    text,
    row=None,
    col=None
):

    kwargs = {}

    if row is not None:
        kwargs["row"] = row
        kwargs["col"] = col

    fig.add_annotation(
        x=x[-1],
        y=y[-1],
        text=text,
        showarrow=True,
        arrowhead=2,
        ax=45,
        ay=-25,
        bgcolor="white",
        bordercolor="#666666",
        **kwargs
    )


# ============================================================
# UNIVARIATE OU ESTIMATION
#
# x[t+1] = a + b*x[t] + error
#
# kappa = -ln(b) / dt
# theta = a / (1-b)
#
# OU stationary sigma:
# sigma_eq = sigma_process / sqrt(2*kappa)
# ============================================================

def estimate_ou(series, periods_per_year=252):

    x = series.dropna()

    if len(x) < 20:
        return None

    x0 = x.iloc[:-1].values
    x1 = x.iloc[1:].values

    X = np.column_stack([
        np.ones(len(x0)),
        x0
    ])

    beta = np.linalg.lstsq(
        X,
        x1,
        rcond=None
    )[0]

    a = beta[0]
    b = beta[1]

    fitted = a + b * x0
    residuals = x1 - fitted

    residual_std = np.std(
        residuals,
        ddof=1
    )

    dt = 1 / periods_per_year

    # Mean-reverting AR coefficient required
    if b <= 0 or b >= 1:

        return {
            "a": a,
            "b": b,
            "kappa": np.nan,
            "theta": np.nan,
            "half_life_days": np.nan,
            "process_sigma": np.nan,
            "equilibrium_sigma": np.nan
        }

    kappa = (
        -np.log(b)
        / dt
    )

    theta = (
        a / (1 - b)
    )

    half_life_years = (
        np.log(2) / kappa
    )

    half_life_days = (
        half_life_years * periods_per_year
    )

    # Convert discrete residual variance
    # into continuous OU process sigma

    process_sigma = (
        residual_std
        * np.sqrt(
            (2 * kappa)
            /
            (1 - np.exp(-2 * kappa * dt))
        )
    )

    equilibrium_sigma = (
        process_sigma
        / np.sqrt(2 * kappa)
    )

    return {
        "a": a,
        "b": b,
        "kappa": kappa,
        "theta": theta,
        "half_life_days": half_life_days,
        "process_sigma": process_sigma,
        "equilibrium_sigma": equilibrium_sigma
    }


# ============================================================
# 60-DAY TRAILING OU
# ============================================================

records = []

for i in range(
    TRAILING_WINDOW,
    len(df) + 1
):

    sample = (
        df["Spread"]
        .iloc[
            i - TRAILING_WINDOW:i
        ]
    )

    est = estimate_ou(sample)

    date = sample.index[-1]

    if est is not None:

        records.append({
            "Date": date,
            "Spread": sample.iloc[-1],
            "Kappa": est["kappa"],
            "Theta": est["theta"],
            "Half Life": est["half_life_days"],
            "OU Sigma": est["equilibrium_sigma"]
        })


rolling = (
    pd.DataFrame(records)
    .set_index("Date")
)


# ============================================================
# TRAILING Z-SCORE
#
# Uses trailing 60-day observed distribution.
# ============================================================

df["Rolling Mean"] = (
    df["Spread"]
    .rolling(TRAILING_WINDOW)
    .mean()
)

df["Rolling Std"] = (
    df["Spread"]
    .rolling(TRAILING_WINDOW)
    .std()
)

df["Z Score"] = (
    (df["Spread"] - df["Rolling Mean"])
    / df["Rolling Std"]
)


# ============================================================
# BULL REGIME
#
# Detect most recent sustained upswing using
# 60-day smoothed spread.
# ============================================================

df["Smooth60"] = (
    df["Spread"]
    .rolling(60)
    .mean()
)

df["Smooth Change"] = (
    df["Smooth60"]
    .diff()
)

# Last local trough in smoothed series
smooth = df["Smooth60"].dropna()

local_troughs = (
    (smooth.shift(1) > smooth)
    &
    (smooth.shift(-1) > smooth)
)

trough_dates = smooth[
    local_troughs
].index

if len(trough_dates) > 0:
    bull_start = trough_dates[-1]
else:
    bull_start = df.index[0]

bull = df.loc[bull_start:].copy()

ytd = df.loc[
    df.index >= pd.Timestamp(YEAR_START)
].copy()


print("\nBull sample begins:")
print(bull_start.date())

print("\nYTD sample begins:")
print(ytd.index.min().date())


# ============================================================
# FULL-SAMPLE OU — NO TRAILING WINDOW
# ============================================================

full_ou = estimate_ou(
    df["Spread"]
)

bull_ou = estimate_ou(
    bull["Spread"]
)

ytd_ou = estimate_ou(
    ytd["Spread"]
)


ou_summary = pd.DataFrame({

    "Full Sample": full_ou,
    "Bull Sample": bull_ou,
    "2026 YTD": ytd_ou

}).T


print("\nOU PARAMETER SUMMARY")

display(
    ou_summary.style.format({
        "a": "{:.5f}",
        "b": "{:.5f}",
        "kappa": "{:.3f}",
        "theta": "{:.3f}",
        "half_life_days": "{:.2f}",
        "process_sigma": "{:.3f}",
        "equilibrium_sigma": "{:.3f}"
    })
)


# ============================================================
# PLOT 1
# SPREAD + 60-DAY MEAN + BULL START
# ============================================================

plot_df = df.dropna(
    subset=["Rolling Mean"]
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=plot_df.index,
        y=plot_df["Spread"],
        mode="lines",
        name="WTI-Brent Spread"
    )
)

fig.add_trace(
    go.Scatter(
        x=plot_df.index,
        y=plot_df["Rolling Mean"],
        mode="lines",
        name="60-Day Mean"
    )
)

fig.add_vline(
    x=bull_start.timestamp() * 1000,
    line_dash="dash",
    annotation_text="Bull sample start"
)

label_last(
    fig,
    plot_df.index,
    plot_df["Spread"].values,
    f"{plot_df['Spread'].iloc[-1]:.2f}"
)

style_fig(
    fig,
    "WTI-Brent Spread — 60-Day Trailing Mean",
    "Spread"
)

iplot(fig)


# ============================================================
# PLOT 2
# 60-DAY TRAILING KAPPA
# SPEED OF MEAN REVERSION
# ============================================================

r = rolling.dropna(
    subset=["Kappa"]
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=r.index,
        y=r["Kappa"],
        mode="lines",
        name="60-Day κ"
    )
)

fig.add_vline(
    x=bull_start.timestamp() * 1000,
    line_dash="dash",
    annotation_text="Bull sample"
)

label_last(
    fig,
    r.index,
    r["Kappa"].values,
    f"κ = {r['Kappa'].iloc[-1]:.2f}"
)

style_fig(
    fig,
    "60-Day Trailing Mean-Reversion Speed (κ)",
    "Annualised κ"
)

iplot(fig)


# ============================================================
# PLOT 3
# HALF-LIFE THROUGH TIME
# ============================================================

h = rolling[
    rolling["Half Life"].notna()
].copy()

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=h.index,
        y=h["Half Life"],
        mode="lines",
        name="Half-Life"
    )
)

fig.add_vline(
    x=bull_start.timestamp() * 1000,
    line_dash="dash"
)

label_last(
    fig,
    h.index,
    h["Half Life"].values,
    f"{h['Half Life'].iloc[-1]:.1f} days"
)

style_fig(
    fig,
    "60-Day Trailing OU Half-Life",
    "Trading Days"
)

iplot(fig)


# ============================================================
# PLOT 4
# TRAILING EQUILIBRIUM THETA
# ============================================================

t = rolling.dropna(
    subset=["Theta"]
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df.index,
        y=df["Spread"],
        mode="lines",
        name="Spread"
    )
)

fig.add_trace(
    go.Scatter(
        x=t.index,
        y=t["Theta"],
        mode="lines",
        name="OU θ"
    )
)

label_last(
    fig,
    t.index,
    t["Theta"].values,
    f"θ = {t['Theta'].iloc[-1]:.2f}"
)

style_fig(
    fig,
    "WTI-Brent Spread vs 60-Day OU Equilibrium",
    "Spread"
)

iplot(fig)


# ============================================================
# PLOT 5
# 60-DAY TRAILING Z-SCORE
# ============================================================

z = df.dropna(
    subset=["Z Score"]
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=z.index,
        y=z["Z Score"],
        mode="lines",
        name="60-Day Z"
    )
)

for level in [
    -3, -2.33, -1.96,
    0,
    1.96, 2.33, 3
]:

    fig.add_hline(
        y=level,
        line_dash="dash"
    )

label_last(
    fig,
    z.index,
    z["Z Score"].values,
    f"Z = {z['Z Score'].iloc[-1]:.2f}"
)

style_fig(
    fig,
    "60-Day Trailing Spread Z-Score",
    "Z-Score"
)

iplot(fig)


# ============================================================
# PDF + CDF FUNCTION
# ============================================================

def distribution_graphs(
    series,
    title
):

    x = series.dropna()

    mu = x.mean()
    sigma = x.std()

    grid = np.linspace(
        x.min(),
        x.max(),
        500
    )

    pdf = stats.norm.pdf(
        grid,
        mu,
        sigma
    )

    cdf = stats.norm.cdf(
        grid,
        mu,
        sigma
    )

    current = x.iloc[-1]

    current_z = (
        (current - mu)
        / sigma
    )

    current_cdf = stats.norm.cdf(
        current,
        mu,
        sigma
    )


    # --------------------------------------------------------
    # PDF
    # --------------------------------------------------------

    fig = go.Figure()

    fig.add_trace(
        go.Histogram(
            x=x,
            histnorm="probability density",
            name="Observed",
            opacity=0.55
        )
    )

    fig.add_trace(
        go.Scatter(
            x=grid,
            y=pdf,
            mode="lines",
            name="Normal PDF"
        )
    )

    fig.add_vline(
        x=current,
        line_dash="dash",
        annotation_text=(
            f"Current = {current:.2f}"
        )
    )

    style_fig(
        fig,
        f"{title} — PDF",
        "Density",
        "Spread"
    )

    iplot(fig)


    # --------------------------------------------------------
    # CDF
    # --------------------------------------------------------

    empirical = np.sort(x)

    empirical_cdf = (
        np.arange(
            1,
            len(empirical) + 1
        )
        / len(empirical)
    )

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=empirical,
            y=empirical_cdf,
            mode="lines",
            name="Empirical CDF"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=grid,
            y=cdf,
            mode="lines",
            name="Normal CDF"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=[current],
            y=[current_cdf],
            mode="markers+text",
            text=[
                f"Current<br>Z={current_z:.2f}"
            ],
            textposition="top left",
            name="Current"
        )
    )

    style_fig(
        fig,
        f"{title} — CDF",
        "Cumulative Probability",
        "Spread"
    )

    iplot(fig)


# ============================================================
# FULL SAMPLE PDF / CDF
# ============================================================

distribution_graphs(
    df["Spread"],
    "Full Sample"
)


# ============================================================
# BULL SAMPLE PDF / CDF
# ============================================================

distribution_graphs(
    bull["Spread"],
    "Bull Sample"
)


# ============================================================
# 2026 YTD PDF / CDF
# ============================================================

distribution_graphs(
    ytd["Spread"],
    "2026 YTD"
)


# ============================================================
# FULL-SAMPLE Z-SCORE — NO TRAILING WINDOW
# ============================================================

full_mean = df["Spread"].mean()
full_std = df["Spread"].std()

df["Full Sample Z"] = (
    (df["Spread"] - full_mean)
    / full_std
)


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df.index,
        y=df["Full Sample Z"],
        mode="lines",
        name="Full-Sample Z"
    )
)

for level in [
    -3, -2.33, -1.96,
    0,
    1.96, 2.33, 3
]:

    fig.add_hline(
        y=level,
        line_dash="dash"
    )

label_last(
    fig,
    df.index,
    df["Full Sample Z"].values,
    f"Z = {df['Full Sample Z'].iloc[-1]:.2f}"
)

style_fig(
    fig,
    "Full-Sample Z-Score — No Trailing Window",
    "Z-Score"
)

iplot(fig)


# ============================================================
# ============================================================
# MULTIVARIATE OU
#
# State vector:
#
# X1 = Spread level
# X2 = Daily change in spread
#
# VAR(1):
# X[t+1] = c + A X[t] + error
#
# Continuous OU speed matrix:
# K = -log(A) / dt
#
# Eigenvectors of K describe independent
# mean-reverting modes.
# ============================================================
# ============================================================

mv = pd.DataFrame(
    index=df.index
)

mv["Level"] = df["Spread"]

mv["Change"] = (
    df["Spread"].diff()
)

mv = mv.dropna()


# Standardise so eigenvectors are not dominated
# simply by units / scale

mv_std = (
    (mv - mv.mean())
    / mv.std()
)


# ============================================================
# FIT VAR(1) MANUALLY
# ============================================================

X_t = mv_std.iloc[:-1].values
X_next = mv_std.iloc[1:].values

X_design = np.column_stack([
    np.ones(len(X_t)),
    X_t
])

B = np.linalg.lstsq(
    X_design,
    X_next,
    rcond=None
)[0]

c = B[0, :]

A = B[1:, :].T


print("\nDiscrete VAR(1) Matrix A")
print(A)


# ============================================================
# CONTINUOUS-TIME OU SPEED MATRIX
# ============================================================

dt = 1 / 252

K = (
    -logm(A)
    / dt
)

K = np.real_if_close(K)

print("\nContinuous OU Speed Matrix K")
print(K)


# ============================================================
# EIGENVALUES / EIGENVECTORS
# ============================================================

eigenvalues, eigenvectors = eig(K)

eigenvalues = np.real_if_close(
    eigenvalues
)

eigenvectors = np.real_if_close(
    eigenvectors
)


eig_table = pd.DataFrame({

    "Eigenvalue / Speed":
        eigenvalues,

    "Half-Life Days":
        np.log(2)
        / eigenvalues
        * 252,

    "Level Weight":
        eigenvectors[0, :],

    "Change Weight":
        eigenvectors[1, :]
})


print("\nMVOU EIGENMODES")

display(
    eig_table.style.format({
        "Eigenvalue / Speed": "{:.3f}",
        "Half-Life Days": "{:.2f}",
        "Level Weight": "{:.3f}",
        "Change Weight": "{:.3f}"
    })
)


# ============================================================
# CREATE EIGENMODE SERIES
#
# Mode = X @ eigenvector
#
# Same sign weights:
# variables move together.
#
# Opposite sign weights:
# variables move against each other.
# ============================================================

mode_series = pd.DataFrame(
    index=mv_std.index
)

for i in range(
    eigenvectors.shape[1]
):

    v = np.real(
        eigenvectors[:, i]
    )

    mode_series[
        f"Eigenmode {i+1}"
    ] = (
        mv_std.values @ v
    )


# ============================================================
# CLASSIFY EIGENVECTORS
# ============================================================

mode_labels = []

for i in range(
    eigenvectors.shape[1]
):

    v = np.real(
        eigenvectors[:, i]
    )

    if np.sign(v[0]) == np.sign(v[1]):

        behaviour = "Same Direction"

    else:

        behaviour = "Opposite Direction"

    mode_labels.append(
        behaviour
    )

    print(
        f"\nEigenmode {i+1}: "
        f"{behaviour}"
    )

    print(
        f"Vector = "
        f"[{v[0]:.3f}, {v[1]:.3f}]"
    )

    print(
        f"Speed = "
        f"{np.real(eigenvalues[i]):.3f}"
    )


# ============================================================
# PLOT 6
# MVOU EIGENMODES
# ============================================================

fig = go.Figure()

for i in range(
    mode_series.shape[1]
):

    name = (
        f"Eigenmode {i+1} "
        f"({mode_labels[i]})"
    )

    fig.add_trace(
        go.Scatter(
            x=mode_series.index,
            y=mode_series.iloc[:, i],
            mode="lines",
            name=name
        )
    )

    label_last(
        fig,
        mode_series.index,
        mode_series.iloc[:, i].values,
        (
            f"M{i+1} = "
            f"{mode_series.iloc[-1, i]:.2f}"
        )
    )


style_fig(
    fig,
    "Multivariate OU Eigenmodes",
    "Eigenmode Value"
)

iplot(fig)


# ============================================================
# PLOT 7
# MVOU EIGENVECTORS
#
# Shows direction of each mean-reverting mode.
# ============================================================

fig = go.Figure()

for i in range(
    eigenvectors.shape[1]
):

    v = np.real(
        eigenvectors[:, i]
    )

    fig.add_trace(
        go.Scatter(
            x=[0, v[0]],
            y=[0, v[1]],
            mode="lines+markers+text",
            text=[
                "",
                f"Mode {i+1}<br>{mode_labels[i]}"
            ],
            textposition="top center",
            name=f"Eigenmode {i+1}"
        )
    )


fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.add_vline(
    x=0,
    line_dash="dash"
)

fig.update_layout(
    title="MVOU Eigenvector Directions",
    xaxis_title="Spread Level Loading",
    yaxis_title="Spread Change Loading",
    plot_bgcolor=BG,
    paper_bgcolor=BG,
    font=dict(color=TEXT),
    width=800,
    height=700
)

fig.update_xaxes(
    showgrid=True,
    gridcolor=GRID
)

fig.update_yaxes(
    showgrid=True,
    gridcolor=GRID
)

iplot(fig)


# ============================================================
# PLOT 8
# TRAILING KAPPA — BULL SAMPLE ONLY
# ============================================================

bull_kappa = rolling.loc[
    rolling.index >= bull_start
].dropna(
    subset=["Kappa"]
)


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=bull_kappa.index,
        y=bull_kappa["Kappa"],
        mode="lines",
        name="Bull Sample κ"
    )
)

if len(bull_kappa) > 0:

    label_last(
        fig,
        bull_kappa.index,
        bull_kappa["Kappa"].values,
        f"κ = {bull_kappa['Kappa'].iloc[-1]:.2f}"
    )

style_fig(
    fig,
    "60-Day Mean-Reversion Speed — Bull Sample",
    "Annualised κ"
)

iplot(fig)


# ============================================================
# PLOT 9
# TRAILING KAPPA — 2026 YTD ONLY
# ============================================================

ytd_kappa = rolling.loc[
    rolling.index >= pd.Timestamp(YEAR_START)
].dropna(
    subset=["Kappa"]
)


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=ytd_kappa.index,
        y=ytd_kappa["Kappa"],
        mode="lines",
        name="2026 YTD κ"
    )
)

if len(ytd_kappa) > 0:

    label_last(
        fig,
        ytd_kappa.index,
        ytd_kappa["Kappa"].values,
        f"κ = {ytd_kappa['Kappa'].iloc[-1]:.2f}"
    )

style_fig(
    fig,
    "60-Day Mean-Reversion Speed — 2026 YTD",
    "Annualised κ"
)

iplot(fig)


# ============================================================
# CURRENT POSITION SUMMARY
# ============================================================

current_date = df.index[-1]
current_spread = df["Spread"].iloc[-1]

current_trailing_z = (
    df["Z Score"].iloc[-1]
)

current_full_z = (
    df["Full Sample Z"].iloc[-1]
)

current_kappa = (
    rolling["Kappa"].dropna().iloc[-1]
)

current_half_life = (
    rolling["Half Life"].dropna().iloc[-1]
)


print("\n")
print("=" * 80)
print("CURRENT SPREAD STATE")
print("=" * 80)

print(
    f"Date:                 "
    f"{current_date.date()}"
)

print(
    f"Spread:               "
    f"{current_spread:.2f}"
)

print(
    f"60-Day Z-Score:       "
    f"{current_trailing_z:.3f}"
)

print(
    f"Full-Sample Z-Score:  "
    f"{current_full_z:.3f}"
)

print(
    f"60-Day Kappa:         "
    f"{current_kappa:.3f}"
)

print(
    f"60-Day Half-Life:     "
    f"{current_half_life:.2f} trading days"
)
```
