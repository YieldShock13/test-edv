```python
# ============================================================
# FOUR-STOCK RELATIVE VALUE / RISK ANALYSIS
# DSBPp, MTMJ, OMU, SLMJ
# Weekly data | 2017 onward
# ============================================================

# ============================================================
# 1. IMPORTS
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from scipy.linalg import logm, expm
from scipy.stats import chi2
from statsmodels.tsa.vector_ar.vecm import coint_johansen

import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
pd.set_option("display.float_format", lambda x: f"{x:,.6f}")


# ============================================================
# 2. SETTINGS
# ============================================================

FILES = {
    "DSYJ": "DSYJ Historical Data.csv",
    "MTMJ":  "MTMJ Historical Data.csv",
    "OMU":   "OMU Historical Data.csv",
    "SLMJ":  "SLMJ Historical Data (1).csv",
}

START_DATE = "2017-01-01"

ROLLING_WINDOW = 52
VAR_ALPHA = 0.05

# Divergence / convergence rules
DIVERGENCE_Z = 1.50
CONVERGENCE_Z = 0.50

TICKERS = list(FILES.keys())


# ============================================================
# 3. LOAD INVESTING.COM CSV FILE
# ============================================================

def parse_number(x):
    if pd.isna(x):
        return np.nan

    s = str(x).strip()
    s = s.replace(",", "")
    s = s.replace("%", "")

    if s in ["", "-", "nan", "None"]:
        return np.nan

    try:
        return float(s)
    except:
        return np.nan


def load_price_file(filename, ticker):

    df = pd.read_csv(filename)

    if "Date" not in df.columns:
        raise ValueError(f"{ticker}: Date column not found.")

    if "Price" not in df.columns:
        raise ValueError(f"{ticker}: Price column not found.")

    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df["Price"] = df["Price"].apply(parse_number)

    df = df[["Date", "Price"]].dropna()
    df = df.sort_values("Date")
    df = df.drop_duplicates("Date", keep="last")

    df = df[df["Date"] >= pd.Timestamp(START_DATE)]

    df = df.set_index("Date")
    df.columns = [ticker]

    return df


price_frames = []

for ticker, filename in FILES.items():

    temp = load_price_file(filename, ticker)

    print(
        f"{ticker}: "
        f"{temp.index.min().date()} to "
        f"{temp.index.max().date()} | "
        f"{len(temp)} observations"
    )

    price_frames.append(temp)


# ============================================================
# 4. ALIGN PRICE SERIES
# ============================================================

prices = pd.concat(price_frames, axis=1, join="inner")
prices = prices.dropna()

print("\nALIGNED PRICE DATA")
print(prices.head())

print("\nAligned start:", prices.index.min())
print("Aligned end:  ", prices.index.max())
print("Aligned observations:", len(prices))

print("\nMissing observations after alignment:")
print(prices.isna().sum())


# ============================================================
# 5. CHECK DATE SPACING
# ============================================================

spacing = prices.index.to_series().diff().dt.days.dropna()

print("\nDATE-SPACING SUMMARY")
print(spacing.describe())

print("\nMost common spacing:")
print(spacing.value_counts().head())


# ============================================================
# 6. WEEKLY RETURNS
# ============================================================

returns = prices.pct_change().dropna()

# Equal-weight, weekly-rebalanced portfolio
returns["EQUAL_WEIGHT"] = returns[TICKERS].mean(axis=1)

print("\nWEEKLY RETURNS")
print(returns.head())

print("\nFULL-SAMPLE RETURN SUMMARY")
print(returns.describe())


# ============================================================
# 7. REBASED PRICE CHART
# ============================================================

rebased = prices / prices.iloc[0] * 100

fig = go.Figure()

for ticker in TICKERS:
    fig.add_trace(
        go.Scatter(
            x=rebased.index,
            y=rebased[ticker],
            mode="lines",
            name=ticker
        )
    )

fig.update_layout(
    title="Rebased Prices: 2017 = 100",
    xaxis_title="Date",
    yaxis_title="Rebased Price",
    template="plotly_white"
)

fig.show()


# ============================================================
# 8. ABNORMAL RETURN / POSSIBLE SPLIT CHECK
# ============================================================

abnormal_rows = []

split_ratios = np.array([
    0.10, 0.20, 0.25, 1/3, 0.50,
    2.00, 3.00, 4.00, 5.00, 10.00
])

for ticker in TICKERS:

    r = returns[ticker]

    median = r.median()
    mad = np.median(np.abs(r - median))

    if mad == 0:
        robust_z = pd.Series(np.nan, index=r.index)
    else:
        robust_z = 0.6745 * (r - median) / mad

    candidates = robust_z[np.abs(robust_z) >= 6]

    for date, rz in candidates.items():

        current_price = prices.loc[date, ticker]
        previous_price = prices[ticker].shift(1).loc[date]

        ratio = current_price / previous_price

        closest_split = split_ratios[
            np.argmin(np.abs(split_ratios - ratio))
        ]

        split_distance = abs(ratio - closest_split) / closest_split

        abnormal_rows.append({
            "Date": date,
            "Ticker": ticker,
            "Weekly Return": r.loc[date],
            "Robust Z": rz,
            "Price Ratio": ratio,
            "Closest Split Ratio": closest_split,
            "Possible Split": split_distance < 0.05
        })


abnormal_df = pd.DataFrame(abnormal_rows)

print("\nPOTENTIAL ABNORMAL OBSERVATIONS")

if len(abnormal_df) == 0:
    print("None detected.")
else:
    print(abnormal_df.to_string(index=False))


# ============================================================
# 9. FULL-SAMPLE WEEKLY RETURN CORRELATION
# ============================================================

return_corr = returns[TICKERS].corr()

print("\nFULL-SAMPLE WEEKLY RETURN CORRELATION MATRIX")
print(return_corr)

fig = go.Figure(
    data=go.Heatmap(
        z=return_corr.values,
        x=return_corr.columns,
        y=return_corr.index,
        zmin=-1,
        zmax=1,
        text=np.round(return_corr.values, 3),
        texttemplate="%{text}"
    )
)

fig.update_layout(
    title="Full-Sample Weekly Return Correlation",
    template="plotly_white"
)

fig.show()


# ============================================================
# 10. ROLLING 52-WEEK RETURN CORRELATIONS
# ============================================================

pairs = []

for i in range(len(TICKERS)):
    for j in range(i + 1, len(TICKERS)):
        pairs.append((TICKERS[i], TICKERS[j]))


rolling_corr = pd.DataFrame(index=returns.index)

for a, b in pairs:

    rolling_corr[f"{a} / {b}"] = (
        returns[a]
        .rolling(ROLLING_WINDOW)
        .corr(returns[b])
    )


fig = go.Figure()

for col in rolling_corr.columns:

    fig.add_trace(
        go.Scatter(
            x=rolling_corr.index,
            y=rolling_corr[col],
            mode="lines",
            name=col
        )
    )

fig.update_layout(
    title="Rolling 52-Week Weekly-Return Correlations",
    xaxis_title="Date",
    yaxis_title="Correlation",
    template="plotly_white"
)

fig.add_hline(y=0, line_dash="dot")

fig.show()


# ============================================================
# 11. LOG PRICE LEVELS
# ============================================================

log_prices = np.log(prices)

print("\nLOG PRICE LEVELS")
print(log_prices.head())


# ============================================================
# 12. JOHANSEN COINTEGRATION TEST
# ============================================================

# det_order = 0:
# constant term in cointegrating relation
#
# k_ar_diff = 1:
# one lag of differenced data

johansen = coint_johansen(
    log_prices[TICKERS],
    det_order=0,
    k_ar_diff=1
)

johansen_table = pd.DataFrame({
    "Rank Tested": np.arange(len(TICKERS)),
    "Trace Statistic": johansen.lr1,
    "90% Critical": johansen.cvt[:, 0],
    "95% Critical": johansen.cvt[:, 1],
    "99% Critical": johansen.cvt[:, 2]
})

print("\nJOHANSEN COINTEGRATION TEST")
print(johansen_table)

# Determine rank at 95%
cointegration_rank = 0

for i in range(len(TICKERS)):

    if johansen.lr1[i] > johansen.cvt[i, 1]:
        cointegration_rank = i + 1
    else:
        break

print("\nEstimated cointegration rank at 95%:",
      cointegration_rank)


# ============================================================
# 13. PACK-RELATIVE LOG-PRICE LEVELS
# ============================================================

# Relative level:
#
# log(P_i) - average log price of all four stocks
#
# This removes the common pack level.
#
# The four deviations sum to zero, so only three are
# independent.

pack_mean = log_prices[TICKERS].mean(axis=1)

relative_levels = log_prices[TICKERS].sub(
    pack_mean,
    axis=0
)

print("\nPACK-RELATIVE LOG-PRICE LEVELS")
print(relative_levels.head())

print("\nMaximum zero-sum error:")
print(relative_levels.sum(axis=1).abs().max())


# ============================================================
# 14. ROLLING 52-WEEK RELATIVE-LEVEL Z-SCORES
# ============================================================

rolling_mean = relative_levels.rolling(
    ROLLING_WINDOW
).mean()

rolling_std = relative_levels.rolling(
    ROLLING_WINDOW
).std()

relative_z = (
    relative_levels - rolling_mean
) / rolling_std


fig = go.Figure()

for ticker in TICKERS:

    fig.add_trace(
        go.Scatter(
            x=relative_z.index,
            y=relative_z[ticker],
            mode="lines",
            name=ticker
        )
    )

fig.add_hline(y=0, line_dash="dot")
fig.add_hline(y=DIVERGENCE_Z, line_dash="dash")
fig.add_hline(y=-DIVERGENCE_Z, line_dash="dash")

fig.update_layout(
    title="Rolling 52-Week Pack-Relative Z-Scores",
    xaxis_title="Date",
    yaxis_title="Z-Score",
    template="plotly_white"
)

fig.show()


# ============================================================
# 15. CURRENT RELATIVE-VALUE DISLOCATION
# ============================================================

latest_z = relative_z.iloc[-1].dropna()

current_dislocation = pd.DataFrame({
    "Z Score": latest_z,
    "Absolute Z": latest_z.abs(),
    "Direction": np.where(
        latest_z >= 0,
        "Above pack",
        "Below pack"
    )
})

current_dislocation = current_dislocation.sort_values(
    "Absolute Z",
    ascending=False
)

print("\nCURRENT PACK-RELATIVE DISLOCATIONS")
print(current_dislocation)


# ============================================================
# 16. SANLAM / OLD MUTUAL DIRECT RELATIVE SPREAD
# ============================================================

slmj_omu_spread = (
    log_prices["SLMJ"] -
    log_prices["OMU"]
)

spread_mean = slmj_omu_spread.rolling(
    ROLLING_WINDOW
).mean()

spread_std = slmj_omu_spread.rolling(
    ROLLING_WINDOW
).std()

slmj_omu_spread_z = (
    slmj_omu_spread - spread_mean
) / spread_std


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=slmj_omu_spread_z.index,
        y=slmj_omu_spread_z,
        mode="lines",
        name="SLMJ - OMU"
    )
)

fig.add_hline(y=0, line_dash="dot")
fig.add_hline(y=DIVERGENCE_Z, line_dash="dash")
fig.add_hline(y=-DIVERGENCE_Z, line_dash="dash")

fig.update_layout(
    title="Sanlam / Old Mutual Relative Log-Price Spread Z-Score",
    xaxis_title="Date",
    yaxis_title="Rolling 52-Week Z-Score",
    template="plotly_white"
)

fig.show()


# ============================================================
# 17. MVOU FUNCTIONS
# ============================================================

# Because the four pack-relative deviations sum to zero,
# use three independent states.
#
# DSBPp is implied by:
#
# DSBPp = -(MTMJ + OMU + SLMJ)

MVOU_COLS = ["MTMJ", "OMU", "SLMJ"]


def fit_var1(data):

    """
    Fit:
        X_(t+1) = c + A X_t + epsilon_(t+1)
    """

    X = np.asarray(data, dtype=float)

    X0 = X[:-1]
    X1 = X[1:]

    regressors = np.column_stack([
        np.ones(len(X0)),
        X0
    ])

    beta = np.linalg.lstsq(
        regressors,
        X1,
        rcond=None
    )[0]

    c = beta[0]
    A = beta[1:].T

    fitted = regressors @ beta
    residuals = X1 - fitted

    cov = np.cov(
        residuals,
        rowvar=False,
        ddof=1
    )

    return c, A, cov, residuals


def var_to_mvou(c, A):

    """
    Exact weekly discrete-to-continuous conversion.

    X_(t+1) = c + A X_t + eps

    dX = K(theta - X)dt + Sigma dW

    A = exp(-K * dt)

    dt = 1 week
    """

    eig_A = np.linalg.eigvals(A)

    if np.any(np.abs(eig_A) >= 1):
        stable = False
    else:
        stable = True

    K = -logm(A)

    K = np.real_if_close(K)

    I = np.eye(A.shape[0])

    try:
        theta = np.linalg.solve(
            I - A,
            c
        )
    except np.linalg.LinAlgError:
        theta = np.full(len(c), np.nan)

    eig_K = np.linalg.eigvals(K)
    eig_K = np.real_if_close(eig_K)

    return K, theta, eig_A, eig_K, stable


def half_lives_from_K(K):

    eigvals = np.linalg.eigvals(K)
    eigvals = np.real_if_close(eigvals)

    rows = []

    for i, eig in enumerate(eigvals):

        real_part = np.real(eig)

        if real_part > 0:
            half_life = np.log(2) / real_part
        else:
            half_life = np.nan

        rows.append({
            "Mode": i + 1,
            "K Eigenvalue": eig,
            "Mean-Reversion Speed": real_part,
            "Half-Life Weeks": half_life
        })

    result = pd.DataFrame(rows)

    return result.sort_values(
        "Mean-Reversion Speed"
    ).reset_index(drop=True)


# ============================================================
# 18. FULL-SAMPLE MVOU
# ============================================================

full_state = relative_levels[
    MVOU_COLS
].dropna()

c_full, A_full, cov_full, resid_full = fit_var1(
    full_state.values
)

K_full, theta_full, eigA_full, eigK_full, stable_full = (
    var_to_mvou(
        c_full,
        A_full
    )
)

full_half_lives = half_lives_from_K(K_full)

print("\n======================================")
print("FULL-SAMPLE MVOU: 2017 TO PRESENT")
print("======================================")

print("\nDISCRETE WEEKLY TRANSITION MATRIX A")
print(
    pd.DataFrame(
        A_full,
        index=MVOU_COLS,
        columns=MVOU_COLS
    )
)

print("\nCONTINUOUS-TIME MVOU K MATRIX")
print(
    pd.DataFrame(
        K_full,
        index=MVOU_COLS,
        columns=MVOU_COLS
    )
)

print("\nLONG-RUN MEAN VECTOR THETA")
print(
    pd.Series(
        theta_full,
        index=MVOU_COLS
    )
)

print("\nINNOVATION COVARIANCE")
print(
    pd.DataFrame(
        cov_full,
        index=MVOU_COLS,
        columns=MVOU_COLS
    )
)

print("\nEIGENVALUES OF A")
print(eigA_full)

print("\nEIGENVALUES OF K")
print(eigK_full)

print("\nDISCRETE SYSTEM STABLE:")
print(stable_full)

print("\nFULL-SAMPLE MVOU HALF-LIVES")
print(full_half_lives)


# ============================================================
# 19. LATEST 52-WEEK MVOU
# ============================================================

latest_state = full_state.iloc[
    -ROLLING_WINDOW:
]

c_52, A_52, cov_52, resid_52 = fit_var1(
    latest_state.values
)

K_52, theta_52, eigA_52, eigK_52, stable_52 = (
    var_to_mvou(
        c_52,
        A_52
    )
)

latest_half_lives = half_lives_from_K(K_52)

print("\n======================================")
print("LATEST 52-WEEK MVOU")
print("======================================")

print("\nDISCRETE WEEKLY TRANSITION MATRIX A")
print(
    pd.DataFrame(
        A_52,
        index=MVOU_COLS,
        columns=MVOU_COLS
    )
)

print("\nCONTINUOUS-TIME MVOU K MATRIX")
print(
    pd.DataFrame(
        K_52,
        index=MVOU_COLS,
        columns=MVOU_COLS
    )
)

print("\nLONG-RUN MEAN VECTOR THETA")
print(
    pd.Series(
        theta_52,
        index=MVOU_COLS
    )
)

print("\nLATEST 52-WEEK MVOU HALF-LIVES")
print(latest_half_lives)


# ============================================================
# 20. ROLLING 52-WEEK MVOU
# ============================================================

rolling_mvou_rows = []

for end in range(
    ROLLING_WINDOW,
    len(full_state) + 1
):

    window = full_state.iloc[
        end - ROLLING_WINDOW:end
    ]

    date = window.index[-1]

    try:

        c, A, cov, resid = fit_var1(
            window.values
        )

        K, theta, eigA, eigK, stable = (
            var_to_mvou(c, A)
        )

        half_lives = half_lives_from_K(K)

        valid_hl = (
            half_lives["Half-Life Weeks"]
            .replace([np.inf, -np.inf], np.nan)
            .dropna()
        )

        if len(valid_hl) > 0:
            slowest_hl = valid_hl.max()
            fastest_hl = valid_hl.min()
            median_hl = valid_hl.median()
        else:
            slowest_hl = np.nan
            fastest_hl = np.nan
            median_hl = np.nan

        rolling_mvou_rows.append({
            "Date": date,
            "Stable": stable,
            "Slowest Half-Life": slowest_hl,
            "Median Half-Life": median_hl,
            "Fastest Half-Life": fastest_hl
        })

    except Exception:

        rolling_mvou_rows.append({
            "Date": date,
            "Stable": False,
            "Slowest Half-Life": np.nan,
            "Median Half-Life": np.nan,
            "Fastest Half-Life": np.nan
        })


rolling_mvou = pd.DataFrame(
    rolling_mvou_rows
).set_index("Date")


print("\nROLLING MVOU SUMMARY")
print(rolling_mvou.describe())


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=rolling_mvou.index,
        y=rolling_mvou["Slowest Half-Life"],
        mode="lines",
        name="Slowest MVOU Mode"
    )
)

fig.add_trace(
    go.Scatter(
        x=rolling_mvou.index,
        y=rolling_mvou["Median Half-Life"],
        mode="lines",
        name="Median MVOU Mode"
    )
)

fig.update_layout(
    title="Rolling 52-Week MVOU Mean-Reversion Half-Life",
    xaxis_title="Date",
    yaxis_title="Weeks",
    template="plotly_white"
)

fig.show()


# ============================================================
# 21. HISTORICAL SLMJ / OMU CONVERGENCE EPISODES
# ============================================================

# IMPORTANT:
#
# Previous code counted repeated threshold crossings during
# the same unresolved divergence as separate events.
#
# This version treats the entire period from first divergence
# until convergence as ONE episode.


def find_convergence_episodes(
    z_df,
    ticker_a="SLMJ",
    ticker_b="OMU",
    divergence_threshold=1.5,
    convergence_threshold=0.5
):

    data = z_df[
        [ticker_a, ticker_b]
    ].dropna()

    episodes = []

    i = 0

    while i < len(data):

        row = data.iloc[i]

        divergent = (
            abs(row[ticker_a]) >= divergence_threshold
            or
            abs(row[ticker_b]) >= divergence_threshold
        )

        if not divergent:
            i += 1
            continue

        start_i = i
        start_date = data.index[i]

        convergence_i = None

        for j in range(i + 1, len(data)):

            future = data.iloc[j]

            converged = (
                abs(future[ticker_a]) <= convergence_threshold
                and
                abs(future[ticker_b]) <= convergence_threshold
            )

            if converged:
                convergence_i = j
                break

        if convergence_i is None:

            episodes.append({
                "Divergence Date": start_date,
                f"{ticker_a} Start Z": row[ticker_a],
                f"{ticker_b} Start Z": row[ticker_b],
                "Convergence Date": pd.NaT,
                "Weeks to Converge": np.nan
            })

            break

        convergence_date = data.index[
            convergence_i
        ]

        weeks = (
            convergence_date -
            start_date
        ).days / 7

        episodes.append({
            "Divergence Date": start_date,
            f"{ticker_a} Start Z": row[ticker_a],
            f"{ticker_b} Start Z": row[ticker_b],
            "Convergence Date": convergence_date,
            "Weeks to Converge": weeks
        })

        # Skip the whole episode.
        i = convergence_i + 1

    return pd.DataFrame(episodes)


convergence_episodes = find_convergence_episodes(
    relative_z,
    ticker_a="SLMJ",
    ticker_b="OMU",
    divergence_threshold=DIVERGENCE_Z,
    convergence_threshold=CONVERGENCE_Z
)

print("\n======================================")
print("SANLAM / OLD MUTUAL HISTORICAL")
print("CONVERGENCE EPISODES")
print("======================================")

print(convergence_episodes.to_string(index=False))


completed_episodes = convergence_episodes[
    "Weeks to Converge"
].dropna()

if len(completed_episodes) > 0:

    print("\nCOMPLETED EPISODES:", len(completed_episodes))

    print(
        "Median weeks to converge:",
        completed_episodes.median()
    )

    print(
        "Mean weeks to converge:",
        completed_episodes.mean()
    )

    print(
        "Minimum weeks:",
        completed_episodes.min()
    )

    print(
        "Maximum weeks:",
        completed_episodes.max()
    )

    print(
        "25th percentile:",
        completed_episodes.quantile(0.25)
    )

    print(
        "75th percentile:",
        completed_episodes.quantile(0.75)
    )


# ============================================================
# 22. HISTORICAL CONVERGENCE DISTRIBUTION
# ============================================================

if len(completed_episodes) > 0:

    fig = go.Figure()

    fig.add_trace(
        go.Histogram(
            x=completed_episodes,
            nbinsx=15,
            name="Historical Episodes"
        )
    )

    fig.update_layout(
        title="Historical Sanlam / Old Mutual Convergence Times",
        xaxis_title="Weeks to Converge",
        yaxis_title="Number of Episodes",
        template="plotly_white"
    )

    fig.show()


# ============================================================
# 23. MODEL EXPECTED PATH FUNCTION
# ============================================================

def expected_mvou_path(
    x0,
    A,
    theta,
    horizon=104
):

    """
    Discrete expected path:

    E[X_(t+h)] =
        theta + A^h (X_t - theta)
    """

    paths = []

    x0 = np.asarray(x0)
    theta = np.asarray(theta)

    for h in range(horizon + 1):

        Ah = np.linalg.matrix_power(A, h)

        expected = (
            theta +
            Ah @ (x0 - theta)
        )

        paths.append(expected)

    return np.asarray(paths)


# ============================================================
# 24. FULL-SAMPLE MVOU EXPECTED PATH
# ============================================================

current_state = full_state.iloc[-1].values

HORIZON = 104

path_full = expected_mvou_path(
    current_state,
    A_full,
    theta_full,
    horizon=HORIZON
)

path_52 = expected_mvou_path(
    current_state,
    A_52,
    theta_52,
    horizon=HORIZON
)


# ============================================================
# 25. NORMALISE MODEL PATH AS FRACTION OF CURRENT GAP
# ============================================================

# 1.0 = today's distance from model equilibrium
# 0.0 = model equilibrium
#
# This avoids the previous problem of converting the MVOU
# forecast back into a rolling historical Z-score whose zero
# was NOT the MVOU equilibrium.


def gap_fraction(path, x0, theta):

    initial_gap = x0 - theta

    output = np.full_like(
        path,
        np.nan,
        dtype=float
    )

    for j in range(path.shape[1]):

        if abs(initial_gap[j]) > 1e-12:

            output[:, j] = (
                path[:, j] - theta[j]
            ) / initial_gap[j]

    return output


gap_full = gap_fraction(
    path_full,
    current_state,
    theta_full
)

gap_52 = gap_fraction(
    path_52,
    current_state,
    theta_52
)


# ============================================================
# 26. FULL SAMPLE VS 52-WEEK EXPECTED CONVERGENCE
# ============================================================

weeks = np.arange(HORIZON + 1)

slmj_idx = MVOU_COLS.index("SLMJ")
omu_idx = MVOU_COLS.index("OMU")

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=weeks,
        y=gap_full[:, slmj_idx],
        mode="lines",
        name="SLMJ — Full Sample"
    )
)

fig.add_trace(
    go.Scatter(
        x=weeks,
        y=gap_full[:, omu_idx],
        mode="lines",
        name="OMU — Full Sample"
    )
)

fig.add_trace(
    go.Scatter(
        x=weeks,
        y=gap_52[:, slmj_idx],
        mode="lines",
        line=dict(dash="dash"),
        name="SLMJ — Latest 52 Weeks"
    )
)

fig.add_trace(
    go.Scatter(
        x=weeks,
        y=gap_52[:, omu_idx],
        mode="lines",
        line=dict(dash="dash"),
        name="OMU — Latest 52 Weeks"
    )
)

fig.add_hline(
    y=0,
    line_dash="dot"
)

fig.add_hline(
    y=0.50,
    line_dash="dot"
)

fig.add_hline(
    y=0.25,
    line_dash="dot"
)

fig.update_layout(
    title="MVOU Expected Closure of Current Dislocation",
    xaxis_title="Weeks Forward",
    yaxis_title="Fraction of Current Gap Remaining",
    template="plotly_white"
)

fig.show()


# ============================================================
# 27. MODEL GAP-CLOSURE TIMES
# ============================================================

def first_gap_below(
    gap_series,
    threshold
):

    vals = np.abs(gap_series)

    idx = np.where(
        vals <= threshold
    )[0]

    if len(idx) == 0:
        return np.nan

    return int(idx[0])


closure_rows = []

for model_name, gap in [
    ("Full Sample", gap_full),
    ("Latest 52 Weeks", gap_52)
]:

    for ticker in ["SLMJ", "OMU"]:

        j = MVOU_COLS.index(ticker)

        closure_rows.append({
            "Model": model_name,
            "Ticker": ticker,
            "50% Gap Closed": first_gap_below(
                gap[:, j],
                0.50
            ),
            "75% Gap Closed": first_gap_below(
                gap[:, j],
                0.25
            ),
            "90% Gap Closed": first_gap_below(
                gap[:, j],
                0.10
            )
        })


closure_table = pd.DataFrame(
    closure_rows
)

print("\nMODEL-IMPLIED GAP CLOSURE")
print(closure_table)


# ============================================================
# 28. CURRENT 95% HISTORICAL VaR
# ============================================================

def historical_var(series, alpha=0.05):

    q = np.quantile(
        series.dropna(),
        alpha
    )

    return max(0, -q)


current_var_rows = []

for ticker in returns.columns:

    recent = returns[ticker].iloc[
        -ROLLING_WINDOW:
    ]

    weekly_var = historical_var(
        recent,
        VAR_ALPHA
    )

    monthly_var = (
        weekly_var *
        np.sqrt(52 / 12)
    )

    annual_var = (
        weekly_var *
        np.sqrt(52)
    )

    current_var_rows.append({
        "Ticker": ticker,
        "Weekly VaR 95%": weekly_var,
        "Monthly VaR 95%": monthly_var,
        "Annual VaR 95%": annual_var
    })


current_var = pd.DataFrame(
    current_var_rows
).set_index("Ticker")

print("\nCURRENT 95% HISTORICAL VaR")
print(current_var)

print(
    "\nNOTE: Monthly and annual values are "
    "square-root-of-time scalings of weekly VaR."
)


# ============================================================
# 29. WALK-FORWARD WEEKLY VaR BACKTEST
# ============================================================

def walk_forward_var(
    series,
    window=52,
    alpha=0.05
):

    series = series.dropna()

    rows = []

    for i in range(window, len(series)):

        historical_window = series.iloc[
            i-window:i
        ]

        realized = series.iloc[i]

        var = historical_var(
            historical_window,
            alpha
        )

        threshold = -var

        exceedance = (
            realized < threshold
        )

        rows.append({
            "Date": series.index[i],
            "Realized Return": realized,
            "VaR": var,
            "VaR Threshold": threshold,
            "Exceedance": exceedance
        })

    return pd.DataFrame(
        rows
    ).set_index("Date")


var_backtests = {}

for ticker in returns.columns:

    var_backtests[ticker] = (
        walk_forward_var(
            returns[ticker],
            ROLLING_WINDOW,
            VAR_ALPHA
        )
    )


# ============================================================
# 30. KUPIEC POF TEST
# ============================================================

def kupiec_test(
    exceedances,
    alpha=0.05
):

    exceedances = np.asarray(
        exceedances,
        dtype=int
    )

    n = len(exceedances)
    x = exceedances.sum()

    if n == 0:
        return np.nan, np.nan

    p_hat = x / n

    eps = 1e-12

    p_hat = np.clip(
        p_hat,
        eps,
        1 - eps
    )

    alpha_c = np.clip(
        alpha,
        eps,
        1 - eps
    )

    log_null = (
        (n - x) * np.log(1 - alpha_c) +
        x * np.log(alpha_c)
    )

    log_alt = (
        (n - x) * np.log(1 - p_hat) +
        x * np.log(p_hat)
    )

    LR = -2 * (
        log_null - log_alt
    )

    p_value = (
        1 -
        chi2.cdf(LR, df=1)
    )

    return LR, p_value


backtest_rows = []

for ticker, bt in var_backtests.items():

    n = len(bt)

    x = int(
        bt["Exceedance"].sum()
    )

    rate = x / n

    LR, p_value = kupiec_test(
        bt["Exceedance"],
        VAR_ALPHA
    )

    backtest_rows.append({
        "Ticker": ticker,
        "Forecasts": n,
        "Exceedances": x,
        "Expected Exceedances": n * VAR_ALPHA,
        "Exceedance Rate": rate,
        "Expected Rate": VAR_ALPHA,
        "Kupiec LR": LR,
        "Kupiec p-value": p_value,
        "Reject Correct Coverage at 5%":
            p_value < 0.05
    })


backtest_summary = pd.DataFrame(
    backtest_rows
).set_index("Ticker")

print("\nVaR WALK-FORWARD BACKTEST")
print(backtest_summary)


# ============================================================
# 31. VaR EXCEEDANCE TABLES
# ============================================================

for ticker, bt in var_backtests.items():

    exceed = bt[
        bt["Exceedance"]
    ]

    print(
        f"\n{ticker} VaR EXCEEDANCES"
    )

    print(
        exceed[
            [
                "Realized Return",
                "VaR Threshold",
                "VaR"
            ]
        ]
    )


# ============================================================
# 32. VaR BACKTEST CHARTS
# ============================================================

for ticker, bt in var_backtests.items():

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=bt.index,
            y=bt["Realized Return"],
            mode="lines",
            name="Realized Weekly Return"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=bt.index,
            y=bt["VaR Threshold"],
            mode="lines",
            name="95% VaR Threshold"
        )
    )

    exceed = bt[
        bt["Exceedance"]
    ]

    fig.add_trace(
        go.Scatter(
            x=exceed.index,
            y=exceed["Realized Return"],
            mode="markers",
            marker=dict(size=8),
            name="VaR Exceedance"
        )
    )

    fig.update_layout(
        title=f"{ticker}: Rolling 52-Week Historical VaR Backtest",
        xaxis_title="Date",
        yaxis_title="Weekly Return",
        template="plotly_white"
    )

    fig.show()


# ============================================================
# 33. ROLLING WEEKLY VaR
# ============================================================

rolling_var = pd.DataFrame(
    index=returns.index
)

for ticker in returns.columns:

    rolling_var[ticker] = (
        returns[ticker]
        .rolling(ROLLING_WINDOW)
        .apply(
            lambda x:
                historical_var(
                    pd.Series(x),
                    VAR_ALPHA
                ),
            raw=False
        )
    )


fig = go.Figure()

for ticker in rolling_var.columns:

    fig.add_trace(
        go.Scatter(
            x=rolling_var.index,
            y=rolling_var[ticker],
            mode="lines",
            name=ticker
        )
    )

fig.update_layout(
    title="Rolling 52-Week Historical 95% Weekly VaR",
    xaxis_title="Date",
    yaxis_title="VaR",
    template="plotly_white"
)

fig.show()


# ============================================================
# 34. FINAL SUMMARY
# ============================================================

print("\n\n======================================")
print("FINAL SUMMARY")
print("======================================")

print("\nDATA")
print(
    f"{prices.index.min().date()} "
    f"to {prices.index.max().date()}"
)
print(
    f"{len(prices)} weekly price observations"
)
print(
    f"{len(returns)} weekly return observations"
)

print("\nCOINTEGRATION")
print(
    "Johansen rank at 95%:",
    cointegration_rank
)

print("\nCURRENT PACK DISLOCATIONS")
print(current_dislocation)

print("\nFULL-SAMPLE RETURN CORRELATION")
print(return_corr)

print("\nFULL-SAMPLE MVOU HALF-LIVES")
print(full_half_lives)

print("\nLATEST 52-WEEK MVOU HALF-LIVES")
print(latest_half_lives)

print("\nHISTORICAL SLMJ / OMU CONVERGENCE")

if len(completed_episodes) > 0:

    print(
        "Completed episodes:",
        len(completed_episodes)
    )

    print(
        "Median:",
        completed_episodes.median(),
        "weeks"
    )

    print(
        "Mean:",
        completed_episodes.mean(),
        "weeks"
    )

else:

    print("No completed episodes found.")


print("\nMODEL-IMPLIED CURRENT GAP CLOSURE")
print(closure_table)

print("\nCURRENT 95% VaR")
print(current_var)

print("\nVaR BACKTEST")
print(backtest_summary)
```
