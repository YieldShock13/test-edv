```python
# ============================================================
# SA RATES RELATIVE-VALUE DATASET
#
# JSE:
#   3M JIBAR
#
# RBOND:
#   Nominal fixed-rate SAGBs ONLY (bond_type == "fixed")
#
# Construct:
#   1Y constant-maturity SAGB  - 3M JIBAR
#   5Y constant-maturity SAGB  - 3M JIBAR
#   10Y constant-maturity SAGB - 3M JIBAR
#
# Frequency:
#   Weekly, Friday-ending
# ============================================================

!pip -q install xlrd openpyxl beautifulsoup4 lxml requests

import io
import re
import warnings
import requests
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup

warnings.filterwarnings("ignore")

START = "2018-01-01"
END   = "2026-09-21"

session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0"
})


# ============================================================
# HELPER
# ============================================================

def numeric(series):

    return pd.to_numeric(
        series.astype(str)
              .str.replace("%", "", regex=False)
              .str.replace(",", "", regex=False)
              .str.strip(),
        errors="coerce"
    )


# ============================================================
# 1. JSE 3M JIBAR
# ============================================================

print("=" * 65)
print("1. DOWNLOADING JSE 3M JIBAR")
print("=" * 65)

jibar_frames = []

for year in range(2018, 2027):

    page_url = (
        "https://clientportal.jse.co.za/downloadable-files"
        f"?RequestNode=%2FSafex%2Fmtmdata%2FJIBAR+{year}"
    )

    try:

        page = session.get(
            page_url,
            timeout=30
        )

        page.raise_for_status()

        soup = BeautifulSoup(
            page.text,
            "html.parser"
        )

        urls = []

        for a in soup.find_all("a", href=True):

            href = a["href"]

            if not re.search(
                r"\.xlsx?$",
                href,
                flags=re.I
            ):
                continue

            if href.startswith("/"):

                href = (
                    "https://clientportal.jse.co.za"
                    + href
                )

            elif not href.startswith("http"):

                href = (
                    "https://clientportal.jse.co.za/"
                    + href.lstrip("/")
                )

            if (
                "jibar" in href.lower()
                or
                "jbar" in href.lower()
            ):
                urls.append(href)

        urls = list(dict.fromkeys(urls))

        print(
            f"{year}: {len(urls)} files"
        )

        for url in urls:

            try:

                r = session.get(
                    url,
                    timeout=30
                )

                r.raise_for_status()

                blob = r.content

                book = pd.ExcelFile(
                    io.BytesIO(blob)
                )

                for sheet in book.sheet_names:

                    raw = pd.read_excel(
                        io.BytesIO(blob),
                        sheet_name=sheet,
                        header=None
                    )

                    header = None

                    for i in range(
                        min(25, len(raw))
                    ):

                        txt = " ".join(
                            raw.iloc[i]
                               .astype(str)
                               .str.lower()
                               .tolist()
                        )

                        if (
                            "date" in txt
                            and
                            "jibar" in txt
                        ):
                            header = i
                            break

                    if header is None:
                        continue

                    df = pd.read_excel(
                        io.BytesIO(blob),
                        sheet_name=sheet,
                        header=header
                    )

                    df.columns = [
                        str(c)
                        .replace("\n", " ")
                        .strip()
                        for c in df.columns
                    ]

                    date_col = next(
                        (
                            c for c in df.columns
                            if "date" in c.lower()
                        ),
                        None
                    )

                    jibar_col = None

                    for c in df.columns:

                        name = c.lower()

                        if "jibar" not in name:
                            continue

                        if "discount" in name:
                            continue

                        if (
                            "3 month" in name
                            or re.search(
                                r"\b3m\b",
                                name
                            )
                        ):
                            jibar_col = c
                            break

                    if (
                        date_col is None
                        or
                        jibar_col is None
                    ):
                        continue

                    temp = pd.DataFrame({

                        "Date":
                            pd.to_datetime(
                                df[date_col],
                                errors="coerce",
                                dayfirst=True
                            ),

                        "JIBAR_3M":
                            numeric(
                                df[jibar_col]
                            )
                    })

                    temp = temp.dropna(
                        subset=[
                            "Date",
                            "JIBAR_3M"
                        ]
                    )

                    if not temp.empty:

                        jibar_frames.append(
                            temp
                        )

            except Exception:
                continue

    except Exception as e:

        print(
            f"{year} FAILED: {e}"
        )


if not jibar_frames:

    raise RuntimeError(
        "No JIBAR observations retrieved."
    )


jibar = pd.concat(
    jibar_frames,
    ignore_index=True
)

jibar = (
    jibar
    .drop_duplicates(
        "Date",
        keep="last"
    )
    .sort_values("Date")
    .set_index("Date")
)

jibar = jibar.loc[
    START:END
]


print("\nJIBAR SUCCESS")

print(
    len(jibar),
    "observations"
)

print(
    jibar.index.min().date(),
    "->",
    jibar.index.max().date()
)


# ============================================================
# 2. RBOND HISTORY
# ============================================================

print("\n" + "=" * 65)
print("2. DOWNLOADING RBOND HISTORY")
print("=" * 65)


rbond_url = (
    "https://rbond.co.za/api/v1/history.csv"
    f"?from={START}&to={END}"
)

r = session.get(
    rbond_url,
    timeout=60
)

r.raise_for_status()

bonds = pd.read_csv(
    io.StringIO(r.text)
)


print("\nBond types downloaded:")

print(
    bonds["bond_type"]
    .value_counts()
)


# ============================================================
# 3. CLEAN RBOND DATA
# ============================================================

bonds["date"] = pd.to_datetime(
    bonds["date"],
    errors="coerce"
)

bonds["years_to_maturity"] = pd.to_numeric(
    bonds["years_to_maturity"],
    errors="coerce"
)

bonds["yield_pct"] = pd.to_numeric(
    bonds["yield_pct"],
    errors="coerce"
)


bonds = bonds.dropna(
    subset=[
        "date",
        "years_to_maturity",
        "yield_pct"
    ]
)


# ============================================================
# 4. CRITICAL FILTER:
#    NOMINAL FIXED-RATE BONDS ONLY
# ============================================================

sagb = bonds[
    bonds["bond_type"]
         .astype(str)
         .str.lower()
         .eq("fixed")
].copy()


# Positive remaining maturity only

sagb = sagb[
    sagb["years_to_maturity"] > 0
].copy()


print("\nFIXED SAGBs ONLY")

print(
    "Observations:",
    len(sagb)
)

print(
    "Unique bonds:",
    sagb["bond"].nunique()
)

print(
    "Bonds:"
)

print(
    sorted(
        sagb["bond"]
        .dropna()
        .unique()
    )
)


print(
    "\nYield range:"
)

print(
    sagb["yield_pct"]
    .describe()
)


# ============================================================
# 5. CONSTANT-MATURITY INTERPOLATION
# ============================================================

print("\n" + "=" * 65)
print("3. BUILDING CONSTANT-MATURITY SAGB CURVE")
print("=" * 65)


def interpolate_yield(group, tenor):

    curve = (
        group[
            [
                "years_to_maturity",
                "yield_pct"
            ]
        ]
        .dropna()
        .sort_values(
            "years_to_maturity"
        )
        .drop_duplicates(
            "years_to_maturity"
        )
    )

    if len(curve) < 2:
        return np.nan

    x = (
        curve["years_to_maturity"]
        .to_numpy()
    )

    y = (
        curve["yield_pct"]
        .to_numpy()
    )

    # NO EXTRAPOLATION
    # Must have an actual bond below and above the tenor.

    if (
        tenor < x.min()
        or
        tenor > x.max()
    ):
        return np.nan

    return np.interp(
        tenor,
        x,
        y
    )


curve_rows = []


for date, group in sagb.groupby("date"):

    curve_rows.append({

        "Date": date,

        "SAGB_1Y":
            interpolate_yield(
                group,
                1.0
            ),

        "SAGB_5Y":
            interpolate_yield(
                group,
                5.0
            ),

        "SAGB_10Y":
            interpolate_yield(
                group,
                10.0
            )
    })


curve = (
    pd.DataFrame(
        curve_rows
    )
    .set_index("Date")
    .sort_index()
)


print(
    "\nNon-null daily observations:"
)

print(
    curve.notna().sum()
)


print(
    "\nLatest curve observations:"
)

display(
    curve.tail(10)
         .round(4)
)


# ============================================================
# 6. SANITY CHECK
# ============================================================

print("\n" + "=" * 65)
print("4. CURVE SANITY CHECK")
print("=" * 65)


for col in [
    "SAGB_1Y",
    "SAGB_5Y",
    "SAGB_10Y"
]:

    x = curve[col].dropna()

    print(
        f"\n{col}:"
    )

    print(
        "min =",
        round(x.min(), 3),
        "| median =",
        round(x.median(), 3),
        "| max =",
        round(x.max(), 3)
    )


# ============================================================
# 7. WEEKLY ALIGNMENT
# ============================================================

print("\n" + "=" * 65)
print("5. WEEKLY ALIGNMENT")
print("=" * 65)


jibar_w = (
    jibar[
        ["JIBAR_3M"]
    ]
    .resample(
        "W-FRI"
    )
    .last()
)


curve_w = (
    curve[
        [
            "SAGB_1Y",
            "SAGB_5Y",
            "SAGB_10Y"
        ]
    ]
    .resample(
        "W-FRI"
    )
    .last()
)


# ============================================================
# 8. MERGE
# ============================================================

data = pd.concat(
    [
        jibar_w,
        curve_w
    ],
    axis=1
)


# Common sample

data = data.dropna(
    subset=[
        "JIBAR_3M",
        "SAGB_1Y",
        "SAGB_5Y",
        "SAGB_10Y"
    ]
).copy()


# ============================================================
# 9. CALCULATE SPREADS
# ============================================================

data["SPREAD_1Y_BP"] = (
    data["SAGB_1Y"]
    -
    data["JIBAR_3M"]
) * 100


data["SPREAD_5Y_BP"] = (
    data["SAGB_5Y"]
    -
    data["JIBAR_3M"]
) * 100


data["SPREAD_10Y_BP"] = (
    data["SAGB_10Y"]
    -
    data["JIBAR_3M"]
) * 100


# Final safety clean

data = data.dropna().copy()


# ============================================================
# 10. FINAL VALIDATION
# ============================================================

print("\n" + "=" * 65)
print("6. FINAL DATASET")
print("=" * 65)


print(
    "\nObservations:",
    len(data)
)

print(
    "Date range:",
    data.index.min().date(),
    "->",
    data.index.max().date()
)


print(
    "\nMissing values:"
)

print(
    data.isna().sum()
)


print(
    "\nLatest 10 observations:"
)

display(
    data.tail(10)
        .round(3)
)


# ============================================================
# 11. SPREAD STATISTICS
# ============================================================

spread_cols = [
    "SPREAD_1Y_BP",
    "SPREAD_5Y_BP",
    "SPREAD_10Y_BP"
]


print(
    "\nSpread statistics (bp):"
)

display(
    data[
        spread_cols
    ]
    .describe()
    .T
    .round(2)
)


print(
    "\nSpread correlations:"
)

display(
    data[
        spread_cols
    ]
    .corr()
    .round(3)
)


# ============================================================
# 12. SAVE
# ============================================================

OUTPUT = (
    "SA_SAGB_fixed_1Y_5Y_10Y_vs_JIBAR3M.csv"
)

data.to_csv(
    OUTPUT
)

print(
    "\nSaved:",
    OUTPUT
)
```

```python
# ============================================================
# SPREAD DYNAMICS:
# ADF + KPSS + ACF + OU/AR(1) + VARIANCE RATIO + HURST
# + 52-WEEK Z-SCORES
#
# Uses existing:
#   jibar_w
#   curve_w
# ============================================================

!pip -q install statsmodels

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller, kpss, acf
import statsmodels.api as sm


# ============================================================
# 1. BUILD PAIRWISE SPREAD SERIES
# ============================================================

spread_map = {
    "1Y - 3M JIBAR": "SAGB_1Y",
    "5Y - 3M JIBAR": "SAGB_5Y",
    "10Y - 3M JIBAR": "SAGB_10Y"
}

spreads = {}

for name, bond_col in spread_map.items():

    temp = pd.concat(
        [
            jibar_w["JIBAR_3M"],
            curve_w[bond_col]
        ],
        axis=1
    ).dropna()

    # Spread in basis points
    s = (
        temp[bond_col]
        - temp["JIBAR_3M"]
    ) * 100

    spreads[name] = s


print("PAIRWISE WEEKLY SAMPLES\n")

for name, s in spreads.items():
    print(
        f"{name:<16} "
        f"N={len(s):>4} | "
        f"{s.index.min().date()} -> {s.index.max().date()}"
    )


# ============================================================
# 2. VARIANCE RATIO
#
# VR < 1  = anti-persistence / mean-reverting tendency
# VR ~ 1  = random-walk-like
# VR > 1  = persistence / trending tendency
# ============================================================

def variance_ratio(series, q):

    x = series.dropna().values

    if len(x) <= q + 1:
        return np.nan

    one_period = np.diff(x)

    q_period = x[q:] - x[:-q]

    var_1 = np.var(
        one_period,
        ddof=1
    )

    var_q = np.var(
        q_period,
        ddof=1
    )

    if var_1 == 0:
        return np.nan

    return var_q / (q * var_1)


# ============================================================
# 3. HURST EXPONENT
#
# Based on scaling of lagged differences.
#
# H < 0.5 = anti-persistent
# H ~ 0.5 = random-walk-like
# H > 0.5 = persistent
# ============================================================

def hurst_exponent(series):

    x = series.dropna().values

    # Multiple horizons, constrained by sample size
    max_lag = min(
        20,
        len(x) // 4
    )

    lags = np.arange(
        2,
        max_lag + 1
    )

    tau = []

    valid_lags = []

    for lag in lags:

        diff = (
            x[lag:]
            - x[:-lag]
        )

        sd = np.std(
            diff,
            ddof=1
        )

        if (
            np.isfinite(sd)
            and sd > 0
        ):
            tau.append(sd)
            valid_lags.append(lag)

    if len(tau) < 3:
        return np.nan

    slope = np.polyfit(
        np.log(valid_lags),
        np.log(tau),
        1
    )[0]

    return slope


# ============================================================
# 4. DIAGNOSTIC FUNCTION
# ============================================================

def analyse_spread(name, series):

    s = series.dropna().astype(float)

    # --------------------------------------------------------
    # ADF
    # H0 = unit root
    # --------------------------------------------------------

    adf_result = adfuller(
        s,
        regression="c",
        autolag="AIC"
    )

    adf_stat = adf_result[0]
    adf_p    = adf_result[1]


    # --------------------------------------------------------
    # KPSS
    # H0 = level stationary
    # --------------------------------------------------------

    try:

        kpss_result = kpss(
            s,
            regression="c",
            nlags="auto"
        )

        kpss_stat = kpss_result[0]
        kpss_p    = kpss_result[1]

    except Exception:

        kpss_stat = np.nan
        kpss_p    = np.nan


    # --------------------------------------------------------
    # ACF 1-12
    # --------------------------------------------------------

    acf_values = acf(
        s,
        nlags=12,
        fft=False
    )


    # --------------------------------------------------------
    # OU / AR(1)
    #
    # S_t = c + phi*S_(t-1) + error
    # --------------------------------------------------------

    y = s.iloc[1:].values

    lag = s.iloc[:-1].values

    X = sm.add_constant(lag)

    model = sm.OLS(
        y,
        X
    ).fit()

    c = model.params[0]
    phi = model.params[1]

    phi_p = model.pvalues[1]

    # Long-run equilibrium mean
    if abs(1 - phi) > 1e-10:

        theta = (
            c /
            (1 - phi)
        )

    else:
        theta = np.nan


    # Half-life in weeks
    if 0 < phi < 1:

        half_life = (
            -np.log(2)
            /
            np.log(phi)
        )

    else:
        half_life = np.nan


    # Continuous-time mean-reversion speed
    # Weekly observations => kappa per week

    if phi > 0:

        kappa = -np.log(phi)

    else:
        kappa = np.nan


    # --------------------------------------------------------
    # Variance ratios
    # --------------------------------------------------------

    vr2  = variance_ratio(s, 2)
    vr4  = variance_ratio(s, 4)
    vr8  = variance_ratio(s, 8)
    vr12 = variance_ratio(s, 12)


    # --------------------------------------------------------
    # Hurst
    # --------------------------------------------------------

    hurst = hurst_exponent(s)


    # --------------------------------------------------------
    # Return everything
    # --------------------------------------------------------

    result = {

        "Spread": name,

        "N": len(s),

        "Mean_BP": s.mean(),

        "Std_BP": s.std(),

        "ADF_Stat": adf_stat,

        "ADF_p": adf_p,

        "KPSS_Stat": kpss_stat,

        "KPSS_p": kpss_p,

        "AR1_phi": phi,

        "AR1_phi_p": phi_p,

        "OU_theta_BP": theta,

        "OU_kappa_weekly": kappa,

        "HalfLife_weeks": half_life,

        "VR_2": vr2,

        "VR_4": vr4,

        "VR_8": vr8,

        "VR_12": vr12,

        "Hurst": hurst
    }

    return result, acf_values


# ============================================================
# 5. RUN ALL THREE
# ============================================================

results = []

acf_results = {}

for name, s in spreads.items():

    result, acf_values = analyse_spread(
        name,
        s
    )

    results.append(result)

    acf_results[name] = acf_values


results_df = (
    pd.DataFrame(results)
    .set_index("Spread")
)


# ============================================================
# 6. MAIN RESULTS TABLE
# ============================================================

print("\n" + "=" * 80)
print("STATISTICAL DIAGNOSTICS")
print("=" * 80)

display(
    results_df.round(4)
)


# ============================================================
# 7. ACF TABLE — LAGS 1 TO 12
# ============================================================

acf_table = pd.DataFrame({

    name:
        values[1:13]

    for name, values
    in acf_results.items()

}, index=range(1, 13))

acf_table.index.name = "Lag_weeks"


print("\n" + "=" * 80)
print("AUTOCORRELATION — WEEKLY LAGS 1 TO 12")
print("=" * 80)

display(
    acf_table.round(4)
)


# ============================================================
# 8. 52-WEEK ROLLING Z-SCORES
# ============================================================

zscore_df = pd.DataFrame()

for name, s in spreads.items():

    rolling_mean = (
        s.rolling(
            52,
            min_periods=26
        )
        .mean()
    )

    rolling_std = (
        s.rolling(
            52,
            min_periods=26
        )
        .std()
    )

    zscore_df[name] = (
        (s - rolling_mean)
        /
        rolling_std
    )


print("\n" + "=" * 80)
print("LATEST 52-WEEK Z-SCORES")
print("=" * 80)

for name in zscore_df.columns:

    x = zscore_df[name].dropna()

    if len(x):

        print(
            f"{name:<16} "
            f"Z = {x.iloc[-1]:>7.3f} | "
            f"Date = {x.index[-1].date()}"
        )


# ============================================================
# 9. ACF PLOTS
# ============================================================

for name in spreads:

    vals = acf_results[name][1:13]

    fig, ax = plt.subplots(
        figsize=(9, 4)
    )

    ax.bar(
        range(1, 13),
        vals
    )

    # Approximate 95% white-noise bounds
    n = len(spreads[name])

    bound = 1.96 / np.sqrt(n)

    ax.axhline(
        bound,
        linestyle="--"
    )

    ax.axhline(
        -bound,
        linestyle="--"
    )

    ax.axhline(
        0,
        linewidth=0.8
    )

    ax.set_title(
        f"{name}: ACF"
    )

    ax.set_xlabel(
        "Lag (weeks)"
    )

    ax.set_ylabel(
        "Autocorrelation"
    )

    ax.set_xticks(
        range(1, 13)
    )

    plt.show()


# ============================================================
# 10. SPREAD + 52-WEEK MEAN
# ============================================================

for name, s in spreads.items():

    rolling_mean = (
        s.rolling(
            52,
            min_periods=26
        )
        .mean()
    )

    fig, ax = plt.subplots(
        figsize=(11, 4)
    )

    ax.plot(
        s.index,
        s,
        label="Spread"
    )

    ax.plot(
        rolling_mean.index,
        rolling_mean,
        label="52-week mean"
    )

    ax.set_title(
        name
    )

    ax.set_ylabel(
        "Spread (bp)"
    )

    ax.legend()

    plt.show()


# ============================================================
# 11. Z-SCORE PLOTS
# ============================================================

for name in zscore_df.columns:

    z = zscore_df[name].dropna()

    fig, ax = plt.subplots(
        figsize=(11, 4)
    )

    ax.plot(
        z.index,
        z
    )

    ax.axhline(0)

    ax.axhline(
        2,
        linestyle="--"
    )

    ax.axhline(
        -2,
        linestyle="--"
    )

    ax.set_title(
        f"{name}: 52-Week Z-Score"
    )

    ax.set_ylabel(
        "Z-score"
    )

    plt.show()
```

```python
# ============================================================
# ROLLING REGIME ANALYSIS — 52-WEEK WINDOWS
#
# For each spread:
#   Rolling mean
#   Rolling volatility
#   Rolling AR(1) phi
#   Rolling OU theta
#   Rolling OU half-life
#
# Uses existing "spreads" object.
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"

WINDOW = 52


# ============================================================
# 1. ROLLING OU / AR(1)
# ============================================================

def rolling_ou(series, window=52):

    s = series.dropna().astype(float)

    rows = []

    for end in range(window, len(s) + 1):

        sample = s.iloc[end-window:end]

        y = sample.iloc[1:].values
        xlag = sample.iloc[:-1].values

        X = sm.add_constant(xlag)

        model = sm.OLS(
            y,
            X
        ).fit()

        c = model.params[0]
        phi = model.params[1]

        # OU equilibrium
        if abs(1 - phi) > 1e-8:
            theta = c / (1 - phi)
        else:
            theta = np.nan

        # Half-life only when OU-style mean reversion exists
        if 0 < phi < 1:

            half_life = (
                -np.log(2)
                / np.log(phi)
            )

            kappa = -np.log(phi)

        else:

            half_life = np.nan
            kappa = np.nan

        rows.append({
            "Date": sample.index[-1],
            "Mean": sample.mean(),
            "Volatility": sample.std(),
            "Phi": phi,
            "Theta": theta,
            "Kappa": kappa,
            "HalfLife": half_life
        })

    return (
        pd.DataFrame(rows)
        .set_index("Date")
    )


rolling_results = {}

for name, s in spreads.items():

    rolling_results[name] = rolling_ou(
        s,
        WINDOW
    )


# ============================================================
# 2. LATEST REGIME PARAMETERS
# ============================================================

print("=" * 90)
print("LATEST 52-WEEK REGIME PARAMETERS")
print("=" * 90)

latest = []

for name, df in rolling_results.items():

    row = df.iloc[-1]

    latest.append({
        "Spread": name,
        "Rolling_Mean_BP": row["Mean"],
        "Rolling_Vol_BP": row["Volatility"],
        "AR1_Phi": row["Phi"],
        "OU_Theta_BP": row["Theta"],
        "OU_Kappa": row["Kappa"],
        "HalfLife_Weeks": row["HalfLife"]
    })


latest_df = (
    pd.DataFrame(latest)
    .set_index("Spread")
)

display(
    latest_df.round(3)
)


# ============================================================
# 3. ROLLING PHI
#
# Phi near 0   -> little persistence
# 0 < phi < 1 -> mean-reverting AR(1)
# Phi near 1   -> highly persistent
# Phi > 1      -> locally explosive
# ============================================================

for name, df in rolling_results.items():

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=df.index,
            y=df["Phi"],
            mode="lines",
            name="Rolling φ"
        )
    )

    fig.add_hline(
        y=1,
        line_dash="dash",
        annotation_text="φ = 1"
    )

    fig.add_hline(
        y=0,
        line_dash="dot"
    )

    fig.update_layout(
        title=f"{name}: Rolling 52-Week AR(1) Persistence",
        xaxis_title="Date",
        yaxis_title="AR(1) φ",
        template="plotly_white",
        hovermode="x unified"
    )

    fig.update_xaxes(
        rangeslider_visible=True
    )

    fig.show()


# ============================================================
# 4. SPREAD VS ROLLING OU EQUILIBRIUM
# ============================================================

for name, df in rolling_results.items():

    s = spreads[name].dropna()

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=s.index,
            y=s.values,
            mode="lines",
            name="Actual spread"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=df.index,
            y=df["Theta"],
            mode="lines",
            name="Rolling OU θ"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=df.index,
            y=df["Mean"],
            mode="lines",
            name="52-week mean"
        )
    )

    fig.update_layout(
        title=f"{name}: Spread and Estimated Equilibrium",
        xaxis_title="Date",
        yaxis_title="Basis points",
        template="plotly_white",
        hovermode="x unified"
    )

    fig.update_xaxes(
        rangeslider_visible=True
    )

    fig.show()


# ============================================================
# 5. ROLLING HALF-LIFE
# ============================================================

for name, df in rolling_results.items():

    plot_df = df[
        np.isfinite(df["HalfLife"])
    ].copy()

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=plot_df.index,
            y=plot_df["HalfLife"],
            mode="lines",
            name="Half-life"
        )
    )

    fig.update_layout(
        title=f"{name}: Rolling OU Half-Life",
        xaxis_title="Date",
        yaxis_title="Weeks",
        template="plotly_white",
        hovermode="x unified"
    )

    fig.update_xaxes(
        rangeslider_visible=True
    )

    fig.show()


# ============================================================
# 6. ROLLING VOLATILITY
# ============================================================

for name, df in rolling_results.items():

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=df.index,
            y=df["Volatility"],
            mode="lines",
            name="52-week volatility"
        )
    )

    fig.update_layout(
        title=f"{name}: Rolling 52-Week Spread Volatility",
        xaxis_title="Date",
        yaxis_title="Basis points",
        template="plotly_white",
        hovermode="x unified"
    )

    fig.update_xaxes(
        rangeslider_visible=True
    )

    fig.show()


# ============================================================
# 7. REGIME SUMMARY
# ============================================================

print("\n" + "=" * 90)
print("ROLLING PHI REGIME SUMMARY")
print("=" * 90)

regime_summary = []

for name, df in rolling_results.items():

    phi = df["Phi"].dropna()

    regime_summary.append({

        "Spread": name,

        "Median_Phi":
            phi.median(),

        "% Phi < 0.5":
            (phi < 0.5).mean() * 100,

        "% Phi 0.5-0.9":
            (
                (phi >= 0.5)
                & (phi < 0.9)
            ).mean() * 100,

        "% Phi >= 0.9":
            (phi >= 0.9).mean() * 100,

        "% Phi >= 1":
            (phi >= 1).mean() * 100
    })


regime_summary_df = (
    pd.DataFrame(regime_summary)
    .set_index("Spread")
)

display(
    regime_summary_df.round(2)
)
```

```python
# ============================================================
# CLEAN SPREAD DIAGNOSTICS
# Uses corrected `spreads` from JIBAR cleaning cell
# ============================================================

import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller, kpss, acf
import statsmodels.api as sm

# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------

def variance_ratio(x, q):
    """
    Variance ratio:
    variance of q-period changes /
    (q * variance of 1-period changes)
    """
    x = np.asarray(x, dtype=float)

    if len(x) <= q + 2:
        return np.nan

    d1 = np.diff(x)
    dq = x[q:] - x[:-q]

    v1 = np.var(d1, ddof=1)
    vq = np.var(dq, ddof=1)

    if v1 == 0:
        return np.nan

    return vq / (q * v1)


def hurst_exponent(x):
    """
    Simple Hurst estimate from scaling of lagged differences.
    Treat as supplementary only.
    """
    x = np.asarray(x, dtype=float)

    lags = np.arange(2, min(20, len(x) // 4))

    if len(lags) < 3:
        return np.nan

    tau = []

    for lag in lags:
        diff = x[lag:] - x[:-lag]
        tau.append(np.std(diff))

    tau = np.asarray(tau)

    valid = (
        np.isfinite(tau)
        & (tau > 0)
    )

    if valid.sum() < 3:
        return np.nan

    slope = np.polyfit(
        np.log(lags[valid]),
        np.log(tau[valid]),
        1
    )[0]

    return slope


# ------------------------------------------------------------
# Run diagnostics
# ------------------------------------------------------------

summary_rows = []
acf_rows = []

for name, series in spreads.items():

    s = (
        pd.to_numeric(series, errors="coerce")
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .sort_index()
    )

    # --------------------------------------------------------
    # ADF
    # H0 = unit root
    # --------------------------------------------------------

    adf_result = adfuller(
        s.values,
        autolag="AIC"
    )

    adf_stat = adf_result[0]
    adf_p = adf_result[1]


    # --------------------------------------------------------
    # KPSS
    # H0 = level stationary
    # --------------------------------------------------------

    try:
        kpss_result = kpss(
            s.values,
            regression="c",
            nlags="auto"
        )

        kpss_stat = kpss_result[0]
        kpss_p = kpss_result[1]

    except Exception:
        kpss_stat = np.nan
        kpss_p = np.nan


    # --------------------------------------------------------
    # AR(1) / discrete OU
    #
    # S_t = c + phi*S_(t-1) + error
    # --------------------------------------------------------

    y = s.iloc[1:].values
    x = s.iloc[:-1].values

    X = sm.add_constant(x)

    model = sm.OLS(y, X).fit()

    c = model.params[0]
    phi = model.params[1]
    phi_p = model.pvalues[1]


    # --------------------------------------------------------
    # OU quantities
    # --------------------------------------------------------

    if phi != 1:
        theta = c / (1 - phi)
    else:
        theta = np.nan

    if 0 < phi < 1:

        kappa = -np.log(phi)

        half_life = (
            np.log(2) / kappa
        )

    else:

        kappa = np.nan
        half_life = np.nan


    # --------------------------------------------------------
    # Variance ratios
    # --------------------------------------------------------

    vr2 = variance_ratio(s.values, 2)
    vr4 = variance_ratio(s.values, 4)
    vr8 = variance_ratio(s.values, 8)
    vr12 = variance_ratio(s.values, 12)


    # --------------------------------------------------------
    # Hurst
    # --------------------------------------------------------

    hurst = hurst_exponent(s.values)


    # --------------------------------------------------------
    # ACF 1–12
    # --------------------------------------------------------

    acf_values = acf(
        s.values,
        nlags=12,
        fft=False
    )

    for lag in range(1, 13):

        acf_rows.append({
            "Spread": name,
            "Lag": lag,
            "ACF": acf_values[lag]
        })


    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    summary_rows.append({

        "Spread": name,

        "N": len(s),

        "Mean_BP": s.mean(),

        "Std_BP": s.std(),

        "Min_BP": s.min(),

        "Max_BP": s.max(),

        "ADF_Stat": adf_stat,

        "ADF_p": adf_p,

        "KPSS_Stat": kpss_stat,

        "KPSS_p": kpss_p,

        "AR1_phi": phi,

        "AR1_phi_p": phi_p,

        "OU_theta_BP": theta,

        "OU_kappa_weekly": kappa,

        "HalfLife_weeks": half_life,

        "VR_2": vr2,

        "VR_4": vr4,

        "VR_8": vr8,

        "VR_12": vr12,

        "Hurst": hurst
    })


# ------------------------------------------------------------
# Tables
# ------------------------------------------------------------

diagnostics = pd.DataFrame(
    summary_rows
).set_index("Spread")

acf_table = (
    pd.DataFrame(acf_rows)
    .pivot(
        index="Lag",
        columns="Spread",
        values="ACF"
    )
)


# ------------------------------------------------------------
# Latest 52-week Z-scores
# ------------------------------------------------------------

z_rows = []

for name, series in spreads.items():

    s = series.dropna().sort_index()

    rolling_mean = s.rolling(52).mean()
    rolling_std = s.rolling(52).std()

    z = (
        (s - rolling_mean)
        / rolling_std
    )

    valid = z.dropna()

    if len(valid) > 0:

        z_rows.append({
            "Spread": name,
            "Date": valid.index[-1],
            "Latest_Spread_BP": s.loc[valid.index[-1]],
            "Z_52W": valid.iloc[-1]
        })


z_table = pd.DataFrame(
    z_rows
).set_index("Spread")


# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("=" * 100)
print("CLEAN DIAGNOSTIC SUMMARY")
print("=" * 100)

display(
    diagnostics.round(4)
)

print("\n" + "=" * 100)
print("AUTOCORRELATION — LAGS 1 TO 12")
print("=" * 100)

display(
    acf_table.round(4)
)

print("\n" + "=" * 100)
print("LATEST 52-WEEK Z-SCORES")
print("=" * 100)

display(
    z_table.round(4)
)
```

```python
# ============================================================
# ROLLING 52-WEEK REGIME ANALYSIS — CLEAN DATA
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"

WINDOW = 52

rolling_results = {}
regime_rows = []


for name, series in spreads.items():

    s = (
        pd.to_numeric(series, errors="coerce")
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .sort_index()
    )

    rows = []

    for i in range(WINDOW, len(s) + 1):

        w = s.iloc[i-WINDOW:i]

        y = w.iloc[1:].values
        x = w.iloc[:-1].values

        X = sm.add_constant(x)
        model = sm.OLS(y, X).fit()

        c = model.params[0]
        phi = model.params[1]

        # OU equilibrium
        if abs(1 - phi) > 1e-6:
            theta = c / (1 - phi)
        else:
            theta = np.nan

        # Half-life only meaningful for 0 < phi < 1
        if 0 < phi < 1:
            kappa = -np.log(phi)
            half_life = np.log(2) / kappa
        else:
            kappa = np.nan
            half_life = np.nan

        rows.append({
            "Date": w.index[-1],
            "Spread": w.iloc[-1],
            "Rolling_Mean": w.mean(),
            "Rolling_Vol": w.std(),
            "Phi": phi,
            "Theta": theta,
            "HalfLife": half_life
        })

    r = pd.DataFrame(rows).set_index("Date")

    rolling_results[name] = r

    # --------------------------------------------------------
    # REGIME SUMMARY
    # --------------------------------------------------------

    regime_rows.append({
        "Spread": name,
        "Median_Phi": r["Phi"].median(),
        "% Phi < 0.5": (r["Phi"] < 0.5).mean() * 100,
        "% Phi 0.5-0.9": (
            (r["Phi"] >= 0.5) &
            (r["Phi"] < 0.9)
        ).mean() * 100,
        "% Phi 0.9-1.0": (
            (r["Phi"] >= 0.9) &
            (r["Phi"] < 1.0)
        ).mean() * 100,
        "% Phi >= 1": (
            r["Phi"] >= 1.0
        ).mean() * 100,
        "Median_HalfLife_Weeks":
            r["HalfLife"].median()
    })


    # ========================================================
    # CHART 1 — SPREAD + 52W ROLLING MEAN
    # ========================================================

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=r.index,
            y=r["Spread"],
            mode="lines",
            name="Spread"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=r.index,
            y=r["Rolling_Mean"],
            mode="lines",
            name="52W Mean"
        )
    )

    fig.update_layout(
        title=f"{name}: Spread vs 52-Week Rolling Mean",
        xaxis_title="Date",
        yaxis_title="Basis Points",
        hovermode="x unified",
        template="plotly_white"
    )

    fig.show()


    # ========================================================
    # CHART 2 — ROLLING PHI
    # ========================================================

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=r.index,
            y=r["Phi"],
            mode="lines",
            name="Rolling Phi"
        )
    )

    fig.add_hline(
        y=1.0,
        line_dash="dash",
        annotation_text="Unit-root boundary"
    )

    fig.add_hline(
        y=0.9,
        line_dash="dot",
        annotation_text="High persistence"
    )

    fig.update_layout(
        title=f"{name}: 52-Week Rolling AR(1) Persistence",
        xaxis_title="Date",
        yaxis_title="Phi",
        hovermode="x unified",
        template="plotly_white"
    )

    fig.show()


    # ========================================================
    # CHART 3 — ROLLING HALF-LIFE
    # ========================================================

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=r.index,
            y=r["HalfLife"],
            mode="lines",
            name="Half-Life"
        )
    )

    fig.update_layout(
        title=f"{name}: 52-Week Rolling Half-Life",
        xaxis_title="Date",
        yaxis_title="Weeks",
        hovermode="x unified",
        template="plotly_white"
    )

    fig.show()


    # ========================================================
    # CHART 4 — ROLLING VOLATILITY
    # ========================================================

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=r.index,
            y=r["Rolling_Vol"],
            mode="lines",
            name="52W Volatility"
        )
    )

    fig.update_layout(
        title=f"{name}: 52-Week Rolling Spread Volatility",
        xaxis_title="Date",
        yaxis_title="Basis Points",
        hovermode="x unified",
        template="plotly_white"
    )

    fig.show()


# ============================================================
# REGIME SUMMARY
# ============================================================

regime_summary = (
    pd.DataFrame(regime_rows)
    .set_index("Spread")
)

print("=" * 100)
print("52-WEEK ROLLING REGIME SUMMARY")
print("=" * 100)

display(regime_summary.round(2))
```

```python
# ============================================================
# CARRY + RELATIVE-VALUE SIGNAL
# 52-week local mean / volatility
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"

WINDOW = 52
rv_signals = {}

for name, series in spreads.items():

    s = (
        pd.to_numeric(series, errors="coerce")
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .sort_index()
    )

    df = pd.DataFrame({"Spread": s})

    # Local equilibrium and volatility
    df["Mean_52W"] = df["Spread"].rolling(WINDOW).mean()
    df["Std_52W"] = df["Spread"].rolling(WINDOW).std()

    # Bands
    df["+1σ"] = df["Mean_52W"] + df["Std_52W"]
    df["-1σ"] = df["Mean_52W"] - df["Std_52W"]

    df["+2σ"] = df["Mean_52W"] + 2 * df["Std_52W"]
    df["-2σ"] = df["Mean_52W"] - 2 * df["Std_52W"]

    # Relative-value Z-score
    df["Z_52W"] = (
        (df["Spread"] - df["Mean_52W"])
        / df["Std_52W"]
    )

    rv_signals[name] = df


    # ========================================================
    # CHART 1 — SPREAD + LOCAL EQUILIBRIUM BANDS
    # ========================================================

    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["Spread"],
        mode="lines",
        name="Spread"
    ))

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["Mean_52W"],
        mode="lines",
        name="52W Mean"
    ))

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["+1σ"],
        mode="lines",
        name="+1σ",
        line=dict(dash="dot")
    ))

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["-1σ"],
        mode="lines",
        name="-1σ",
        line=dict(dash="dot")
    ))

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["+2σ"],
        mode="lines",
        name="+2σ",
        line=dict(dash="dash")
    ))

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["-2σ"],
        mode="lines",
        name="-2σ",
        line=dict(dash="dash")
    ))

    fig.update_layout(
        title=f"{name}: Carry / RV Spread",
        xaxis_title="Date",
        yaxis_title="Spread (bp)",
        hovermode="x unified",
        template="plotly_white"
    )

    fig.show()


    # ========================================================
    # CHART 2 — RELATIVE-VALUE Z-SCORE
    # ========================================================

    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=df.index,
        y=df["Z_52W"],
        mode="lines",
        name="52W Z-score"
    ))

    for level, dash in [
        (2, "dash"),
        (1, "dot"),
        (0, "solid"),
        (-1, "dot"),
        (-2, "dash")
    ]:
        fig.add_hline(
            y=level,
            line_dash=dash
        )

    fig.update_layout(
        title=f"{name}: 52-Week Relative-Value Z-Score",
        xaxis_title="Date",
        yaxis_title="Z-score",
        hovermode="x unified",
        template="plotly_white"
    )

    fig.show()


# ============================================================
# CURRENT SIGNAL TABLE
# ============================================================

current_rows = []

for name, df in rv_signals.items():

    latest = df.dropna().iloc[-1]

    current_rows.append({
        "Spread": name,
        "Spread_BP": latest["Spread"],
        "52W_Mean_BP": latest["Mean_52W"],
        "52W_Std_BP": latest["Std_52W"],
        "Z_52W": latest["Z_52W"]
    })

current_signals = (
    pd.DataFrame(current_rows)
    .set_index("Spread")
)

print("=" * 80)
print("CURRENT CARRY / RELATIVE-VALUE SIGNALS")
print("=" * 80)

display(current_signals.round(2))
```
