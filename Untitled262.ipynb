```python
# ============================================================
# WEEKLY RETURN OU / MVOU ANALYSIS
#
# Assets:
#   BRK.B
#   IAK
#   SPY
#
# DATA:
#   Investing.com weekly CSV exports
#
# IMPORTANT:
#   - Uses CHANGE IN PRICE (returns), NOT price levels
#   - No forward filling
#   - Individual returns calculated first
#   - Then inner-joined
#   - Then explicit dropna()
#   - Rolling window = 26 WEEKS
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd

from scipy import stats
from scipy.linalg import logm

from sklearn.linear_model import LinearRegression

from statsmodels.tsa.stattools import (
    adfuller,
    kpss,
    acf
)

import plotly.graph_objects as go
import plotly.express as px


# ============================================================
# 2. SETTINGS
# ============================================================

ROLLING_WINDOW = 26

MIN_ROLLING_OBS = 26

TRADING_WEEKS_PER_YEAR = 52

ASSET_NAMES = [
    "BRKB",
    "IAK",
    "SPY"
]


# ============================================================
# 3. FILE NAMES
#
# The Berkshire file was supplied without ".csv".
# The loader below automatically adds ".csv" if necessary.
# ============================================================

FILES = {

    "BRKB":
        "Berkshire Hathaway B Stock Price History (1)",

    "IAK":
        "IAK ETF Stock Price History (1).csv",

    "SPY":
        "SPY ET Z.csv"

}


# ============================================================
# 4. CHECK FILES
# ============================================================

def resolve_filename(filename):

    # Exact filename exists
    if os.path.exists(filename):
        return filename

    # Try adding .csv
    if not filename.lower().endswith(".csv"):

        candidate = filename + ".csv"

        if os.path.exists(candidate):
            return candidate

    # Try case-insensitive directory search
    target = filename.lower()

    if not target.endswith(".csv"):
        target += ".csv"

    for f in os.listdir("."):

        if f.lower() == target:
            return f

    raise FileNotFoundError(
        f"\nCould not find:\n{filename}\n\n"
        f"Files currently visible in Colab:\n"
        f"{os.listdir('.')}"
    )


# ============================================================
# 5. ROBUST INVESTING.COM CSV PARSER
# ============================================================

def load_investing_csv(path):

    path = resolve_filename(path)

    print(f"\nLoading: {path}")

    # --------------------------------------------------------
    # Read CSV
    # --------------------------------------------------------

    df = pd.read_csv(path)

    # Clean column names
    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
    )

    print("Columns:", list(df.columns))

    # --------------------------------------------------------
    # FIND DATE COLUMN
    # --------------------------------------------------------

    date_col = None

    # First look for explicit Date-type names
    date_name_candidates = [

        c for c in df.columns

        if "date" in c.lower()

    ]

    for c in date_name_candidates:

        parsed = pd.to_datetime(
            df[c],
            errors="coerce"
        )

        if parsed.notna().mean() >= 0.80:

            date_col = c
            break

    # If not found, test every column
    if date_col is None:

        for c in df.columns:

            parsed = pd.to_datetime(
                df[c],
                errors="coerce"
            )

            if parsed.notna().mean() >= 0.80:

                date_col = c
                break

    if date_col is None:

        raise ValueError(
            f"Could not identify date column in {path}"
        )

    # --------------------------------------------------------
    # FIND PRICE COLUMN
    # --------------------------------------------------------

    price_col = None

    preferred_names = [

        "Price",
        "Close",
        "Adj Close",
        "Adjusted Close",
        "Last"

    ]

    # Exact preferred column
    for candidate in preferred_names:

        for c in df.columns:

            if c.strip().lower() == candidate.lower():

                price_col = c
                break

        if price_col is not None:
            break

    # Fallback: numeric-looking columns
    if price_col is None:

        for c in df.columns:

            if c == date_col:
                continue

            numeric_test = (

                df[c]
                .astype(str)
                .str.replace(",", "", regex=False)
                .str.replace(" ", "", regex=False)
                .str.replace("%", "", regex=False)

            )

            numeric_test = pd.to_numeric(
                numeric_test,
                errors="coerce"
            )

            if numeric_test.notna().mean() >= 0.80:

                price_col = c
                break

    if price_col is None:

        raise ValueError(
            f"Could not identify price column in {path}"
        )

    print("Date column :", date_col)
    print("Price column:", price_col)

    # --------------------------------------------------------
    # CREATE CLEAN DATAFRAME
    # --------------------------------------------------------

    dates = pd.to_datetime(
        df[date_col],
        errors="coerce"
    )

    prices = (

        df[price_col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace(" ", "", regex=False)

    )

    prices = pd.to_numeric(
        prices,
        errors="coerce"
    )

    clean = pd.DataFrame({

        "Date": dates,
        "Price": prices

    })

    # --------------------------------------------------------
    # CLEAN
    # --------------------------------------------------------

    clean = clean.dropna(
        subset=["Date", "Price"]
    )

    # Prices must be positive
    clean = clean[
        clean["Price"] > 0
    ]

    # Remove duplicate dates
    clean = clean.drop_duplicates(
        subset="Date",
        keep="last"
    )

    # Sort
    clean = clean.sort_values(
        "Date"
    )

    # Index
    clean = clean.set_index(
        "Date"
    )

    # --------------------------------------------------------
    # PRICE CHANGE / RETURN
    #
    # This is the variable used for all OU work.
    # --------------------------------------------------------

    clean["Return"] = (
        clean["Price"].pct_change()
    )

    # Remove infinities
    clean["Return"] = clean[
        "Return"
    ].replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Explicitly drop invalid returns
    clean = clean.dropna(
        subset=["Return"]
    )

    # --------------------------------------------------------
    # FINAL SORT
    # --------------------------------------------------------

    clean = clean.sort_index()

    return clean


# ============================================================
# 6. LOAD ALL THREE ASSETS
# ============================================================

assets = {}

for ticker, filename in FILES.items():

    assets[ticker] = load_investing_csv(
        filename
    )

    print(
        f"{ticker}: "
        f"{assets[ticker].index.min().date()} "
        f"to "
        f"{assets[ticker].index.max().date()} "
        f"| "
        f"{len(assets[ticker])} observations"
    )


# ============================================================
# 7. ALIGN WEEKLY RETURNS
#
# Each asset's return is calculated FIRST.
#
# Then:
#   inner join
#   replace infinities
#   dropna
#
# No forward filling.
# ============================================================

returns = pd.concat(

    [

        assets["BRKB"]["Return"].rename("BRKB"),

        assets["IAK"]["Return"].rename("IAK"),

        assets["SPY"]["Return"].rename("SPY")

    ],

    axis=1,

    join="inner"

)

returns = returns.replace(
    [np.inf, -np.inf],
    np.nan
)

returns = returns.dropna(
    how="any"
)

returns = returns.sort_index()


# ============================================================
# 8. DATA QUALITY REPORT
# ============================================================

print("\n")
print("=" * 70)
print("FINAL ALIGNED WEEKLY RETURN DATASET")
print("=" * 70)

print(
    f"Observations : {len(returns)}"
)

print(
    f"Start        : {returns.index.min().date()}"
)

print(
    f"End          : {returns.index.max().date()}"
)

print(
    f"Missing      : {returns.isna().sum().sum()}"
)

print("\nReturn summary:")
print(
    returns.describe().T
)


# ============================================================
# 9. CUMULATIVE RETURN VISUAL
# ============================================================

cumulative_returns = (
    1 + returns
).cumprod() - 1

fig = go.Figure()

for col in returns.columns:

    fig.add_trace(
        go.Scatter(
            x=cumulative_returns.index,
            y=cumulative_returns[col] * 100,
            mode="lines",
            name=col
        )
    )

fig.update_layout(

    template="plotly_white",

    title="Cumulative Return — Weekly Data",

    xaxis_title="Date",

    yaxis_title="Cumulative Return (%)",

    hovermode="x unified"

)

fig.show()


# ============================================================
# 10. CORRELATION MATRIX
# ============================================================

corr = returns.corr()

fig = px.imshow(

    corr,

    text_auto=".3f",

    aspect="auto"

)

fig.update_layout(

    template="plotly_white",

    title="Weekly Return Correlation Matrix"

)

fig.show()


# ============================================================
# 11. HURST EXPONENT
# ============================================================

def hurst_exponent(x):

    x = np.asarray(x)

    x = x[
        np.isfinite(x)
    ]

    n = len(x)

    # Adaptive maximum lag
    max_lag = min(
        100,
        n // 2
    )

    if max_lag < 6:
        return np.nan

    lags = []
    tau = []

    for lag in range(
        2,
        max_lag
    ):

        differences = (
            x[lag:] - x[:-lag]
        )

        sd = np.std(
            differences
        )

        if (

            np.isfinite(sd)
            and sd > 0

        ):

            lags.append(lag)

            tau.append(
                np.sqrt(sd)
            )

    if len(lags) < 5:
        return np.nan

    slope = np.polyfit(

        np.log(lags),

        np.log(tau),

        1

    )[0]

    return slope * 2


# ============================================================
# 12. REVERSION SMOOTHNESS
#
# Path smoothness based on direction changes.
#
# Higher = fewer reversals / smoother path.
# ============================================================

def reversion_smoothness(x):

    x = np.asarray(x)

    x = x[
        np.isfinite(x)
    ]

    if len(x) < 4:
        return np.nan

    dx = np.diff(x)

    direction = np.sign(dx)

    # Ignore zero changes
    direction = direction[
        direction != 0
    ]

    if len(direction) < 2:
        return 1.0

    turning_points = np.sum(

        direction[1:]
        != direction[:-1]

    )

    return (

        1
        -
        turning_points
        /
        (len(direction) - 1)

    )


# ============================================================
# 13. ZERO CROSSINGS
# ============================================================

def zero_crossings(x):

    x = np.asarray(x)

    x = x[
        np.isfinite(x)
    ]

    if len(x) < 2:
        return np.nan

    centred = (
        x - np.mean(x)
    )

    return np.sum(

        np.sign(centred[1:])
        !=
        np.sign(centred[:-1])

    )


# ============================================================
# 14. OU ESTIMATION
#
# Discrete OU representation:
#
# X_t = a + b X_(t-1) + epsilon_t
#
# Continuous mean-reversion speed:
#
# kappa = -ln(b)
#
# Half-life:
#
# ln(2) / kappa
#
# Since observations are WEEKLY:
# half-life is in WEEKS.
# ============================================================

def ou_metrics(x):

    x = np.asarray(x)

    x = x[
        np.isfinite(x)
    ]

    if len(x) < 20:

        return {

            "mu": np.nan,
            "kappa_per_week": np.nan,
            "half_life_weeks": np.nan,
            "half_life_years": np.nan,
            "sigma": np.nan,
            "R2": np.nan,
            "ADF_p": np.nan,
            "KPSS_p": np.nan,
            "Hurst": np.nan,
            "Smoothness": np.nan,
            "Zero_crossings": np.nan,
            "Zero_crossings_per_year": np.nan,
            "ACF_1": np.nan

        }

    y = x[1:]

    lag = x[:-1]

    model = LinearRegression()

    model.fit(

        lag.reshape(-1, 1),

        y

    )

    a = model.intercept_

    b = model.coef_[0]

    predictions = model.predict(

        lag.reshape(-1, 1)

    )

    residuals = (
        y - predictions
    )

    # --------------------------------------------------------
    # OU PARAMETERS
    # --------------------------------------------------------

    if (
        np.isfinite(b)
        and 0 < b < 1
    ):

        kappa = -np.log(b)

        half_life_weeks = (
            np.log(2)
            /
            kappa
        )

    else:

        kappa = np.nan

        half_life_weeks = np.nan

    # Long-run mean
    if abs(1 - b) > 1e-12:

        mu = (
            a / (1 - b)
        )

    else:

        mu = np.nan

    # Residual volatility
    sigma = np.std(
        residuals,
        ddof=1
    )

    # --------------------------------------------------------
    # ADF
    # --------------------------------------------------------

    try:

        adf_p = adfuller(
            x,
            autolag="AIC"
        )[1]

    except:

        adf_p = np.nan

    # --------------------------------------------------------
    # KPSS
    # --------------------------------------------------------

    try:

        kpss_p = kpss(
            x,
            regression="c",
            nlags="auto"
        )[1]

    except:

        kpss_p = np.nan

    # --------------------------------------------------------
    # ACF
    # --------------------------------------------------------

    try:

        acf_1 = pd.Series(
            x
        ).autocorr(
            lag=1
        )

    except:

        acf_1 = np.nan

    # --------------------------------------------------------
    # ZERO CROSSINGS
    # --------------------------------------------------------

    zc = zero_crossings(x)

    if np.isfinite(zc):

        zc_year = (
            zc
            /
            (len(x) / 52)
        )

    else:

        zc_year = np.nan

    # --------------------------------------------------------
    # RETURN
    # --------------------------------------------------------

    return {

        "mu": mu,

        "kappa_per_week":
            kappa,

        "half_life_weeks":
            half_life_weeks,

        "half_life_years":
            (
                half_life_weeks / 52
                if np.isfinite(
                    half_life_weeks
                )
                else np.nan
            ),

        "sigma": sigma,

        "R2":
            model.score(
                lag.reshape(-1, 1),
                y
            ),

        "ADF_p":
            adf_p,

        "KPSS_p":
            kpss_p,

        "Hurst":
            hurst_exponent(x),

        "Smoothness":
            reversion_smoothness(x),

        "Zero_crossings":
            zc,

        "Zero_crossings_per_year":
            zc_year,

        "ACF_1":
            acf_1

    }


# ============================================================
# 15. PAIRWISE SPREAD FUNCTION
#
# IMPORTANT:
#
# spread(A, B) = Return_A - Return_B
#
# Therefore:
#
# BRKB - SPY:
#
#   positive spread
#       = BRKB outperforming SPY
#       = BRKB relatively rich
#       = SPY relatively cheap
#
#   negative spread
#       = BRKB underperforming SPY
#       = BRKB relatively cheap
#       = SPY relatively rich
# ============================================================

def make_spread(a, b):

    return (
        returns[a]
        -
        returns[b]
    ).dropna()


# ============================================================
# 16. PAIRWISE ANALYSIS
# ============================================================

pairs = [

    ("BRKB", "IAK"),

    ("BRKB", "SPY"),

    ("IAK", "SPY")

]


pair_results = {}

pair_metrics_table = []


for a, b in pairs:

    spread = make_spread(
        a,
        b
    )

    pair_name = (
        f"{a} - {b}"
    )

    pair_results[pair_name] = spread

    metrics = ou_metrics(
        spread.values
    )

    row = {

        "Pair": pair_name,

        **metrics

    }

    pair_metrics_table.append(
        row
    )

    # --------------------------------------------------------
    # PRINT METRICS
    # --------------------------------------------------------

    print("\n")
    print("=" * 70)
    print(
        f"PAIRWISE OU: {a} - {b}"
    )
    print("=" * 70)

    print(
        f"Interpretation: "
        f"positive = {a} rich / {b} cheap"
    )

    print(
        f"                "
        f"negative = {a} cheap / {b} rich"
    )

    print()

    for key, value in metrics.items():

        if isinstance(
            value,
            (float, np.floating)
        ):

            print(
                f"{key:30s}: "
                f"{value:.6f}"
            )

        else:

            print(
                f"{key:30s}: "
                f"{value}"
            )

    # ========================================================
    # 17. FULL-SAMPLE SPREAD + SIGMA BANDS
    # ========================================================

    mu = spread.mean()

    sigma = spread.std()

    fig = go.Figure()

    fig.add_trace(

        go.Scatter(

            x=spread.index,

            y=spread,

            mode="lines",

            name=pair_name

        )

    )

    fig.add_hline(

        y=mu,

        line_dash="solid",

        annotation_text="Mean"

    )

    fig.add_hline(

        y=mu + sigma,

        line_dash="dot",

        annotation_text="+1σ"

    )

    fig.add_hline(

        y=mu - sigma,

        line_dash="dot",

        annotation_text="-1σ"

    )

    fig.add_hline(

        y=mu + 2 * sigma,

        line_dash="dash",

        annotation_text="+2σ"

    )

    fig.add_hline(

        y=mu - 2 * sigma,

        line_dash="dash",

        annotation_text="-2σ"

    )

    fig.update_layout(

        template="plotly_white",

        title=(
            f"{pair_name} — Weekly Return Spread"
        ),

        xaxis_title="Date",

        yaxis_title="Weekly Return Difference",

        hovermode="x unified"

    )

    fig.show()

    # ========================================================
    # 18. 26-WEEK ROLLING Z-SCORE
    # ========================================================

    rolling_mean = (
        spread
        .rolling(
            ROLLING_WINDOW,
            min_periods=MIN_ROLLING_OBS
        )
        .mean()
    )

    rolling_std = (
        spread
        .rolling(
            ROLLING_WINDOW,
            min_periods=MIN_ROLLING_OBS
        )
        .std()
    )

    rolling_z = (
        spread - rolling_mean
    ) / rolling_std

    rolling_z = rolling_z.replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    fig = go.Figure()

    fig.add_trace(

        go.Scatter(

            x=rolling_z.index,

            y=rolling_z,

            mode="lines",

            name="26-week Z-score"

        )

    )

    fig.add_hline(
        y=0,
        line_dash="solid"
    )

    fig.add_hline(
        y=1,
        line_dash="dot"
    )

    fig.add_hline(
        y=-1,
        line_dash="dot"
    )

    fig.add_hline(
        y=2,
        line_dash="dash"
    )

    fig.add_hline(
        y=-2,
        line_dash="dash"
    )

    fig.update_layout(

        template="plotly_white",

        title=(
            f"{pair_name} — "
            f"26-Week Rolling Z-Score"
        ),

        xaxis_title="Date",

        yaxis_title="Z-Score",

        hovermode="x unified"

    )

    fig.show()

    # ========================================================
    # 19. 26-WEEK ROLLING OU METRICS
    # ========================================================

    rolling_kappa = []
    rolling_half_life = []
    rolling_smoothness = []
    rolling_hurst = []

    rolling_dates = []

    for i in range(
        ROLLING_WINDOW,
        len(spread) + 1
    ):

        window = spread.iloc[
            i - ROLLING_WINDOW:i
        ].values

        m = ou_metrics(
            window
        )

        rolling_dates.append(
            spread.index[i - 1]
        )

        rolling_kappa.append(
            m["kappa_per_week"]
        )

        rolling_half_life.append(
            m["half_life_weeks"]
        )

        rolling_smoothness.append(
            m["Smoothness"]
        )

        rolling_hurst.append(
            m["Hurst"]
        )

    rolling_metrics = pd.DataFrame(

        {

            "Kappa":
                rolling_kappa,

            "Half_Life_Weeks":
                rolling_half_life,

            "Smoothness":
                rolling_smoothness,

            "Hurst":
                rolling_hurst

        },

        index=pd.DatetimeIndex(
            rolling_dates
        )

    )

    # --------------------------------------------------------
    # ROLLING KAPPA
    # --------------------------------------------------------

    fig = go.Figure()

    fig.add_trace(

        go.Scatter(

            x=rolling_metrics.index,

            y=rolling_metrics["Kappa"],

            mode="lines",

            name="κ"

        )

    )

    fig.update_layout(

        template="plotly_white",

        title=(
            f"{pair_name} — "
            f"26-Week Rolling Mean-Reversion Speed"
        ),

        xaxis_title="Date",

        yaxis_title="κ per week",

        hovermode="x unified"

    )

    fig.show()

    # --------------------------------------------------------
    # ROLLING HALF-LIFE
    # --------------------------------------------------------

    fig = go.Figure()

    fig.add_trace(

        go.Scatter(

            x=rolling_metrics.index,

            y=rolling_metrics[
                "Half_Life_Weeks"
            ],

            mode="lines",

            name="Half-life"

        )

    )

    fig.update_layout(

        template="plotly_white",

        title=(
            f"{pair_name} — "
            f"26-Week Rolling OU Half-Life"
        ),

        xaxis_title="Date",

        yaxis_title="Half-Life (weeks)",

        hovermode="x unified"

    )

    fig.show()

    # --------------------------------------------------------
    # ROLLING SMOOTHNESS
    # --------------------------------------------------------

    fig = go.Figure()

    fig.add_trace(

        go.Scatter(

            x=rolling_metrics.index,

            y=rolling_metrics[
                "Smoothness"
            ],

            mode="lines",

            name="Smoothness"

        )

    )

    fig.update_layout(

        template="plotly_white",

        title=(
            f"{pair_name} — "
            f"26-Week Rolling Reversion Smoothness"
        ),

        xaxis_title="Date",

        yaxis_title="Smoothness",

        hovermode="x unified"

    )

    fig.show()

    # ========================================================
    # 20. HISTOGRAM / DISTRIBUTION
    # ========================================================

    fig = px.histogram(

        spread,

        nbins=60,

        marginal="box",

        title=(
            f"{pair_name} — "
            "Spread Distribution"
        )

    )

    fig.update_layout(
        template="plotly_white"
    )

    fig.show()

    # ========================================================
    # 21. YEARLY BOX PLOTS
    # ========================================================

    yearly_df = pd.DataFrame({

        "Date": spread.index,

        "Spread": spread.values

    })

    yearly_df["Year"] = (
        yearly_df["Date"]
        .dt.year
        .astype(str)
    )

    fig = px.box(

        yearly_df,

        x="Year",

        y="Spread",

        points=False,

        title=(
            f"{pair_name} — "
            "Yearly Spread Distribution"
        )

    )

    fig.update_layout(
        template="plotly_white"
    )

    fig.show()

    # ========================================================
    # 22. QQ PLOT
    # ========================================================

    qq = stats.probplot(
        spread.values,
        dist="norm"
    )

    theoretical = qq[0][0]

    ordered = qq[0][1]

    slope = qq[1][0]

    intercept = qq[1][1]

    line = (
        intercept
        +
        slope * theoretical
    )

    fig = go.Figure()

    fig.add_trace(

        go.Scatter(

            x=theoretical,

            y=ordered,

            mode="markers",

            name="Observations"

        )

    )

    fig.add_trace(

        go.Scatter(

            x=theoretical,

            y=line,

            mode="lines",

            name="Normal reference"

        )

    )

    fig.update_layout(

        template="plotly_white",

        title=(
            f"{pair_name} — "
            "Normal QQ Plot"
        ),

        xaxis_title="Theoretical Quantiles",

        yaxis_title="Observed Quantiles"

    )

    fig.show()

    # ========================================================
    # 23. ACF
    # ========================================================

    max_acf_lag = min(
        26,
        len(spread) // 4
    )

    if max_acf_lag >= 2:

        acf_values = acf(

            spread,

            nlags=max_acf_lag,

            fft=True

        )

        lags = np.arange(
            len(acf_values)
        )

        fig = go.Figure()

        fig.add_trace(

            go.Bar(

                x=lags,

                y=acf_values,

                name="ACF"

            )

        )

        fig.add_hline(y=0)

        fig.update_layout(

            template="plotly_white",

            title=(
                f"{pair_name} — "
                "Autocorrelation"
            ),

            xaxis_title="Lag (weeks)",

            yaxis_title="Autocorrelation"

        )

        fig.show()


# ============================================================
# 24. PAIRWISE RESULTS TABLE
# ============================================================

pair_metrics = pd.DataFrame(
    pair_metrics_table
)

print("\n")
print("=" * 70)
print("PAIRWISE OU SUMMARY")
print("=" * 70)

display(
    pair_metrics.round(6)
)


# ============================================================
# 25. MVOU ESTIMATION
#
# Weekly return vector:
#
# X_t = c + A X_(t-1) + epsilon_t
#
# Continuous-time:
#
# dX = K(mu - X)dt + Sigma dW
#
# where:
#
# K = -log(A)
#
# mu = (I-A)^(-1)c
#
# ============================================================

X = returns[
    ASSET_NAMES
].values

X_lag = X[:-1]

X_next = X[1:]


# ============================================================
# 26. ESTIMATE DISCRETE VAR(1) WITH INTERCEPT
# ============================================================

X_design = np.column_stack(

    [

        np.ones(
            len(X_lag)
        ),

        X_lag

    ]

)

coefficients = np.linalg.lstsq(

    X_design,

    X_next,

    rcond=None

)[0]

# Intercept
c = coefficients[0]

# Transition matrix
A = coefficients[1:]


# ============================================================
# 27. MVOU K MATRIX
# ============================================================

A_log = logm(A)

K_complex = -A_log

K = np.real_if_close(
    K_complex
)

K = np.asarray(
    K,
    dtype=float
)


# ============================================================
# 28. MVOU LONG-RUN MEAN
# ============================================================

I = np.eye(
    len(A)
)

try:

    mu_mvou = np.linalg.solve(

        I - A,

        c

    )

except:

    mu_mvou = np.linalg.pinv(
        I - A
    ) @ c


# ============================================================
# 29. MVOU EIGENVALUES / EIGENVECTORS
# ============================================================

eigvals, eigvecs = np.linalg.eig(
    K
)

eigvals = np.real(
    eigvals
)

eigvecs = np.real(
    eigvecs
)


# ============================================================
# 30. MVOU EIGENMODE HALF-LIVES
# ============================================================

mvou_modes = []

for i, eigenvalue in enumerate(
    eigvals
):

    if (
        np.isfinite(eigenvalue)
        and eigenvalue > 0
    ):

        half_life_weeks = (
            np.log(2)
            /
            eigenvalue
        )

        half_life_years = (
            half_life_weeks
            /
            52
        )

    else:

        half_life_weeks = np.nan

        half_life_years = np.nan

    mvou_modes.append({

        "Mode":
            i + 1,

        "Eigenvalue_K":
            eigenvalue,

        "Half_Life_Weeks":
            half_life_weeks,

        "Half_Life_Years":
            half_life_years

    })


mvou_modes = pd.DataFrame(
    mvou_modes
)


# ============================================================
# 31. MVOU RESIDUAL COVARIANCE
#
# This is the discrete weekly residual covariance.
# ============================================================

X_fitted = (
    X_design
    @ coefficients
)

mvou_residuals = (
    X_next
    - X_fitted
)

residual_covariance = np.cov(

    mvou_residuals.T

)


# ============================================================
# 32. MVOU R-SQUARED
# ============================================================

mvou_r2 = {}

for j, asset in enumerate(
    ASSET_NAMES
):

    y = X_next[:, j]

    yhat = X_fitted[:, j]

    ss_res = np.sum(
        (y - yhat) ** 2
    )

    ss_tot = np.sum(
        (y - y.mean()) ** 2
    )

    mvou_r2[asset] = (

        1
        -
        ss_res / ss_tot

    )


# ============================================================
# 33. PRINT MVOU RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("MULTIVARIATE OU — MVOU")
print("=" * 70)

print("\nTransition matrix A:")
print(
    pd.DataFrame(
        A,
        index=ASSET_NAMES,
        columns=ASSET_NAMES
    ).round(8)
)

print("\nIntercept c:")
print(
    pd.Series(
        c,
        index=ASSET_NAMES
    ).round(8)
)

print("\nMean-reversion matrix K:")
print(
    pd.DataFrame(
        K,
        index=ASSET_NAMES,
        columns=ASSET_NAMES
    ).round(8)
)

print("\nLong-run mean μ:")
print(
    pd.Series(
        mu_mvou,
        index=ASSET_NAMES
    ).round(8)
)

print("\nEigenvalues:")
print(
    mvou_modes.round(8)
)

print("\nMVOU R²:")
print(
    pd.Series(
        mvou_r2
    ).round(6)
)

print("\nWeekly residual covariance:")
print(
    pd.DataFrame(
        residual_covariance,
        index=ASSET_NAMES,
        columns=ASSET_NAMES
    ).round(10)
)


# ============================================================
# 34. MVOU K MATRIX HEATMAP
# ============================================================

fig = px.imshow(

    K,

    x=ASSET_NAMES,

    y=ASSET_NAMES,

    text_auto=".4f",

    aspect="auto",

    title="MVOU Mean-Reversion Matrix K"

)

fig.update_layout(
    template="plotly_white"
)

fig.show()


# ============================================================
# 35. MVOU EIGENVALUES
# ============================================================

fig = go.Figure()

fig.add_trace(

    go.Bar(

        x=[
            f"Mode {i}"
            for i in mvou_modes["Mode"]
        ],

        y=mvou_modes[
            "Eigenvalue_K"
        ],

        text=[
            f"{x:.4f}"
            for x in mvou_modes[
                "Eigenvalue_K"
            ]
        ],

        textposition="auto"

    )

)

fig.update_layout(

    template="plotly_white",

    title=(
        "MVOU Mean-Reversion Eigenvalues"
    ),

    xaxis_title="Eigenmode",

    yaxis_title="Eigenvalue κ"

)

fig.show()


# ============================================================
# 36. MVOU HALF-LIVES
# ============================================================

fig = go.Figure()

fig.add_trace(

    go.Bar(

        x=[
            f"Mode {i}"
            for i in mvou_modes["Mode"]
        ],

        y=mvou_modes[
            "Half_Life_Weeks"
        ],

        text=[
            (
                f"{x:.2f} weeks"
                if np.isfinite(x)
                else "N/A"
            )

            for x in mvou_modes[
                "Half_Life_Weeks"
            ]
        ],

        textposition="auto"

    )

)

fig.update_layout(

    template="plotly_white",

    title=(
        "MVOU Eigenmode Half-Lives"
    ),

    xaxis_title="Eigenmode",

    yaxis_title="Half-Life (weeks)"

)

fig.show()


# ============================================================
# 37. MVOU EIGENVECTORS
#
# Shows what combination of assets each mean-reverting
# eigenmode represents.
# ============================================================

eigenvector_df = pd.DataFrame(

    eigvecs,

    index=ASSET_NAMES,

    columns=[
        f"Mode {i+1}"
        for i in range(
            eigvecs.shape[1]
        )
    ]

)

print("\n")
print("=" * 70)
print("MVOU EIGENVECTORS")
print("=" * 70)

print(
    eigenvector_df.round(6)
)


fig = px.imshow(

    eigenvector_df,

    text_auto=".3f",

    aspect="auto",

    title="MVOU Eigenvectors"

)

fig.update_layout(
    template="plotly_white"
)

fig.show()


# ============================================================
# 38. MVOU STATE SPACE — BRK.B vs IAK
# ============================================================

fig = px.scatter(

    returns,

    x="BRKB",

    y="IAK",

    hover_data={
        "SPY": True
    },

    title=(
        "Weekly Return State Space — "
        "BRK.B vs IAK"
    )

)

fig.update_layout(
    template="plotly_white"
)

fig.show()


# ============================================================
# 39. MVOU STATE SPACE — BRK.B vs SPY
# ============================================================

fig = px.scatter(

    returns,

    x="BRKB",

    y="SPY",

    hover_data={
        "IAK": True
    },

    title=(
        "Weekly Return State Space — "
        "BRK.B vs SPY"
    )

)

fig.update_layout(
    template="plotly_white"
)

fig.show()


# ============================================================
# 40. MVOU STATE SPACE — IAK vs SPY
# ============================================================

fig = px.scatter(

    returns,

    x="IAK",

    y="SPY",

    hover_data={
        "BRKB": True
    },

    title=(
        "Weekly Return State Space — "
        "IAK vs SPY"
    )

)

fig.update_layout(
    template="plotly_white"
)

fig.show()


# ============================================================
# 41. MVOU 3D STATE SPACE
# ============================================================

fig = px.scatter_3d(

    returns,

    x="BRKB",

    y="IAK",

    z="SPY",

    title=(
        "MVOU 3D Weekly Return State Space"
    )

)

fig.update_layout(
    template="plotly_white"
)

fig.show()


# ============================================================
# 42. FINAL INDIVIDUAL-ASSET WEEKLY STATISTICS
# ============================================================

asset_summary = []

for asset in ASSET_NAMES:

    x = returns[asset]

    asset_summary.append({

        "Asset":
            asset,

        "Mean_Weekly_Return":
            x.mean(),

        "Weekly_Volatility":
            x.std(),

        "Annualized_Return":
            (1 + x.mean()) ** 52 - 1,

        "Annualized_Volatility":
            x.std() * np.sqrt(52),

        "Skew":
            x.skew(),

        "Kurtosis":
            x.kurtosis(),

        "Hurst":
            hurst_exponent(
                x.values
            ),

        "ADF_p":
            adfuller(
                x.values
            )[1],

        "ACF_1":
            x.autocorr(
                lag=1
            )

    })


asset_summary = pd.DataFrame(
    asset_summary
)


print("\n")
print("=" * 70)
print("INDIVIDUAL WEEKLY RETURN STATISTICS")
print("=" * 70)

display(
    asset_summary.round(6)
)


# ============================================================
# 43. FINAL PAIRWISE TABLE
# ============================================================

print("\n")
print("=" * 70)
print("FINAL PAIRWISE OU RESULTS")
print("=" * 70)

display(
    pair_metrics.round(6)
)


# ============================================================
# 44. FINAL MVOU TABLE
# ============================================================

print("\n")
print("=" * 70)
print("FINAL MVOU EIGENMODE RESULTS")
print("=" * 70)

display(
    mvou_modes.round(6)
)


# ============================================================
# 45. SAVE RESULTS TO CSV
# ============================================================

pair_metrics.to_csv(
    "pairwise_OU_results.csv",
    index=False
)

mvou_modes.to_csv(
    "MVOU_eigenmodes.csv",
    index=False
)

asset_summary.to_csv(
    "individual_weekly_statistics.csv",
    index=False
)

eigenvector_df.to_csv(
    "MVOU_eigenvectors.csv"
)

pd.DataFrame(
    K,
    index=ASSET_NAMES,
    columns=ASSET_NAMES
).to_csv(
    "MVOU_K_matrix.csv"
)

print("\n")
print("=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)

print("\nFiles saved:")

print("  pairwise_OU_results.csv")
print("  MVOU_eigenmodes.csv")
print("  individual_weekly_statistics.csv")
print("  MVOU_eigenvectors.csv")
print("  MVOU_K_matrix.csv")
```
