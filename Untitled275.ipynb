```python
import pandas as pd
import requests
import xml.etree.ElementTree as ET

# ============================================================
# SETTINGS
# ============================================================

VALUATION_DATE = pd.Timestamp("2026-09-21")

URL = (
    "https://home.treasury.gov/resource-center/data-chart-center/"
    "interest-rates/pages/xml"
    "?data=daily_treasury_yield_curve"
    "&field_tdr_date_value_month=202609"
)

# ============================================================
# DOWNLOAD OFFICIAL TREASURY XML
# ============================================================

r = requests.get(URL, timeout=30)
r.raise_for_status()

root = ET.fromstring(r.content)

# Treasury XML namespaces
ns = {
    "atom": "http://www.w3.org/2005/Atom",
    "m": "http://schemas.microsoft.com/ado/2007/08/dataservices/metadata",
    "d": "http://schemas.microsoft.com/ado/2007/08/dataservices",
}

rows = []

for entry in root.findall("atom:entry", ns):

    props = entry.find("atom:content/m:properties", ns)

    if props is None:
        continue

    row = {}

    for child in props:
        name = child.tag.split("}")[-1]
        row[name] = child.text

    rows.append(row)

raw = pd.DataFrame(rows)

# ============================================================
# FIND 22 SEP 2026
# ============================================================

date_col = "NEW_DATE"

raw[date_col] = pd.to_datetime(raw[date_col]).dt.tz_localize(None)

row = raw.loc[raw[date_col] == VALUATION_DATE]

if row.empty:
    raise ValueError(f"No Treasury curve found for {VALUATION_DATE.date()}")

# ============================================================
# PAR CURVE NODES
# ============================================================

nodes = {
    1/12:  "BC_1MONTH",
    1.5/12:"BC_1_5MONTH",
    2/12:  "BC_2MONTH",
    3/12:  "BC_3MONTH",
    4/12:  "BC_4MONTH",
    6/12:  "BC_6MONTH",
    1:     "BC_1YEAR",
    2:     "BC_2YEAR",
    3:     "BC_3YEAR",
    5:     "BC_5YEAR",
    7:     "BC_7YEAR",
    10:    "BC_10YEAR",
    20:    "BC_20YEAR",
    30:    "BC_30YEAR",
}

curve = pd.DataFrame({
    "Maturity_Years": list(nodes.keys()),
    "Par_Yield_pct": [
        pd.to_numeric(row[col].iloc[0], errors="coerce")
        for col in nodes.values()
    ]
})

curve = curve.dropna().reset_index(drop=True)

print(f"Official Treasury Par Curve — {VALUATION_DATE.date()}")
display(curve)
```

```python
import numpy as np
import pandas as pd
from scipy.interpolate import PchipInterpolator
import plotly.graph_objects as go

# ============================================================
# CELL 2 — BOOTSTRAP SEMIANNUAL TREASURY SPOT CURVE
# ============================================================

# Use Treasury par nodes from 6M onward
par_nodes = curve[curve["Maturity_Years"] >= 0.5].copy()

maturity = par_nodes["Maturity_Years"].to_numpy(dtype=float)
par_yield = par_nodes["Par_Yield_pct"].to_numpy(dtype=float) / 100

# ------------------------------------------------------------
# 1. INTERPOLATE PAR CURVE USING PCHIP
# ------------------------------------------------------------

par_interp = PchipInterpolator(maturity, par_yield)

# Semiannual grid: 0.5Y to 30Y
t_grid = np.arange(0.5, 30.0 + 0.5, 0.5)

par_grid = par_interp(t_grid)

# ------------------------------------------------------------
# 2. BOOTSTRAP DISCOUNT FACTORS + SPOT RATES
# ------------------------------------------------------------

discount_factors = []
spot_rates = []

for i, (t, y) in enumerate(zip(t_grid, par_grid)):

    # Semiannual coupon payment per $100 face
    coupon = 100 * y / 2

    if i == 0:
        # 6M par Treasury has one remaining payment
        df = 100 / (100 + coupon)

    else:
        # PV of all coupons before maturity
        previous_coupon_pv = coupon * sum(discount_factors)

        # Solve final discount factor from par pricing equation
        df = (100 - previous_coupon_pv) / (100 + coupon)

    discount_factors.append(df)

    # Nominal annual spot rate, compounded semiannually
    spot = 2 * (df ** (-1 / (2 * t)) - 1)

    spot_rates.append(spot)

# ------------------------------------------------------------
# 3. OUTPUT TABLE
# ------------------------------------------------------------

spot_curve = pd.DataFrame({
    "Maturity_Years": t_grid,
    "Par_Yield_pct": par_grid * 100,
    "Spot_Yield_pct": np.array(spot_rates) * 100,
    "Discount_Factor": discount_factors
})

print("=" * 80)
print("BOOTSTRAPPED U.S. TREASURY SPOT CURVE")
print(f"Valuation Date: {VALUATION_DATE.strftime('%d %B %Y')}")
print("=" * 80)

display(
    spot_curve.round({
        "Maturity_Years": 2,
        "Par_Yield_pct": 4,
        "Spot_Yield_pct": 4,
        "Discount_Factor": 6
    })
)

# ------------------------------------------------------------
# 4. INTERACTIVE PLOTLY CHART — COLAB COMPATIBLE
# ------------------------------------------------------------

fig = go.Figure()

# Treasury observed nodes
fig.add_trace(
    go.Scatter(
        x=par_nodes["Maturity_Years"],
        y=par_nodes["Par_Yield_pct"],
        mode="markers",
        name="Treasury Par Nodes",
        marker=dict(size=9)
    )
)

# Interpolated par curve
fig.add_trace(
    go.Scatter(
        x=spot_curve["Maturity_Years"],
        y=spot_curve["Par_Yield_pct"],
        mode="lines",
        name="Interpolated Par Curve"
    )
)

# Bootstrapped spot curve
fig.add_trace(
    go.Scatter(
        x=spot_curve["Maturity_Years"],
        y=spot_curve["Spot_Yield_pct"],
        mode="lines",
        name="Bootstrapped Spot Curve"
    )
)

fig.update_layout(
    title=(
        "U.S. Treasury Par and Bootstrapped Spot Curves — "
        f"{VALUATION_DATE.strftime('%d %b %Y')}"
    ),
    xaxis_title="Maturity (Years)",
    yaxis_title="Yield (%)",
    hovermode="x unified",
    template="plotly_white",
    width=1000,
    height=600,
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="center",
        x=0.5
    )
)

fig.update_xaxes(
    showgrid=True,
    dtick=2
)

fig.update_yaxes(
    showgrid=True,
    ticksuffix="%"
)

fig.show()
```

```python
# ============================================================
# CELL 6 — CASH TREASURY VALUATION
# Market YTM Price vs Spot-Curve DCF Fair Value
# ============================================================

import numpy as np
import pandas as pd
from scipy.interpolate import PchipInterpolator
import plotly.graph_objects as go

# ------------------------------------------------------------
# 1. THREE ZN DELIVERABLES
# ------------------------------------------------------------

bonds = pd.DataFrame({
    "CUSIP": ["91282CQW4", "91282CJJ1", "91282CLW9"],
    "Coupon_pct": [4.250, 4.500, 4.250],
    "Maturity_Date": pd.to_datetime([
        "2033-06-30",
        "2033-11-15",
        "2034-11-15"
    ]),
    "Conversion_Factor": [0.9069, 0.9177, 0.8927],
    "Market_YTM_pct": [4.870, 4.912, 4.939]
})

VAL_DATE = pd.Timestamp("2026-09-21")

# ------------------------------------------------------------
# 2. DISCOUNT-FACTOR INTERPOLATOR
# ------------------------------------------------------------

log_df_interp = PchipInterpolator(
    spot_curve["Maturity_Years"],
    np.log(spot_curve["Discount_Factor"]),
    extrapolate=True
)

# ------------------------------------------------------------
# 3. COUPON SCHEDULE
# ------------------------------------------------------------

def coupon_schedule(maturity, valuation_date):

    dates = []
    d = maturity

    while d > valuation_date:
        dates.append(d)
        d = d - pd.DateOffset(months=6)

    return sorted(dates), d

# ------------------------------------------------------------
# 4. VALUE TREASURY
# ------------------------------------------------------------

def value_treasury(row):

    coupon_rate = row["Coupon_pct"] / 100
    ytm = row["Market_YTM_pct"] / 100
    maturity = row["Maturity_Date"]

    coupon = 100 * coupon_rate / 2

    payment_dates, previous_coupon = coupon_schedule(
        maturity,
        VAL_DATE
    )

    next_coupon = payment_dates[0]

    # Accrued interest
    days_period = (next_coupon - previous_coupon).days
    days_accrued = (VAL_DATE - previous_coupon).days

    accrued_fraction = days_accrued / days_period
    accrued_interest = coupon * accrued_fraction

    # Fractional coupon period to next payment
    w = 1 - accrued_fraction

    # --------------------------------------------------------
    # MARKET PRICE FROM YTM
    # --------------------------------------------------------

    dirty_market = 0.0

    for j, pay_date in enumerate(payment_dates):

        periods = w + j
        cashflow = coupon

        if pay_date == maturity:
            cashflow += 100

        dirty_market += (
            cashflow /
            (1 + ytm / 2) ** periods
        )

    clean_market = dirty_market - accrued_interest

    # --------------------------------------------------------
    # SPOT-CURVE DCF FAIR VALUE
    # --------------------------------------------------------

    dirty_dcf = 0.0

    for pay_date in payment_dates:

        t = (pay_date - VAL_DATE).days / 365.25

        df = np.exp(
            float(log_df_interp(t))
        )

        cashflow = coupon

        if pay_date == maturity:
            cashflow += 100

        dirty_dcf += cashflow * df

    clean_dcf = dirty_dcf - accrued_interest

    # --------------------------------------------------------
    # RELATIVE VALUE
    # --------------------------------------------------------

    mispricing = clean_dcf - clean_market

    return pd.Series({
        "Accrued_Interest": accrued_interest,
        "Market_Clean": clean_market,
        "Market_Dirty": dirty_market,
        "DCF_Clean": clean_dcf,
        "DCF_Dirty": dirty_dcf,
        "DCF_minus_Market": mispricing,
        "Price_Diff_bp": mispricing * 100
    })

# ------------------------------------------------------------
# 5. RUN VALUATION
# ------------------------------------------------------------

results = bonds.join(
    bonds.apply(value_treasury, axis=1)
)

print("=" * 100)
print("ZN DELIVERABLE CASH TREASURY VALUATION")
print(f"Valuation Date: {VAL_DATE.strftime('%d %B %Y')}")
print("=" * 100)

display(
    results[[
        "CUSIP",
        "Coupon_pct",
        "Maturity_Date",
        "Conversion_Factor",
        "Market_YTM_pct",
        "Accrued_Interest",
        "Market_Clean",
        "Market_Dirty",
        "DCF_Clean",
        "DCF_Dirty",
        "DCF_minus_Market",
        "Price_Diff_bp"
    ]].round(4)
)

# ============================================================
# 6. INTERACTIVE LINE CHART
# ============================================================

# Sort chronologically by maturity
plot_df = results.sort_values("Maturity_Date")

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=plot_df["Maturity_Date"],
        y=plot_df["Market_Clean"],
        mode="lines+markers",
        name="Market Clean Price",
        customdata=plot_df["CUSIP"],
        hovertemplate=(
            "CUSIP: %{customdata}<br>"
            "Maturity: %{x|%d %b %Y}<br>"
            "Market: %{y:.4f}"
            "<extra></extra>"
        )
    )
)

fig.add_trace(
    go.Scatter(
        x=plot_df["Maturity_Date"],
        y=plot_df["DCF_Clean"],
        mode="lines+markers",
        name="Spot-Curve DCF",
        customdata=plot_df["CUSIP"],
        hovertemplate=(
            "CUSIP: %{customdata}<br>"
            "Maturity: %{x|%d %b %Y}<br>"
            "DCF: %{y:.4f}"
            "<extra></extra>"
        )
    )
)

fig.update_layout(
    title=(
        "ZN Deliverables: Market vs Spot-Curve Fair Value — "
        f"{VAL_DATE.strftime('%d %b %Y')}"
    ),
    xaxis_title="Maturity",
    yaxis_title="Clean Price per $100 Face",
    template="plotly_white",
    width=1000,
    height=600,
    hovermode="x unified"
)

fig.update_xaxes(showgrid=True)
fig.update_yaxes(showgrid=True)

fig.show()
```

```python
# ============================================================
# CELL 12 — CORRECTED IMPLIED REPO + SAMPLE CTD
# ============================================================

import pandas as pd
import numpy as np
import plotly.graph_objects as go

TRADE_DATE = pd.Timestamp("2026-09-21")
DELIVERY_DATE = pd.Timestamp("2026-12-31")

F = FUTURES_PRICE
REPO = BGCR
DAY_BASIS = 360

# ------------------------------------------------------------
# ACCRUED INTEREST
# ------------------------------------------------------------

def accrued_interest(coupon_pct, maturity, date):

    semi_coupon = coupon_pct / 2  # per $100 face

    d = maturity

    while d > date:
        next_coupon = d
        d = d - pd.DateOffset(months=6)

    previous_coupon = d

    period_days = (next_coupon - previous_coupon).days
    accrued_days = (date - previous_coupon).days

    return semi_coupon * accrued_days / period_days


# ------------------------------------------------------------
# COUPONS BETWEEN PURCHASE AND DELIVERY
# ------------------------------------------------------------

def coupons_to_delivery(coupon_pct, maturity):

    semi_coupon = coupon_pct / 2

    dates = []
    d = maturity

    while d > TRADE_DATE:
        dates.append(d)
        d = d - pd.DateOffset(months=6)

    dates = sorted([
        x for x in dates
        if TRADE_DATE < x <= DELIVERY_DATE
    ])

    coupon_future_values = []

    for coupon_date in dates:

        reinvest_days = (
            DELIVERY_DATE - coupon_date
        ).days

        fv = semi_coupon * (
            1 + REPO * reinvest_days / DAY_BASIS
        )

        coupon_future_values.append({
            "date": coupon_date,
            "coupon": semi_coupon,
            "reinvest_days": reinvest_days,
            "future_value": fv
        })

    return coupon_future_values


# ------------------------------------------------------------
# IMPLIED REPO
# ------------------------------------------------------------

def calculate_irr(row):

    dirty_purchase = row["Market_Dirty"]
    cf = row["Conversion_Factor"]
    coupon_pct = row["Coupon_pct"]
    maturity = row["Maturity_Date"]

    # Delivery accrued interest
    delivery_ai = accrued_interest(
        coupon_pct,
        maturity,
        DELIVERY_DATE
    )

    # Futures invoice amount
    invoice = (
        F * cf
        + delivery_ai
    )

    # Interim coupons + reinvestment
    coupon_info = coupons_to_delivery(
        coupon_pct,
        maturity
    )

    coupon_cash = sum(
        x["coupon"]
        for x in coupon_info
    )

    coupon_fv = sum(
        x["future_value"]
        for x in coupon_info
    )

    coupon_reinvestment = (
        coupon_fv - coupon_cash
    )

    days = (
        DELIVERY_DATE - TRADE_DATE
    ).days

    # Total terminal proceeds
    terminal_value = (
        invoice
        + coupon_fv
    )

    # Annualised implied repo
    irr = (
        (terminal_value / dirty_purchase - 1)
        * DAY_BASIS / days
    )

    spread_bp = (
        irr - REPO
    ) * 10000

    return pd.Series({
        "Delivery_AI": delivery_ai,
        "Invoice_Price": invoice,
        "Coupon_Income": coupon_cash,
        "Coupon_Reinvestment": coupon_reinvestment,
        "Terminal_Proceeds": terminal_value,
        "Implied_Repo_pct": irr * 100,
        "BGCR_pct": REPO * 100,
        "IRR_minus_BGCR_bp": spread_bp
    })


# ------------------------------------------------------------
# RUN MODEL
# ------------------------------------------------------------

corrected = basis.copy()

irr_detail = corrected.apply(
    calculate_irr,
    axis=1
)

corrected = pd.concat(
    [corrected, irr_detail],
    axis=1
)

# Rank by implied repo
corrected["IRR_Rank"] = (
    corrected["Implied_Repo_pct"]
    .rank(ascending=False, method="min")
    .astype(int)
)

sample_ctd = corrected.loc[
    corrected["Implied_Repo_pct"].idxmax()
]

print("=" * 105)
print("ZN DEC-26 — CORRECTED IMPLIED REPO ANALYSIS")
print(f"Trade Date     : {TRADE_DATE.date()}")
print(f"Delivery Date  : {DELIVERY_DATE.date()}")
print(f"ZN Price       : {F:.4f}")
print(f"BGCR           : {REPO*100:.3f}%")
print("=" * 105)

display(
    corrected[[
        "CUSIP",
        "Market_Dirty",
        "Conversion_Factor",
        "Delivery_AI",
        "Invoice_Price",
        "Coupon_Income",
        "Coupon_Reinvestment",
        "Implied_Repo_pct",
        "BGCR_pct",
        "IRR_minus_BGCR_bp",
        "IRR_Rank"
    ]].round(4)
)

print()
print(
    "SAMPLE CTD CANDIDATE:",
    sample_ctd["CUSIP"]
)

print(
    f"Implied Repo: "
    f"{sample_ctd['Implied_Repo_pct']:.4f}%"
)

print(
    f"Spread vs BGCR: "
    f"{sample_ctd['IRR_minus_BGCR_bp']:+.2f} bp"
)


# ============================================================
# INTERACTIVE CHART
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=corrected["CUSIP"],
        y=corrected["IRR_minus_BGCR_bp"],
        text=corrected["IRR_minus_BGCR_bp"].map(
            lambda x: f"{x:+.1f} bp"
        ),
        textposition="outside",
        name="IRR − BGCR"
    )
)

fig.add_hline(
    y=0,
    line_dash="dash",
    line_width=1
)

fig.update_layout(
    title=(
        "ZN Dec-26: Implied Repo Spread vs BGCR — "
        "21 Sep 2026"
    ),
    xaxis_title="CUSIP",
    yaxis_title="Spread to BGCR (bp)",
    template="plotly_white",
    width=1000,
    height=550
)

fig.show()
```

```python
# ============================================================
# CELL 15 — CFA-STYLE 1Y TREASURY RETURN DECOMPOSITION
#
# Total Return ≈
# Carry + Rolldown + Duration Impact + Convexity Impact
#
# Duration / convexity:
#   calculated from each bond's own cash flows + market YTM
#
# Rate scenarios:
#   parallel Treasury curve shift of -50 / 0 / +50 bp
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go

CUSIPS = [
    "91282CQW4",
    "91282CJJ1",
    "91282CLW9"
]

SCENARIOS_BP = [-50, 0, 50]

HORIZON_DATE = TRADE_DATE + pd.DateOffset(years=1)


# ============================================================
# 1. FIND EXISTING SPOT CURVE
# Used ONLY for rolldown / DCF information
# ============================================================

def find_spot_curve():

    maturity_candidates = [
        "maturity_years", "maturity", "tenor_years",
        "tenor", "years", "t"
    ]

    spot_candidates = [
        "spot_rate_pct", "spot_pct", "spot_rate",
        "spot_yield_pct", "spot_yield", "spot"
    ]

    for obj_name, obj in list(globals().items()):

        if not isinstance(obj, pd.DataFrame):
            continue

        cols = {
            str(c).lower().strip(): c
            for c in obj.columns
        }

        m_col = next(
            (cols[x] for x in maturity_candidates if x in cols),
            None
        )

        s_col = next(
            (cols[x] for x in spot_candidates if x in cols),
            None
        )

        if m_col is None or s_col is None:
            continue

        temp = obj[[m_col, s_col]].copy()

        temp.columns = ["Years", "Spot"]

        temp["Years"] = pd.to_numeric(
            temp["Years"], errors="coerce"
        )

        temp["Spot"] = pd.to_numeric(
            temp["Spot"], errors="coerce"
        )

        temp = (
            temp.dropna()
            .sort_values("Years")
            .drop_duplicates("Years")
        )

        if len(temp) < 5:
            continue

        if temp["Spot"].median() < 1:
            temp["Spot"] *= 100

        print(
            f"Spot curve found: {obj_name} | "
            f"{m_col} + {s_col}"
        )

        return temp

    raise ValueError(
        "Could not locate bootstrapped spot curve."
    )


spot = find_spot_curve()

spot_years = spot["Years"].to_numpy(float)
spot_rates = spot["Spot"].to_numpy(float)


def spot_rate(t):

    return np.interp(
        t,
        spot_years,
        spot_rates
    ) / 100


# ============================================================
# 2. SPOT-CURVE DCF
# Used for fair value + unchanged-curve rolldown
# ============================================================

def spot_dcf_price(
    coupon_pct,
    maturity,
    valuation_date
):

    coupon_payment = coupon_pct / 2

    payment_dates, previous_coupon = coupon_schedule(
        maturity,
        valuation_date
    )

    dirty = 0.0

    for pay_date in payment_dates:

        t = (
            pay_date - valuation_date
        ).days / 365.25

        if t <= 0:
            continue

        cf = coupon_payment

        if pay_date == maturity:
            cf += 100

        r = spot_rate(t)

        df = (
            1 + r / 2
        ) ** (-2 * t)

        dirty += cf * df

    next_coupon = payment_dates[0]

    period_days = (
        next_coupon - previous_coupon
    ).days

    accrued_days = (
        valuation_date - previous_coupon
    ).days

    accrued_fraction = (
        accrued_days / period_days
    )

    ai = (
        coupon_payment
        * accrued_fraction
    )

    clean = dirty - ai

    return clean, dirty, ai


# ============================================================
# 3. CONVENTIONAL BOND RISK METRICS
#
# Macaulay Duration
# Modified Duration
# Convexity
#
# All based on each bond's OWN market YTM.
# ============================================================

def conventional_risk_metrics(
    coupon_pct,
    ytm_pct,
    maturity,
    valuation_date
):

    coupon_payment = coupon_pct / 2

    y = ytm_pct / 100
    semi_yield = y / 2

    payment_dates, previous_coupon = coupon_schedule(
        maturity,
        valuation_date
    )

    next_coupon = payment_dates[0]

    period_days = (
        next_coupon - previous_coupon
    ).days

    accrued_days = (
        valuation_date - previous_coupon
    ).days

    accrued_fraction = (
        accrued_days / period_days
    )

    # Fraction of semiannual period until next coupon
    w = 1 - accrued_fraction

    dirty_price = 0.0
    weighted_pv_time = 0.0
    convexity_numerator = 0.0

    for j, pay_date in enumerate(payment_dates):

        # Semiannual periods from settlement
        n = w + j

        cf = coupon_payment

        if pay_date == maturity:
            cf += 100

        pv = (
            cf /
            (1 + semi_yield) ** n
        )

        dirty_price += pv

        # Time in years
        t_years = n / 2

        weighted_pv_time += (
            t_years * pv
        )

        # Conventional semiannual convexity
        convexity_numerator += (
            cf
            * n
            * (n + 1)
            /
            (1 + semi_yield) ** (n + 2)
        )

    macaulay_duration = (
        weighted_pv_time /
        dirty_price
    )

    modified_duration = (
        macaulay_duration /
        (1 + semi_yield)
    )

    convexity = (
        convexity_numerator /
        dirty_price /
        4
    )

    ai = (
        coupon_payment
        * accrued_fraction
    )

    clean_price = (
        dirty_price - ai
    )

    dv01_per_100 = (
        modified_duration
        * dirty_price
        * 0.0001
    )

    return {
        "YTM_Clean": clean_price,
        "YTM_Dirty": dirty_price,
        "Macaulay_Duration": macaulay_duration,
        "Modified_Duration": modified_duration,
        "Convexity": convexity,
        "DV01_per_100": dv01_per_100
    }


# ============================================================
# 4. BUILD RETURN DECOMPOSITION
# ============================================================

rows = []

for cusip in CUSIPS:

    bond = results.loc[
        results["CUSIP"] == cusip
    ].iloc[0]

    coupon = float(
        bond["Coupon_pct"]
    )

    ytm = float(
        bond["Market_YTM_pct"]
    )

    maturity = pd.Timestamp(
        bond["Maturity_Date"]
    )

    market_clean = float(
        bond["Market_Clean"]
    )

    # --------------------------------------------------------
    # CONVENTIONAL DURATION / CONVEXITY
    # --------------------------------------------------------

    risk = conventional_risk_metrics(
        coupon,
        ytm,
        maturity,
        TRADE_DATE
    )

    mac_duration = risk[
        "Macaulay_Duration"
    ]

    mod_duration = risk[
        "Modified_Duration"
    ]

    convexity = risk[
        "Convexity"
    ]

    dv01 = risk[
        "DV01_per_100"
    ]

    # --------------------------------------------------------
    # SPOT DCF TODAY
    # --------------------------------------------------------

    spot_clean_today, _, _ = spot_dcf_price(
        coupon,
        maturity,
        TRADE_DATE
    )

    # --------------------------------------------------------
    # CARRY
    # Actual coupon cash received during next year
    # --------------------------------------------------------

    payment_dates, _ = coupon_schedule(
        maturity,
        TRADE_DATE
    )

    coupons_received = [
        d for d in payment_dates
        if TRADE_DATE < d <= HORIZON_DATE
    ]

    coupon_cash = (
        coupon / 2
    ) * len(coupons_received)

    carry_pct = (
        coupon_cash /
        market_clean *
        100
    )

    # --------------------------------------------------------
    # ROLLDOWN
    #
    # Age bond one year and revalue using the
    # SAME unchanged spot curve.
    # --------------------------------------------------------

    rolled_clean, _, _ = spot_dcf_price(
        coupon,
        maturity,
        HORIZON_DATE
    )

    rolldown_pct = (
        rolled_clean
        - spot_clean_today
    ) / market_clean * 100

    # --------------------------------------------------------
    # TREASURY CURVE SCENARIOS
    #
    # Parallel curve move translated through this bond's
    # OWN modified duration and convexity.
    # --------------------------------------------------------

    for shock_bp in SCENARIOS_BP:

        dy = (
            shock_bp /
            10000
        )

        duration_impact_pct = (
            -mod_duration
            * dy
            * 100
        )

        convexity_impact_pct = (
            0.5
            * convexity
            * dy**2
            * 100
        )

        yield_impact_pct = (
            duration_impact_pct
            + convexity_impact_pct
        )

        total_return_pct = (
            carry_pct
            + rolldown_pct
            + yield_impact_pct
        )

        rows.append({

            "CUSIP": cusip,
            "Coupon_pct": coupon,
            "Maturity": maturity,
            "Market_YTM_pct": ytm,
            "Market_Clean": market_clean,
            "Spot_DCF": spot_clean_today,

            "Macaulay_Duration":
                mac_duration,

            "Modified_Duration":
                mod_duration,

            "Convexity":
                convexity,

            "DV01_per_100":
                dv01,

            "Carry_pct":
                carry_pct,

            "Rolldown_pct":
                rolldown_pct,

            "Yield_Shock_bp":
                shock_bp,

            "Duration_Impact_pct":
                duration_impact_pct,

            "Convexity_Impact_pct":
                convexity_impact_pct,

            "Yield_Impact_pct":
                yield_impact_pct,

            "Total_Return_pct":
                total_return_pct
        })


decomp = pd.DataFrame(rows)


# ============================================================
# 5. BASE + SCENARIO TABLE
# ============================================================

base = decomp[
    decomp["Yield_Shock_bp"] == 0
].copy()


scenarios = (
    decomp.pivot(
        index="CUSIP",
        columns="Yield_Shock_bp",
        values="Total_Return_pct"
    )
    .reset_index()
)

scenarios = scenarios.rename(
    columns={
        -50: "TR_-50bp",
        0: "TR_0bp",
        50: "TR_+50bp"
    }
)


summary = (
    base[[
        "CUSIP",
        "Coupon_pct",
        "Maturity",
        "Market_YTM_pct",
        "Market_Clean",
        "Spot_DCF",
        "Macaulay_Duration",
        "Modified_Duration",
        "Convexity",
        "DV01_per_100",
        "Carry_pct",
        "Rolldown_pct"
    ]]
    .merge(
        scenarios,
        on="CUSIP"
    )
)


summary["Maturity"] = pd.to_datetime(
    summary["Maturity"]
).dt.strftime("%d %b %Y")


summary.columns = [
    "CUSIP",
    "Coupon %",
    "Maturity",
    "YTM %",
    "Clean Price",
    "Spot DCF",
    "Mac Duration",
    "Mod Duration",
    "Convexity",
    "DV01 / $100",
    "Carry %",
    "Rolldown %",
    "TR −50bp",
    "TR Unchanged",
    "TR +50bp"
]


print("=" * 145)
print("U.S. TREASURY — 1-YEAR RETURN DECOMPOSITION")
print(
    f"Valuation Date: {TRADE_DATE.date()} | "
    "Parallel Treasury Curve Scenarios: −50 / 0 / +50 bp"
)
print("=" * 145)


display(
    summary.round({
        "Coupon %": 3,
        "YTM %": 3,
        "Clean Price": 4,
        "Spot DCF": 4,
        "Mac Duration": 3,
        "Mod Duration": 3,
        "Convexity": 2,
        "DV01 / $100": 4,
        "Carry %": 3,
        "Rolldown %": 3,
        "TR −50bp": 2,
        "TR Unchanged": 2,
        "TR +50bp": 2
    })
)


# ============================================================
# 6. DECOMPOSITION CHARTS
# ============================================================

for cusip in CUSIPS:

    d = decomp[
        decomp["CUSIP"] == cusip
    ].copy()

    info = base[
        base["CUSIP"] == cusip
    ].iloc[0]

    labels = [
        f"{x:+.0f} bp"
        for x in d["Yield_Shock_bp"]
    ]

    fig = go.Figure()

    # Carry
    fig.add_trace(
        go.Bar(
            x=labels,
            y=d["Carry_pct"],
            name="Carry",
            text=d["Carry_pct"].map(
                lambda x: f"{x:.2f}%"
            ),
            textposition="inside"
        )
    )

    # Rolldown
    fig.add_trace(
        go.Bar(
            x=labels,
            y=d["Rolldown_pct"],
            name="Rolldown",
            text=d["Rolldown_pct"].map(
                lambda x: f"{x:+.2f}%"
            ),
            textposition="inside"
        )
    )

    # Duration
    fig.add_trace(
        go.Bar(
            x=labels,
            y=d["Duration_Impact_pct"],
            name="Duration Impact",
            text=d["Duration_Impact_pct"].map(
                lambda x: f"{x:+.2f}%"
            ),
            textposition="inside"
        )
    )

    # Convexity
    fig.add_trace(
        go.Bar(
            x=labels,
            y=d["Convexity_Impact_pct"],
            name="Convexity Impact",
            text=d["Convexity_Impact_pct"].map(
                lambda x: f"{x:+.2f}%"
            ),
            textposition="inside"
        )
    )

    # Total
    fig.add_trace(
        go.Scatter(
            x=labels,
            y=d["Total_Return_pct"],
            mode="lines+markers+text",
            name="Total Return",
            text=d["Total_Return_pct"].map(
                lambda x: f"{x:.2f}%"
            ),
            textposition="top center",
            line=dict(width=3),
            marker=dict(size=10)
        )
    )

    fig.add_hline(
        y=0,
        line_dash="dash",
        line_width=1
    )

    fig.update_layout(

        title=(
            f"{cusip} — "
            f"{info['Coupon_pct']:.3f}% U.S. Treasury Note "
            f"due {info['Maturity'].strftime('%d %b %Y')}"
            f"<br><sup>"
            f"YTM {info['Market_YTM_pct']:.3f}% | "
            f"Mac Duration {info['Macaulay_Duration']:.2f} | "
            f"Mod Duration {info['Modified_Duration']:.2f} | "
            f"Convexity {info['Convexity']:.2f}"
            f"</sup>"
        ),

        xaxis_title="Parallel Treasury Curve Move",

        yaxis_title="1-Year Return Contribution (%)",

        barmode="relative",

        template="plotly_white",

        width=1050,
        height=600,

        legend_title=""
    )

    fig.update_yaxes(
        ticksuffix="%"
    )

    fig.show()


# ============================================================
# 7. FOOTBALL FIELD
# ============================================================

football = (
    decomp.groupby("CUSIP")
    .agg(
        Low_Return=("Total_Return_pct", "min"),
        High_Return=("Total_Return_pct", "max")
    )
    .reset_index()
)


football = football.merge(
    base[[
        "CUSIP",
        "Total_Return_pct"
    ]],
    on="CUSIP"
)


football = football.rename(
    columns={
        "Total_Return_pct":
            "Base_Return"
    }
)


fig = go.Figure()


for _, row in football.iterrows():

    fig.add_trace(
        go.Scatter(

            x=[
                row["Low_Return"],
                row["High_Return"]
            ],

            y=[
                row["CUSIP"],
                row["CUSIP"]
            ],

            mode="lines",

            line=dict(
                width=12
            ),

            showlegend=False,

            hovertemplate=(
                f"{row['CUSIP']}<br>"
                "Return: %{x:.2f}%"
                "<extra></extra>"
            )
        )
    )


fig.add_trace(
    go.Scatter(

        x=football["Base_Return"],

        y=football["CUSIP"],

        mode="markers+text",

        marker=dict(
            size=15,
            symbol="diamond"
        ),

        text=football[
            "Base_Return"
        ].map(
            lambda x: f"{x:.2f}%"
        ),

        textposition="top center",

        name="Unchanged Curve"
    )
)


fig.add_vline(
    x=0,
    line_dash="dash",
    line_width=1
)


fig.update_layout(

    title=(
        "Treasury 1-Year Return Football Field — "
        "±50 bp Parallel Curve Scenario"
    ),

    xaxis_title="1-Year Total Return (%)",

    yaxis_title="",

    template="plotly_white",

    width=1050,

    height=500,

    legend_title=""
)


fig.update_xaxes(
    ticksuffix="%"
)


fig.show()
```
