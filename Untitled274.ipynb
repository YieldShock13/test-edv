```python
import pandas as pd
import requests
import xml.etree.ElementTree as ET
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "colab"

# ============================================================
# US TREASURY DIRECT DATA
# Daily Par Yield Curve: 1Y and 10Y
# ============================================================

frames = []

for year in range(2018, 2027):

    url = (
        "https://home.treasury.gov/resource-center/data-chart-center/"
        "interest-rates/pages/xml"
        f"?data=daily_treasury_yield_curve&field_tdr_date_value={year}"
    )

    r = requests.get(url, timeout=30)
    r.raise_for_status()

    root = ET.fromstring(r.content)

    rows = []

    # Treasury XML uses Atom + Microsoft data namespaces
    ns = {
        "atom": "http://www.w3.org/2005/Atom",
        "m": "http://schemas.microsoft.com/ado/2007/08/dataservices/metadata",
        "d": "http://schemas.microsoft.com/ado/2007/08/dataservices"
    }

    for entry in root.findall("atom:entry", ns):

        props = entry.find("atom:content/m:properties", ns)

        if props is None:
            continue

        date = props.find("d:NEW_DATE", ns)
        y1   = props.find("d:BC_1YEAR", ns)
        y10  = props.find("d:BC_10YEAR", ns)

        rows.append({
            "Date": date.text if date is not None else None,
            "1Y": y1.text if y1 is not None else None,
            "10Y": y10.text if y10 is not None else None
        })

    frames.append(pd.DataFrame(rows))


# Combine years
ust = pd.concat(frames, ignore_index=True)

# Clean
ust["Date"] = pd.to_datetime(ust["Date"], errors="coerce")
ust["1Y"] = pd.to_numeric(ust["1Y"], errors="coerce")
ust["10Y"] = pd.to_numeric(ust["10Y"], errors="coerce")

ust = (
    ust[
        (ust["Date"] >= "2018-09-01")
    ]
    .dropna(subset=["Date", "1Y", "10Y"])
    .sort_values("Date")
    .drop_duplicates("Date")
    .reset_index(drop=True)
)

# 10Y minus 1Y in basis points
ust["10Y_1Y_Spread_bp"] = (ust["10Y"] - ust["1Y"]) * 100

last_date = ust["Date"].max()

print(ust.tail(10))
print(f"\nFinal Treasury observation: {last_date:%d %B %Y}")


# ============================================================
# PLOT
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=ust["Date"],
        y=ust["10Y_1Y_Spread_bp"],
        mode="lines",
        name="10Y − 1Y",
        line=dict(width=2),
        hovertemplate=(
            "%{x|%d %b %Y}"
            "<br>10Y − 1Y: %{y:.1f} bp"
            "<extra></extra>"
        )
    )
)

fig.add_hline(
    y=0,
    line_dash="dash",
    line_width=1
)

fig.update_layout(

    title=dict(
        text=(
            "<b>U.S. Treasury 10Y–1Y Yield Spread</b>"
            f"<br><sup>September 2018 – {last_date:%d %B %Y}</sup>"
        ),
        x=0.05
    ),

    xaxis=dict(
        title="Date",
        range=[
            ust["Date"].min(),
            last_date + pd.Timedelta(days=20)
        ]
    ),

    yaxis=dict(
        title="10Y − 1Y Spread (bp)"
    ),

    hovermode="x unified",
    template="plotly_white",

    width=1200,
    height=600,

    margin=dict(
        l=90,
        r=60,
        t=110,
        b=70
    )
)

fig.show(renderer="colab")
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# ============================================================
# PERSISTENCE: AR(1) PHI ACROSS MULTIPLE LOOKBACK WINDOWS
# ============================================================

series_map = {
    "1Y Treasury − BGCR": "1Y_minus_BGCR_bp",
    "10Y Treasury − BGCR": "10Y_minus_BGCR_bp"
}

windows = {
    "3M": 60,
    "6M": 120,
    "1Y": 252,
    "2Y": 504,
    "3Y": 756
}

results = []

for name, col in series_map.items():

    s = spreads[["Date", col]].dropna().sort_values("Date")

    for label, n in windows.items():

        sample = s.tail(n).copy()

        y = sample[col].iloc[1:].values
        x = sample[col].iloc[:-1].values

        X = sm.add_constant(x)
        model = sm.OLS(y, X).fit()

        phi = model.params[1]

        # AR(1) implied half-life
        if 0 < phi < 1:
            half_life = -np.log(2) / np.log(phi)
        else:
            half_life = np.nan

        results.append({
            "Spread": name,
            "Lookback": label,
            "N": len(sample),
            "Phi": phi,
            "Half-life (days)": half_life
        })

persistence = pd.DataFrame(results)

print(
    persistence.to_string(
        index=False,
        formatters={
            "Phi": "{:.4f}".format,
            "Half-life (days)": lambda x:
                f"{x:.1f}" if pd.notna(x) else "N/A"
        }
    )
)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller, kpss
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio
import warnings

warnings.filterwarnings("ignore")
pio.renderers.default = "colab"

# ============================================================
# SETTINGS
# ============================================================

series_map = {
    "1Y Treasury − BGCR": "1Y_minus_BGCR_bp",
    "10Y Treasury − BGCR": "10Y_minus_BGCR_bp"
}

windows = {
    "3M": 60,
    "6M": 120,
    "1Y": 252,
    "2Y": 504,
    "3Y": 756
}

# ============================================================
# HURST
# ============================================================

def hurst_exponent(series):
    x = np.asarray(series, dtype=float)
    x = x[np.isfinite(x)]

    max_lag = min(100, len(x) // 3)

    if max_lag < 10:
        return np.nan

    lags = np.arange(2, max_lag)

    tau = np.array([
        np.std(x[lag:] - x[:-lag])
        for lag in lags
    ])

    valid = (tau > 0) & np.isfinite(tau)

    return np.polyfit(
        np.log(lags[valid]),
        np.log(tau[valid]),
        1
    )[0]


# ============================================================
# CALCULATE
# ============================================================

results = []

for spread_name, col in series_map.items():

    s = spreads[["Date", col]].dropna().sort_values("Date")

    for lookback, n in windows.items():

        sample = s.tail(n)[col].values

        # AR(1)
        y = sample[1:]
        x = sample[:-1]

        model = sm.OLS(
            y,
            sm.add_constant(x)
        ).fit()

        phi = model.params[1]

        # Half-life
        half_life = (
            -np.log(2) / np.log(phi)
            if 0 < phi < 1
            else np.nan
        )

        # ADF
        adf_p = adfuller(
            sample,
            autolag="AIC"
        )[1]

        # KPSS
        kpss_p = kpss(
            sample,
            regression="c",
            nlags="auto"
        )[1]

        # Hurst
        hurst = hurst_exponent(sample)

        results.append({
            "Spread": spread_name,
            "Lookback": lookback,
            "N": len(sample),
            "AR(1) φ": phi,
            "Half-life": half_life,
            "ADF p": adf_p,
            "KPSS p": kpss_p,
            "Hurst": hurst
        })


diag = pd.DataFrame(results)

order = ["3M", "6M", "1Y", "2Y", "3Y"]

diag["Lookback"] = pd.Categorical(
    diag["Lookback"],
    categories=order,
    ordered=True
)

diag = diag.sort_values(
    ["Spread", "Lookback"]
).reset_index(drop=True)


# ============================================================
# PRESENTATION CHART
# ============================================================

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        "AR(1) Persistence",
        "Implied Mean-Reversion Half-Life"
    )
)

for spread_name in series_map.keys():

    d = diag[diag["Spread"] == spread_name]

    fig.add_trace(
        go.Scatter(
            x=d["Lookback"],
            y=d["AR(1) φ"],
            mode="lines+markers",
            name=spread_name,
            line=dict(width=3),
            marker=dict(size=9),
            hovertemplate=(
                "Lookback: %{x}<br>"
                "φ: %{y:.4f}"
                "<extra></extra>"
            )
        ),
        row=1,
        col=1
    )

    fig.add_trace(
        go.Scatter(
            x=d["Lookback"],
            y=d["Half-life"],
            mode="lines+markers",
            name=spread_name,
            showlegend=False,
            line=dict(width=3),
            marker=dict(size=9),
            hovertemplate=(
                "Lookback: %{x}<br>"
                "Half-life: %{y:.1f} days"
                "<extra></extra>"
            )
        ),
        row=1,
        col=2
    )


fig.update_yaxes(
    title_text="AR(1) φ",
    range=[0.80, 1.01],
    row=1,
    col=1
)

fig.update_yaxes(
    title_text="Trading Days",
    rangemode="tozero",
    row=1,
    col=2
)

fig.update_xaxes(
    title_text="Lookback Window"
)

fig.update_layout(
    title=dict(
        text=(
            "<b>Treasury–BGCR Spread Persistence</b>"
            "<br><sup>Persistence and implied decay across multiple estimation windows</sup>"
        ),
        x=0.5,
        xanchor="center"
    ),

    template="plotly_white",

    width=1200,
    height=560,

    legend=dict(
        orientation="h",
        x=0.5,
        xanchor="center",
        y=1.06
    ),

    margin=dict(
        l=80,
        r=50,
        t=130,
        b=70
    )
)

fig.show(renderer="colab")


# ============================================================
# CLEAN DIAGNOSTIC TABLE
# ============================================================

table = diag.copy()

table["AR(1) φ"] = table["AR(1) φ"].map(lambda x: f"{x:.4f}")
table["Half-life"] = table["Half-life"].map(lambda x: f"{x:.1f}")
table["ADF p"] = table["ADF p"].map(lambda x: f"{x:.4f}")
table["KPSS p"] = table["KPSS p"].map(lambda x: f"{x:.4f}")
table["Hurst"] = table["Hurst"].map(lambda x: f"{x:.3f}")

table = table.rename(columns={
    "Half-life": "Half-life (days)"
})

display(
    table.style
    .hide(axis="index")
    .set_caption("Persistence Diagnostics")
    .format(precision=4)
)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.io as pio
import warnings

warnings.filterwarnings("ignore")
pio.renderers.default = "colab"

# ============================================================
# SETTINGS
# ============================================================

ROLL = 252

SERIES = {
    "1Y Treasury − BGCR": "1Y_minus_BGCR_bp",
    "10Y Treasury − BGCR": "10Y_minus_BGCR_bp"
}

WINDOWS = {
    "3M": 60,
    "6M": 120,
    "1Y": 252,
    "2Y": 504,
    "3Y": 756
}


# ============================================================
# FUNCTIONS
# ============================================================

def get_phi(x):
    x = np.asarray(x, dtype=float)

    y = x[1:]
    xlag = x[:-1]

    X = sm.add_constant(xlag)
    model = sm.OLS(y, X).fit()

    return float(model.params[1])


def get_half_life(phi):
    if 0 < phi < 1:
        return -np.log(2) / np.log(phi)
    return np.nan


def get_adf_p(x):
    try:
        return float(
            adfuller(
                np.asarray(x, dtype=float),
                autolag="AIC"
            )[1]
        )
    except Exception:
        return np.nan


# ============================================================
# STORAGE FOR SUMMARY TABLE
# ============================================================

summary_rows = []


# ============================================================
# LOOP THROUGH 1Y AND 10Y SPREADS
# ============================================================

for spread_name, column in SERIES.items():

    d = (
        spreads[["Date", column]]
        .dropna()
        .sort_values("Date")
        .reset_index(drop=True)
        .copy()
    )

    # --------------------------------------------------------
    # Rolling mean
    # --------------------------------------------------------

    d["Rolling_Mean"] = (
        d[column]
        .rolling(ROLL)
        .mean()
    )


    # --------------------------------------------------------
    # Rolling AR(1) phi
    # --------------------------------------------------------

    d["Rolling_Phi"] = (
        d[column]
        .rolling(ROLL)
        .apply(get_phi, raw=True)
    )


    # --------------------------------------------------------
    # Rolling ADF p-value
    # --------------------------------------------------------

    d["Rolling_ADF_p"] = (
        d[column]
        .rolling(ROLL)
        .apply(get_adf_p, raw=True)
    )


    # ========================================================
    # FIGURE
    # ========================================================

    fig = make_subplots(
        rows=3,
        cols=1,

        shared_xaxes=True,

        row_heights=[
            0.46,
            0.27,
            0.27
        ],

        vertical_spacing=0.10,

        subplot_titles=[
            "Spread and 252-Day Rolling Mean",
            "252-Day Rolling AR(1) Persistence",
            "252-Day Rolling ADF Test"
        ]
    )


    # ========================================================
    # PANEL 1
    # ========================================================

    fig.add_trace(
        go.Scatter(
            x=d["Date"],
            y=d[column],

            mode="lines",

            name="Spread",

            line=dict(
                width=1.4
            ),

            hovertemplate=(
                "%{x|%d %b %Y}<br>"
                "Spread: %{y:.1f} bp"
                "<extra></extra>"
            )
        ),

        row=1,
        col=1
    )


    fig.add_trace(
        go.Scatter(
            x=d["Date"],
            y=d["Rolling_Mean"],

            mode="lines",

            name="252D mean",

            line=dict(
                width=2.5
            ),

            hovertemplate=(
                "%{x|%d %b %Y}<br>"
                "252D mean: %{y:.1f} bp"
                "<extra></extra>"
            )
        ),

        row=1,
        col=1
    )


    fig.add_hline(
        y=0,
        line_dash="dash",
        line_width=1,
        row=1,
        col=1
    )


    # ========================================================
    # PANEL 2
    # ========================================================

    fig.add_trace(
        go.Scatter(
            x=d["Date"],
            y=d["Rolling_Phi"],

            mode="lines",

            name="AR(1) φ",

            showlegend=False,

            line=dict(
                width=2
            ),

            hovertemplate=(
                "%{x|%d %b %Y}<br>"
                "AR(1) φ: %{y:.3f}"
                "<extra></extra>"
            )
        ),

        row=2,
        col=1
    )


    fig.add_hline(
        y=1.0,

        line_dash="dash",
        line_width=1,

        row=2,
        col=1
    )


    # ========================================================
    # PANEL 3
    # ========================================================

    fig.add_trace(
        go.Scatter(
            x=d["Date"],
            y=d["Rolling_ADF_p"],

            mode="lines",

            name="ADF p-value",

            showlegend=False,

            line=dict(
                width=2
            ),

            hovertemplate=(
                "%{x|%d %b %Y}<br>"
                "ADF p-value: %{y:.3f}"
                "<extra></extra>"
            )
        ),

        row=3,
        col=1
    )


    fig.add_hline(
        y=0.05,

        line_dash="dash",
        line_width=1,

        row=3,
        col=1
    )


    # ========================================================
    # AXES
    # ========================================================

    fig.update_yaxes(
        title_text="Spread (bp)",
        row=1,
        col=1
    )

    fig.update_yaxes(
        title_text="AR(1) φ",
        row=2,
        col=1
    )

    fig.update_yaxes(
        title_text="ADF p-value",
        range=[0, 1],
        row=3,
        col=1
    )

    fig.update_xaxes(
        title_text="Date",
        row=3,
        col=1
    )


    # ========================================================
    # LAYOUT
    # ========================================================

    fig.update_layout(

        title=dict(
            text=f"<b>{spread_name} — Persistence Analysis</b>",
            x=0.5,
            xanchor="center"
        ),

        font=dict(
            family="Arial",
            size=13
        ),

        template="plotly_white",

        width=1250,
        height=950,

        hovermode="x unified",

        legend=dict(
            orientation="h",
            x=0.5,
            xanchor="center",
            y=1.02,
            yanchor="bottom"
        ),

        margin=dict(
            l=100,
            r=60,
            t=130,
            b=70
        )
    )


    # Format Plotly's own subplot titles only.
    # No coordinate manipulation.
    for annotation in fig.layout.annotations:

        annotation.font = dict(
            family="Arial",
            size=14
        )


    fig.show(renderer="colab")


    # ========================================================
    # MULTI-WINDOW SUMMARY
    # ========================================================

    for window_name, n in WINDOWS.items():

        sample = d[column].tail(n).values

        phi = get_phi(sample)
        hl = get_half_life(phi)
        adf_p = get_adf_p(sample)

        summary_rows.append({
            "Spread": spread_name,
            "Window": window_name,
            "N": len(sample),
            "AR(1) φ": phi,
            "Half-life (days)": hl,
            "ADF p-value": adf_p
        })


# ============================================================
# SUMMARY TABLE
# ============================================================

summary = pd.DataFrame(summary_rows)

window_order = [
    "3M",
    "6M",
    "1Y",
    "2Y",
    "3Y"
]

summary["Window"] = pd.Categorical(
    summary["Window"],
    categories=window_order,
    ordered=True
)

summary = (
    summary
    .sort_values(["Spread", "Window"])
    .reset_index(drop=True)
)


# Presentation copy
table = summary.copy()

table["AR(1) φ"] = table["AR(1) φ"].map(
    lambda x: f"{x:.4f}"
)

table["Half-life (days)"] = table[
    "Half-life (days)"
].map(
    lambda x: f"{x:.1f}"
    if np.isfinite(x)
    else "—"
)

table["ADF p-value"] = table[
    "ADF p-value"
].map(
    lambda x: f"{x:.4f}"
)


display(
    table.style
    .hide(axis="index")
    .set_caption(
        "Treasury–BGCR Persistence Summary"
    )
    .set_table_styles([

        {
            "selector": "caption",
            "props": [
                ("font-family", "Arial"),
                ("font-size", "18px"),
                ("font-weight", "bold"),
                ("text-align", "left"),
                ("padding-bottom", "12px")
            ]
        },

        {
            "selector": "th",
            "props": [
                ("font-family", "Arial"),
                ("font-size", "12px"),
                ("text-align", "center"),
                ("padding", "8px")
            ]
        },

        {
            "selector": "td",
            "props": [
                ("font-family", "Arial"),
                ("font-size", "12px"),
                ("text-align", "center"),
                ("padding", "8px")
            ]
        }
    ])
)
```
